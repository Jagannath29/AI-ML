## Semantic Similarity
This section will be a short recap of the chapter: Word Similarity and vector semantics discussed in the earlier module.

Semantics is the study of the meaning of words. Words can acquire different meanings based on the context of use, which gave us the concept of **word senses**. Though the word remains the same, the sense might change if the context changes.   
The notion of word similarity can be expressed in two parts: word relations and semantic similarity. Understanding each of these parts can help gain the overall meaning of a phrase or a sentence.

Two senses can have a **synonymous relation** that expresses identical meanings like car/automobile, cellphone/mobile, vomit/throwup, or anonymous relation that express opposite meanings. Similarly, These words can have all the word relations like homonymy, meronymy, polysemy, etc., as we discussed earlier.   

Moving from word relations to similarity, **semantic similarity** deals with the closeness of words with a measurement metric. Two words with entirely different definitions (gloss) or meanings can be somehow related or similar.

For example, car and road are not synonymous words, but they are similarly based on the context of use. Car and gasoline are never used in place of each other, but it is quite likely they appear with each other a lot. One of such relatedness can be expressed using **semantic field**. A semantic field is a set of words of a particular semantic domain. For example, A semantic field of coffee will have (tea, cup, sugar, coffee, brown), house (door, kitchen, bedroom, family) words where these words are considered semantically related.


### Computation of word similarity
There are two broad approaches to calculating word similarity metrics:



#### Thesaurus based approach
This approach makes the use of lexical resources like thesaurus and wordnets. Words are categorized similarly based on the similarity of gloss, senses, concepts, and the closeness of two words in the hypernym hierarchy.
We stuidied path based similarity calculations like:
- **Pathlength**   
  $\text{pathlen}(c_1, c_2)= 1 +\text{number of edges}\\ \text{ in the shortest path in hypernym graph between } c_1 \text{ and } c_2$   

- **Sim Path**   
$\text{sim}_{path}(c_1,c_2)=\frac{1}{\text{pathlen}(c_1,c_2)}$

- **WordSim**   
$\text{wordsim}(w_1,w_2)=\text{max} \big\{  \text{sim}(c_1,c_2): c_1 \epsilon \text{ senses } (w_1), c_2 \epsilon \text{ senses } (w_2) \big\}$

#### Distributional approach or Vector Semantics
The calculations in the thesaurus based approach are only as good as the resources available and make no use of the contextual information. Unlike that, the distributional approach solely takes contextual information into account. The distributional approach's fundamental idea is that the two words are similar if they have similar contexts.   
Lets take an example <sub>[[1]](#Book)</sub> to understand the distributional:

Suppose you don't know the meaning of the word "**Ongchoi**," but you see it being used in the following sentences.
- Ongchoi is delicious sauteed with garlic
- Ongchoi is superb over rice
- ...Ongchoi leaves with salty sauces...

Although we don't know the actual meaning of the word "Ongchoi," since it is frequently used with the words like _delicious, rice, garlic_, we can deduce that it is some form of an edible food/thing. It can be a leafy green vegetable similar to spinach as one would use spinach like the way Ongchoi is being used here.

We can understand the semantics even by counting words (vectors) in the context of Ongchoi. The words like _delicious, rice and garlic_ will score high frequency, which helps us discover the similarity of these words with the word "Ongchoi."

## Vector Semantics
Vector Semantics precisely captures the same distributional and vector representation intuition. It takes the context of words or only neighboring words into account and represents them in the form of a vector to understand the semantics.    
Ideally, this vector representation is supposed to everything about the word's semantics, including the meaning and the context.

### Word Representations
The sole idea of vector semantics is to represent any word in a multidimensional semantic space using vectors for each word.


#### 1. One hot Representations
One hot representation is one of the simplest methods of representing words in a corpus.

Let us consider a corpus contains $|V|=10000$ vocabulary of words (unique words) where   

$V=[a, aa, b,...zone, <UNK>]$

_$<UNK>$ represents unknown values_

In this vocabulary, let the word ```Man``` be in the 120$^{th}$ index and the word ```Woman``` be in the 600$^{th}$ index. The one-hot representation for the word ```Man``` will be a vector of dimension ```10000 X 1```, with all the values being 0 other than at 120$^{th}$ index, which will be 1.   
i.e.   

Man(120)=$\begin{bmatrix}0
\\0\\0\\ \vdots \\1\\\vdots\\0\\0\\0
\end{bmatrix}$
Woman(600)=$\begin{bmatrix}0
\\0\\0\\0\\ \vdots \\1\\\vdots\\0\\0
\end{bmatrix}$
Car(1500)=$\begin{bmatrix}0
\\0\\0\\0\\0\\ \vdots \\1\\\vdots\\0
\end{bmatrix}$
King(10)=$\begin{bmatrix}0
\\0\\1\\0\\0\\ \vdots \\0\\\vdots\\0
\end{bmatrix}$
Queen(9000)=$\begin{bmatrix}0
\\0\\0\\0\\0\\ 0\\\vdots\\1\\0
\end{bmatrix}$


Like `Man,` the value of 600$^{th}$ index will be 1, and the rest of all the values will be 0 for the word `Woman.` Each word in the vocabulary is represented in the same way as man and woman.

This representation treats one word as a completely independent unit than another word. The value at 120$^{th}$ index being 1 has nothing to do with women's representation. It is challenging for any algorithm to know that the word `Man` might be more similar to `Woman` than `Car` based on this representation.


#### 2. Featured Representations or Embeddings
Instead of one-hot representations, we can represent any words using featured representations where each word learns values for a set of pre-defined features.
The features can be **a noun, or an action or a verb or something else**.

For example, Similar to one-hot representation, let us take a corpus with $|V|=10000$ vocabulary of words  where,   

$V=[a, aa, b,...zone, <UNK>]$.

Now in featured representation, we will have a set of features **{gender, fruit, size, machine, $\cdots$}** which will be the basis for assigning values to the words—considering $300$ such features.

||Man(120)|woman(600)|Car(1500)|King(10)|Queen(9000)|
|---|---|---|---|---|---|
|Gender|-1|1|0.00|0.00|0.02|
|Food|0.05|0.04|0.00|0.02|-0.01|
|Royal|0.01|0.02|0.03|0.93|0.95|
|Machines|0.03|0.02|0.85|0.01|-0.02|
|$\vdots$|$\vdots$|$\vdots$|$\vdots$|$\vdots$|$\vdots$|
|Alive|1|1|-1|1|1|
|Costs|0.95|0.95|0.2|0.98|0.96|

<br>
<center><figcaption>Table 1: Featurised Word Representations</figcaption></center>

The table above contains different features represented in rows and different words represented in columns. Each cell corresponds to the value that represents how much a word is associated with a feature. The high value represents the high association and vice versa.

Taking ```Man(120)``` column, we will have ```300 X 1``` dimensional vector containing values for 300 different features for the word `Man`. It is represented by **$E_{120}=[-1, 0.05, 0.01, 0.03, \cdots ]$**. Similarly,  King(10) will be represented by **$E_{10}=[0.00, 0.02, 0.93, 0.01, \cdots ] $** and so on for all words.

Such representation of words based on a set of features is known as **Word Embeddings.** It is called  "Embeddings" because each word in the vocabulary is embedded in a 300-dimensional vector space, and the similar words are embedded closer to each other.

## Word Embeddings
**The representation of a word in some multidimensional vector space is known as word embeddings.**   
The vector space dimension is defined based on the number of features taken to embed each word.

### Why word embeddings?
Unlike one-hot representations, word embeddings are much denser and can represent in a lower dimension. If the words' vocabulary is `10000`, the dimension of representation of each word in one-hot representation is `10000 X 1`. In contrast, the representation of each word in word embeddings depends on the number of features used.    
The values in one-hot representation were sparse. It contained mostly `0` values apart from single `1.` Still, in word embeddings, each value corresponds to the word's association with features and rarely contains `0`.

An increase in dimension of representation of words in one-hot representation only increases computation cost and sparsity, where an increase in dimension in word embeddings increases the potential to represent each word and understand the semantic nature accurately.   
**Ideally, word embeddings with enough feature dimension and carefully calculated values for each word can capture the complete semantics of the word.**



### Types of word embedding:
On the basis of how embedding values are calculated for each words, we divide word embeddings into count based models and predictive based models.

#### 1. Count Based Models
These models count the number of times the words appear in a particular context.
##### **1.1. Documents as contexts**
Calculations of word embeddings with documents as contexts are based on the term-document matrix. In this matrix, each row represents words in the vocabulary for which we need to find the embeddings and columns represents a document from the collection of documents, as shown in the table below:

||Machine Learning for all| Hands-on deep learning|binary systems|
|---|---|---|---|
|machine|1900|1600|900|
|deeplearning|1100|2304|1020|
|binary|503|905|2600|

<br>
<center>
<figcaption>Table 2: Term-Document Matrix</figcaption>
</center>

Each cell in this matrix represents the number of times a word (defined by row) appears in a document (defined by column). According to the table above, the word `machine` appears $1900$ times in the document `Machine Learning for all.`

The vector representation of the document `Machine learning for all` is $[1900, 1100, 503]$, represented as a point in a three-dimensional vector space. Similarly, we can vectorically represent all the documents as shown in <sub>[[figure 4]](#termdocumentmatrix)</sub>.

**Term-Document Matrix is used in Information Retrieval and document classification. Similar documents have similar vectors, as they have similar words.**

##### **1.2. Words as contexts**
We used documents as context earlier, but now we will be using words as contexts (smaller context than documents). The calculation of embeddings with words as contexts is based on the term-term matrix, also known as the word-word matrix.

Each cell will represent the number of times the row (target word), **co-occur** with the column (context word). So the word-word matrix represents a co-occurrence matrix. Rather than taking individual words, we will be looking into a window of words. For example: **With a window of 2, the value of each cell represents the number of times the column word occurs in such a** $\pm2$ **word window around the row word.**

Unlike the term-document matrix, the word vector will be a row vector in the word-word matrix rather than a column vector. So each of the vectors will be of length |V|, the size of the vocabulary.

|---|---|---|---|---|---|---|---|---|---|
|cherry|0|$\cdots$|2|8|9|442|25|$\cdots$|3|
|strawberry|0|$\cdots$|0|0|1|60|19|$\cdots$|1|
|digital|0|$\cdots$|1670|1683|85|5|4|$\cdots$|2|
|information|0|$\cdots$|3325|3982|378|5|13|$\cdots$|1|

<br>
<center><figcaption>Table 3: Word-Word Matrix</figcaption></center>
According to table 3, the value of the cell [cherry, computer] is $2$ indicating that the word `cherry` and `computer` occurs twice in a window of $\pm2$. Similar are the other cell values.    

**We will be talking the row vectors as embeddings, so the embeddings for the word** `cherry` is $[0, \cdots, 2, 8, 9, 442, 25, \cdots]$. The embedding vector size will be equal to the size of the vocabulary, |V|.

#### Predictive based models
Unlike count-based models, where vectors' values were based upon either co-occurrences or appearances in documents, in predictive-based models, the embeddings' values are learned using an algorithm.

##### **Embedding Matrix Calculations (Neural Networks)**
Neural networks primarily focus on language modeling, more explicitly predicting the sequence of words, but the input to the neural network is not the word itself; instead, it is word embeddings. So, we use an **embedding matrix** that converts words into their corresponding embeddings in neural networks.

**Embedding matrix contains embeddings weights that are optimized during the training of the neural network.**

 According to <sub>[[figure 4]](#predictivemodels)</sub>, in a neural network (predictive model), an embedding matrix of dimension `d X |V|` is multiplied with the One-hot Representation of a given word of size `1 X |V|` to output word embeddings of size `1 X d' for that word.   
i.e.   
$Word Embeddings = One hot Representation * (Embedding Matrix)^T$

#####**Word2vec**
Similar to neural networks, word2vec is an efficient method to create word embeddings. The fundamental notion of word2vec is to **represent words in dense low dimensional space so that similar words have similar vectors and are closer in the vector space.**

Word2vec uses a deeper neural network model to learn word associations and generate embeddings for each word. Word2vec **involves two methods: Skip Gram and Continuous Bag of Words (CBOW) embeddings.**

The training of a word2vec model involves:
- Moving through the training corpus using a sliding window where each word is a prediction problem
- The objective of the model is to predict the current word based on the previous words
- Based on the predicted word, we determine whether to adjust the current word vector or not. The optimal word vector is then selected as the embedding vector.

## Key Takeaways
- The notion of word similarity can be expressed in two parts: word relations and semantic similarity
- Semantic similarity deals with the closeness of words with a metric of measurement.
- Thesaurus based  approach makes the use of lexical resources like thesaurus and wordnets but is limited by the availability of resources
- The fundamental idea behind the distributional approach is that two words are similar if they have similar contexts.
- Vector semantics represents any word in a multidimensional semantic space using vectors for each word, also known as word embeddings or featured representations.
- Count-based models use raw counts and co-occurrence to generate word embeddings taking documents or windows of words as contexts.
- Predictive models focus on generating learned embeddings using embedding matrix in neural networks or word2vec.
