## Intuition and Motivation

In Feed Forward Neural Network Language Model, the inputs of the neural network were the context words from the past and the softmax output predicted the next word for the sequence.

The general architecture of the feed forward neural net language model is shown below. It consists of a input layer, projection layer, hidden layer and the final output layer. The task of the model is to learn embedding matrix.

#### **Limitation of the feed forward NNLM**
The major disadvantage of NNLM architecture is that the projection layers are dense(as each word vectors are concatenated) so the computation between projection and the hidden layer becomes complex.

#### **What we do?**
We try to simplify the model architecture by removing the **extra hidden layers** and averaging the word vectors instead of concatenation. We also give context for the model from the future(in addition to past context words).

## CBOW model

### Introduction

In CBOW model, we train the neural network to learn the target word, given the context words. The context words are the words from the past and from the future timesteps.


<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1KtPP3gM9iU3cl24acts5WO-VrBXY_V2z"  > -->
<img src="https://i.postimg.cc/GmQVKqMZ/image.png"  >
<figcaption align="center"> </figcaption>

In the above example, we have a window size of $\pm$ 2. It looks at two timestep back in the past of the target word $w_n$ which are $w_{n-1}$ and $w_{n-2}$ and two timestep into the future i.e $w_{n+1}$ and $w_{n+2}$. So these four context words are the inputs to the neural network. The models tries to learn the target word $w_n$.

Observe that the **extra hidden layer is removed** in the model and **the projection layer is shared**.


<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1T41gFP2Wu-8ZJJxOcpd0TZqq-RiMmir9" > -->
<img src="https://i.postimg.cc/KvXYh8Zm/cbow-model.png" >
<figcaption align="center">Figure 2: CBOW model </figcaption>


The objective of this model is to learn the embedding matrix (which is the weight matrix) between input layer and the projection layer.

In the next section, let's discuss how this model learns those embeddings.

The more general architecture of the CBOW model for a window of size $\pm$ 1 is given below. The model has a input layer, a projection layer and the output layer.


<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1xOVho9pS-V-2FFFlqiDQ1OGv2Mps5782" > -->
<img src="https://i.postimg.cc/mZP0N3Wr/image.png" >
<figcaption align="center">Figure 5: CBOW model </figcaption>




#### **Input Layer:**
In input layer we pass the one hot representation of the context words. Each of the context words have a dimension of Vocabulary size $V$.

#### **Projection layer:**
The projection layer has the size of $N$ which means it has $N$ nodes. The weight matrix $W_{V*N}$ is the learnable parameter between the input layers and the projection layer. The output of the projection layer is the average of the word vectors.

i.e
$$
h = \frac{1}{C}W^{T}(\text{one hot representation}(w_{n-1}) + \text{one hot representation}(w_{n+1}))
$$

$$
h = \frac{1}{2}(v_{n-1} + v_n)^T
$$

Here, $v_{n-1}$ and $v_{n}$ are the word vectors of the words $w_{n-1}$ and $w_{n}$ respectively and $C=2$ is the context length.

<!-- **Note**: We use the terms projection layer and hidden layer interchangebly while describing the model. This projection/hidden layer should not be confused with removed "extra hidden layers". -->


#### **Final Layer Output:**
The output layer is the softmax outputs for the target word. So, it also has a dimension of size $V$. The learnable parameters between the hidden layer and the output layer is $W_{N*V}^{'}$.

#### **The Fake task**
We train the model but we do not use it for what it is trained for. Instead, once we optimize the model, our main objective is to get the embedding matrix (or the weight matrix) $W_{V*N}$. The embedding matrix is a look up embeddings for each words in the vocabulary. That is why this learning is often called the fake task.


The learned embedding matrix with the size $V$ rows and $N$ columns is given below. The dimension of the embedding for each words in the vocabulary will be of size $N$.

$$
{W}_{V*N} =
\begin{pmatrix}
w_{1,1} & w_{1, 2} &&...&& w_{1, N}\\
w_{2,1} & w_{2, 2} &&...&& w_{2, N}\\
... & ... &&...&&...\\
w_{k,1} & w_{k, 2} &&...&& w_{k, N}\\
... & ... &&...&&...\\
w_{V,1} & w_{V, 2} &&...&& w_{V, N}
\end{pmatrix}_{V*N}
$$


#### **Lookup embeddings**
The embedding for word $1$ in vocabulary is:
$$
\begin{pmatrix}
w_{1,1} & w_{1, 2} &&...&& w_{1, N}\\
\end{pmatrix}_{1*N}
$$


The embedding for word $k$ in vocabulary is:
$$
\begin{pmatrix}
w_{k,1} & w_{k, 2} &&...&& w_{k, N}\\
\end{pmatrix}_{1*N}
$$


### Comparision with Feed Forward Neural Net Language Model

* In feed forward NNLM only projection matrix was shared whereas in cbow model projection layer is also shared.
* The hidden layers(after the projection layers) are removed in CBOW models which makes the model more simple.
* In CBOW models, the context words are taken also from the future.
* In CBOW model the order of words in history does not influence projection whereas the order of words influences in feed forward NNLM.

## Skip gram model






### Introduction

In skip gram model, we train the neural network to learn the context words, given the target word.


<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1Gces2HvZKvWUC5LntCBMqgYYOVHSAefg"  > -->
<img src="https://i.postimg.cc/FHTMVTz7/image.png"  >
<figcaption align="center"> </figcaption>


For the window size of two, the input is the center target word $w_n$. The outputs of the model are context words $w_{n-1}, w_{n-2}, w_{n+1}, w_{n+2}$.


<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=10TvliZhxXEo5FxSJO8M-U3ZQmlV0QjoG"  > -->

<figcaption align="center">Figure 6: Skip-gram model </figcaption>



As shown in figure, the target word is passed to the projection layer and the final outputs are the context words.

The dataset can be generated in the similar fashion as the CBOW model. In this case the target word is the input to the model and context words are the outputs of the model. The sample training samples for the skip gram model with a window of size 1 is shown below.

The general architecture of skip gram model for a window of size 1 is shown below. One hot representation of the target word is passed as the input to the model and it is passed to the projection layer and the the final layer learns the context words.

<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1_U8fY_jI9G4UvscZ8Rw7hpiRjMCjpnkI" > -->
<img src="https://i.postimg.cc/TwbmcHdx/image.png" >
<figcaption align="center">Figure 8: Skip gram model </figcaption>



#### **Input Layer:**
In input layer we pass the one hot representation of the target word. Thus the dimension of the input layer is $V$.

#### **Projection layer:**
The projection layer has $N$ nodes. The weight matrix $W_{V*N}$ is the learnable parameter between the input layers and the projection layer.


#### **Final Layer Output:**
The output layer is the softmax outputs for each of the context words word. So, each context has a dimension of size $V$. The learnable parameters between the hidden layer and the output layer is $W_{N*V}^{'}$.

#### **The Fake task**
As in CBOW model, our main goal is to learn embedding for each words in the vocabulary. After we get the embedding matrix $W_{V*N}$, we can simply look for the embeddings of each words.

For instance, the embedding for word $k$ in vocabulary of size $V$ is:

$$
\begin{pmatrix}
w_{k,1} & w_{k, 2} &&...&& w_{k, N}\\
\end{pmatrix}_{1*N}
$$

## Key Takeaways
* In Feed Forward NNLM, the major limitation is computation complexity as the projection layer is dense and it has extra hidden layers.
* In the CBOW model, the projection layer is shared and the hidden layer is removed. Also the inputs contain context words from the future.
* In the CBOW model, the projection layer takes the average of word vectors as inputs
* In the skip-gram model, the inputs are target words and the outputs are context words.
* The weight matrix between input and the hidden layer is the embedding matrix.
