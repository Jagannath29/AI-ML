## Feed Forward Neural Networks
Before starting neural language models, we will do a quick recap of neural networks and their components.

### 1. Computational Units
A neural network's fundamental building block is a single computational unit that takes input from a set of real-valued numbers and produces an output. **The output is the weighted sum of the input added with a bias term and activated with an activation function, a.**

Let us consider inputs $x_1,x_2,x_3, \cdots, x_n$ with corresponding weights $w_1, w_2, w_3, \cdots, w_n$ and a bias $b$. Then the weighted sum $z$ can be mathematically represented as:

$$z = \sum _{i=1}^{n} w_i \cdot x_i + b $$

This weighted sum is then passed through **a non-linear activation function**, $a$ to compute the computational unit's overall output.

$$y=a=f(z)$$

### 2. Feed-Forward Neural Network
Feedforward NN is one of the simplest multilayer neural networks in which the **outputs from each unit are passed on to all the higher layer units and never to the lower layers**. <sub>[[Figure:2]](#mlnn)</sub> is an excellent example of multi-layered feedforward neural network with three hidden layers (layer one and layer two), including the output layer. Each output of units is passed on to its higher layer units and finally to the output layer.

### 3. Training a Neural Network
We need a **loss function** that maps the predicted output with the actual output to calculate the error for training a neural network. This error is minimized by optimizing the weights using the **gradients** of the loss function.    
**Overall Combination of all these steps is also known as error backpropagation or reverse differentiation**




## Neural Language Models

The intuition behind language modeling is to predict upcoming words from prior word context. Language models can be primarily divided into two types:
1. Statistical Language Models: N-grams, Hidden Markov Models (HMM)
2. Neural Language Models: Neural Networks, RNN, LSTM

Neural Language modeling uses neural networks to make that prediction of the upcoming words. For language modeling, neural language models have a much higher predictive accuracy than the n-gram language model, with some difficulties in training time and machine.

### Applications of Language Modelling
Language Modelling is used in applications like:
- Optical Character and Handwriting Recognition
- Speech Recognition
- Machine Translation
- Text Summarization
- Image Captioning
- Spelling Correction

### Why Neural Language Modelling?

N-gram models have a very sparse representation of language because the probabilities calculations are based on words' co-occurrence. It is the simplest language model, and with simplicity comes the lack of fluency and generalization.

_For example_:
Lets take a training sentence: <sub>[[1.1]](#Book)</sub>

`I have to make sure I get home to feed my cat.`


The prior contexts in n-gram models are represented with the words itself, whereas in neural language models, we represent the prior context with embeddings, giving better generalization.

When statistical language models like N-grams are trained in such sentences, they will only predict `cat` but not `dog.` But a neural language model will be able to generalize that `cat` and `dog` have similar embeddings and `dog` will have similar probability like `cat` to get predicted.

### Feed Forward Neural Language Model

Based on whether the embeddings for each word are trained or calculated from pre-trained embeddings, we divide neural language models into two parts:

#### 1. Pretrained Embeddings

In this neural language model, input words will be directly converted to their corresponding embeddings using a pre-trained embedding algorithm like word2vec. The embeddings are then concatenated in the projection layer and passed on to the higher layers(hidden).

The size (no of units) and the number of hidden layers is a hyperparameter and optimized for better prediction.

The final output layer consists of a softmax operation over |V| to find the word with the highest probability, which will be the next predicted word.

#### 2. Training Embeddings with Network

Such Feedforward neural language model consists of an **Embedding Matrix, E** ('d X |V`) that converts a one-hot representation of the input words into its corresponding embeddings.

These embeddings (`of size 1 X d') is input to the higher layers (hidden) to predict the next word in the sequence.


_For an example_:   
Let us consider `Man` to be an input word and is placed in $120^{th}$ position in the vocabulary, $V$. The **one-hot representation of the word `Man` will contain $0$ at all position except $120$, where it will have $1$.**  

This 1-hot representation of `Man` (`1 X |V|`) is now multiplied with Embeddings Matrix (`d X |V|`) to get embeddings, $E_{\text{man}} or E_{120}$ of dimension (`1 X d`).   
Such dimensions are concatenated to form the first hidden layer, also known as the **projection layer**.

Mathematically,

$$E_{Man} or E_{120} = \text{1-Hot Representation(Man)} \space X \space \text{Embedding Matrix}^T$$

The output layer is the same as in the previous network with pre-trained embeddings. The output layer consists of a softmax over all the vocabulary of words, V to predict a word with the highest probability.

### Training Neural Language Model

To train a neural language model is to perform gradient descent using error backpropagation to adjust all the weights, including the **Embedding Matrix (E)**. We predict the next upcoming word and learn the embeddings, E for each word that best predicts the upcoming words.

At each word, w$_t$, we calculate the cross-entropy (negative log-likelihood) loss given by:

$$L=- \log p (w_t|w_{t-1},  \cdots, w_{t-n+1})$$

and the weight updates is given by:

$$\theta_{t+1}=\theta_t - \alpha \frac{\partial(-logp(w_t|w_{t-1}, \cdots, w_{t-n+1}))}{\partial \theta}$$

_where_,   
$\alpha$ = learning rate of the model    
$p(w_t|w_{t-1})$ = Probability of occurance of $w_t$ with occurance of words ($w_{t-1}, \cdots, w_{t-n+1}$)

Such gradients are backpropagated through all the layers and update weights  including Embedding matrix, E, to minimize the loss and eventually learn a new set of Embeddings, E.


## Key Takeaways
- Each computational unit in a neural network multiplies input values by a weight vector, adds a bias, and then applies a non-linear activation function.
- Neural Networks are trained using optimization algorithms like gradient descent to backpropagate errors through all the learnable parameters like (weights, bias)
- Language Modelling predicts upcoming words based on prior words(context)
- Neural Language Models like ANN, RNN makes the use of neural networks to make such predictions
- Embeddings of corresponding words can be either learned (Embedding matrix) or generated from embedding algorithms like word2vec.
