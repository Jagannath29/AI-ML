In the previous chapter, we discussed about word2vec model. Word2vec model is a predictive model which optimizes the loss of predicting target words from the context words and thereby learning the word vectors as a by-product.

The notable consequence of the learnt word vectors from word2vec is that the similar words are located together in the same vector space. Also, another feature of word vectors from word2vec model is that the arthematic operations on these vectors poses semantic or syntactic relationships.
For e.g:
```
“king” - “man” + “woman” -> “queen”

“better” - “good” + “bad” -> “worse”.
```

The count based models such as LDA which estimates word vectors through dimensionality reduction on the co-occurrence matrix of the words was unable to capture this type of linear relationship.

The Glove is a count-based cum predictive based model to estimate word vectors. The Glove leverages the advantages of count-based method such as faster computation and parallelization over a large amount of data. The Glove also taks the benefits of predictive based model to learn the semantic or syntactic relationships (linear relationship) applying the predictive task on co-occurrence matrix explicitly.

## Glove Embeddings

### Architecture

The general architecture for the glove embedding is shown below. We pass two word tokens as the input and pass it to the embedding layer. The idea for the glove embedding is that it assumes that the dot product of the word vectors and the log of the co-occurrence counts of those word pairs are comparable. So, the error is calculated as the difference between dot product of the word vectors and the log of the co-occurrence counts of the input word pairs. We optimize this error and update the learnable parameters which are the glove word vectors. So, glove architecture tries to capture the intuition of count based method using co-occurrence of the words and the intuition of predictive based models by using the optimization of error to learn word embeddings.

#### **Input**
We pass word pairs $w_i$ and $w_j$ as the input to the model. The input is passed to the embedding layer.

#### **Embedding Vectors**
The embedding layer captures the vector representation of the words. We are assuming that the embedding dimension is $k$. That means word vectors has a dimension of $k$. The embeddings for words $w_i$ and $w_j$ are $v_i$ and $v_j$ respectively.

For instance,

Embedding for $w_i$:
$$v_i= \begin{pmatrix} a_1& a_2 & a_3&...&a_k\end{pmatrix}_{1*k}$$

Embedding for $w_j$:
$$v_j= \begin{pmatrix} b_1& b_2 & b_3&...&b_k\end{pmatrix}_{1*k}$$

#### **Dot Product**
The next layer is just the dot products of the embedding vectors. The dot product gives a scalar quantity.

i.e

$$v_i.v_j = \begin{pmatrix} a_1& a_2 & a_3&...&a_k\end{pmatrix}. \begin{pmatrix} b_1& b_2 & b_3&...&b_k\end{pmatrix}$$

$$v_i.v_j = a_1b_1 + a_2b_2 + ... + a_kb_k$$

#### **Error Calculation**
As we previously mentioned the dot product of word vectors and the count of co-occurrence of word pair are comparable. So the error term is simply the difference between dot product of word vectors and log counts of co-occurrence word pairs. We take the square difference for the error.

**Error term:**

$$error = (v_i.v_j - log(X_{ij}))^2$$

Here,
$$X_{ij} = \text{number of times words } w_i \text{ and } w_j \text{ co-occurred together in the corpus}$$

**Cost:**

The total cost of the model is given below.
$$
J = \sum_{i, j=1}^{V}(v_i.v_j - log(X_{ij}))^2
$$

After calculating the cost for the model, we backprogate the error and update the glove embeddings for the word vectors $v_i$ and $v_j$.


## Optional (Additional Reading)

In a more improved implementation, we also take account for the weighting function $f(X_{ij}).$ So the cost of the glove model becomes:

$$
J = \sum_{i, j=1}^{V}f(X_{ij})(v_i.v_j - log(X_{ij}))^2
$$

The reason for the weighting function is that to compensate for the possible cases which are listed below.

#### **Case 1:**
If co-occurrence of words count is 0 (if you look at the co-occurrence matrix you will see counts = 0), then log of co-occurrence count is undefined.

i.e
$$X_{ij} = 0 \rightarrow log(X_{ij}) \text {  is undefined} $$

In this case, we make the weighting function to be 0. So, overall it makes that term 0 without having to calculate the log term.

i.e

$$f(X_{ij}) = 0 \text {, if } X_{ij}=0$$

#### **Case 2:**

If value of $X_{ij}$ is very large i.e in the case of frequently occuring word pairs (eg: stop words such as `the, a an, which, those, this` etc), the model may give more priority (or overweight) these words. To compensate this we make $f(X_{ij})$ to be relatively small.


## Key Takeaways
* The embeddings of word2vec capture the co-occurrence of the words.

* In the glove model, we instead look at co-occurrence of the words first and then learn embeddings.

* The general idea of the glove model is that it assumes there is some kind of comparison between dot product of word vectors and their co-occurrence counts.
