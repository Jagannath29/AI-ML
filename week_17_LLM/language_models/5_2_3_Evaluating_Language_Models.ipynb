# Intuition and motivation
Every statistical model needs an evaluation. Evaluation is a measure of how good a model is and how well it is able to generalize the unseen data. In a n-gram language models, evaluation is done either extrinsically or intrinsically. In this chapter, we will discuss both type of evaluation measures and introduce an intrinsic language model evaluation called **perplexity**.

## Extrinsic Evaluation

**How much does the method or model help the application in which it is embedded?**

Extrinsic evaluation are end to end evaluations where a language model is built for a particular application and it is evaluated on the basis of how well it performs on that specific application. It is an application specific evaluation where the evaluation is done by embedding it in a particular applications such as spelling correction, machine translation or speech recognizer etc.

For example, we build two models namely model A and model B for the application of spelling correction. We evaluate the models by comparing their accuracy in that application. And we choose the model which gives more correct spellings looking at the context. In this case the winner is model A having the accuracy 0.8 which is greater than that of model B having accuracy of 0.6.

### Limitations of Extrinsic evaluations


* Extrinsic evaluations are costly for big NLP systems.
* It is application specific.

Due to these limitations, it would be nice to have an evaluation metric which is both application independent and can be used to quickly evaluate the models and compare it's performances. We will introduce intrinsic evaluation which overcomes these limitations.

## Intrinsic Evaluation

**How does the method or model score with respect to a given evaluation measure?**



Intrinsic evaluation is an application independent evaluation metric which measures the model's quality, compares it's performances and quickly evaluate the probable improvements in a language model.

The concept of intrinsic evaluation is simple that:
* We divide the corpus into three parts namely training set/corpus, test set/corpus and dev set/corpus.
* We find the probabilities of an n-gram model using the training set.
* We tune the hyperparameters of our model using dev set.
* We measure the quality of the model on unseen data called test set.
* We choose the model which gives the best accuracy in the test set.

The test set is also referred to as **hold out** set as we hold them from the training. For better evaluation, we should make sure that training set should not contain data from **hold out** set. Otherwise, the model will be biased and will give more priorities/probabilites to this data i.e **hold out set** while evaluating the model. It will make the model probabilities too high and thus will cause inaccurate accuracies in **perplexity** and will not be able to generalize well.

**Perplexity** is a common type of intrinsic evaluation metrics which we will discuss in the next section.


## Perplexity

### Intuition of Perplexity

As discussed earlier, the best model is the one that gives the best accuracy in the test data. Stated otherwise, the best model is that model which best predicts the test data.

Suppose we have a language model which predicts the next word. We have an unseen data **I am afraid of cats and __** and we need the best word after the word **"and"**.


<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1-rjgkgScfGZpIH7679K-GwRfrHPYke2B"> -->

The word **"dogs"** has a high probability followed by the probable words such as **"rats"**, "**tigers**" etc. The word **"and"** has the lowest of the probability as the word after **"and"** after **"and"** is meaningless.

This seems to be rather a convincing word predicted for the task. The better model is that model which assigns higher probability to the word which is most probable. Our evaluation measure should be such that, it should assign a high probability to the word which actually occurs or is the most probable.

With this understanding, let's formally define perplexity.

**Perplexity is defined as the inverse probability of the test set normalized by the number of words**.

We will see how to normalize the test set probability to form the evaluation metric perplexity in the mathmatical explanation of perplexity.

### Mathematical Explanation of Perplexity
Let's recall perplexity again:

Perplexity is the inverse test set probabilites with it's normalization with the number of words in the test set.


Suppose we have a sentence,
<br><br>
$$
S = w_1w_2w_3...w_N
$$
<br><br>
where $w_1$ refers to the first word of the sentence, $w_2$ refers to the second word of the sentence and so on and $w_N$ refers to the last word of the sentence.

The perplexity of the model on the test sentence $S$ denoted by $PP(S)$ thus can be expressed as the inverse probability of the test sentence $S$ normalized by the number of words in the test sentence $N$ as follows:

<br><br>
$$
PP(S) = P(S)^{\frac{-1}{N}} =  \sqrt[N]{\frac{1}{P(S)}} \tag{i}
$$
<br><br>

Since $P(S)$ is the **Joint Probability** of occurance of the sentence $S$, we can apply **Chain rule of Probability** to expand it.

i.e
<br><br>
$$
P(S) = P(w_1w_2w_3...w_N) = \prod \limits_{i=1}^{N} P(w_i | w_1w_2...w_{i-1})  \tag{ii}
$$
<br><br>

### Zero Order Markov Approximation/ Unigram model

We can approximate eqn $(ii)$ with unigram as:
$$
P(S) = \prod \limits_{i=1}^{N} P(w_i) \tag{iii}
$$
<br><br>

i.e
$$
P(S) = P(w_1) * P(w_2) * .... * P(w_{N}) \tag{iv}
$$
<br><br>

From eqn $(ii)$, we have,
$$
PP(S) = \sqrt[N]{\frac{1}{P(S)}}
$$

Thus,
<br><br>
$$
PP(S) = \sqrt[N]{\frac{1}{P(w_1) * P(w_2) * .... * P(w_{N}) }} \tag{v}
$$

In eqn $(iv)$ the higher the conditional probability of the word sequence the lower is the perplexity. Stated otherwise, the less the perplexity, the better is our model predictions. So our goal in language model evaluation is to minimize perplexity to maximize the model probabilities.

## Example of perplexity calculation


Perplexity can also explained in terms of **weighted average branching factor**. The branching factor of a language is the number of possible words that can follow any word. Let's understand this with an example.

Suppose we have the following training set.

Training set

```
eat sleep play repeat
eat sleep play repeat
```

There are total of 4 words which are equally likely to happen. So the probability of each words in training set are below:

$$
P(eat) = P(sleep) = P(play) = P(repeat) = \frac{2}{8} = \frac{1}{4}
$$


We are trying to find the **perplexity** in the test set and suppose there are total of 4 words which are also equally likely.

The test set is given below.

```
eat sleep play repeat
```
Here, $$N = 4$$


**Let's find the joint probability of the test set using approximation of unigram model.**

i.e
$$
P(S) = P(eat)\,P(sleep)\,P(play)\,P(repeat)
$$

Thus,

$$
P(S) = \frac{1}{4}\cdot\frac{1}{4}\cdot\frac{1}{4}\cdot\frac{1}{4} = \frac{1}{256}
$$

We have

$$
PP(S) = \sqrt[N]{\frac{1}{P(S)}}
$$

Thus,

$$
PP(S) = \sqrt[4]{\frac{1}{\frac{1}{256}}} = \sqrt[4]{256} = 4
$$


What does this value of $PP(S) = 4$ means? It means there are 4 words which are likely to follow any words.

Suppose the first word that occured is **eat**, then there are 4 words **eat, play, sleep, repeat** which are equally likely to occur after the word **eat**.

### Second scenerio

Let's see what happens when we change the training set with word **eat** repeating 5 times and other words occuring only one time as follows:

Training set
```
eat eat eat eat
eat sleep play repeat
```

The probability of occurance of words are:
$$
P(eat) = \frac{5}{8}
$$

$$
P(play) =P(sleep) =P(repeat) = \frac{1}{8}
$$


Suppose the

Test set

```
eat eat eat eat
```
Here,
$$N = 4$$

<br><br>
$$
P(S) = P(eat) * P(eat) * P(eat) * P(eat)
$$

<br><br>
$$
P(S) = \frac{5}{8} * \frac{5}{8} * \frac{5}{8} * \frac{5}{8}
$$

<br><br>
$$
P(S) = 0.625^4 = 0.15258
$$

<br><br>
$$
PP(S) = \sqrt[4]{\frac{1}{0.15258}}
$$

<br><br>
$$
PP(S) = 1.6
$$


The branching factor is still 4 (i.e there are still 4 probable words) but the **weighted branching factor** is reduced to 1.6. It means now the model has learned about the word **eat** that it is more likely than other words in the training set.

It is important to construct the model probability $P$ without any kind of prior information of the test set or it's vocabulary. Any such prior information about the test set causes perplexity to be artificially low.


Also, intrinsic evaluation should only be used as a quick check on the algorithm and it should be noted that intrinsic improvement does not guarantee the extrinsic improvement on a particular applications although intrinsic evaluation often correlates with extrinsic evaluation. We should always confirm the model's performance by end-to-end evaluations on real applications.

### (Optional) First Order Markov Approximation/ Bigram model
Let's approximate $(ii)$ with first order Markov assumption or bigram model.

<br><br>
$$
P(S) = \prod \limits_{i=1}^{N} P(w_i | w_{i-1}) \tag{vi}
$$
<br><br>

Substuting $(vi)$ in $(i)$, we get,
<br><br>
$$
PP(S) =  \sqrt[N]{\frac{1}{P(S)}}
$$
<br><br>
$$
PP(S) = \sqrt[N]{\frac{1}{\prod \limits_{i=1}^{N} P(w_i | w_{i-1})}} \tag{vii}
$$
<br><br>

The equation $(iv)$ is the perplexity for bigram model, we can similarly derive perplexity for other n-gram models using Markov approximations and substuting it in eqn $(i)$.


## Key Takeaways
* Extrinsic evaluation is an end to end evaluation of a language model on a particular application.
* Intrinsic evaluation measures performance of a language model independent of any particular application.
* Intrinsic evaluation considers splitting of the corpus into training, test and dev sets.
* Perplexity is the evaluation metrics of the language model.
* Perplexity is defined as the inverse probability of a test set normalized by the number of words in the test set.
* Minimizing perplexity is the same as maximizing model probabilities.
* Intrinsic evaluation does not guarantee extrinsic improvements. We should always confirm a model's performance by end-to-end evaluations on real applications.
