## Generalization and Zeros

As we discussed in earlier chapters, we train n-gram model in the training set and we evaluate it's performance in the test set. We want our model to learn in training data and capture it's distribution well and adapt to new and unseen data drawn from the same distribution.

**Generalization is the model's ability to adapt to new and unseen data which are drawn from the same distribution as the training data.**

In this chapter, we will discuss generalization and issues of generalization in ngram models.

## Generalization in n-grams

There are following generalization issues in n-gram models:
* Training on one genre of corpus and testing on another genre
* Effect of value of N in n-grams
* Zeros as generalization types

### Effect of different genres in training and test set

Suppose, our training corpus is a sport related corups and it contains sentences related to sports such as:
```
Messi scores a goal. Ronaldo took an hat-trick. Dhoni hits a Six.
```

The probability for sports related pharses/words in this senerio will be high. However, after training the n-gram model on the sport corpus, if we evaluate it in a politics corpus, then it assigns a very low probability to the politics related words or pharses such as:
```
GPD is increased by 10 percent.
```

n-gram model is thus dependent on training corpus and probabilities often encode specific facts about the training corpus. Infact, the semantics and grammer in the n-gram depends on the training set. n-grams only work well for word prediction if the test corpus looks like the training corpus. So, we should always be careful to match the genre of training and test set for the better generalization of n-gram models.

### Effect of n in n-gram



As N in n-gram increases, it does a better job of generalization as it is capable of learning the context of the words. The lower value of N in gram captures less context and higher value of N captures more context of the word.

Let's look at an example:
```
                 men
             for men
        step for men
  small step for men
A small step for men
```

Suppose, we have a prediction task for occurance of word **men**. For this task there is no unigram context, the word **for** is the bigram context, sequence of words **step for** is the trigram context is , sequence of words **small step for** is the 4-gram context and sequence of words **A small step for** is the 5-gram context. Clearly, in terms of context 5-gram is the winner as it captures all the information of the word **men**.

However, if we capture all the context in the training set, it does a better generalization in the training corpus but it can not adapt in the new unseen scenarios of word predictions and it will thus lead to overfitting.

```
...
with Amy
with Country
with dog
with dreams
...
``` -->

Sparsity in n-grams happen because of limited corpus. As we discussed earlier, language is artistic in nature and new data are being generated over time, we can never be assured of enoguh data. And also, even if we have a huge corpus some perfectly acceptable words will be missing from it.

Let's look at real case senerio of **Shakespear corpus**:

Total Number of words appeared in the corpus are, $N = 787,434$

The unique vocab size is, $V = 26,431$

The possible bigram from the vocabulary will be of size:
2-gram, $V^2= 698,597,761$

<!-- 4-gram, $V^4 = 4.8803* 10^{17}$ -->

Let's look at a particular senerio of bigrams taken from the **Shakespear corpus**. While creating bigrams all texts were converted in lowercase and puntuation marks were treated as words. The table shows the bigram count for a particular sentence **```my mistress with a monster is in love```**.


<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1T3vl2oyh221QQbQ8Q2ZdmpKkuHyF9Are" > -->
<img src="https://i.postimg.cc/D0G4CcWk/image.png" >
<Figcaption>Fig 1: Bigram for a test sentence from Shakespear corpus and valid bigrams with zero counts</Figcaption>




You can see from the table that, even in this large corpus, bigrams such as **is monster**, **in monster** etc have zero counts. It is because not all the bigram sequence from the vocabulary appear in the training set. The individual words are in the vocabulary, but the related bigram sequence are missing in the training corpus. In the figure you can see some perfectly acceptable bigrams which have zero counts in the training corpus.

<!-- ```
is monster 0
in monster 0
love monster 0
with monster 0
``` -->

Zeros happen because of the missing of n-grams words in the training data.

## Zeros as a generalization type

In the above example, bigrams such as **{is monster}**, **{in monster}** etc have never occured in training data. But they are perfectly acceptable sequence of words, and they can appear in the test set.

However if we calculate their probability, we get 0 as their count in the training set.

i.e
<br>
$$
P(monster | is) = 0
$$

$$
P(monster | in) = 0
$$


Let's see at another example. Suppose our training set contains 4-gram sequences like in figure below:


<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1WJogeVtUMebqIrjltejraWkqieRBYYDM" > -->
<img src="https://i.postimg.cc/RFBbXgpK/image.png" >
<Figcaption> Fig 2: Zeros as a generalization type example</Figcaption>



We can correctly estimate the joint probability as well as conditional probability such as,


$$
P(bus | looked \;at\; a)
$$

$$
P(looked \;at\; a\; bus)
$$


but if our test set contains sequence of 4-gram sequence which are missing in the training set such as **...looked at a tiger...**:

The joint probability of the sequence becomes zero.
$$
P(tiger | looked \;at\; a) = 0  
$$

$$
P(looked \;at\; a\; tiger) = 0  
$$

As their probability is zero, the perplexity of the test set containing these n-grams becomes undefined.


This is an generalization issue called zeros which we will address in our next chapter called **Smoothing and Backoff**.

## Unknown words
In the previous section, we saw the issue zero probability for bigrams. These words were in the train set. But there are situations where we have the words which we have never seen before. These words are called unknown words. Let's understand the intuition of unknown words with the concept of closed and open vocabulary.

### Closed Vocabulary

In a closed vocabulary system, the test set contains words only from the known vocabulary so there will be no unknown words. for eg. machine translation uses a look up tables for the translation of the words and these lookup table have a fixed set of vocabulary.

### Open Vocabulary
In most of the cases, we have to deal with words which we have not seen before. These words are called unknown words or **Out of Vocabulary(OOV)** words.

In **Open vocabulary** system, we model these unknown words or OOV words appearing in the test set by adding a $<unk>$ token.

There are two ways of estimating the probabilities of OOV words.

#### First Method
1. Choose a vocabulary (word list fixed in advance)
2. In the text normalization step, assign any words which is not in the the fixed vocabulary we created in the step 1 by $<unk >$ token.
3. Estimate probabilities for $<unk>$ from its count like any other word in the training set.


#### Second Method
If we do not have prior fixed vocabulary, we could implicitly create such vocabulary from the training set based on the frequency.


For example, we could replace by $<unk>$ token if the word count occurs lesser than n times in the training set. Or we could fix a vocabulary size in advance, let's say $V = 50,000$ unique words and we could select top $V$ words by frequency and replace the other by $<unk>$.

<!-- **A language model can achieve low perplexity by choosing a small vocabulary and
assigning the unknown word a high probability. For this reason, perplexities should only be compared across language models with the same vocabularies **(Buck et al., 2014).** -->

# Smoothing

Zeros occur when a known word happens in a unseen context. That means the word is in the vocabularly but it does not appear in a particular context. These unseen events assign zero probabilities to these n-grams. In this chapter will we see how we can address this issue with smoothing or discounting.

Smoothing or discounting is clipping some probability masses from more frequent events and distributing them to the unseen events.

Let's look at the intuition for smoothing with the figure below.

<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1_GSQNZ8xqZ0nbaXpHCYWmw3g0rKSVXGh" > -->
<img src="https://i.postimg.cc/66Dz0j73/image.png" >
<Figcaption>Fig 1: Intuition of smoothing</Figcaption>

Suppose we have a context **looked at a** and the words **bus, car, man, person** follows the context in the training corpus. However, we have other words such as **tiger, lion, bicycle** etc in the vocabulary which are probable words but they do not appear after the context.

What we could do is trim the counts or probability mass of appeared words such as **bus, car, man, person** etc and assign it to not appearing words such as **tiger, lion, bicycle** etc. So, when in test set if particular n-grams such as **looked at a tiger**, **looked at a lion** etc appear the n-gram model does not assign them a zero probability.

## Types of smoothing

There are various types of smoothing techniques. They are as follows:

* add-1 or laplace smoothing
* add-k-smoothing
* stupid backoff
* Kneser-Ney smoothing

In this course, we will discuss add-1-smoothing and add-k-smoothing techniques. However, advanced smoothing concepts like stupid backoff and Kneser-Ney smoothing are not discussed in this course. Students are highly encourged to explore these techniques on their own.

Let's discuss the basic types of smoothing such as add-1 and add-k smoothing in the next sections.

## Laplace smoothing (add-1-smoothing)

In Laplace smoothing we increase the count of the n-gram by 1. That means if the count is 0, it becomes 1, if the count is 1 it becomes 2 and so on. After increasing the counts to estimate the probability, we normalize by taking account of extra vocabulary size. Laplace smoothing does not perform well enough to be used in modern n-gram models, but it introduces various concepts for smoothing and it can be used as a baseline estimation. However, it can be used in practical senerio such as **text classification**.

### Laplace smoothing for Unigram

Recall for a unigram model, the unsmoothed probability estimation is given as:
$$
P(w_n) = \frac{count(w_n)}{N}
$$
Here, $N$ refers to the number of word tokens.

In Laplace smoothing, we add one to each word, so we have to take account for extra $V$ which is the vocabulary size.
$$
P_{Laplace}^{*}(w_n) =\frac{count(w_n) + 1}{N + V}
$$

Typically, we assume $$V= \{ w:count(w)>0 \} ∪ \{<unk>\} $$

This means that the vocabulary is the set of words whose count is greater than 0 plus the unknown words.

Let's see with an example:

Suppose, we have a small corpus of size $N=20$ and vocabulary of $V=4$.

There are four unique words **eat, play, sleep, repeat** in the vocabulary. Here, the word **repeat** is an unknown word which does not appear in the training sequences but it is in the vocabulary due to its high possibility of occurance.

```
eat eat eat play play sleep sleep
eat eat eat eat
play play play sleep sleep sleep
eat eat eat
```

The smoothed and unsmoothed probabilities for the unigram words are shown below.


<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1a3gzpexko_bbUO4IG__lzLXZ8K5L4yJj"> -->
<img src="https://i.postimg.cc/zG3BGjjC/image.png">
<Figcaption>Fig 2: unigram smoothing</Figcaption>

From the table, we can see that for unsmoothed unigram we have $P(repeat) = 0$. However after smoothing the probability becomes non-zero. i.e $P(repeat)=0.0416$.

### Laplace smoothing for Bigram (Optional)

In case of bigrams, we add 1 to the bigram counts and we augument the previous word(i.e unigram) by the total number of word types in the vocabulary $V$.


For unsmoothed bigram,
$$
P(w_n|w_{n-1}) = \frac{count(w_{n-1}w_n)}{count(w_{n-1})}
$$

After Laplace smoothing,
$$
P_{Laplace}^{*}(w_n|w_{n-1}) = \frac{count(w_{n-1}w_n) + 1}{count(w_{n-1}) + V}
$$


<!-- Typically, we assume $$V= \{ w:count(w)>0 \} ∪ \{<unk>\} $$  -->


Let's look at the previous example of **Shakespear corpus** and see how add-1-smoothing addresses zeros.

<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1WFUeFgfSQTpY05HEXed9Q8Q1n-2qiaaT" > -->
<img src="https://i.postimg.cc/8CpVCPDq/image.png" >
<Figcaption>Fig 3: Bigram and unigram counts for bigram smoothing</Figcaption>

Before smoothing,
$$
P(monster | is) = \frac{count(is \; monster)}{count(monster)} = \frac{0}{77} = 0
$$


After add-1-smoothing,

$$
P(monster | is) = \frac{count(is \; monster) + 1}{count(monster) + V} = \frac{1}{77 + 26431} = 0.000037725
$$

## Add k smoothing

In add-k smoothing, instead of adding 1 to each count, we add a fractional count k. The purpose is same as add-1-smoothing, however in this scenerio, the probability mass of frequent sequence is clipped to distribute for unseen sequences a bit less with a factor k. k can be a fraction such as 0.1, .5, 0.6 etc.

For unigram,
$$
P_{Add-k}^*(w_n) = \frac{count(w_n) + k}{N + kV}
$$

### For bigram (Optional)

$$
P_{Add-k}^*(w_n | w_{n-1}) = \frac{count(w_{n-1}w_n) + k}{count(w_{n-1}) + kV}
$$


The choice of k is a hyperparameter and it needs to be optimized in a devset. Add-k smoothing still does not perform well for language models. However, it can be applied to tasks such as text classifications.

## Backoff

### Intuition for Backoff

The smoothing techniques we discussed in the earlier sections solves the problem of zero frequency in n-grams. However, we can look into additional source of knowledge. Backoff is an attempt of solving the zero frequency problem in another way.

### Introduction to Backoff

If we are trying to compute the probability $w_n$ given sequences $w_{n-2}$ and $w_{n-1}$ then if there is no particular example of the trigram sequence, we could estimate it by the bigram probaility  $P(w_n | w_{n-1})$ and if there is no particular example of the bigram sequence $w_{n-1}w_n$ then we can compute the probability using its unigram probability i.e $P(w_n)$.

<br>
$$
P(w_n | w_{n-2}w_{n-1}) \;\; \rightarrow \;\; P(w_n | w_{n-1}) \;\; \rightarrow \;\; P(w_n)
$$
<br>


Consider the following trigram.

```
languish for her
```

We find that the counts for this trigram is zero in training corpus. So we have to backoff the trigram to either bigram or to unigram. The counts for the unigram, bigram and unigram are as follows:

```
languish for her 0
for her 32
her 41
```

The backoff for this example is below.

<br>
$$
P(her | languish \; for) \;\; \rightarrow \;\; P(her | for)
$$
<br>

Note: in this case, we have non-zero counts for **`for her`**, so we backoff the trigram to bigram only.

<br>

Using backoff smoothing, we can generalize for context that the model has not learned much about by using the less context. We estimate the higher order model's probability, if the evidence for higher order n-gram is sufficient for probability estimation and if the evidence for higher n-gram models is zero we **"back off"** to lower order n-grams.

## Key Takeways

- n-gram has three main types of generalization issues - training and test set of different genres, effect of N in n-grams and zeros as generalization types.

- n-gram is dependent on training corpus and probabilities often encode specific facts about the training corpus

- n-grams only work well for word prediction if the test corpus looks like the training corpus

- High value of n tends to generalize the training data well but it will lead to overfitting.

- Smoothing or discounting is clipping some probability masses from more frequent events and distributing them to the unseen events.

- Laplace smoothing adds one to the counts of n-grams.

- Add-k smoothing adds a fractional count k. k is a hyperparameter which is tuned in the devset.

- In backoff, we estimate the higher order model's probability, if the evidence for higher order n-gram is sufficient for probability estimation and if the evidence for higher n-gram is zero we "back off" to lower order n-grams.

# References

<a name="Book"></a>
1. Book
   * [Dan Jurafsky, James H. Martin, Speech and Language Processing, (2019) - Chapter 3 ](https://web.stanford.edu/~jurafsky/slp3/3.pdf)

<a name="Youtube videos"></a>
2. Youtube Videos
  * Yeung Hau. Sep 24, 2019. Natural Language Processing by Michael Collins, Columbia University P9 01. Line retrieved from https://www.youtube.com/watch?v=uqsJ3ZS1J6I

<a name="Tutorial slides"></a>
3. Tutorial Slides
 * https://web.stanford.edu/~jurafsky/slp3/slides/LM_4.pdf

<a name="Tutorial videos"></a>
4. Tutorial Videos
  * Pieter Abbeel. May 11, 2012. Laplace smoothing.
 Retreived from https://www.youtube.com/watch?v=gCI-ZC7irbY
