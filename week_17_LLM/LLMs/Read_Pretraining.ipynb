# Pre-Training

## Introduction to Pre-Training

Pretraining is the training process of deep learning models where the neural network is trained on a large and diverse dataset before fine-tuning it for a specific task. A neural network learns and encodes a deep statistical representation of language to extract meaningful and valuable features from the text data. It learns the word embeddings or contextualized representations of words and sentences.

Pre Training enables transfer learning, which is the process of using knowledge learned from one task to improve performance on a different but related task. The weights and parameters of a neural network are initialized during pretraining. The model weights get updated during pretraining to minimize the loss of a training objective. And these initializations is from where the network starts its finetuning phase.

The data used in pretraining is from a wide variety of text sources from the internet, including books, articles, and websites. These data are usually in large amounts, usually Terabyte or Petabyte. Pretrained models are trained usually in an unsupervised or semi-supervised learning approach. This way small task specific dataset is enough to finetune the model and achieve a considerable performance.

In NLP, models like GPT (Generative Pre-trained Transformer) and BERT (Bidirectional Encoder Representations from Transformers) have become famous examples of pretrained models that have achieved state-of-the-art performance on various downstream tasks by leveraging pretraining.


### Architecture

The pretraining can be done in multiple ways. For instance, the model could be trained to predict subsequent words for the given sequence of words. For example, GPT models. Also, the model could be trained using the masked language modeling where it needs to guess randomly masked words in a sentence. For example, BERT.

There are mainly three variants of the transformer model.
1. Encoder Only
2. Decoder Only
3. Encoder-Decoder

Each of these models are trained on a different objective. So, it learns how to carry out different tasks.


#### Encoder Only

Encoder-only models are also known as autoencoding models. They are pre trained using masked language modeling (MLM), where some of the words are masked (hide) in a sentence and then the model is trained to predict those masked words based on the context provided by the rest of the sentence. So, the main objective is the reconstruction of text (aka “denoising”).

They maintain the context by paying attention to all the words in the sentence simultaneously which is achieved with multi-head self attention, bidirectional understanding and positional embedding. They build bidirectional representations of the input sequence i.e. model has understanding of the full context of token (not just of the words that comes before)

#### Decoder Only

Decoder-only models are also known as autoregressive models. They are pre trained using Causal Language Modeling (CLM). The main objective is to predict the next token based on the previous sequence of tokens. At each time step, the prediction is dependent only on the past and not on the future. The task of predicting next tokens is sometimes called full language modeling

The context is unidirectional as input tokens are masked and can only see the input till the generation has been completed (example; till the teacher …). The decoder-only architecture does not have an explicit encoder to summarize the input information. Instead, the information is encoded implicitly in the hidden state of the decoder, which is updated at each step of the generation process.

#### Encoder-Decoder

Encoder Decoder models are also known as sequence to sequence models. They use both the encoder and decoder part of original Transformer Architecture. The training objectives vary from model to model. Popular T5 model pretrains encoder using Span Corruption.

In span corruption, masks random sequences of input tokens, then those masked sequences are replaced with unique sentinel tokens. Sentinel tokens are special tokens added to vocabulary but do not correspond to any actual word from the input text. Decoder reconstructs masked token sequences autoregressively and output is a sentinel token followed by predicted tokens.

### Transfer Learning

The training of the model that can make the basic understanding of the language or any domain requires a huge amount of data. In many real-world scenarios, obtaining labeled data for training a machine learning model can be challenging and expensive. And often, tasks are related or have some common underlying structure. That's where transfer learning capitalizes on these relationships.

Transfer learning is a machine learning paradigm that involves training a model on one task and then leveraging the knowledge gained from that task to perform well on a different but related task. The idea is to transfer the learned features, representations, or knowledge from one domain to another, thus saving computational resources and time compared to training a model from scratch.

For analogy, In computer vision, for example, neural networks usually try to detect edges in the earlier layers, shapes in the middle layer and some task-specific features in the later layers. In transfer learning, the early and middle layers are used and we only retrain the latter layers.

Types of Transfer Learning:
1. Inductive Transfer Learning: The source and target tasks are different, but there is an assumption that some knowledge gained in the source task can be useful for the target task.

2. Transductive Transfer Learning: The source and target tasks are the same, but the distributions of data might differ.

Some of the examples of transfer learning models are: pre trained word embeddings like Word2Vec, GloVe, used as features for NLP tasks, CNN pretrained on ImageNet often used as feature extractors for various computer vision tasks.

#### ULMFiT

The idea of transfer learning was first introduced in Computer Vision where the pretraining of the deep neural network on a large dataset of images such as ImageNet and then the pretrained model was used to extract the features of a new dataset of images.

ULMFiT stands for Universal Language Model Fine-tuning. It is a natural language processing (NLP) technique introduced by Jeremy Howard and Sebastian Ruder in 2018. One of the first papers was using language models to do an effective transfer learning for text classification. It is designed to address the challenge of transfer learning in the context of NLP tasks. Transfer learning involves pre-training a model on a large dataset and then fine-tuning it on a specific task with a smaller dataset.


image.png

Previously, the state-of-the-art on many NLP tasks were achieved by training the models from scratch that required large datasets and days to converge. This paper proposes discriminative fine-tuning, slanted triangular learning rates,and gradual unfreezing, novel techniques to retain previous knowledge and avoid catastrophic forgetting during fine-tuning.

**Discriminative Fine-tuning:** ULMFiT uses a technique called discriminative fine-tuning, where different layers of the neural network are fine-tuned at different learning rates. The authors found that fine-tuning earlier layers with lower learning rates and later layers with higher learning rates was effective in practice.

**Slanted Triangular Learning Rates:** ULMFiT employs a learning rate schedule called slanted triangular learning rates, which starts with a low learning rate, increases it quickly, and then gradually decreases it.

**Gradual Unfreezing:** The model is gradually "unfrozen" during fine-tuning. This starts fine-tuning with only the last layer, then gradually unfreeze and fine-tune earlier layers. This allows the model to adapt to the specific task without losing the knowledge gained during pre-training. We first unfreeze the last layer and fine-tune all unfrozen layers for one epoch. We then unfreeze the next lower frozen layer and repeat, until we fine tune all layers until convergence at the last iteration.

ULMFiT has been successful in achieving state-of-the-art results on various NLP tasks, even when the amount of task-specific data is limited. It demonstrated that pre-training a language model on a diverse range of text can lead to better generalization and performance on downstream tasks.

For further detailed information on every stage of pre-training and fine-tuning of language models, you can refer to the original paper. [Source]

#### Meta Learning

Metalearning aka learning to learn is a subfield of ML that focuses on developing algorithms and models that can learn from and adapt to new tasks more efficiently and effectively. The goal of meta-learning is to design systems that can generalize knowledge and learn new tasks with minimal amounts of data.

The motivation behind meta-learning is to mimic the way humans learn. Humans can leverage knowledge gained from learning one task to quickly adapt and learn new, related tasks. Meta-learning aims to imbue machine learning models with a similar capability for rapid adaptation and generalization.

This paradigm provides an opportunity to tackle many conventional challenges of deep learning, like the need for large datasets for training, high operational costs due to many trials/experiments during the training phase, and the need for experiments that take a long time to find the best model which performs best for a certain dataset.

Popular Meta-learning algorithms
- Metric learning
- Model-Agnostic Meta learning

GPT models, particularly those like GPT-3, are known for their ability to perform few-shot learning. Users can provide a few examples of a task, and GPT can generate outputs based on that limited input. This aligns with the concept of meta-learning, where the model generalizes from a few examples to perform a task.


### Generative Pre-Trained Model (GPT)

There are two main architectures that have been the successor of the Transformer architecture that was introduced in 2017 by Google. BERT, which has the architecture designed with stacking of the encoder layers and GPT, which has the architecture designed with stacking of the decoder layers. Each of these architectures have created their own lines of research. Both of these architectures use the transfer learning principle to perform the tasks. They are first trained on some tasks such that the model has some basic knowledge of the domain tasks, which is also called pretraining and then use the base knowledge of the model to further train the model on another task, which is also called finetuning.

We hope you have fully understood the intuition behind the transformer architecture and about the various components used in the architecture from the Transformer section of NLP. In this reading material, I will discuss the progression of GPT architecture and briefly talk about the findings in each version of GPT.

GPT-1 is the first large language model released by OpenAI in February 2018. GPT-1 has 117 million parameters and was trained by gathering training data BookCorpus Dataset. BooksCorpus consists of about 7000 unpublished books which helped in training the language model on unseen data. This corpus also contained long stretches of contiguous text, which assisted the model in processing long-range dependencies.

The key ideas from the introduction of GPT-1 are:
- we can stack the bunch of transformer decoders
- Semi-supervised pre training helps a lot for transfer learning and achieving great performance on tasks
- We can take pre trained network and finetune on several tasks

GPT-1 explored how to take advantage of massive unlabelled text dataset to finetune them on limited supervised dataset. This paper introduced that large gains on different natural language processing tasks can be realized by generative pre-training of a language model on a diverse corpus of unlabeled text, followed by discriminative fine-tuning on each specific task. This paper also showed that general task-agnostic models (task-aware input transformations during finetuning) outperforms discriminatively trained models that use architectures specifically crafted for each task.

In terms of its architecture GPT-1 applies the 12-layer decoder of the transformer architecture with 12 attention heads. This generative language model was able to learn large range dependencies and acquire vast knowledge on a diverse corpus of contiguous text and long stretches.

They performed a major of four tasks: text classification, entailment, text similarity and multiple choice using 12 supervised datasets. They observed that fine tuning using the pretrained model showed improved performance. Another interesting thing they found was they showed good results on the tasks even without having to see the data for those tasks at all. They showed to be a zero shot learner, i.e. ability to generalize on a new task without having to train on them. This very idea formed the base for the future improvement in GPT versions.

Humans can learn from a few examples. And, the GPT required some thousands of examples for the specific tasks. But, we want the field of deep learning to take direction along the lines of human intelligence. That's where the GPT2 focuses on.


GPT-2 was released using an even larger dataset and additional parameters. It was trained with 1.5 billion parameters. It introduced the concept of meta learning. GPT2 uses a similar approach to GPT architecture but it focuses on zero shot learning more than the fine tuning approach. In zero shot learning, we don't make any updates to the parameters once the models have been pretrained.  While making inference, we pass in the input as we usually do. But, here we also add the prompt, that is the instruction of what should be done with the input.

The issue with the zero shot learning is that it's hard for the small model to generalize on the task. We need to scale up the architecture to capture as many language patterns as possible during pre-training. So, this gave the motivation for the GPT-3 model.

The main difference between the three GPT models is their size. GPT-3 is trained upon a 500-billion-word data set (known as “Common Crawl”) collected from the vast internet and content repository. It uses the same strategy of meta learning but scales the architecture even more. The GPT3 model is trained with 175 billion parameters. Here, instead of one of the meta learning, it could perform zero shot learning, one shot learning, and few shot learning. The learning is limited by the context window of the model which limits the number of few-shot examples from which the model can learn.

ChatGPT is a conversation friendly model which is not a model trained from scratch, but rather a refined version of GPT-3.5, which itself was a refined version of GPT-3. The training follows a similar approach to that of the InstructGPT paper where the model is geared towards following instructions (user provides a single request and model outputs the response). But, ChatGPT extends to more interactive dialogues with back and forth messages where models can retain and use earlier context.

ChatGPT involves three phases of  training:
- Generative Pretraining
- Supervised Fine Tuning
- Reinforcement Learning with Human Feedback

ChatGPT cannot produce or understand images, GPT-4 can accept both images and text as inputs or prompts. GPT-4 is based on the multimodal model that makes it capable of analyzing and responding to images submitted alongside prompts.

#### Rotary positional embeddings (RoPE)

The self-attention operation, which is implemented in every layer of a language model’s decoder-only transformer architecture, does not naturally consider the position of each token within a sequence. As such, we must inject position information (e.g., via an additive positional embedding) into this operation for each token.

This has the limitation that the relative position between the tokens are not captured. Position 1 token and position 3 token are quite related to each other than the token in Position 1 token and Position 500. And, also because of the absolute positional embedding the max length is bounded i.e. words beyond that fails to be encoded.

Rotary positional embeddings are an alternative that incorporates both the absolute (i.e., global position of a token in the sequence) and relative (i.e., defines position based on distances between tokens) position of each token into self-attention by:

- Encoding absolute position with a rotation matrix
- Adding relative position information directly into the self-attention operation

This approach is found to yield a balance between absolute and relative position information, which benefits performance especially on tasks that require longer sequence lengths.

For further detailed information, you can refer to the original paper of RoPE.


#### Multi-Query Attention

Original multi-head self-attention operation can be replaced with multi-query attention.
Multi-query attention just shares key and value vectors between each of a layer’s attention heads. Instead of performing a separate projection for each head, all heads share the same projection matrix for keys and the same projection layer for values.


image.png

The multi-head attention takes up many computations which becomes the bottleneck. To resolve this problem, a KV cache was introduced. The KV cache solved the problem of the number of computations required but arose a new problem of memory access. Further, to solve this problem multi-query attention was introduced.

Multi-query attention is still multi-head attention but only with respect to query Q. we have multiple heads for Query Q but the Key and value will be shared by all the heads. Although this change does not make training any faster, it significantly improves the inference speed of the resulting LLM.

For further detailed information, you can refer to the paper that introduced the multi-query attention.


#### Grouped Multi-query Attention

In multi-query attention, we have multiple heads for the queries and only one head for keys and value. In grouped-query, we divide the queries into groups and for each group we have one different head of key and value.

Multi-head attention is of high quality and computationally slow and multi-query attention has loss in quality and is computationally fast. But, the grouped multi-query attention has a good compromise between quality and speed.

#### Falcon

Falcon is an open source multilingual large language model created by  Technology Innovation Institute (TII).

**Dataset** \\
They have used the data from REFINED WEB, a high-quality five trillion tokens web-only English pre training dataset. [Paper]

**Architecture Details** \\
Both Falcon-7B and Falcon-40B models use a modified decoder-only transformer architecture. Modifications made to this architecture include:

- Flash Attention
- RoPE embeddings
- Multi-Query Attention
- Parallel Attention and Feed-Forward Layers

These modifications drastically improve Falcon’s inference speed. In fact, Falcon-40B is 5X faster than GPT-3 at inference time. Both Falcon-7B and Falcon-40B are trained with a sequence length of 2K tokens.

#### LLaMA

LLaMA is an open source multilingual large language model developed by Meta AI.

**Dataset** \\
They have used publicly available dataset compatible with open sourcing.
They have not used the data from Meta’s product and services. And also, they have removed data from the sites that contain high volumes of personal information about individuals.
- English CommonCrawl [67%]
- C4 [15%]
- Github [4.5%]
- Wikipedia [4.5%]
- Gutenberg and Books3 [4.5%]
- ArXiv [2.5%]
- Stack Exchange [2%]


**Architecture** \\
They leverage various improvements that were subsequently proposed, and used in different models such as PaLM. Modification from the original transformer architecture include:
- Pre-Normalization (RMSNorm) (GPT3)
- SwiGLU Activation function (PaLM)
- Rotary Embeddings (GPTNeo)

The primary architectural differences from Llama 1 in Llama 2 include increased context length and grouped-query attention (GQA).
