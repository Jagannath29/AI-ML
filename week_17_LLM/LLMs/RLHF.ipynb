## 1. Introduction

### What is RLHF?

Reinforcement Learning from Human Feedback (RLHF) is an approach that involves training a "reward model" directly using human feedback. This model is then employed as a reward function to fine-tune and optimize an agent's policy through reinforcement learning (RL). It is a type of machine learning that involves training an agent to make decisions based on feedback from its environment. In RLHF, the agent also receives feedback from humans in the form of ratings or evaluations of its actions, which can help it learn more quickly and accurately.


The idea of RLHF is to use methods from reinforcement learning to directly optimize a language model with human feedback. RLHF has enabled language models to begin to align a model trained on a general corpus of text data to that of complex human values. RLHF is used to enhance the text generation capabilities of language models, particularly large pre-trained language models like GPT-3 and InstructGPT.

### Key Components of RLHF:

RLHF involves the training of three key models: the Language Model, Reward Model, and Policy Model. These models work together to optimize the language model's performance based on human feedback.

- **LLM Training** - updating parameters of the language model only

- **RLHF Training** - training parameters of three different models
1. Language Model
2. Reward Model
3. Policy Model (Agent)

### The Need for Training Beyond Maximum Likelihood:

Historically, language models were trained solely by optimizing their parameters using maximum likelihood. RLHF breaks free from this traditional approach by introducing reinforcement learning, allowing for the fine-tuning of models and learning an objective function directly from data.

# 2. Data Collection for Instruction and Preference Data:

Two types of human feedback are used in RHLF:

### a. Instruction Data (aka Demonstration Data):
Instruction data comprises pairs of inputs and outputs that demonstrate how the model should behave in response to specific inputs. This data plays a crucial role in training language models to follow desired behavior, and creating high-quality instruction data is a meticulous process.

### b. Preference Data:
Preference data is used for training the reward model, which involves humans ranking different model-generated outputs based on their preferences. This data provides training signals that differentiate between right and wrong behavior.

# 3. Role of Instruction Data in Model Behavior:

### Effectiveness of Instruction Data:
Using instruction data to train language models significantly improves their performance, as demonstrated by comparing models trained with and without instruction data. Well-crafted instruction data is essential for achieving desired model behavior.

### Prompt Alignment with Training Data:
Aligning user prompts with the model's instruction data results in more effective model responses. Prompts similar to the model's training data can reduce trial-and-error in generating desired outputs.

### Addressing Hallucinations:
While being helpful is crucial, it can lead to hallucinations in model responses. Balancing priorities such as truthfulness and harmlessness can mitigate hallucinations.

### Handling Uncertainty in Model Responses:
Incentivizing models to express uncertainty is an important step in reducing hallucinations and improving user safety.

# 4. Three Steps of RLHF:

### Supervised Fine-Tuning (SFT) Initialization:
SFT initializes the language model's weights by training on instruction data. This phase primes the model for reinforcement learning and is crucial for downstream task performance.

### Training the Reward Model:
The reward model is trained to return a scalar representing human preferences for pairs of input and output. The reward model's job is to differentiate between preferred and non-preferred model-generated responses.

### Optimizing a Policy against the Reward Model using RL:
In this step, language model parameters and the policy are jointly optimized to maximize expected rewards from the reward model. This process fine-tunes the model's behavior.

## Let us see in detail How RLHF Works:

### Step 1: Unsupervised Pre-Training

- Start with a pre-trained language model, like the original GPT-3.
- This pre-trained model serves as a next-token predictor.

### Step 2: Supervised Fine-Tuning

- Create a set of commands and pair each command with a human-written response.
- This forms a training dataset of <prompt, ideal generation> pairs.
- Fine-tune the pre-trained model to generate responses that match these human-written responses.

### Step 3: Training a "Human Feedback" Reward Model

- Build a reward model that scores the quality of language model (LLM) outputs for given prompts.
- This reward model can be another language model with its final layers removed, which takes a prompt and generated text as input and produces a scalar reward.
- Collect a new set of commands and generate machine-generated responses to these commands.
- Ask human evaluators to score or rank the quality of these responses.
- Use this dataset to train the reward model, which assigns quality scores to <prompt, generation> pairs.
- The loss function of the reward model is given as follows:

$$ loss(θ) = - \frac{1}{\begin{pmatrix}
    K \\
    2 \\
\end{pmatrix}} E_{(x, y_w, y_l)∼D}[log(σ(r_θ(x,y_w)-r_θ(x,y_l)))]
$$

$x$ = input, question, or prompt

$y_w$ = winning output

$y_l$ = losing output

$K$ = number of outputs

$θ$ = reward model parameters being trained

$r_θ$ = reward score from model (scalar)

### Step 4: Train a Reinforcement Learning Policy

- Train a Reinforcement Learning (RL) policy, essentially an algorithm that generates text, to optimize based on the reward model.
- The RL policy's goal is to generate text that the reward model thinks humans prefer.
- Initially, initialize the RL policy with the fine-tuned LLM from Step 2.
- In a loop:
  - Provide a prompt and use the RL policy to generate an output.
  - Calculate a reward for this generated output using the reward model, simulating how a human would score it.
  - Update the RL policy based on the reward, helping the policy learn to generate better responses.

By following these steps, an RLHF model is trained, resulting in a state-of-the-art large language model, like ChatGPT. This process combines unsupervised pre-training, supervised fine-tuning, and reinforcement learning to enhance language model performance.

# 5. Why RL in LLM?

### a. RL vs. Maximum Likelihood:
RLHF incorporates reinforcement learning as it offers more flexibility compared to traditional maximum likelihood optimization. RL enables the training of not only model parameters but also the reward function and policy.

### b. The Flexibility of RL in Language Model Training:
RL allows for the development of a learnable loss function tailored to the end goal. This flexibility is critical in achieving sophisticated language models.

### c. Formulating RL in RLHF:
RLHF frames the language generation problem as an RL problem, with components like the agent, action space, policy, environment, and reward. This formulation enables models to interact with the reward model for performance improvement.

# 6. The Components of the RL Problem:

- **Agent**: The language model acts as the RL agent, learning to generate text based on the reward system.
- **Action Space**: The action space encompasses all possible language outputs the model can generate.
- **Policy**: The policy is the probability distribution over possible model outputs at each generation step.
- **Environment**: The environment provides feedback to the agent based on a human preference model.
- **Reward**: The reward is a scalar signal indicating the quality of the model's output, as determined by human preferences.
