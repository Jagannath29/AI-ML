````python
import os
import openai
from dotenv import load_dotenv, find_dotenv

load_dotenv()
openai.api_key = os.environ['OPENAI_API_KEY']
````

````python
def get_completion(prompt, model="gpt-3.5-turbo"):
    messages = [{"role": "user", "content": prompt}]
    response = openai.ChatCompletion.create(
        model=model,
        messages=messages,
        temperature=0,
    )
    return response.choices[0].message["content"]
````

### Chatbot

````python
# set the role of the chatbot according to your use case in the system role content then, the user query goes to the user role and the response from the AI goes to the assitant role.

history = [{"role": "system", "content":""},
           {"role": "user", "content": ""},
           {"role": "assistant", "content": ""}
           ]
````

````python
# messages contains the history of the conversation. It is a list of messages between the user and assistant.

def get_response(input):
  history.append({"role": "user", "content": input})
  completion = openai.ChatCompletion.create(
    model="gpt-3.5-turbo",
    messages= history
  )
  response = completion.choices[0].message['content']
  history.append({"role": "assistant", "content": response})
  return response
````

````python
while True:
  prompt = input("Human: ")
  if prompt == 'exit':
    break

  response = get_response(prompt)
  print(f'Response: {response}')
````
