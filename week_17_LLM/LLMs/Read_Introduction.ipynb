# Large Language Model (LLM)

## Introduction


- A type of Artificial Intelligence model that is capable of understanding and generating the human language by being trained on massive text data from various internet sources using deep learning techniques and neural networks.  
- They have a huge number of parameters (in billions) to capture the complex patterns and connections between words in language enabling higher degree of understanding and generation of text.
- They have a wide domain knowledge and very high generalization ability since they are trained on a huge dataset and can perform a wide range of tasks (completion, summarization, translation, sentiment analysis, etc)
- They act as a reasoning engine.

The LLM training process mainly has two steps: pre-training and fine-tuning. In pre-training, the model is trained on a massive corpus of unlabeled text data providing the model ability to predict the next word in a sequence. And during fine-tuning, the model is trained on specific tasks providing the labeled data for the tasks.

### Evolution

1. Early Language Models and Statistical Language Models: \\
The history of language models starts from the time when the field of computer science started where researchers first began to develop rule-based systems to process and generate text. Then, with the advent of statistics, statistical language models emerged, which relied on probabilistic models to estimate the likelihood of word sequences. During this time, n-gram models, which calculate the probability of a word based on the preceding n-1 words, became popular.

2. Neural Networks Language Models: \\
Then, the advent of deep learning and neural networks revolutionized the field of Natural Language Processing (NLP). In 2010, researchers started to use neural networks, specifically Recurrent Neural Network (RNNs) and Long Short-Term Memory (LSTM) networks, for language modeling. These performed better than traditional statistical models but had the limitations of capturing long-range dependencies in text.

3. Transformer Architecture:\\
In 2017, with the paper “Attention is all you need” a Transformer architecture was introduced. This architecture used a self-attention mechanism that significantly improved the ability to capture context in text and enabled parallelization of computation. And, this was the very architecture that formed the foundation for large language models.

4. Emergence of Large Pretrained Models: \\
Based on the architecture of Transformers, many pre-trained models started to emerge as a derivative of this noble architecture. Large pre-trained models like OpenAI’s GPT (Generative Pretrained Transformer) in 2018 and BERT (Bidirectional Encoder Representations from Transformers) in 2019 gained popularity. These models were pre trained on massive amounts of text data and fine-tuned for specific downstream tasks.

Nowadays, a significant trend in the evolution of large language models has been the continuous scaling up of model sizes. Models like ChatGPT, GPT-4, Falcon have huge domain understanding and generate more coherent and relevant texts.

### How do LLMs work?

#### Architecture

Large Language Models (LLM) are a type of neural network that are designed to perform the Natural Language Processing (NLP) tasks. LLMs are based on transformer architecture.

The Transformer architecture relies on the self-attention mechanism, which allows a large language model (LLM) to collectively process all aspects of a text input. This enables the model to assess the significance of individual words in the input, discern connections among words within a sequence, and consequently produce precise and cohesive outputs. Further, The Transformer architecture is highly parallelizable, allowing it to handle numerous data segments at the same time. This capability has paved the way for the creation of larger language models like OpenAI's GPT-3, which has an astonishing 175 billion parameters, as it can process vast volumes of data simultaneously.

LLMs work in two primary stages. They are pre-training and fine tuning.

**Pre-Training:**
In the pre-training phase, the model is trained on an extensive dataset that encompasses a wide variety of text sources from the internet, including books, articles and websites. These steps help the model to learn the patterns of language including grammar, syntax and semantics. This understanding is achieved through unsupervised learning.

The pretraining can be done in multiple ways. For instance, the model could be trained to predict subsequent words for the given sequence of words. For example, GPT models. Also, the model could be trained using the masked language modeling where it needs to guess randomly masked words in a sentence. For example, BERT.

This is the most expensive and time consuming phase in the model building process.

**Fine Tuning:**
In the Fine Tuning phase, the pre-trained model is taken and is further trained on a smaller, task specific dataset. This finetuning is achieved through supervised learning, where the examples are labeled. This procedure typically employs methods like gradient descent and backpropagation to adjust the model's parameters and enhance its performance on the given task.


#### How do language models perform next word prediction?


Nowadays, LLM are mainly used for generation purposes where the next token is predicted based on the previous tokens. Using the working of Transformer architecture, the decoder plays an important role in generating the next word. It does so by using the information encoded in the input sequence and the previously generated words in the output sequence in an auto-regressive manner. Let's see how that is achieved.

At first, the input text or prompt is tokenized and converted into the embedding for the vector representation of the text. Our input tokens undergo a couple of encoding steps: they’re encoded using an Embedding layer, followed by a Positional Encoding layer, and then the two encodings are added together. Next, our encoded inputs go through a sequence of N decoding steps, followed by a normalization layer. And finally, we send our decoded data through a linear layer and a softmax, ending up with a probability distribution that we can use to select the next token.


image.png

The decoder generates text one word at a time. It applies a self-attention mechanism to the words generated so far in the output sequence. The decoder uses feedforward neural networks to process the attended information and make predictions for the next word. The Linear layer takes the vector generated by the stack of decoders and transforms it into a much larger vector known as a logits vector. The size of the logit vector will be equal to the vocabulary learned from the training dataset. This would make the logits vector with vocabulary size width with each cell corresponding to the score of a unique word.

Next, the softmax layer takes these scores and converts them into probabilities, ensuring they are all positive and sum up to 1.0. The cell with the highest probability is chosen, and the word associated with it is produced as the output for the particular time step.

Then, the input to the decoder will be updated concatenating the predicted word which becomes the context for generating the next word in the sequence. The decoder repeats the autoregressive generation process for each word in the output sequence until a stopping criterion is met (e.g., a maximum sequence length or an end-of-sentence token).

By repeatedly attending to and processing the input sequence and the previously generated words, the decoder helps in generating coherent and contextually relevant text.

### Limitations of LLMs

1. Hallucination: LLMs can generate text that is factually incorrect or nonsensical with confidence because in some cases the data they are trained on is often incomplete or contradictory. Also, if enough context and constraints are not provided, the model tends to generate the content that is irrelevant to the input data.

2. Knowledge cut-off: The models are trained on a dataset till a certain point of time. This lacks the model to have the updated information. For example, ChatGPT has a knowledge cutoff data of September 2021, and if we ask about anything that has been discovered or updated after that time then, the model will not  be able to answer that question.

3. Biasness: LLMs can be biased in their response because of the training data that is littered with inconsistencies, inaccuracies, and discriminatory bias. For example, the real-world dataset may contain the gender and racial bias and when the language models are trained on them, then LLM perpetuates these biases in its predictions.

4. Reasoning Errors: LLMs fail to give the correct result to the queries that require reasoning and logical thinking. They could be solving the complex reasoning math problem or problem that requires many steps to reach the solution. A lot of research is going on in this. Chain-of-Thought (CoT) is one prompt technique that can handle this to some extent.

5. Prompt Injection: When someone bypasses filters or manipulates the LLM using carefully crafted prompts making models to ignore previous instructions and perform unintended actions, it is called as prompt injection.This opens LLMs to the world of security exploits. There are various techniques to prompt injections like: Jailbreaking, obfuscation, code injection, Prompt leakage, etc.

A lot of research and work are being conducted to handle the limitations of LLM and develop the area of making a responsible AI.
