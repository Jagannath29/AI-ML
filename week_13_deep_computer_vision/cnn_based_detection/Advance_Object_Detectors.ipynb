## Two Phase detectors Vs Single Phase detectors

State Of The Art(SOTA) object detectors are broadly divided into two categories. Two-Phase detectors and Single Phase detectors.

As the name suggests, Two-Phase detectors process the entire object detection pipeline in two phases. Firstly, it generates regions of interest from the input image then secondly feeds the proposed region for object classification and bounding box coordinates prediction( regression ). Faster R-CNN (Region-based Convolutional Neural Networks) is an example of two-phase object detectors. Such detector algorithms achieve high accuracy but are slower.

In Contrast to Two-phase detectors, Single-Phase detectors look at an entire image at once, feeding through Convolution Neural Network at the same time predicting multiple bounding boxes coordinate and class probabilities for the located object.YOLO (You Only Look Once) is an example of Single-phase object detectors. Single-Phase detectors are Real-time detectors with compromised accuracy compared to Two-phase detectors.

## Two-Phase Detectors : Faster RCNN

The main problem in object detection, using simple architecture like Convolution Neural Network as backbone network with Regression and Classification head attached at the top is, if we have an image consisting of more than one object to locate and classify, it fails. As mentioned earlier, simple architecture works for images where a single object is to be located and classified. Computer-Vision researchers have proposed different methods to tackle the problem faced on detecting multiple objects at a single image. Here, we will be discussing the method proposed in late 2015 by [(Ross Girshick)](https://www.rossgirshick.info/) and his team, named [Faster R-CNN: Towards Real-Time Object
Detection with Region Proposal Networks](https://arxiv.org/pdf/1506.01497.pdf).

Faster-RCNN falls under the Two-Phase Detectors category, so the entire workflow is divided into two tasks:
- First, finding the region of interest in the entire image where an object can be located.
- Processing such obtained region independently to locate and classify objects of interest in an image.

For Two-phase object detectors, we need first to extract region proposals (i,e region in the image where an object can be located), which gives a rough idea where the objects can be located in the input image. This is done by the network named Region Proposal Network (RPN) in Faster-RCNN. The RPN plays as an attention mechanism for another network, giving another network hints of where to pay attention or focus.

The second network, sometimes called Detector Network or Fast RCNN, uses the region proposals calculated by the RPN network to select or pay attention to the feature map obtained from Convolution layers corresponding to that region. Then some special pooling operation named ROI pooling is performed to produce a fixed-shaped output and feedforward to generate the final class and bounding box.





## Single-Phase Detectors : YOLO - You Only Look Once

YOLO is State Of The Art(SOTA) object detectors with real-time object detection performance. In contrast to Faster RCNN, a single Convolutional Neural Network simultaneously predicts multiple bounding boxes and class probabilities associated with the object. So, it's named You Only Look Once. Since YOLO is a single shot, it's comparatively faster than Faster RCNN but falls back in accuracy. YOLO trains on full images and directly optimizes detection performance.

YOLO  divides the input image into an $S × S $ grid where each of the cells in the grid predicts $ B $ bounding boxes, a confidence score that measures the certainty of object presence and $ C $ class probabilities is like a regular classifier that outputs probability distribution associate with the classes. An example of $7 \times 7$ map in the input image is shown in figure 2 (left). A color map shown at the bottom corresponds to classes predicted. Figure 2 (top) shows the predicted bounding boxes where the box's intensity represents the confidence for the boxes. Since the Algorithm predicts multiple bounding boxes for an object, those predicted bounding boxes are feed to Non-Max Suppression, which filters the unwanted bounding boxes using some threshold value and finally outputs the final detection regions(right).

## Key Takeaways

- Object detectors are broadly divided into Two-Phase detectors and Single-Phase detectors. The former finds the region where an object can be located and processes such a proposed region to locate and classify multiple objects within the image. In contrast, the latter passes an input image in a convolution network to predicts multiple bounding boxes and class probabilities for those boxes in a single shot.

- The faster-RCNN consists of an RPN network to generate a region where the image can be located. The detector network generates the final class and bounding box for the proposed region.

- The RPN plays as an attention mechanism for another network, giving another network hints of where to pay attention or focus.

- YOLO uses a single convolutional neural network that simultaneously predicts multiple bounding boxes and class probabilities associated with the object.
