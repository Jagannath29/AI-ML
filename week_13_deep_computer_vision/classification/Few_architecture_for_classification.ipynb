## **VGG**

VGG is a Convolution Neural Network architecture, it was proposed by Karen Simonyan and Andrew Zisserman of Oxford Robotics Institute in 2014. VGG stands for Visual Geometry Group. It was submitted to the Large Scale Visual Recognition Challenge 2014(ILSVRC 2014). ILSVRC is the challenge contest run by ImageNet. VGG model achieves 92.7 % top-5 test accuracy in ImageNet.which is a dataset of over 14 million images belonging to 1000 classes. The key point of VGG is the evaluation of networks of increasing depth using an architecture with a very small ($3 \times 3$) convolution filter. VGG is composed by small components called VGG blocks.

### **VGG Block**

The basic building block of VGG networks is a sequence of convolution and max-polling layers. VGG block uses $3 \times 3$  filters at stride one, pad one for the convolution layer and $2 \times 2$ filters at stride two for max polling layer as shown figure 1 below.

## **ResNet**



In the VGG net, we have seen nothing but an increasing number of layers with same small size of filter in the design and achieving better performance. But with the network depth increasing, accuracy gets saturated (which might be unsurprising) and then degrades rapidly.ResNet addressed this problem using skip connections(a.k.a  shortcut connections, residuals) while building deeper models.

Before diving into the ResNet, let us understand what is skip connection/residual block.

### **Residual Block**




Residual block insert shortcut connection, skips that contain nonlinearities (ReLU) and batch normalization in between. Shortcut connection simply perform identity mapping, and their outputs are added to the outputs of stack layers. Identity shortcut connections add neither extra parameter nor computational complexity.

The degradation problem is address by deep residual network as shown figure 3 above.In-
stead  of  hoping  each  few  stacked  layers  directly  fit  a
desired  underlying  mapping,  we  explicitly  let  these  layers fit a residual mapping.

Formally, denoting the desire underlying mapping as $H(x)$, we let the stacked nonlinear-layers fit another mapping of $f(x)=H(x)-x$. The original mapping becomes $f(x)+x$, which is easier to optimize the residual mapping than to optimize the original mapping.

### **ResNet Architecture**

ResNet presents a residual learning framework to ease the training of networks that are substantially deeper than those used previously. ResNet explicitly reformulates the layers as learning residual functions with reference to the layer inputs, instead of learning unreferenced function. ResNet provides comprehensive empirical evidence showing that these residual networks are easier to optimize and can gain accuracy from considerably increased depth. On the ImageNet dataset, this method was evaluated with residual nets with a depth of up to 152 layers-8$\times$ deeper than VGG nets but still having lower complexity.
