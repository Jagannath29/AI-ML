# Introduction to Transfer Learning
Before diving into the technical part, let's understand Transfer Learning in a laymens' term. To put things simply, Transfer Learning just transferring the knowlegde gained by one network to another network.

In deep learning, "Knowledge" refers to the weights that the neurons aquired during the trainig process. You have already learned about the weights in the nueurons. here we use the pre-learned weights on a new network for the task in hand.

When we want to develop a working model using transfer Learning, we simply use a pre-trained model developed for some task and repurpose that model as a starting model for our required task.

We use Pre-trained model as a starting model or starting point mostly on Computer Vision(CV), Natural Language Processing(NLP). It is because developing a neural network model for such task requires a vast amount of time and resources. The knowlegde acquired by the pre-trained models can be reused into the another model hence the name "Transfer Learning" is derived. The data generalization that the network has already learnd gives us huge leap in skill that they provide on related problems.



## Steps in Transfer Learning
Now, let's discuss how we develop a model using **Transfer Learning** technique. Firstly we take a model that has already been trained, and then we repurpose the learned features (or transfer them) to a second target network to be trained on a target dataset and task.
<!-- this method works only when features are general, meaning suitable to both base and target tasks, instead of specific to the base task. -->

As mentioned before, while adapting the transfer learning process, we stack the custom layers on the top of the pre-trained layer. and what we do is train the custom layer. It is not compulsory that we need to train only the custom layer by freezing the pre-trained layer. we can also re-train the pre-trained layer. The decision for doing so depends upon the type of problem we are trying to solve and the architecture of the network.

Here upto this point we discussed only training the final few layers of the new model. but there are few different ways we can train our network. Basicly we can train our new model by three different ways:
1. Training only the last layers:  
 Here we train only the few of the last layers
2. Training layers in between  
 Here we train the in between layers in our model, or may be some blocks.
3. Training only similar types of layers:
 Here we train only the similar types of layer in our model.(for eg: batch norm in multiple levels)

## When to use Transfer Learning

One important thing to know before adopting Transfer Learning is that we can use this method can only when the model features learned from the first task in general i.e. model was trained previously with task that is somehow similar with the task that we want to accomplish. for example, Consider we want to classify cars but we have a pretrained model of buses classifier. If we were to make cars classifier we have to make a model from scratch and train it with collosol amount of data. instead what we can do is we make our cars classifier on the top of buses classifier. since the cars and buses share similar features like wheels, Silhouette, stance, etc, we can use pretrained model as a feature extractor, Since the idea is to reuse the common aspect

If we wish to make use of transfer learning, we have a have a lot of pre-trained model that are available to us. few of the popular model for the image data are VGG model, Inception Net, Resnet, etc.

If we have to develop the complex model like VGG model and others mentioned above, it might take days or even months to train ad lets not forget the computing power required. so we can take the advatage of the transfer learning and develop our required model in the fraction of the time it takes to develop those models form scratch.  



## Benefits of Transfer Learning

Now lets talk about the benefits of the transfer learning,

1. Higher Start  
  The initial skill on the cource model is higher than it otherwise would be
1. Higher scope  
  The rate of improvement of skill during training of the source model is steeper than it otherwosw would be
1. Higher asymptote:  
  The convergeed skill of the trained models is better than its otherwise would be.
1. Transfer Learning can be used to train model if not a lot of dataset is available.
