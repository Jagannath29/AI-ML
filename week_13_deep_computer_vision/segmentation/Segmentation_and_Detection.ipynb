# Segmentation and Detection

In this chapter, we will discuss about instance segmentation. Instance segmentation lies at the intersection of object detection and semantic segmentation. It involves finding each instance of the objects we are interested in, which resembles object detection. But instead of a bounding box for each object, we are interested in a mask for each object. This mask gives us the silhouette of the object in the image.

We can break down instance segmentation into three tasks:

1. Find the regions in the image which may contain objects.

2. In each of the Region of Interest from step 1, classify the object in the region.

3. In each of the RoI from step 1, estimate a binary mask for the object.

## Mask R-CNN

Mask RCNN is a variant in the RCNN family of architectures used in object detection. It is an extension to the Faster RCNN model that adds a branch that also predicts the mask for each detected object. Mask R-CNN uses an FCN that segments the main object in each region.

Lets first take a look at the Faster R-CNN architecture.

Note the three distinct steps in the Faster RCNN workflow. First, we take an image and pass it through a convolutional neural network to get a high-level feature map of the image. Then, we use that feature map to find out interesting regions in the image that might contain objects. Then, we pass each of these regions through a classifier that predicts the class of the object in the region.

Mask RCNN builds off of Faster RCNN, adding a couple of modifications to help in segmentation.

<!-- ![image.png](https://docs.google.com/uc?export=download&id=1BbBcbR2SZDTGpewbt5v5xVdjK4chF15f) -->
![image.png](https://i.postimg.cc/bvY8rDJn/image.png)

<center>
    Figure 3: Mask RCNN Architecture
</center>

The first one is that it replaces the ROI pooling layer in Faster RCNN with an RoI align layer. Segmentation masks need to be very precise, and it turns out that RoI pooling involves quantizing values, which reduces the precision of the information in each layer. In contrast, RoI-Align uses bilinear interpolation to create feature maps for each region.

The second one is that it adds a fully convolutional network, which takes each of the feature maps and outputs a mask that separates the object in the feature map. We take the feature map for each region proposal, and feed each of them through a network that predicts the class of the object in the region, and the mask of the object. As with Faster RCNN, the network may also predict a bounding box that is more precise than the one given by the region proposal network.

## Panoptic Segmentation

So far, we have looked at semantic and instance segmentation. It turns out that the third segmentation problem that we discussed, panoptic segmentation, can be done by combining instance and panoptic segmentation. Let's see how we can do that.



A simple but helpful way of thinking about instance and semantic segmentation is that instance segmentation tries to segment things, and semantic segmentation tries to segment stuff. 'Things' includes countable entities like people, cats, and trees. On the other hand, 'stuff' includes uncountable entities, such as water, grass, and the sky.

When you think about it, most real-world images contain entities that are either 'stuff', or 'things.' Combining instance segmentation and semantic segmentation should give us information about what class each of the pixels belongs to, and which object belongs to which instance.

To combine results from instance segmentation with semantic segmentation, we can use the pixel-wise labels assigned by semantic segmentation to the pixels that do not belong to any instances of entities. That is, we can start with the outputs from instance segmentation. Then, for each pixel that does not exist in any object masks, we get its label from semantic segmentation.

## Bounding Box from Instance Segmentation

It is easy to recover the bounding box for each instance of object from its mask. You can simply threshold the mask at some confidence value between 0 and 1, which should give you a binary mask. Then, you can simple take the minimum and maximum x and y values of the mask as the top left and bottom right vertices of the bounding box.

Most implementations of Mask R-CNN also contain the bounding box regressor stage from Faster RCNN, so if you are using Mask R-CNN, you get the bounding boxes from the model too.
