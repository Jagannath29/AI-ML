## Segmentation

Segmentation algorithms divide an image into a set of pixels or regions. The primary purpose of dividing is to understand better what the image represents.

Segmentation is a handy tool in any Computer Vision practitioner's toolbox. In the last few years, there have been breakthroughs in segmentation models due to Deep Learning. Naturally, a lot of research effort over the previous few years has been focused on building better segmentation models.
We will begin by discussing the semantic segmentation network:




## Semantic Segmentation Network

Semantic segmentation is a pixel-wise classification task.

The simple classifier predicts the class of given data points by generating a `one-hot vector` as output.
In semantic segmentation, the classification runs simultaneously with the number of pixels of input image.For each pixels, it generates its `one-hot vector` prediction with the dimension equal to the number of object categories of interest or the product of image Height and Width. (`H*W`)

## Segmentation using FCNs

Finally, we understood the concept of semantic segmentation; we can now introduce a neural network pixel-wise classifier.


It turns out that we can use Fully Convolutional Networks(FCNs) to perform pixel-level semantic segmentation. The main idea of FCNs is to use multiple scales of feature maps in generating the final prediction. First, let's take a look at what FCNs are.

### Fully Convolutional Networks

Fully convolutional networks are similar to convolutional neural networks, with one crucial distinction. The CNN contains a few dense layers at their ends, fully convolutional networks(FCNs), as their name suggests, do not include the final dense layers.

A fully convolutional network outputs a volume of shape `(HxWxC)`, where `H` is the height, `W ` is its width, and `C` is the number of channels in the volume.

Segmentation is a per-pixel classification task. That is, we want our model to classify every single pixel into a class. In conventional neural networks, we would use an output vector representing the probabilities of the input data belonging to each of the categories(Class). Similarly, we now need a vector for each pixel containing the pixel's possibilities belonging to a class.

Now, because we have `HxW` pixels, we need `HxW` vectors that each represent the classification probabilities for one pixel. We can arrange these vectors into a volume of shape `(H, W, C)`, where `C` is the number of classes we have. FCNs can be naturally used for this task because they output volumes.



### Using  FCNs for segmentation

In this section, we will take a look at a very simple FCN architecture that was published in the paper [Fully Convolutional Networks for Semantic Segmentation](https://arxiv.org/abs/1411.4038). In fact, it was the first work to try to use Fully Convolutional Networks for segmentation tasks. The figure below shows the architecture of this network.

In the architecture given in the image above, each convolutional layer extracts features from its inputs and outputs a spatially smaller feature map. In the end, we get the network to output a feature map with 21 channels. These 21 channels correspond to the 21 classes in the dataset that the authors used. The softmax function is applied to every single pixel, which gives us the classification for each pixel.

### FIGURE

In FCN, as shown as the above figure the FC layer is turn to $1*1$ convolutional layers. Due to max pooling, we get the size of output as smaller than the input size. And if we upsample the $1*1$  then we can calculate the pixelwise output i.e., the label map.

Deconvolution is the process of changing the small size input to the larger output size. It is also known as up convolution or transposed convolution.The main goal of upsampling via deconvolution  in FCN is to recreate the label of input image.  All these three FCN architecture follows the similar downsampling path but are differ in upsamplig paths.

- FCN-32: The segmentation map is produced directly form conv7 followed by  deconvolution.

- FCN-16:Sums the 2x upsampled prediction from conv7 (using a transposed convolution with stride 2) with pool4 and then produces the segmentation map, by using a transposed convolution layer with stride 16 on top of that.

- FCN-8: Sums the 2x upsampled conv7 (with a stride 2 transposed convolution) with pool4, upsamples them with a stride 2 transposed convolution and sums them with pool3, and applies a transposed convolution layer with stride 8 on the resulting feature maps to obtain the segmentation map.

## State of the art architectures

Different state-of-the-art approaches further enhance the semantic segmentation network. Let's view a few of them,

**1. U-Net:**

First insight, it is `U` Shaped architecture because of it's symmetricity. The U-Net architecture consists of two primary path- contracting path /downsampling path (to capture the context of the image) and the expanding path/ Upsampling path (to enable precise localization using transposed convolutions).



<!-- ![](http://drive.google.com/uc?export=view&id=1Bk_2oYtVaWqMnm4FrYsGhbCNIDjdbMNR) -->
![](https://i.postimg.cc/MGV0nNfg/image.png)

<center>

Figure 4: Unet Architecture([Source](https://arxiv.org/pdf/1505.04597.pdf))

</center>

U-Net architecture is an end to end FCN and modified in a way that yields better segmentation in medical imaging. The key points of U-Net which are different than FCN variants are;

- U-Net is Symmetric nature. So, consist of a network with a large number of feature map in the upsampling path to transfer information.

- Skip connection between downsampling and upsampling path; helps provide local information to global information while upsampling.

You can go through the [Paper](https://arxiv.org/pdf/1505.04597.pdf) for more detail.




**2. Pyramid Scene Parsing Network (PSPNet):**

PSPNet highly focuses on global information for better performance. As the name
**Scene Parsing** defines the category label of all pixels within the image.

The figure below shows the architecture overview of PSPNet. We passed the input Image using CNN to get the feature map of the last convolutional layer. Then a pyramid parsing module is applied to extract different sub-region representations followed by upsampling and concatenation layers. Finally, we passed through the convolution layer to get the per-pixel prediction.


<!-- ![](http://drive.google.com/uc?export=view&id=10FaIq7uHgHML7cvNhLJQQzcel1DWSwOg) -->
![](https://i.postimg.cc/YSdrzqFF/image.png)

<center>

Figure 5 : PSPNet Architecture([Source](https://arxiv.org/pdf/1612.01105.pdf))

</center>

The key points of PSPNet which are different than FCN variants are:

- PSPNet uses co-occurrent visual patterns to collect the contextual information and helps to decrease the chance of miss classification.

- It considers the relationship between confusing classes like mountain and hill; Wall, house, and building.


- It provides more attention to sub-region and helps to improve performance. E.g., a small object like streetlight and signboard is hard to find, which is essential.

Go through the paper, [PSPNet](https://arxiv.org/pdf/1612.01105.pdf) for more detail.

DeepLab uses dilated Convolution and Fully Connected. The atrous i.e dilated convolution is applied at Encoder, i.e, downsampling to extract contextual information and the decoder, i.e., Upsampling decodes segments pixel prediction. The key difference point is ;

- The depthwise convolution with striding is used instead of max pooling operations.

- After each 3 x 3 depthwise convolution, new batch normalization and ReLU activation are added.



For more details, go through the [Paper](https://arxiv.org/pdf/1802.02611.pdf).

## Key Takeways

- Segmentation algorithms divide an image into a set of pixels or regions.

- Semantic segmentation classifies each pixels.

-  In semantic segmentation, the classification runs simultaneously with the number of pixels of input image.

- CNNs consist few dense layer at it's end whereas FCNs doesn't.

- There are other state of art architecture for semantic segmentation like Unet, PSPNet, and DeepLab.
