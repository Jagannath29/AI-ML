## **What is  CNN?**

CNN is the class of deep neural networks that uses at least a convolution layer. These convolution layers are a particular type of linear operation which uses convolution operation instead of the traditional matrix multiplication.CNN is also known as a "ConvNet".


## **Architecture of CNN**

A CNN usually takes an order three tensor as its input, e.g., an input image with H rows, W columns, and three channels (R, G, B color channels). Higher-order tensor inputs, however, can be similarly handled by CNN. The input then sequentially goes through several processes. One processing step is called a layer, a convolution layer, a pooling layer, an activation layer, a loss layer, a fully connected layer, etc.

We will learn the details of these layers in the next chapter. For now, let us look at an abstract architecture of CNN first.

The above figure 3 shows how a CNN runs layer by layer in a forward pass. The input image order three tensors. It goes through the processing in the first layer, which is the first box. The output of the first layer acts as the input to the second layer processing and so on.


CNN is formed by the combination of feature learning and classification components. Feature learning components contain convolution, pooling, and activation layers. The classification component contains a fully connected layer.
