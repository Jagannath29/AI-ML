## **Convolutional layer**

In literature, the implementation is cross-correlation but is called convolution. We had already learned convolution and cross-correlation in module three. In this chapter, we follow this convention of calling both operations convolution. If you are confused or any problem in convolution, you can go through module three.

The convolution layer is an essential building block of a CNN. The layer's parameters consist of a set of learnable filters. Each filter is convolved during the forward pass in deep neural networks across the width and height of input volume. During the backward pass, the network learns filters.

General steps for convolution operation (talking for a 2D convolution, with a 3-dimensional input),
* Take an input volume
* Applies a kernel(a.k.a filter) at every position of the input
* Output another volume

## **Relationship between inputs, filters and outputs**

The main goal of convolution is feature extraction from the give image data and reduce model parameters. In this section, we will learn how convolution reduces model parameters by establishing the relationship between the inputs, filters, and outputs.

The following quantities are taken into account while describing the relationship between the inputs, filters, and outputs.

* Input volume of size $w_1 \times h_1 \times d_1$

* Hyperparameters:

    * Number of filters $k$,
    * The spatial extent of filter $f$(the depth of each filter is the same as the depth of each input: Convolution over Volume)
    * The stride $s$
    * The number of zero paddings $p$
* Output volume of size $w_2 \times h_2 \times d_2$

Note: Do not worry about the filter, padding; we will learn this topic step by step below.

Stride controls how the kernel convolves around the input area. For an input volume of $w_1 \times h_1 \times d_1$:

Output volume after stride with convolution is calculated as,

* $w_2 = \frac{h_1-f}{s} + 1$
* $d_2$ remains unchanged



### **Padding**

Generally, kernels have width and height greater than one. We get considerably smaller outputs after applying successive convolution than our input image and obliterating exciting information on the original image's boundaries. Padding is the tool for handling this issue.

Padding is adding layers of pixels to an image when it is being processed by the kernel of a CNN to avoid the above issue.

#### **Effect of padding on output size**

## **Pooling layer**

The Pooling layer reduces the height and width of the input. It helps reduce computation and helps make feature detectors more invariant to its position in the input. The two types of pooling layers are:


* Max-pooling layer: slides a window over the input and stores the window's max value in the output.

* Average-pooling layer: slides a window over the input and stores the window's average value in the output.



Consider the following example:

A $2 \times 2$  pooling with a stride of $2$ is applied to the input. In max pooling, the max value from each region is picked up, and on average pooling, the average value of each region is computed.

For example, in the first region, which is marked blue, the max value, i.e., 9 is picked.

> $max(1, 0, 9, 4) = 9$


Then a stride of 2 passes the operation to the next region, marked yellow in the figure. The max value here is 8.

> $max(2, 3, 6, 8) = 8$

Similary, the average value of the blue region is 3.5,

> $avg(1, 0, 9, 4) = \frac{1+0+9+4}{4} = 3.5$

For the yellow region, the average value is 4.75.

> $avg(2, 3, 6, 8) = \frac{2+3+6+8}{4} = 4.75$


The intuition behind using the max pool is that the highest values characterize the essential features. So, if an important feature has been detected in the input region, the max pool operations keep a high value.



#### **Effect of pooling on output size**

**In general,**

For an input volume of $w_1 \times h_1 \times d_1$:


  * $w_2 = \frac{w_1 - f}{s} + 1$

  * $h_2 = \frac{h_1 - f}{s} + 1$

  * $d2$ remains unchanged

So the final output volume after pooling is $w_2 \times h_2 \times d_1$
