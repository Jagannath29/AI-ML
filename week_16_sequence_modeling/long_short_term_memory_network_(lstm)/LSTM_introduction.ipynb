## Overview: Recurrent Neural Networks (RNNs) and the Vanishing Gradient Problem

### 1. Recurrent Neural Networks (RNNs)
When people think, they build on what came before. As you follow a story or a conversation, each new piece of information makes sense only because you remember what was said earlier. You don’t reset your mind every second—you carry context forward.

Standard neural networks, however, don’t naturally have this ability. They process inputs independently, without remembering what came before. This makes them less suited for tasks where sequence and history matter—like understanding language, tracking a melody, or analyzing events unfolding over time.

Recurrent neural networks (RNNs) were designed to overcome this limitation. They include feedback connections that let information from previous steps influence future steps, giving the network a kind of memory. This persistence of information allows RNNs to model sequences and contexts much more effectively than traditional feedforward networks.

In the figure above, we see the basic structure of a **Recurrent Neural Network (RNN)**.



#### Loop and General Memory State
- On the **left side**, the RNN cell is shown with a loop.  
- This loop represents how the network passes information from one step to the next.  
- The **hidden state** (`h`) acts like a memory, storing information about what has been seen so far.  
- At each step, the hidden state is updated based on:
  - The current input (`x_t`)  
  - The previous hidden state (`h_{t-1}`)  

This allows the RNN to "remember" context across time instead of starting from scratch at each input.



#### Unrolling Through Time
- On the **right side**, the loop is "unrolled" to show the RNN processing a sequence of inputs `x_{t-1}, x_t, x_{t+1}, ...`).  
- At each time step:
  - The input (`x_t`) and the previous memory (`h_{t-1}`) combine to form a new hidden state (`h_t`).  
  - This hidden state is then used to generate an output (`O_t`).  

At each time step *t*, the RNN updates its hidden state and produces an output using the following equations:

$
h_t = f(W h_{t-1} + U x_t + b_h)
$

$
O_t = g(Vh_t + b_o)
$


#### Where:
- \(h_t\): hidden state (memory) at time *t*  
- \(x_t\): input at time *t*  
- \(O_t\): output at time *t*  

#### Weight matrices:
- \(U\): input → hidden weights  
- \(W\): hidden → hidden (recurrent) weights  
- \(V\): hidden → output weights  

#### Other terms:
- \$(b_h, b_o\$): bias terms  
- \$(f\$): non-linear activation (e.g., tanh, ReLU)  
- \$(g\$): output activation (e.g., softmax)  
####  Weight Matrices
- **\(U\)** : Input-to-hidden weights  
  - Controls how the current input \$(x_t\$) influences the hidden state.  

- **\(W\)** : Hidden-to-hidden (recurrent) weights  
  - Controls how the previous hidden state \$(h_{t-1}\$) affects the current hidden state \$(h_t\$).  
  - This is what gives the RNN its ability to "remember" past information.  

- **\(V\)** : Hidden-to-output weights  
  - Maps the hidden state \$(h_t\$) to the output \$(O_t\$).  
#### Key Idea
- The hidden state carries information forward, giving the RNN a kind of **memory**.  
- This makes RNNs well-suited for sequential tasks like language modeling, speech recognition, or time-series prediction.




### 2. Vanishing Gradient Problem
RNNs are trained using **Backpropagation Through Time (BPTT)**, where the gradients are propagated backward across many time steps.  
During this process, the gradient at each step is multiplied by the recurrent weights:

$$
\frac{\partial L}{\partial W} \approx \prod_{t=1}^{T} \frac{\partial h_t}{\partial h_{t-1}}
$$


If $\left| \frac{\partial h_t}{\partial h_{t-1}} \right| < 1$, then the gradient shrinks exponentially.


the gradient **shrinks exponentially** as \(T\) increases which makes it
difficult for the network to learn **long-term dependencies**. Also early time-step information has little to no effect on weight updates which leads to slow convergence or complete failure in capturing distant context.


RNNs are powerful for sequential modeling, but their reliance on repeated gradient multiplication causes the **vanishing gradient problem**, making it hard to learn patterns spanning long time intervals. This limitation motivated the development of architectures like **LSTMs** and **GRUs** that can preserve information over longer sequences.




**Why LSTM?**

One of the major drawbacks of a Recurrent Neural Network (RNN) is that it struggles to retain information when processing sequences with a large number of time steps. This happens because RNNs suffer from the vanishing gradient problem during training — as the sequence length grows, the gradients shrink exponentially, making it difficult for the network to learn long-term dependencies.

**How LSTM Solves This ?**

***Long Short-Term Memory (LSTM)*** networks were specifically designed to address this issue.

- LSTMs introduce a cell state that can carry information across many time steps with minimal loss.

- They use gates (input, forget, and output) to control what information is stored, discarded, or output at each step.

- This mechanism allows LSTMs to retain important context for long durations, making them highly effective for sequences with long-term dependencies.

**RNN vs. LSTM in Practice**

Think of LSTM as an **improved version of RNN**, where the simple recurrent unit is replaced with a more sophisticated memory cell that can decide:

- **What to remember**

- **What to forget**

- **When to output** information



## Introduction to Long Short-Term Memory (LSTM)

**Long Short-Term Memory (LSTM)** networks, They were introduced by [Hochreiter & Schmidhuber (1997)](https://www.bioinf.jku.at/publications/older/2604.pdf), and were refined and popularized by many people in following work.
They work tremendously well on a large variety of problems, and are now widely used. RNNs are a specialized type of recurrent neural network (RNN) designed to overcome the limitations of vanilla RNNs, particularly the **vanishing gradient problem**.  

In traditional RNNs, gradients tend to **exponentially decay** during **Backpropagation Through Time (BPTT)**, making it difficult to learn dependencies that span more than a few time steps. This severely limits their ability to capture long-term patterns in sequential data.  LSTMs resemble standard recurrent neural networks but here each ordinary recurrent node is replaced by a memory cell. Each memory cell contains an internal state, i.e., a node with a self-connected recurrent edge of fixed weight 1, ensuring that the gradient can pass across many time steps without vanishing or exploding.

All recurrent neural networks have the form of a chain of repeating modules of neural network. In standard RNNs, this repeating module will have a very simple structure, such as a single tanh layer.
