Recurrent Neural Networks (RNNs) are powerful for modeling sequential data, such as time series.
However, when applied to long or complex time series tasks, they suffer from important limitations.
In this notebook, we will explore these limitations and understand why more advanced models (e.g., LSTMs, GRUs, Transformers) were introduced.





## 1. Vanishing and Exploding Gradients
---

RNNs are trained using **Backpropagation Through Time (BPTT)**. During training, gradients are propagated backward across many time steps.

- **Vanishing gradients**: When time series are long, repeated multiplications by values < 1 cause gradients to shrink toward zero. This makes the network unable to learn long-term dependencies.

- **Exploding gradients**: Conversely, when weights > 1, gradients can blow up exponentially. This destabilizes training.

Formula for gradient across time steps:

  $$ ∂L/∂W ≈ Π (∂h_t / ∂h_{t-1}) $$

Where $Π$ denotes multiplication across time steps.

If $|∂h_t / ∂h_{t-1}| < 1$ the gradients are vanishing.

If $|∂h_t / ∂h_{t-1}| > 1$ the gradients are exploding.

Due to the vanishing (and exploding) gradient problems, RNNs are only able to learn recent patterns (last few time steps), high-frequency variations and immediate cause-effect relationships.

Long-term temporal dependencies like seasonal patterns spanning months/years, long-term trends, distant but relevant events, etc. are lost by RNNs. This can be mitigated by the use of LSTMs, but even LSTMs have their own issues.

## 2. Long-Term Dependency Challenge

Due to the RNN's struggle with vanishing gradient problems, thye have difficulties with long-term dependencies. RNNs are inherently biased toward **recent time steps** because each new hidden state is a function of the previous one.

Like we learned in the pooint before, RNNs excel at short-term forecasting, yet fail to remember events from hundreds of steps ago (e.g., yearly seasonality).

Example: Predicting electricity demand might depend on both recent days and annual cycles — standard RNNs struggle to capture the annual cycle.


## 3. Memory Bottleneck

The RNN hidden state is a **fixed-size vector** summarizing the entire past. This creates a **compression bottleneck**, as rich histories are forced into a limited capacity. The more complex the time series, the more information is lost. Time series problems are too complex and have too many different variables for RNNs to be able to give due importance to each variable.

Formula: $$h_t = f(W_h h_{t-1} + W_x x_t)$$

Here, $h_t$ is a single vector at each time step, regardless of how much past information exists. For time series data, which has different importance of variables depending on time, and which has dynamic interactions between variables, this method does not yield satisfactory results.

## 4. Computational Inefficiency

- RNNs as well as LSTMs are inherently **sequential**: each hidden state $h_t$ depends on $h_{t-1}$.

- This prevents parallelization across time steps during training.

- For long time series, this results in slow training and high computational cost.

Example: Training a transformer on a GPU can process many time steps in parallel, but an RNN cannot.

## 5. Difficulty with Irregular Sampling & Missing Data

Many real-world time series are **irregularly sampled** or contain missing values.

- RNNs as well as LSTMs assume fixed step sizes between inputs.
- Handling irregular data often requires preprocessing or additional mechanisms.

 Example: Patient medical records with irregular hospital visits.

## Solutions

These limitations of RNNs (and LSTMs) can be overcome with the help of Transformers, which leverage attention to better handle time series forecasting. We will learn more about them in the next notebook.
