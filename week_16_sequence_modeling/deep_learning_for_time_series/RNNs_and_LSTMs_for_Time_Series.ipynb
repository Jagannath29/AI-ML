# RNNs and LSTMs for Time Series

## Learning Objectives
- Understand how RNNs and LSTMs are applied to time series forecasting.  
- Learn their strengths and limitations in modeling temporal dependencies in time series data.

---

## Prerequisites
Before starting, you should be comfortable with:
- Classical time series forecasting (e.g., ARIMA, moving averages).  
- The basics of Recurrent Neural Networks (RNNs) and LSTMs.  
- The concept of temporal dependencies in data.  

## 1. Sequence Prediction in Time Series
Time series forecasting can be framed as a **sequence-to-value** or **sequence-to-sequence** problem:
- **One-step forecasting**: Predict the next value given past observations.

  Eg: Given daily stock prices over the last 30 days, predict tomorrow’s price

- **Multi-step forecasting**: Predict multiple future values.

  Eg: Given daily stock prices over the last 30 days, predict price for the next seven days.

In practice, we often start with one-step forecasting, then extend to multi-step forecasting.

## 2. RNNs for Time Series
As we have learned before, Recurrent Neural Networks (RNNs) process sequential data by carrying forward information from one step to the next, maintaining a hidden state that carries information forward. This makes them a natural choice for time series tasks.

 <img src="https://i.postimg.cc/L6nCtwDw/RNN-Unroll.png" >
</p>

At each time step, the RNN takes the current observation (eg., today’s stock price) and combines it with the information it has “remembered” from previous timesteps. The model then produces the output (eg., the forecast for tomorrow’s stock price or next hour’s traffic flow).
- At day $t$, the RNN looks at today’s observation (today’s stock price).
- It combines this with its “memory” of the past (previous hidden state).
- It produces a forecast for day $t+1$ (tomorrow’s stock price).

The updated hidden state acts as short-term memory, storing the recent trends or fluctuations in the data. By updating the hidden state at each step, the RNN can **carry forward trends, patterns, and correlations** in the data.

The vanilla RNN network “remembers” recent observations, which is critical for forecasting tasks like:
  - Next-day sales prediction
  - Short-term temperature forecasting

We know that vanilla RNNs struggle with **long sequences**, due to vanishing gradients and exploding gradients.

RNNs often fail to capture seasonal or long-range trends.

They are well-suited for **short-term dependencies** in time series.


## 3. LSTMs for Time Series
Long Short-Term Memory (LSTM) networks improve upon RNNs by introducing a **memory cell** that retains information across longer horizons.

 <img src="https://i.postimg.cc/xTKJdZgF/lstm-1-drawio.png" >
</p>

### Why LSTMs work well for time series
- **Long-term memory**: They maintain a cell state that acts as long-term memory, allowing the model to preserve information across weeks, months, or even years of data.  
- **Selective forgetting**: Through gates (forget, input, output), LSTMs can decide which patterns to keep and which to discard. For example:
  - Forget irrelevant daily fluctuations when forecasting seasonal electricity demand.
  - Keep track of yearly cycles in retail sales or agricultural yield.
- **Handling irregular patterns**: They are robust in the presence of noise and irregular patterns, making them suitable for real-world data where anomalies are common.


### Applications in forecasting
- **Short-term predictions**: Daily or hourly forecasts using recent observations, such as:
  - Predicting the next few hours of traffic flow
  - Hourly energy demand for the following day.
- **Long-term predictions**: Weekly, monthly, or multi-step ahead forecasts that require memory of past trends and cycles, such as:
  - Predicting monthly electricity demand over the next year, where seasonal effects play a major role.
- **Multivariate time series**: Capturing interactions between multiple variables over time, such as:
  - Using multiple related features (e.g., temperature, humidity, wind speed) to predict future weather conditions.

LSTMs are much better at modeling **long-term dependencies** (e.g., yearly seasonality). They handle irregular patterns and noise more effectively. Overall, they extend the forecasting horizon and accuracy by capturing both short-term dynamics and long-term seasonality. LSTMs are commonly used in domains like **energy demand forecasting, traffic flow prediction, and healthcare time series**.

## 4. How They Are Used


### 1. **Sliding Window Approach**

This is the approach used by classical models for time series analysis and forecasting. This approach handles time series by taking a fixed window of past n observations as input.

At each time step, the window slides forward by one observation.

The RNN processes this window sequentially, updating its hidden state, and outputs the prediction for the next time step.

This is the method used in one-step forecasting.

Example: Let's take an example of a stock price forecasting RNN where the sliding window is of length $7$.

- Given the stock prices of the last $7$ days, the RNN predicts the stock price for the $8^{th}$ day.

- For predicting the stock price for the $9^{th}$ day, the RNN takes the last $7$ stock prices. So, it ignores the $1^{st}$ day, takes input from  the $2^{nd}$ day to the newly predicted $8^{th}$ day, and gives us the output as prediction of the $9^{th}$ day.

- Again, for predicting the stock price for the $10^{th}$ day, the RNN takes the last $7$ stock prices (from the $3^{rd}$ to the $9^{th}$), and so on.

### 2. **Sequence-to-Sequence (Seq2Seq) Forecasting**

Sequence to Sequence Forecasting is used for sequence-to-sequence tasks like machine translation or text summarization as well as time series forecasting.

Here, the RNN/LSTM takes an input sequence and generates an output sequence, which can be of the same or different length. The architecutre takes two sequence models(can be same models or two different models) as encoder and decoder.

The **encoder** processes the input sequence and compresses it into a fixed-size context vector or a series of hidden states, capturing the essential information.

The **decoder** takes this encoded information and generates multiple outputs step by step.

For eg:
- **Encoder**: Reads the past 30 days of stock prices and encodes them into a context vector.

- **Decoder**: Uses that information to predict the next 7 days of stock prices.

This method is useful for multi-step predictions like forecasting the next 24 hours of energy consumption or predicting future words in language modeling.
