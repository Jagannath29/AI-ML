## 1. Classical Model: ARIMA

AutoRegressive Integrated Moving Average (ARIMA) represents the classical statistical approach to time series analysis, built on decades of proven mathematical foundations.

At its core, ARIMA models assume that future values can be predicted as a linear combination of *past values*, *past forecast errors*, and *differenced values* to achieve **stationarity**. The components of ARIMA that determine the future values are:

- **AutoRegressive component (AR):** Predicts future values from past observations.
- **Integrated component (I):** Applies differencing to remove trends and make data stationary.
- **Moving Average component (MA):** Models errors from past time steps.


Formula:
$$Y_t = c + \phi_1Y_{t-1} + \phi_2Y_{t-2} + … + \phi_pY_{t-p} + \varepsilon_t$$


### Strengths:
- ARIMA is interpretable and mathematically elegant.
- It works well for short-term and linear relationships.
- It is fast to train on small datasets.
- It captures autocorrelation, trend, seasonality (with SARIMA)


### Limitations:
- ARIMA assumes linearity, and cannot capture complex non-linear relationships that are common in modern time series data.
- The data needs stationarity (may require heavy preprocessing).
- It struggles with complex, nonlinear, or long-term patterns.

ARIMA is ideal when the time series dataset is small, and the series is linear and stationary, and ideally univariate. In cases where interpretability is important, use of ARIMA is recommended.

## 2. Deep Learning Methods: LSTM

Long Short-Term Memory(LSTMs) revolutionized time series modeling by introducing the ability to learn complex, non-linear temporal patterns. Designed to handle **long-term dependencies** better than vanilla RNNs, they use gating mechanisms *forget*, *input*, and *output* gates to decide what information to retain or discard over time.

The strength of LSTMs lies in their flexibility and automatic feature learning capabilities.  They handle multivariate inputs naturally, can process multiple time series simultaneously, and learn representations that capture complex temporal dynamics.


### Strengths:
- Unlike ARIMA, LSTM captures nonlinear as well as long-range dependencies without requiring explicit specification of the mathematical form.
- LSTMs handle seasonality and complex dynamics better than ARIMA.
- LSTMs are adaptable to multivariate time series where multiple interacting factors influence the target variable.


### Limitations:
- The sequential processing requirement (each time step must be computed before the next can begin) creates a fundamental bottleneck, making training slow and preventing parallelization.
- They are computationally expensive, prone to overfitting and require large datasets.
- Training instability, sensitivity to hyperparameters, and the fixed-size hidden state bottleneck mean that LSTMs often struggle to live up to their theoretical promise, especially when dealing with the long-range dependencies common in real-world time series data.


LSTMs are best when you need to capture nonlinear and moderately long dependencies on a reasonably large dataset.

## 3. Deep Learning Methods: Transformers

Transformers revolutionized NLP, and now they represent the current frontier in time series modeling, adapting the revolutionary attention mechanism from natural language processing to temporal data. Unlike the sequential processing of RNNs, transformers create direct connections between any two points in a sequence through **self-attention**, allowing them to capture long-range dependencies without the information degradation that plagues recurrent architectures.

The attention mechanism provides transformers flexibility in modeling temporal relationships. Rather than compressing entire histories into fixed-size hidden states, attention allows the model to dynamically focus on relevant parts of the input sequence for each prediction. This selective attention often provides interpretability, as attention weights can reveal which historical periods the model considers most important for specific predictions.

Variants of transformer like Temporal Fusion Transformer, Informer, and Autoformer have specifically addressed time series challenges, incorporating variable selection, efficient attention mechanisms, and explicit seasonality modeling.

However, transformers come with their own set of challenges that make them unsuitable for all scenarios, though recent innovations like sparse attention partially address this. The complex architecture means that there are many hyperparameters and design choices, making them more difficult to tune and deploy.


### Strengths:
- Transformers possess parallel processing capability, resulting in faster training.
- They are built to handle extremely long sequences.
- Transformers learn flexible dependencies (not just recent ones) thanks to attention mechanisms.
- Transformer based models like TFT, Inormer, Autoformer, FEDformer, etc are state-of-the-art in many forecasting competitions.

### Limitations:
- The quadratic scaling of attention with sequence length creates computational and memory bottlenecks for extremely long sequences.
- They need lots of data to perforn well, they can easily overfit on small datasets where ARIMA would excel.
- Transformer models are less interpretable than ARIMA, even the TFT model which was built for interpretability.
- Transformers must artificially inject positional information to encode temporal relationships, as it has no inherent understanding of temporal order. This is still an active area of research.

Transformers dominate when the datasets are massive, long-term patterns and parallel training are needed, and computing resources are available.
