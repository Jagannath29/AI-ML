## Prerequisites
---

*   Classical Time series Forecasting (covered in Module 9)






As we learned previously in module 9, a **time series** is a collection of data points or observations indexed in the order of time. Any sequence of experimental observations taken at equally spaced timestamps can be called a time series. Thus, it is a sequence of discrete-time data.

A time-series $X$ of length $T+1$ can be represented as
$$X=\{x_t\}_{t=0}^{T}$$
where $x_t$ is the observation or value at timestamp $t$ and $T$ is the time-stamp for the last observation.

Unlike cross-sectional data, which is collected at a single point in time and captures only a snapshot of different entities without any consideration of time, the main focus of time series data is to identify and analyze temporal patterns such as trends, seasonality, and cycles. It helps in understanding how a variable behaves over time and is commonly used for forecasting future values.

In Module 9, we explored classical time series forecasting.

For short-term, linear, low-dimensional time series, **classical models (ARIMA, SARIMA, ETS, VAR)** often perform as well or better, with less training data and lower computational cost.

However, when the data is large, nonlinear, multivariate, and long-range dependent, **deep learning** is a better choice as it outperforms classical time series forecasting models.

## Time Series Characteristics

### 1. **Trends**
---

The trend component represents the long-term movement or direction in a time series. It shows whether the data is increasing, decreasing, or stable over a long period. Trends can be linear or non-linear, and is affected by smoothing techniques like moving averages.

It reflects sustained growth or decline and is not affected by short-term fluctuations.
Common causes of trends include economic growth, technological progress, or demographic changes.

Trends introduce **non-stationarity** in the data. Models like RNNs, LSTMs, and Transformers can learn non-linear, dynamic trends directly from data. Instead of assuming a fixed trend form, they approximate it flexibly using hidden states.

### 2. **Seasonality**
---

Seasonality refers to **regular and predictable patterns** that repeat over fixed time intervals, such as weeks, months, or quarters.

It is caused by recurring events such as holidays, weather patterns, or cultural habits.

<center>
<img src="https://i.postimg.cc/KY5q70B3/tsa-seasonality.png" width=730/>
<br>Figure: Seasonality Component
</center>

Seasonality is **consistent in both timing and magnitude** (e.g., sales always peak in December).

Classical methods usually assume fixed periodic patterns, but real-world seasonality often shifts in intensity, phase, or frequency.

Attention-based models (Informer, Autoformer, FEDformer) can capture variable and multiple overlapping seasonal patterns across different frequencies. We will learn more about these in the upcoming notebooks.

### 3. **Cyclic fluctuations**
---
Cyclic fluctuations are **irregular oscillations** in the time series that occur over long, variable periods. These are not tied to specific calendar intervals like seasonality. They do not have fixed frequency or amplitude.
- They typically arise from economic or political factors, such as booms and innovations.
- Cycles are **more complex and less predictable** than seasonal patterns.


<center>
<img src="https://i.postimg.cc/25Jw7g36/tsa-cyclic.png" width=730/>
<br>Figure: Cyclic Component
</center>

Cycles require models to capture long-term dependencies. LSTMs with long horizons or attention-based models like Transformers are often suited for data with a lot of cyclic fluctuations.

### 4. **Irregularity**
---
Irregularity represents unexpected, short-term, and non-recurring variations in a time series.

An irregular time series is a sequence of data points collected or recorded at uneven time intervals, with uneven sampling frequencies and maybe even missing data. Unlike regular time series, where the data is stationary and data points are equidistant in time, irregular time series have varying intervals between observations and pattern changes over time. Examples of irregular time series include:

* Event logs with timestamps
* Financial transactions recorded at different times
* Environmental measurements taken at irregular intervals

Unlike cyclic fluctuations, irregularity has no recognizable structure or recurrence.

Neural nets can sometimes learn useful signals hidden within noise (e.g., sudden demand spikes due to holidays). Attention models can focus on informative irregular points while ignoring true noise. Deep learning models are better at handling non-stationary, abrupt shifts than classical models.

### 5. **Error / Noise / Residue**
---
The error or noise or residue component captures the **random, unpredictable variations** in time series data that cannot be explained by trend, seasonality, or cycles. In the real world, unforeseen events cause such instabilities, such as natural disasters, terrorist attacks, and strikes. It represents the "leftover" after other components are accounted for.

In practical terms, the residual component is what remains after the estimation of T(Trend) and S(Seasonality) components, and their removals from the time series. We can define the residue R of a period t of a series Z, in agreement with the additive and multiplicative models, as $$R_t = Z_t − (T_t + S_t)$$ $$&$$ $$R_t = Z_t ÷ (T_t × S_t)$$ respectively.


This noise can **distort forecasts** and must be minimized or smoothed for better accuracy. It can be reduced by using techniques like exponential smoothing or filtering.

**Characteristics of errors**
- The residuals should have a constant zero mean, indicating no systematic bias.
- The residuals should exhibit constant variance (homoscedasticity), meaning the spread of residuals remains consistent over time.

## Types of Time Series

### Univariate Time Series
A univariate time series consists of observations of a single variable collected at regular time intervals. It captures how one quantity changes over time, and all forecasting is done based solely on the past values of that variable. Since only one variable is involved, univariate time series analysis focuses on identifying internal patterns like trends, seasonality, or cyclic behavior.

Classical Models such as ARIMA, Exponential Smoothing, and Simple Moving Averages are commonly used for analyzing univariate time series.

In case of deep learning, simple RNN models like LSTMs and GRUs are enough for univariate time series

**Example:** Daily temperature readings, Monthly sales of a product, Hourly electricity consumption

### Multivariate Time Series
A multivariate time series involves **two or more variables** recorded over time. Unlike univariate series, multivariate analysis considers **interactions between variables**, capturing how they influence each other. These series are useful when one variable’s behavior depends on or is influenced by others—for example, sales affected by advertising spend and season.

Multivariate models aim to analyze both the individual trends and the relationships between variables. Techniques like Vector AutoRegression (VAR), Vector Error Correction Models (VECM), or Multivariate LSTMs are commonly used for forecasting.

Attention mechanisms are very powerful for multivariate tasks, especially with heterogeneous features. Transformer-based models like Temporal Fusion Transformers(TFTs) and FEDFormer do well in multivariate time series forecasting.

**Examples:**
- Weather data: temperature, humidity, pressure
- Stock data: prices, volumes, volatility
- Business metrics: sales, ad spend, customer visits
