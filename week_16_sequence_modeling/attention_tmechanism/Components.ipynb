## *In this notebook we will learn about Query, Key and Value components of attention mechanism.*

In the attention mechanism, **Query (Q), Key (K), and Value (V)** are different ways of representing information that help the model decide what to focus on when processing a sequence. You can imagine it like searching in a dictionary: the Query is what you are looking for, the Keys are all the possible dictionary entries, and the Values are the actual meanings you get once you find the right entry. The model compares the Query with all the Keys to measure how closely they match (similarity). These comparisons produce attention weights, which tell the model how important each word (or input element) is for the current task. Finally, the Values—linked with the Keys—are combined using these weights to create a context vector that highlights the most relevant information.

For example, if we want to translate the word “love” in the sentence “I love dogs”, the Query comes from the decoder asking, “Which part of the input should I pay attention to right now?” The Keys represent the words I, love, and dogs. The model finds that the Query matches most strongly with the Key for love. The Value associated with love is then given the most weight, so the context vector emphasizes that word, helping the decoder generate the correct translation.

**The Library Analogy**

Imagine you're in a library looking for information about "machine learning":

**Query (Q)**:

Your question - "I want to learn about machine learning"

**Keys (K):**

Book titles/descriptions on library shelves

**Values (V):**

The actual content inside each book

**Attention:**

The process of finding which books are most relevant to your query. The attention mechanism compares your query with all available keys, calculates relevance scores, then retrieves a weighted combination of the corresponding values.


These components play a key role in computing the attention scores and aggregating information from different parts of the input. The ”query,” “key,” and “value” vectors are components used in the computation of attention scores and in aggregating information across sequences of data. These vectors are typically linear projections of the input data and play specific roles in the attention mechanism:

### Step 1: Input Embeddings
- Each word is first converted into a vector (embedding).  
- Example:  
  - `X1` → Embedding for "Thinking"  
  - `X2` → Embedding for "Machines"  


### Step 2: Linear Transformations
- Each embedding is multiplied with **learned weight matrices** to create three different representations:
### Attention Mechanism — Query, Key, Value  

- **Queries (Q):**  
  $$
  Q = X \cdot W^{Q}
  $$

- **Keys (K):**  
  $$
  K = X \cdot W^{K}
  $$  

- **Values (V):**  
  $$
  V = X \cdot W^{V}
  $$  


### Explanation  

- $W^{Q}$, $W^{K}$, and $W^{V}$ are **learnable weight matrices** (training parameters).  
- Their role is to **project the same input $X$** into three different vector spaces:  
  - $W^{Q}$ → transforms input into **queries** (what we are looking for).  
  - $W^{K}$ → transforms input into **keys** (what we have).  
  - $W^{V}$ → transforms input into **values** (the actual information we retrieve).  
- During training, these matrices are optimized through backpropagation so the model learns **how to best represent queries, keys, and values** for attention calculation.  

 In short: $W^{Q}, W^{K}, W^{V}$ are parameters the model learns, and they decide **how the input data gets mapped** into the Query–Key–Value spaces that drive the attention mechanism.  



### Step 3: Role of Q, K, V
- **Query (Q):** Represents what a word is *looking for*.  
- **Key (K):** Represents what a word *offers* (its identity/meaning).  
- **Value (V):** The actual information carried forward.  


### Intuition
- When computing attention, each **Query** from one word is compared with all **Keys** to see **how much focus** that word should place on others.  
- The resulting scores are used to weight the **Values (V)**, which are then summed to produce the final representation.  

In short:  
- **Q (Query):** "What am I looking for?"  
- **K (Key):** "Do I have what you’re looking for?"  
- **V (Value):** "Here’s the information if you need me."



### 1. Query Vector
- The **query** represents what we are currently focusing on or trying to understand.  
- It usually comes from the current word or the output of the previous layer.  
- The query is compared with **keys** to measure similarity and find relevant information.  

 **Example:**  
Suppose you’re translating a sentence from English to French, and you’re at a particular word in the English sentence (the query). The keys are representations of all words in the English sentence, and the values are their corresponding translations in French. Example:

“apple” (the word you want to translate)

### 2. Key Vector
- Each word in the input sequence has a **key vector** that describes its features.  
- Keys are used to calculate **relevance scores** with the query (often using a dot product).  
- These scores tell us how much attention the model should give to each word.  

**Example:**  
Keys could be the set of English words: `["cat", "apple", "tree", "juice"]`.  
The query ("apple") will match most strongly with the key "apple".  


### 3. Value Vector
- Each word also has a **value vector**, which contains the actual useful information.  
- After computing attention scores between the query and keys, these scores are used to **weight the values**.  
- The weighted sum of values is the final output of attention.  

 **Example:**  
Values could be the French translations: `["chat", "pomme", "arbre", "jus"]`.  
If the query is "apple," the model assigns a high weight to "pomme," making it the most important in the output.

In this example the **attention mechanism** decides which parts of the input are most important for producing an output. It works using three main components:


### 1. Queries (Q)
A **query** is like a **question** — it asks:  
*"Which information should I focus on?"*


### 2. Keys (K)
**Keys** are like **labels or addresses** for information.  
Each key tells us what kind of content is stored in its matching value.



###  3. Values (V)
**Values** hold the **actual information**.  
When the query finds a matching key, the system retrieves the corresponding value.


### How Attention Works
1. The **query (Q)** is compared with all **keys (K1, K2, …, Km)** → this gives **attention scores** (similarity).

2. Apply **softmax** to these scores → turns them into weights (probabilities).
3. Use these weights to combine the **values (V1, V2, …, Vm)** → this gives the **final weighted sum (output).**

*We wll learn more about score function/ attention weight mathematically in next notebook.*

### Intuition
- The **query** asks the question.  
- The **keys** tell where to look.  
- The **values** give the actual answer.  

So, **attention = weighted lookup of values based on query-key similarity.**


### Simple Example
If you’re reading the sentence:

*"The cat sat on the mat."*

- Query: "Who sat?"  
- Keys: ["The", "cat", "sat", "on", "the", "mat"]  
- Values: Same words, but with meaning attached  
- Attention highlights **"cat"** most strongly → output focuses on "cat".
