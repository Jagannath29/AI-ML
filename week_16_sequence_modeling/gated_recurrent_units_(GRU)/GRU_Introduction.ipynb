## **Variants on Long Short Term Memory (LSTM)**

Variants of LSTM are modified versions of the original LSTM architecture that are developed to overcome certain limitations or to enhance performance for specific tasks. These modifications may involve changes in the gate structure, additional connections, or the inclusion of mechanisms such as attention or peepholes. By introducing these changes, the variants make LSTMs more effective at handling long-term dependencies, improving context representation, or increasing efficiency. Each variant is designed with a focus on solving particular sequence modeling challenges found in areas like speech recognition, language translation, or time-series forecasting.


## 1. Peephole LSTM  
One popular LSTM variant, introduced by Gers & Schmidhuber (2000), is adding “peephole connections.” where the **cell state is directly connected to the gates** (input, forget, and output gates) helping the network learn **precise timings** for sequences like speech and rhythm modeling.
This allows each gate to "peek" into the cell state, giving the model more precise control over memory updates and outputs.  


## Gate Equations with Peepholes

For a given time step *t*:

- **Forget Gate**  
$
f_t = \sigma(W_f \cdot [h_{t-1}, x_t] + V_f \cdot c_{t-1} + b_f)
$

- **Input Gate**  
$
i_t = \sigma(W_i \cdot [h_{t-1}, x_t] + V_i \cdot c_{t-1} + b_i)
$

- **Candidate Cell State**  
$
\tilde{c}_t = \tanh(W_c \cdot [h_{t-1}, x_t] + b_c)
$

- **Cell State Update**  
$
c_t = f_t \odot c_{t-1} + i_t \odot \tilde{c}_t
$
- **Output Gate**  
$
o_t = \sigma(W_o \cdot [h_{t-1}, x_t] + V_o \cdot c_t + b_o)
$

- **Hidden State**  
$
h_t = o_t \odot \tanh(c_t)
$

Where:  
- \( W \) are weight matrices  
- \( V \) are peephole weight matrices (connections from cell state)  
- \( b \) are biases  
- $( \sigma $) is the sigmoid function  
- $( \odot $) is element-wise multiplication  



## Key Difference from Vanilla LSTM
- In Vanilla LSTM: gates depend only on input and previous hidden state.  
- In Peephole LSTM: **cell state \(c_t\)** also influences the gates.

## 2. Bidirectional LSTM (BiLSTM)  
.  
A Bidirectional Long Short-Term Memory (Bi-LSTM) is an extension of LSTM that processes sequential data in both **forward and backward directions**. Unlike a normal LSTM, which only looks at past context, Bi-LSTM also considers future context, making it powerful for tasks where both sides of the sequence matter (e.g., language modeling, speech recognition, and named entity recognition).

**Key Properties of Bi-LSTM**

**1. LSTM Core** → LSTM solves the vanishing gradient problem by using memory cells and gates (input, forget, output) to retain long-term dependencies.

**2. Bidirectional Processing** → Bi-LSTM has two LSTMs:

 - Forward LSTM → reads the sequence from start to end.

- Backward LSTM → reads the sequence from end to start.

**3. Context from Both Sides** → At each time step, hidden states from forward and backward LSTMs are combined (often concatenated), giving the model access to both past and future information.

**4. Better Sequence Understanding** → This makes Bi-LSTMs very effective in NLP and speech tasks where context from both directions improves accuracy.

**Workflow**

Input sequence → embeddings → Bi-LSTM layer (forward + backward) → combined hidden states → output layer (e.g., classification, sequence labeling, text generation).

 [Original Paper: "Bidirectional RNN" (Schuster & Paliwal, 1997)](https://ieeexplore.ieee.org/document/650093)

## 5. Gated Recurrent Unit (GRU)   


The previous chapters talked about RNNs and their ability to process sequential data. This helped us make predictions based on the past experiences. However, when there are long term dependencies, RNNs do not perform well because of their vanishing gradient. So, we used LSTM to preserve the long term dependencies. But, LSTM has larger number of gates and took a long time to train.

So, the Gated Recurrent Unit was introduced by [Kyunghyun Cho et al](https://arxiv.org/pdf/1412.3555) in 2014 as a modification to the LSTM cell. In a nutshell, the GRU uses only two gates, which makes the model simpler, with fewer parameters to learn. This simpler model is still able to match the performance of LSTMs in many tasks, which is why GRUs have a growing popularity in deep learning.

There are also other changes they did to LSTMs. They got rid of the cell state and replaced it with the hidden state itself. We will discuss all these changes in detail.

# **Architecture of Gated Recurrent Units (GRU)**

In simple language as shown in above figure, At each timestamp t, it takes an input Xt and the hidden state Ht-1 from the previous timestamp t-1. Later it outputs a new hidden state Ht which again passed to the next timestamp.

Now there are primarily **two gates in a GRU as opposed to three gates in an LSTM** cell. The first gate is the **Reset gate** and the other one is the **update gate**.

## **Reset Gate and Update Gate**

Here, the **LSTM’s three gates** are replaced by **two gates**: the **reset gate** and the **update gate**.  
- Both gates use a **sigmoid activation**, ensuring their values lie between 0 and 1.  
- The **reset gate** decides how much of the **past information to forget**.
- The **update gate** decides how much of the **past information to keep** and **how much to update** with new information.  

The following figure illustrates the inputs for both the reset and update gates in a GRU, given the input of the current time step and the hidden state of the previous time step. The outputs of the gates are produced by two fully connected layers with a sigmoid activation function.

<div align="center">

[![GRU-gate.png](https://i.postimg.cc/K8sPkLn4/GRU-gate.png)](https://postimg.cc/NKm9ZKCw)
*figure: Reset gate and Update gate in GRU
</div>



Mathematically, for a given time step *t*, suppose that the input is a minibatch  
$
\mathbf{X}_t \in \mathbb{R}^{n \times d}
$
(number of examples = *n*; number of inputs = *d*) and the hidden state of the previous time step is  $
\mathbf{H}_{t-1} \in \mathbb{R}^{n \times h}
$
(number of hidden units = *h*).  

Then the reset gate $(\mathbf{R}_t \in \mathbb{R}^{n \times h})$

and update gate $(\mathbf{Z}_t \in \mathbb{R}^{n \times h}\$) are computed as follows:

$
\mathbf{R}_t = \sigma(\mathbf{X}_t \mathbf{W}_{xr} + \mathbf{H}_{t-1}\mathbf{W}_{hr} + \mathbf{b}_r)
$

$
\mathbf{Z}_t = \sigma(\mathbf{X}_t \mathbf{W}_{xz} + \mathbf{H}_{t-1}\mathbf{W}_{hz} + \mathbf{b}_z),
$

where
$
(\mathbf{W}_{xr}, \mathbf{W}_{xz} \in \mathbb{R}^{d \times h}),  
(\mathbf{W}_{hr}, \mathbf{W}_{hz} \in \mathbb{R}^{h \times h})$ are weight parameters, and  
$
(\mathbf{b}_r, \mathbf{b}_z \in \mathbb{R}^{1 \times h})$ are bias parameters.  

- If $R_t \approx 0$, the previous hidden state ($h_{t-1}$) is mostly **ignored** → reset happens.  
- If $R_t \approx 1$, more past information **flows** into the candidate hidden state.  

- If $Z_t \approx 1$, the model **keeps most** of the past hidden state ($h_{t-1}$).  
- If $Z_t \approx 0$, the model **updates more** with the new candidate state.  



## **Candidate Hidden State**


<div align="center">

[![GRU.png](https://i.postimg.cc/FFGrmxbt/GRU.png)](https://postimg.cc/dDkPjrSn)
*figure: GRU Architecture with candidate hidden state*
</div>

Next, we integrate the reset gate as $\mathbf{R}_t$
 with the regular updating mechanism  leading to the following candidate hidden state $\tilde{\mathbf{H}}_t \in \mathbb{R}^{n \times h}$
 at time step $t$:

$\tilde{\mathbf{H}}_t = \tanh(\mathbf{X}_t \mathbf{W}_{\textrm{xh}} + \left(\mathbf{R}_t \odot \mathbf{H}_{t-1}\right) \mathbf{W}_{\textrm{hh}} + \mathbf{b}_\textrm{h})$,

where $\mathbf{W}_{\textrm{xh}} \in \mathbb{R}^{d \times h}$
 and $\mathbf{W}_{\textrm{hh}} \in \mathbb{R}^{h \times h}$
 are weight parameters, $\mathbf{b}_\textrm{h} \in \mathbb{R}^{1 \times h}$
 is the bias, and the symbol $\odot$
 is the Hadamard (elementwise) product operator. Here we use a $tanh$ activation function.

This equation represents how the candidate hidden state  
$(\tilde{\mathbf{H}}_t)$ in a GRU is calculated. The input $(\mathbf{X}_t$) is multiplied with its weight matrix $(\mathbf{W}_{xh})$,  
and the previous hidden state $(\mathbf{H}_{t-1})$, controlled by the reset gate $(\mathbf{R}_t)$, is multiplied with another weight matrix $(\mathbf{W}_{hh})$.  

These contributions, along with a bias term $(\mathbf{b}_h)$,  are summed up and passed through a $(\tanh)$ activation to produce the candidate hidden state.  

*In short, it combines the current input and selectively resets past information to form a new candidate state.*



The result is a candidate, since we still need to incorporate the action of the update gate. The influence of the previous states can now be reduced with the elementwise multiplication of $\mathbf{R}_t$ and $\mathbf{H}_{t-1}$. Whenever the entries in the reset gate $\mathbf{R}_t$ are close to 1, we recover a vanilla RNN. For all entries of the reset gate $\mathbf{R}_t$
 that are close to 0, the candidate hidden state is the result of an MLP with
 $\mathbf{X}_t$ as input. Any pre-existing hidden state is thus reset to defaults.

Bellow figure illustrates the computational flow after applying the reset gate.




# **Hidden State**
Finally, we need to incorporate the effect of the update gate $\mathbf{Z}_t$
.This determines the extent to which the new hidden state $\mathbf{H}_t \in \mathbb{R}^{n \times h}$ matches the old state $\mathbf{H}_{t-1}$ compared with how much it resembles the new candidate state $\tilde{\mathbf{H}}_t$ The update gate $\mathbf{Z}_t$ can be used for this purpose, simply by taking elementwise convex combinations of $\mathbf{H}_{t-1}$ and $\tilde{\mathbf{H}}_t$
. This leads to the final update equation for the GRU:
          
<div align= "center">
 $\mathbf{H}_t = (1 - \mathbf{Z}_t) \odot \mathbf{H}_{t-1}  + \mathbf{Z}_t \odot \tilde{\mathbf{H}}_t.$
 </div>

Whenever the update gate $\mathbf{Z}_t$ is close to 1, we simply retain the old state. In this case the information from
 is ignored, effectively skipping time step
 in the dependency chain. By contrast, whenever $\mathbf{Z}_t$ is close to 0, the new latent state $\mathbf{H}_t$ approaches the candidate latent state $\tilde{\mathbf{H}}_t$.

In summary, GRUs have the following two distinguishing features:

- Reset gates help capture short-term dependencies in sequences.

- Update gates help capture long-term dependencies in sequences.
