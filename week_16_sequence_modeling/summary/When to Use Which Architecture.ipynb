# When to Use Which Architecture

## Learning Objectives
- When to use RNNs, LSTMs, GRUs, or Transformers for sequence modeling tasks.

## Prerequisites
- Knowledge about RNNs, LSTMs, GRUs and Transformers


Sequence modeling is not a one-size-fits-all problem, and the right choice depends on factors like data size, dependency length, and computational resources.

Different neural architectures have evolved to handle sequence data:

- RNNs are the first models to handle sequential dependencies, and vanilla RNNs are also the simplest of the models, which come with their own sets of problems.
- LSTMs/GRUs: When Vanilla RNNs started suffering from vanishing (or exploding) gradient problem for longer sequences, LSTMs and GRUs were developed.
- Transformers: Transformers use self-attention for parallelization and long-range dependency modeling.

Choosing the right model depends on **task complexity, sequence length, and data availability**.

**Strengths:**
- RNNs model temporal dependencies directly.
- They are computationally lighter than LSTMs/Transformers.

**Limitations:**
- They struggle with long-term dependencies (vanishing gradients).
- The sequential processing nature of RNN slows training.

**Best Use Cases:**
- Toy problems / educational purposes.
- Small-scale time series forecasting.
- Short-sequence tasks (e.g., word-level classification).


## LSTMs

<p align='center'>
 <img src="https://i.postimg.cc/xTKJdZgF/lstm-1-drawio.png" >
 <figcaption align='center'>Fig: LSTM</figcaption>
</p>

**Strengths**:
- A LSTM retains information over longer time spans via a cell state.
- Gating mechanisms (input, forget, output) regulate what to keep, update, or discard.
- It address the vanishing gradient problem better than vanilla RNNs.

**Limitations**:
- LSTMs still process data step by step, slower compared to parallel models like Transformers.
- LSTM has a complex architecture, meaning more parameters and longer training times.

**Best Use Cases**:
- Time Series Forecasting: Capturing both short-term dynamics and seasonal effects (e.g., energy demand, traffic flow).
- Healthcare: Monitoring patient vital signs over time.




## GRUs
<p align='center'> <img src="https://i.postimg.cc/Qdg9Q7Gj/GRU-final.png" > <figcaption align='center'>Fig: GRU architecture</figcaption> </p>

**Strengths**:
- GRUs are simpler and faster than LSTMs due to fewer gates (update and reset).
- They require fewer parameters, making them lighter for smaller datasets.
- They handle vanishing gradient issues effectively.

**Limitations**:
- They are less expressive than LSTMs, may not capture very complex patterns (even compared to LSTMs).
- GRUs are still sequential models, so not efficient for very long sequences.

**Best Use Cases**:
- Multivariate Time Series: Forecasting with multiple related signals.
- Speech / Audio Recognition: Phoneme or speaker identification.
- Quick Prototyping: When training speed and fewer parameters are priorities.

## Transformers

<p align="center">
 <img src="https://i.postimg.cc/bvWnp3x4/transformer.png" >
 <figcaption align='center'> Fig: Transformer Architecture</figcaption>
</p>

**Strengths:**
- The self-attention mechanism of transformers captures global dependencies.
- Faster training on GPUs/TPUs due to parallelization.
- The model scales well for large datasets.

**Limitations:**
- The model requires large datasets, does not do well in smaller datasets.
- It has a high computational cost (quadratic in sequence length).

**Best Use Cases:**
- Machine translation.
- Summarization, text generation.
- Long-range time series forecasting (e.g., Temporal Fusion Transformer).
- Multimodal data (vision + language, audio + text).

## Comparison Table

| Architecture | Strengths | Weaknesses | Best Use Cases |
|--------------|-----------|------------|----------------|
| RNN | Lightweight, short dependencies | Vanishing gradients, sequential bottleneck | Short sequences, small-scale tasks |
| LSTM / GRU | Moderate dependencies, better memory | Slower than Transformers | NLP tagging, speech, multivariate forecasting |
| Transformer | Long-range dependencies, parallelizable | Large data & compute needed | Translation, summarization, long sequences |

---

## Summary

For simple, short-sequence tasks which require limited compute, vanlla RNNs are best suited, as they do the job fine.

However, when moderate-length dependencies are involved, and RNNs can't handle them, LSTMs and GRUs are best suited, as their gating mechanisms helps with memory. If the sequence is long and has complex dependencies, LSTM is used. If a moderate sized sequence input is given and speed is important, GRU works better.

Still, LSTMs and GRUs are sequential, so when long-range dependencies and large-scale data come to play, they become slower and less reliable. That is where we use Transformers, as they process data and train model in parallel, accomodating large data, and also leverage self-attention, which helps with long-range dependencies.
