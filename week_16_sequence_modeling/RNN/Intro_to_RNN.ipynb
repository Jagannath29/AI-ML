RNNs are widely used in:
- Natural Language Processing (NLP) → text generation, translation
- Time Series Analysis → stock prediction, weather forecasting
- Speech Recognition → transcribing audio


RNNs are different from Feed Forward Networks(ANNs) because unlike ANNs, they have a memory of past inputs, making them suitable for tasks where order and context matter.

## Basic Structure of an RNN

An RNN processes sequences **step-by-step**, maintaining a hidden state $( h_t )$ that carries information from previous time steps.

**Key Components:**
- Input at time $( t ): ( x_t )$
- Hidden state at time $( t ): ( h_t )$
- Output at time $( t ): ( y_t )$

The formula for hidden state of RNN is:

  $$h_t = f(h_{t-1}, x_t ; \theta)$$

This formula demonstrates that the hidden state $h_t$ at the current time step is the function of the previous hidden state $h_{t-1}$ and current input $x_t$. And, $\theta$ are the parameters of the function $f$.

Now let’s see how implementing this kind of neural network supports our requirement:

1. Since we want the same function to be executed at each timestep, we simply share the same network i.e. we use the same parameters at each timestep. The parameters of the network are $U, W$ and $V$. Sharing parameters across different parts of a model makes it possible to extend and apply the model to examples of different forms (different lengths, here) and generalize across them.

2. As a consequence of this, parameter sharing ensures that the network becomes agnostic to the size of the input. Since we are simply going to compute the same function with the same parameters at each timestep, the number of timesteps doesn’t matter. We just create multiple copies of the network and execute them at each timestep. This takes care of the variable-sized input.

3. Our final concern is how do we account for the case of multiple dependent inputs and not just one independent input? The solution for this is to add a recurrent connection(horizontal arrow) in the network.

## **Working of Vanilla RNNs**

The RNN accepts an input vector $x$ and predicts an output vector $\widehat{y}$. It scans through data from left to right and passes activation at each time step to next. Parameters used by RNN for each time step are shared. In Figure 2, the parameters governing the connection from $x_t$ to the hidden layer will be some set of learnable parameters, $U$. The same parameter $U$ will be used for every time step. Activations (or horizontal connections) are be governed by a set of parameters $W$ .The same parameters $W$ is also used in every time step. The sum $V$ governs the output predictions in similar way. During predictions information from previous inputs are also passed.

One drawback with the type of cell is that it can pass information earlier in the sequence but not later ones. For example if we have to predict whether certain word is a name entity or not, we have to understand the context first like in following sentences:

* He said "Teddy Roosevelt was a dynamic President of US."
* He said "Teddy bears make children happy."

In above sentences, we can see that the word Teddy is a name entity in the first sentence and not a name entity in next one. But from the earlier sequence of words, it is undistinguishable. For such purposes we need to use bidirectional RNNs (or BRNNs) that lets us take information from both earlier and later in the sequence.
