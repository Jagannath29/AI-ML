### **Overview:**

Computational graphs are the formalized representations of the structure of set of computations that are involved in mapping the parameters while traversing from input to the outputs and loss.

<p align="center">Figure 1. Computational Graphs (Unfolded)</p>

On a computational graph architecture, the Recurrent Neural Network consists of an Input layer, Hidden Layer and a output layer.

$U, V$ and $W$ are the weight parameters that represent the weights associated between different layers. In this figure, $U$ represents the weights associated between the inputs and the hidden layer, $W$ represents the weights associated with between hidden layers and $V$ represents the weights associated between hidden layers and output layers.

The definition formula for RNN is:

$h_t$ = $f(h_{t-1}, x_t ; \theta)$ ----------(1)

This formula demonstrates that the hidden state $h_t$ at the current time step is the function of the previous hidden state $h_{t-1}$ and current input $x_t$. And, $\theta$ are the parameters of the function $f$.

Moreover, for a finite number of time steps $T$, the computational graphs as depicted in Figure 1. can be unfolded by applying the definition equation (1) $T-1$ times. For example, if we unfold the equation for $T = 4$ time steps, we get:

$h_4 = f(h_3, x_4 ;\theta)$ ----------(2)

Thus, the unfolded computational graph of Figure 1. is represented as a traditional directed acyclic computational graph with repeated time steps as defined by the equation (1).

<p align="justify">
The given Figure 1. illustrates the computational graph to compute the training loss of a recurrent network. The recurrent network maps an input sequence of $x$ values $x_i$ to a corresponding sequence of output values $\hat{y_i}$. Loss $\mathcal{L}$ measures how far each $\hat{y_i}$ is from the corresponding training target $y$. Loss function depends on the nature of problem but for simple tasks like predicting the next word or character, cross-entropy loss is often used.

At each time step, the RNN produces a prediction vector $\hat{y_t}$ using the softmax activation:

$$\hat{y_t} =softmax(o_t) = softmax(Vh_t+b_h)$$

This prediction is then compared with the true label $y_t$ using a loss function. Typically, **categorical cross-entropy loss** is used.  

$$\mathcal{L}_t = -\sum_{k=1}^{C} y_t^{(k)} \log \hat{y_t}^{(k)}$$

where

$C$ = no. of classes

$y_t^{(k)}$ = $1$ if class is $k$, else $0$,

$\hat{y_t}^{(k)}$ = predicted probability for class $k$


The **total loss** across $T$ time steps is simply the sum of loss at each timestep.  

$$\mathcal{L}(\theta) = \sum_{t=1}^{T} \mathcal{L}_t(\theta)$$  

</p>

> Equations:

$$ a_t = W h_{t-1} + U x_t + b_a $$

$$ h_t = tanh(a_t)$$

$$ o_t = V h_t + b_h$$

$$ \hat{y_t} = softmax(o_t)$$

$$ \mathcal{L}(\theta) = \sum_{t=1}^{T}\mathcal{L_t}(\theta) $$

In the Figure 1. on the left shows the RNN and its loss drawn with recurrent connections. In the right, the same network is seen as a time-unfolded computational graph, where each node is now associated with one particular time instance.
