Here, let us write the notations in the matrix notations.

$x$ is the input vector, $\in$ $\mathcal{R}^{n*t}$

- *n* is the number of feature vectors, and *t* is the number of words in the vocabulary.

$U$ - is the weight parameter that connects input to hidden layer. $\in$ $\mathcal{R}^{d*n}$

- *d*, is the number of units present in the hidden layer, and *n* units present in the input vector.

$V$ - is the weight parameter that connects hidden to the Output layer, $\in$ $\mathcal{R}^{k*d}$

- *k*, is the number of units present in the Output layer, and *d* units present in the hidden layer.

$W$ - is the weight parameter that connects Hidden to Hidden layer, $\in$ $\mathcal{R}^{d*d}$

*d*, is the number of units present in the hidden layer, and *d* units present in the hidden layer.

$h$ is the hidden state vector, $\in$ $\mathcal{R}^{d*t}$

$y$ is the output Vector. $\in$ $\mathcal{R}^{k*t}$

$L$ is the loss at each time step.



## Forward Propagation

Lets write out equations for forward propagation in RNN.

*We will ignore the bias term for now.*

Hidden state:

$h_t$ = $tanh(U x_t + W h_{t-1})$

Output :

$z_t$ =  $V h_t$

$\widehat{y_t} = softmax(z_t)$


Loss :

This is the multi-class cross-entropy loss.

$L_t = - y_t log ( \widehat{y_t} )$

Now, the total loss in all the timesteps is the sum of loss in each time step.

$L$ = $\sum_j^T L_j$


## Back propagation

Before getting into what is backpropagation through time, first, let's see if normal backpropagation used in a Feed-forward Neural network can be used here or not in Recurrent Neural Network.

We have three different weights for RNN i.e., U, W, and V.

We also have a Loss function  $L$. Now, with normal backpropagation, the steps are,

- Calculate the gradients of Loss with respect to weight,

- Update the weight using Gradient descent.

Let's use this concept here,


Let's write down the equations to calculate the **derivative of Loss with respect to weight $V$**

$$ \frac{\partial \mathcal{L}}{\partial V} = \sum_i^T \underbrace{\frac{\partial \mathcal{L_i}}{\partial \widehat y_i} }_1 \underbrace{ \frac{\partial \mathcal{\widehat y_i}}{\partial  z_i} }_2 \underbrace{ \frac{\partial \mathcal{z_i}}{\partial V}}_3 $$



We are summing over all the time-steps since, in RNN, the total Loss is the sum of Loss at all time-steps.

Suppose, for first time step the equation is written as :

$$ \frac{\partial \mathcal{L}}{\partial V} =  \underbrace{\frac{\partial \mathcal{L_1}}{\partial \widehat y_1} }_1 \underbrace{ \frac{\partial \mathcal{\widehat y_1}}{\partial  z_1} }_2 \underbrace{ \frac{\partial \mathcal{z_1}}{\partial V}}_3 $$

This is simply the normal backpropagation. The single time-step in RNN is simply a single feed-forward Network, and we already know how backpropagation works for Feed-forward Neural Network. Here, we are sum up the gradients of all time -steps before updating the weight.


Next, let's calculate the **derivative of Loss with respect to $W$**

The total loss is sum of loss at all time-steps which means we need to calculate gradients at each time stpe,

$$ \frac{\partial \mathcal{L}}{\partial W} = \sum_i^T \frac{\partial \mathcal{L_i}}{\partial W} $$

For ease of demonstration, let's calculate the derivative of Loss at time step 4.

i.e, $ \frac{\partial \mathcal{L_4}}{\partial W} $

Now, as normal backpropagation, we need to write the chain rule. Here, this starts to complicate.

The loss at fourth time step is dependent upon current time-step $h_4$ and $h_4$ is dependent on $W$, but $h_4$ is also dependent on $h_3$ and $h_3$ is depended again on $W$ which means, $L_4$ is indirecly dependent on on $W$ through $h_3$, and similar story goes for $h_2$ and $h_1$.

So, we write the chain rule as,


$$ \frac{\partial \mathcal{L_4}}{\partial W} = \frac{\partial \mathcal{L_4}}{\partial h_4}  \frac{\partial \mathcal{h_4}}{\partial W} $$

$$ \frac{\partial \mathcal{h_4}}{\partial W} = \frac{\partial \mathcal{h_4}}{\partial W} +  \frac{\partial \mathcal{h_4}}{\partial h_3}\frac{\partial \mathcal{h_3}}{\partial W}+ \frac{\partial \mathcal{h_4}}{\partial h_3} \frac{\partial \mathcal{h_3}}{\partial h_2} \frac{\partial \mathcal{h_2}}{\partial W} + \frac{\partial \mathcal{h_4}}{\partial h_3} \frac{\partial \mathcal{h_3}}{\partial h_2} \frac{\partial \mathcal{h_2}}{\partial h_1} \frac{\partial \mathcal{h_1}}{\partial W}$$


This is called **Backpropagation through time** because you sum up the derivatives across all the time steps. This is different from Normal backpropagation because, in Normal backpropagation, we do not go in the previous time steps.

In the next section, we will go thoroughly deriving all these derivatives.

## Derivative of loss $L$ with respect to $V$.


Let us calculate the derivative of loss $L$ with respect to $V$.

$$ \frac{\partial \mathcal{L}}{\partial V} $$

Here, V is the weight parameter that connects hidden to the Output layer.

Since, the paramter V is shared across all time-steps in RNN, the total derivative is the sum of derivatives in all time steps i.e given as :

$$ \frac{\partial \mathcal{L}}{\partial V} = \frac{\partial \mathcal{L_1}}{\partial V} + \frac{\partial \mathcal{L_2}}{\partial V} + ... + \frac{\partial \mathcal{L_T}}{\partial V}$$

Now, lets calcualte the derivative :


$$ \frac{\partial \mathcal{L}}{\partial V} = \sum_i^T \frac{\partial \mathcal{L_i}}{\partial V} $$


Lets recall once again  the equations for forward propagation :

$h_t$ = $tanh(U x_t + W h_{t-1})$

$z_t$ =  $V h_t$

$\widehat{y_t} = softmax(z_t)$

Loss :

$L_t = - y_t log ( \widehat(y_t) )$


Now, Using Chain ule we have :

$$ \frac{\partial {L}}{\partial V} = \sum_i^T \underbrace{\frac{\partial {L_i}}{\partial \widehat y_i} }_1 \underbrace{ \frac{\partial \mathcal{\widehat y_i}}{\partial  z_i} }_2 \underbrace{ \frac{\partial \mathcal{z_i}}{\partial V}}_3 ......... equation 1$$

We have three parts here. Let us calculate each of them one by one. First, we start with $1$

$$ \frac{\partial \mathcal{L_i}}{\partial \widehat y_i}    = \frac{\partial \mathcal{ - y_i log \widehat y_i}}{\partial \widehat y_i} =  \frac{- y_i}{\widehat y_i}   $$







Let us move to the second part $2$ i.e


$ \frac{\partial \mathcal{\widehat y_i}}{\partial  z_i}  $

We have,

$\widehat y_i$ = $ softmax(z_i) = \frac{e^{z_i}}{\sum_{k=1} ^ K e^{z_k}}$

Where $K$ is the number of classes.

Now, we need to take the Derivative of this softmax function, which is :

**We have two cases for this, when $ i= k $
 and when $i != k$**

First, let us start with

**Case 1 i.e $ i =k $**


$$ \frac{\partial \mathcal{\widehat y_i}}{\partial  z_i}  = \frac{\partial \mathcal{ \frac{e^{z_i}}{\sum_{k=1} ^ K e^{z_k}}}}{\partial  z_i}  $$

*Using Quotient rule of Derivative.*

$$ \frac{\partial \mathcal{\widehat y_i}}{\partial  z_i}  = \frac{ \mathcal{ {e^{z_i}}{\sum_{k=1} ^ K e^{z_k}}} - e^{z_i} e^{z_i} } {(\sum_{k=1} ^K e^{z_k})^2}  $$

$$ \frac{\partial \mathcal{\widehat y_i}}{\partial  z_i}  = \frac{ \mathcal{ {e^{z_i}} ( {\sum_{k=1} ^ K e^{z_k}}} - e^{ z_i} ) } {(\sum_{k=1} ^K e^{z_k}) (\sum_{k=1} ^K e^{z_k})}  $$

Now with simple algebraic manipulations, we can write the equation as :


$$ \frac{\partial \mathcal{\widehat y_i}}{\partial  z_i}  = \widehat y_i (1 - \widehat y_i)  $$

*Note: We changed the notation from k to i because in this case, we are considering for i = k*


















**Case 2 i.e $i$ $!=$ $k$**



$$ \frac{\partial \mathcal{\widehat y_i}}{\partial  z_i}  = \frac{\partial \mathcal{ \frac{e^{z_i}}{\sum_{k=1} ^ K e^{z_k}}}}{\partial  z_i}  $$

*Using Quotient rule of Derivative.*

$$ \frac{\partial \mathcal{\widehat y_i}}{  z_i}  = \frac{ { {{}}{{}  {0}}} - e^{z_k} e^{z_i} } {(\sum_{k=1} ^K e_{z_k})^2}  $$

$$ \frac{\partial \mathcal{\widehat y_i}}{\partial  z_i}  = \frac{ \mathcal{ {{}} {{}{}}} - e^{z_k}   e^{z_i}} {(\sum_{k=1} ^K e^{z_k}) (\sum_{k=1} ^K e^{z_k})}  $$

Now with simple algebraic manipulations, we can write the equation as :


$$ \frac{\partial \mathcal{\widehat y_i}}{\partial  z_i}  = - \widehat y_k (\widehat y_i)  $$

Now, lets substitute the values 1, and 2 in equation 1:


$$ \frac{\partial \mathcal{L}}{\partial V} = \sum_i^T \underbrace{\frac{\partial \mathcal{L_i}}{\partial \widehat y_i} }_1 \underbrace{ \frac{\partial \mathcal{\widehat y_i}}{\partial  z_i} }_2 \underbrace{ \frac{\partial \mathcal{z_i}}{\partial V}}_3 $$


$$ \frac{\partial \mathcal{L}}{\partial V} = \sum_i^T {\frac{ \mathcal{y_i}}{ \widehat y_i} } { }   \begin{cases}
      \widehat y_i (1 - \widehat y_i) & i=k  \\
      -\widehat y_i \widehat y_k & i != k\\
       &   
   \end{cases}
\hspace{0.3cm} ( { \frac{\partial \mathcal{z_i}}{\partial V}} )
$$

$$ \frac{\partial \mathcal{L}}{\partial V} = \sum_i^T {{ \mathcal{}}{ } } { }   \begin{cases}
       - y_i (1 - \widehat y_i) & i=k  \\
      y_i \widehat y_k & i != k\\
       &   
   \end{cases}
\hspace{0.3cm} ( { \frac{\partial \mathcal{z_i}}{\partial V}} )
$$

Now, to generalize this equation, lets focus on this only,

\begin{cases}
       - y_i (1 - \widehat y_i) & i=k  \\
      y_i \widehat y_k & i != k\\
       &   
\end{cases}

$$ -y_i + y_i  \widehat y_i + \sum_{ i != k} y_i \widehat y_k$$


$$ -y_k + y_k  \widehat y_k + \sum_{ i != k} y_i \widehat y_k$$


$$ -y_k + \widehat y_k [ y_k + \sum_{ i != k}y_i ]$$

$$ -y_k + \widehat y_k [ y_k + \sum_i^K y_i ]$$

Here, the $ \sum_i^K y_i  $ is 1, because $y_i$ is bascially one-hot encoded vector.
So, we write as :

$$ -y_k + \widehat y_k $$

Som now lets plug this equation in equation 1 :

$$ \frac{\partial \mathcal{L}}{\partial V} = \sum_i^T {{ \mathcal{}}{ } } { ( -y_k + \widehat y_k)}   
\hspace{0.3cm} ( { \frac{\partial \mathcal{z_i}}{\partial V}} )
$$

Now, the only thing remaining is $ { \frac{\partial \mathcal{z_i}}{\partial V}}$ which is simple and calculated as,

We have, $z_i = V h_t$

$$ { \frac{\partial \mathcal{z_i}}{\partial V}}= h_t $$

Then,

$$ \frac{\partial \mathcal{L}}{\partial V} = \sum_i^T {{ \mathcal{}}{ } } {   ( \widehat y_i -y_i )}   
\hspace{0.3cm} \circ h_i
$$

## Derivative of loss $L$ with respect to $W$.

This is an important step; in this step, we would have backpropagation through time.

Lets recall the forward propagation steps:

$h_t$ = $tanh(U x_t + W h_{t-1})$

$z_t$ =  $V h_t$

$\widehat{y_t} = softmax(z_t)$

Loss :

$L_t = - y_t log ( \widehat(y_t) )$



We have, the partial derivative of Loss with respect to W as :

$$ \frac{\partial \mathcal{L}}{\partial W} = \frac{\partial \mathcal{L_1}}{\partial W} + \frac{\partial \mathcal{L_2}}{\partial W} + ... + \frac{\partial \mathcal{L_T}}{\partial W}$$


$$ \frac{\partial \mathcal{L}}{\partial W} = \sum_i^T \frac{\partial \mathcal{L_i}}{\partial W} $$

First,
**lets calculate for $L_1$**

i.e. ${}  \frac{\partial \mathcal{L_1}}{\partial W}$

We again need to use the chain rule here. First we caluclate the derivative of  $L_1$ with respect to $y_1$, then derivative of $y_1$ with respect $h_1$ and then $h_1$ with respect to $W$ which is :

$$ \frac{\partial \mathcal{L_1}}{\partial W} = \underbrace{ \frac{\partial \mathcal{L_1}}{\partial \widehat y_1} }_1 \underbrace{\frac{\partial \mathcal{\widehat y_1}}{\partial h_1} }_2 \underbrace{ \frac{\partial \mathcal{h_1}}{\partial W}}_3$$

The first two terms, $1$ and $2$, are already calculated, so the only remaining is  $3$, which is just the derivative of tan hyperbolic:

We have,

$h_1$ = $tanh(U x1 + W h_0)$

Then,

$$ \frac{\partial \mathcal{h_1}}{\partial W} = h_0(1 - tanh^2(Ux1 + W h_0))$$

Then, we find the derivative of loss of $L1$ with respect to $W$ and plugging into the equation.


This is for the first time step. Let's see how gradients flow.

Now, we need to find the **derivative of Loss $L_2$ with respect $W$**

Now, thigs starts to complicate. Lets see,

$$ \frac{\partial \mathcal{L_2}}{\partial W} = \underbrace{ \frac{\partial \mathcal{L_2}}{\partial \widehat y_2} }_1 \underbrace{\frac{\partial \mathcal{\widehat y_2}}{\partial h_2} }_2 \underbrace{ \frac{\partial \mathcal{h_2}}{\partial W}}_3$$

Here, $h_2$ = $tanh(U x_2 + W h_1)$

Here, $h_1$ is not the constant, $h_1$ is again the function of $W$ i.e $h_1$ = $tanh(U x1 + W h_0)$

There is an inherent implicitness going on this equation. So, we need to divide into two parts explicit and implicit part for taking a derivative of $h_2$ with respect to $W$.

$$ \frac{\partial \mathcal{h_2}}{\partial W} = \underbrace{ \frac{\partial \mathcal{h_2^+}}{\partial W} }_{explicit} \underbrace{\frac{\partial \mathcal{h_2}}{\partial h_1}  { \frac{\partial \mathcal{h_1}}{\partial W}}}_{implicit}$$

So, now we have derivative of  $L_2$ with respect to $W$ as,


$$ \frac{\partial \mathcal{L_2}}{\partial W} = { \frac{\partial \mathcal{L_2}}{\partial \widehat y_2} } {\frac{\partial \mathcal{\widehat y_2}}{\partial h_2} } [ \underbrace{ \frac{\partial \mathcal{h_2^+}}{\partial W} }_{explicit} \underbrace{\frac{\partial \mathcal{h_2}}{\partial h_1}  { \frac{\partial \mathcal{h_1}}{\partial W}}}_{implicit}]$$

*In the explicit part, the argument is treated as constant, whereas in the implicit part argument is not treated as constant.*

$$ \frac{\partial \mathcal{L_2}}{\partial W} = { \frac{\partial \mathcal{L_2}}{\partial \widehat y_2} } {\frac{\partial \mathcal{\widehat y_2}}{\partial h_2} }  { \frac{\partial \mathcal{h_2^+}}{\partial W} }_{} + { \frac{\partial \mathcal{L_2}}{\partial \widehat y_2} } {\frac{\partial \mathcal{\widehat y_2}}{\partial h_2} } {\frac{\partial \mathcal{h_2}}{\partial h_1}  { \frac{\partial \mathcal{h_1}}{\partial W}}}$$




Now, moving forward, let us calculate the **derivative of Loss $L_3$ with respect to $W$.**

We follow the same steps as earlier, dividing into the implicit and explicit parts. We would not write the whole process. It is the same as a derivative of $L_2$ with respect to $W$


$$ \frac{\partial \mathcal{L_3}}{\partial W} = \underbrace{ \frac{\partial \mathcal{L_3}}{\partial \widehat y_3} }_1 \underbrace{\frac{\partial \mathcal{\widehat y_3}}{\partial h_3} }_2 \underbrace{ \frac{\partial \mathcal{h_3}}{\partial W}}_3$$

Here, $h_3$ is the function of $h_2$ and $h_2$ is the function of $h_1$ and $h_1$ is the function of $W$.

Finally, we get ;

$$ \frac{\partial \mathcal{L_3}}{\partial W} = { \frac{\partial \mathcal{L_3}}{\partial \widehat y_3} } {\frac{\partial \mathcal{\widehat y_3}}{\partial h_3} } { \frac{\partial \mathcal{h_3^+}}{\partial W}}+
{ \frac{\partial \mathcal{L_3}}{\partial \widehat y_3} } {\frac{\partial \mathcal{\widehat y_3}}{\partial h_3} } { \frac{\partial \mathcal{h_3}}{\partial h_2}}{ \frac{\partial \mathcal{h_2^+}}{\partial W}} + { \frac{\partial \mathcal{L_3}}{\partial \widehat y_3} } {\frac{\partial \mathcal{\widehat y_3}}{\partial h_3} } { \frac{\partial \mathcal{h_3}}{\partial h_2}}{ \frac{\partial \mathcal{h_2}}{\partial h_1}}{ \frac{\partial \mathcal{h_1^+}}{\partial W}}$$

$$ \frac{\partial \mathcal{L_3}}{\partial W} = { \frac{\partial \mathcal{L_3}}{\partial \widehat y_3} } {\frac{\partial \mathcal{\widehat y_3}}{\partial h_3} } [ { \frac{\partial \mathcal{h_3}}{\partial h_3}} { \frac{\partial \mathcal{h_3^+}}{\partial W}}+
   { \frac{\partial \mathcal{h_3}}{\partial h_2}}{ \frac{\partial \mathcal{h_2^+}}{\partial W}} +  \underbrace{{ \frac{\partial \mathcal{h_3}}{\partial h_2}}{ \frac{\partial \mathcal{h_2}}{\partial h_1}}}{ \frac{\partial \mathcal{h_1^+}}{\partial W}}]$$

We can generalzie the  equation as :

$$ \frac{\partial \mathcal{L_3}}{\partial W} = { \frac{\partial \mathcal{L_3}}{\partial \widehat y_3} } {\frac{\partial \mathcal{\widehat y_3}}{\partial h_3} }  [\sum_{k=1}^3 { \frac{\partial \mathcal{h_3}}{\partial h_k} } {\frac{\partial ^+ \mathcal{h_k}}{\partial W} } ]$$

$\sum_{k=1}^i$, is the summation over hidden steps.

Before that lets recap, what is the derivative of a vector with respect to the vector. Lets take an exampple,

$z_1$ = $x^2 + y^2$

$z_2$ = $x^3 + y^3$

$ Z = [z_1, z_2 ]$

$\theta = [x, y]$

So, now derivative of $Z$ with respect to $\theta$ is


\begin{bmatrix}
 \frac{dz_1}{x}  &  \frac{dz_2}{x}  \\
 \frac{dz_1}{y}  &  \frac{dz_2}{y}  \\
 &  &
\end{bmatrix}



Here, $Z$ and $\theta$ is the two dimensional vector and the derivative of vector with respect to vector is the matrix.

So, in our case $h_m$ is the hidden state vector and $h_{m-1}$ is also the hidden state vector. The derivative of $h_m$ with respect to $h_{m-1}$ is

$$ \frac{\partial \mathcal{h_m}}{\partial h_{m-1}} =  { \frac{\partial \mathcal{tanh(U x_m + W h_{m-1})}}{\partial (U x_m + W h_{m-1})} } {\frac{\partial \mathcal{(U x_m + W h_{m-1})}}{\partial h_{m-1}}  } $$

$ { \frac{\partial \mathcal{tanh(U x_m + W h_{m-1})}}{\partial (U x_m + W h_{m-1})}}$ is actually the derivative of vector with respect to vector which gives
actually a matrix which is as follows :




\begin{bmatrix}
 \  { \frac{\partial \mathcal{tanh(U x_1 + W h_{0})}}{\partial (U x_1 + W h_{0})}} &  { \frac{\partial \mathcal{tanh(U x_2 + W h_{1})}}{\partial (U x_1 + W h_{0})}}   &  { \frac{\partial \mathcal{tanh(U x_3 + W h_{2})}}{\partial (U x_1 + W h_{0})}} .... \\

*Note: This is an explicit part, so argument inside is constant*




Now, the derivative of Loss (L) with respect to $W$ is written as from equation (a):

$$ \frac{\partial \mathcal{L}}{\partial W} = \sum_{i=1}^T \sum_{k=1}^i \underbrace{{ \frac{\partial \mathcal{L_i}}{\partial \widehat y_i} }}_1 \underbrace{{\frac{\partial \mathcal{\widehat y_i}}{\partial h_i} }}_2 \underbrace{{ \frac{\partial \mathcal{h_i}}{\partial h_k} } }_3 \underbrace{{\frac{\partial ^+ \mathcal{h_k}}{\partial W} }}_4  $$

$$ \frac{\partial \mathcal{L}}{\partial W} = \sum_{i=1}^T \sum_{k=1}^i( \widehat y_i - y_i)  \prod_{m = k+1}^i  W^T diag(1 - tanh^2(W h_{m-1} + U x_m)) * h_{t-1}$$



## Derivative of loss $L$ with respect to $U$.

Derivaive of Loss with respect to $U$ also follows the same patters as for W, we  need to traverse all the way back to $h_0$,  the final expression we obtain is :


$$ \frac{\partial {L}}{\partial U} = \sum_{i=1}^T \sum_{k=1}^i( \widehat y_i - y_i)  \prod_{m = k+1}^i  W^T diag(1 - tanh^2(W h_{m-1} + U x_m)) * x_k $$

The only thing different is,


$$ \frac{\partial \mathcal{^+ h_k}}{\partial U} = x_k$$

Since, $h_k$ = $tanh(U x_k + W h_{k-1})$

## Key takeaways :

- The hidden state of RNN carries the information from the past, thus preserving long-term dependency.

- Normal backpropagation cannot be used in RNN, we need another algorithm called Backpropagation through time.

- In Backpropagation through time,gradients are accumulated from the previous timesteps.
