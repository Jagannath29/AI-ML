## Temporal dependencies
---

Temporal dependencies in sequential data describe how past events shape present states as well as future possibilities. They create complex webs of influence where past events shape present states and future possibilities.

Temporal dependencies can be local, where adjacent elements strongly influence each other, or they can span long distances where events separated by many time steps remain causally connected. The two types of temporal dependencies are :
- Local Dependency
- Long-range Dependency

In natural language, local dependencies might involve subject-verb agreement within a sentence, for eg:
**"The dog barks"** vs **"The dogs bark"**

Here, we can tell if the third word would be 'bark' or 'barks' depending upon if the second work is 'dogs' or 'dog'.

Long-range dependencies on the other hand could involve coreference resolution where a pronoun refers back to a noun mentioned paragraphs earlier.

"Sarah went to the library. She borrowed a book." (She refers to Sarah)

Think of a novel, where the actions of the protagonist at the start usually influences the happenings of the novel later on. That is an example of Long-range dependency.

In Time Series Data like weather forecasting, today's temperature being similar to yesterday's is a local dependency, while a long-range dependency would refer to seasonal patterns eg. July tends to be hotter than January.

## Variable-length inputs/outputs
---

Sequential data is different from tabular or image data, where input size is usually fixed. Models must be flexible and generalizable enough to handle inputs/outputs of arbitrary length.

Variable length input and output are common characteristics of sequence modeling tasks, particularly in fields like Natural Language Processing (NLP) and speech recognition. They present fundamental challenges for machine learning architectures that traditionally expect fixed-size inputs.

A sequence model must handle data where the length of the input sequence (e.g., a sentence, a speech utterance) and the output sequence (e.g., a translated sentence, a transcript) can vary significantly. This variability requires models to be inherently flexible, capable of processing and producing outputs of arbitrary length without degrading performance or requiring artificial padding or truncation.

The ability to handle variable length sequences is crucial for many applications, including:
- Machine Translation: Input sentences and output translations rarely have the same number of words or letters.
- Text Summarization: Input documents are typically much longer than the desired output summaries.
- Speech Recognition: Spoken utterances vary in duration, and their corresponding text transcripts vary in length.
- Chatbots: User queries and bot responses can have diverse lengths.

The relationship between input and output lengths adds another layer of complexity. Some tasks require fixed-length outputs regardless of input length (sentiment classification of variable-length text), while others demand outputs that scale with inputs (machine translation where sentence lengths vary across languages). More complex scenarios involve many-to-many mappings with different input-output length ratios, such as automatic speech recognition where acoustic features must be aligned with words, or video captioning where temporal visual content must be condensed into concise descriptions.



### Key aspects of handling variable length input/output in sequence modeling:
---
- Encoder-Decoder Architecture:
This is a widely used framework, especially for sequence-to-sequence tasks like machine translation or text summarization. It is also called seq2seq model, and it uses two sequence models ( RNNs, LSTMs, Transformers, etc) as encoder and decoder.

  Here is a brief overview of Encoder-Decoder Architecture. We will learn about this topic in detail in later notebooks.
    - Encoder: Processes the variable-length input sequence and compresses it into a fixed-size context vector or a series of hidden states, capturing the essential information.
    - Decoder: Takes this encoded information and generates a variable-length output sequence, one element at a time.

  <p align="center">
  <img src= "https://i.postimg.cc/6538n3XV/encdec.png" alt="seq2seq Diagram" width="1000"/>
</p>

  Here we can see a model using encoder to process input questions and a decoder to give output answers. Both the input and the output can be of variable length.


- Padding and Masking:

  To efficiently process batches of variable-length sequences, especially in deep learning frameworks, shorter sequences within a batch are extended with special "padding" tokens to match the length of the longest sequence in that batch. This creates uniform-sized inputs for batch processing.

  A corresponding mask is typically used to indicate which parts of the padded sequence are actual data and which are padding. This ensures that the model ignores the padded values during calculations, preventing them from influencing the learning process.
