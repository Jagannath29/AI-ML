# Chapter II: Environment Setup
This chapter focus on environment setup. Notebook provides the commands and the snippet.

### Objective: Get a local Kubernetes cluster running with necessary tools.

## Step-by-Step Guide Overview:
- **Install Docker:** Download from docker.com and verify with `docker --version.`

- **Install Minikube:** Follow minikube.sigs.k8s.io/docs/start/. Start with `minikube start --driver=docker.`

- **Install kubectl:** `curl -LO "https://dl.k8s.io/release/$(curl -L -s https://dl.k8s.io/release/stable.txt)/bin/linux/amd64/kubectl"` and add to PATH. Verify with kubectl version.

- **Install MLflow:** `pip install mlflow.`

- **Install Airflow:** `pip install apache-airflow` (use standalone mode for demo: airflow standalone).

- **Set up MinIO (local S3 equivalent):** Run in Docker docker run -p 9000:9000 -p 9001:9001 minio/minio server /data --console-address ":9001". Create bucket via UI.

- **Install other libs:** `pip install scikit-learn boto3 kubernetes` (for Python interactions).

## Set up MinIO (local S3 equivalent)
### Explanation of MinIO setup steps for a local S3 equivalent:

#### 1. Run MinIO in Docker
`docker run -p 9000:9000 -p 9001:9001 minio/minio server /data --console-address ":9001"`

**-p 9000:9000** → Exposes MinIO’s S3 API on port 9000.

**-p 9001:9001** → Exposes MinIO web console on port 9001.

**minio/minio** → Official MinIO Docker image.

**server /data** → Tells MinIO to store data in the /data folder inside the container.

**--console-address ":9001"** → Makes the web-based admin console accessible at port 9001.

After running this, you’ll have a fully functional local S3-compatible storage.

---
#### 2. Access the MinIO Console

Open your browser and go to:

http://localhost:9001


Login using the credentials shown in the Docker logs (usually MINIO_ROOT_USER and MINIO_ROOT_PASSWORD).

---

#### 3. Create a Bucket

Inside the MinIO console, click “Create Bucket”.

Give it a name (e.g., ml-datasets) and save.

This bucket acts like an S3 bucket where you can upload datasets, model artifacts, and logs.
