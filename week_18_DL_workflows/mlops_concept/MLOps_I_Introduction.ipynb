## What is MLOps?
MLOps, short for Machine Learning Operations, is a set of practices that aims to deploy and maintain machine learning models in production reliably and efficiently.

It is like DevOps for Machine Learning. While DevOps streamlined the process of building and releasing software, MLOps does the same for AI/ML systems. It's the bridge between the experimental, research-oriented work of data scientists and the operational, stability-focused world of IT and engineering.

The idea is to automate and monitor all steps of the ML system lifecycle to ensure continuous delivery of high-performance models. The tasks of MLOps are also experiment tracking, model deployment, monitoring and retraining.

# What is Kubernates?
Kubernetes (often called K8s) is an open-source container orchestration platform. It helps you deploy, scale, and manage containerized applications (like Docker containers) automatically.

# Why Kubernetes?

Kubernetes is used in MLOps because it helps manage and scale ML workloads efficiently. In short:

- Scalability → run and scale ML training/inference across multiple nodes. Adds or removes containers depending on workload.

- Automation → automates deployment, scaling, and resource allocation.

- Portability → ensures models and services run consistently across environments.

- Resource Management → allocates GPUs/CPUs effectively for ML tasks.

- Orchestration → manages multiple ML pipelines, experiments, and services. automates running many containers across multiple machines.

- Self-healing → Restarts failed containers, replaces unhealthy ones.

- Portability → Works across cloud providers and on-premises servers.

- Load balancing → Distributes traffic across containers.

In MLOps, Kubernetes is the backbone for reliable, reproducible, and scalable ML model training and deployment.

# Some MLOps related Concepts and Terms

Orchestration is the automated management, coordination, and scaling of multiple services or containers in a system.

In MLOps or software development, it ensures that all components (like ML models, data pipelines, and microservices) run in the right order, at the right time, and reliably.

Example: Airflow orchestrates ML workflows, scheduling tasks like data preprocessing → training → evaluation automatically.

The concepts and terms related to MLOps are briefly described below:

### Basic Concepts
- MLOps → Practices to streamline ML model development, deployment, and monitoring.

- CI/CD/CT → Continuous Integration, Continuous Deployment, Continuous Training for ML.

- Pipeline → Automated steps for data prep, training, evaluation, and deployment.

- Experiment Tracking → Recording hyperparameters, metrics, and results of model runs.

- Versioning → Tracking changes in data, models, and code.

- Model Registry → Central repository to store and manage trained models.

### Infrastructure & Tools

- Containerization (Docker) → Package ML code and dependencies.

- Orchestration (Kubernetes) → Scale/manage ML workloads.

- Feature Store → Central place to store, reuse, and serve features.

- Monitoring → Track model performance (drift, accuracy, latency) after deployment.
- Scaling → Handle growing data and user demand (training or inference).

### Model Lifecycle Terms

- Model Training → Process of fitting model to data.

- Model Evaluation → Measuring performance with metrics like accuracy, F1-score, AUC.

- Deployment → Making a model available for predictions in production.

- Drift (Data Drift / Concept Drift) → When input data or relationships change over time.

- Rollback → Reverting to a previous stable model version.

# Understanding of Drift Detection Automation

Here’s a concise explanation of how Drift Detection works in MLOps:

**1. Drift Detection:**

The system continuously monitors incoming data or model predictions.

**Data drift:** input features change over time.

**Concept drift:** relationship between input and output changes.

Metrics like KL divergence, PSI (Population Stability Index), or accuracy drop can signal drift.

**2. Trigger Retraining:**

Once drift is detected beyond a threshold, a retraining pipeline is automatically triggered.

The pipeline fetches the latest data, applies preprocessing, and trains a new model version.

**3.Evaluation & Comparison:**

The new model is evaluated on a validation set or shadow production data.

Key metrics (accuracy, F1-score, AUC, etc.) are compared with the current production model.

**4. Auto-deploy:**

If the new model outperforms the existing model, CI/CD pipelines can automatically deploy it.

Tools like MLflow, Kubeflow, or Airflow handle versioning, testing, and deployment.

**5. Safety Measures:**

Often, canary deployment or shadow testing is used first to ensure stability before full rollout.

In short: Drift detection → retrain model → evaluate → deploy automatically if better.

## MLflow (Versioning)
MLflow is a tool for tracking experiments, versioning models, and managing the machine learning lifecycle. It has four main components:

- MLflow Tracking – Logs parameters, metrics, and artifacts of experiments.

- MLflow Projects – Packages code in a reusable format for reproducibility.

- MLflow Models – Stores, serves, and deploys models in different formats.

- MLflow Registry – Centralized repository for model versioning and lifecycle management.

So in general understanding MLflow is the Lab Notebook for our data science experiments.
The MLflow Model Registry is a centralized model store, set of APIs and a UI designed to collaboratively manage the full lifecycle of a model. It tracks experiments, metrics, and manages models.

**Some basic setup**

Install MLflow:
`pip install mlflow`

Run tracking server:
`mlflow ui`



# Airflow (Orchestration)
Airflow is a workflow orchestration tool that helps you schedule, manage, and automate complex ML/ETL pipelines. It doesn't care if the workflow is for ML, data analytics, or ETL (Extract, Transform, Load); its job is to ensure tasks run in the correct order, at the right time, and handle failures gracefully. Think of it as a “manager” that runs tasks in the right order and handles dependencies.

So in nutshell: Automates ML pipelines (ETL → Training → Deployment).

## Key Concepts

**1. DAG (Directed Acyclic Graph):**

- Represents your workflow.

- Nodes = tasks (like training, preprocessing, or evaluation).

- Edges = dependencies (order of execution).

**2. Scheduling & Monitoring:**

- Schedule training jobs daily/weekly/monthly.
- Monitor pipeline runs via the Airflow UI.
- Retry failed tasks, set alerts, and view logs.
- Runs tasks according to the DAG schedule.

**3. Operators**:

Airflow has built-in and community-contributed operators for.  Operators define what a task actually does. The KubernetesPodOperator is key for ML—it lets Airflow tell Kubernetes to run a task in an isolated Docker container. In general term, Airflow is the Factory Foreman that orchestrates the assembly line. Predefined actions for tasks:

- PythonOperator → run Python code

- BashOperator → run shell commands

- DockerOperator → run containerized tasks

**Basic Setup Example**

 Install Airflow:
`pip install apache-airflow`

Initialize database:
`airflow db init`

 Start webserver and scheduler

`airflow webserver --port 8080`

`airflow scheduler`

# GitHub Actions

GitHub Actions makes it easy to automate all your software workflows, now with world-class CI/CD. Build, test, and deploy your code right from GitHub. Make code reviews, branch management, and issue triaging work the way you want [2].

GitHub Actions is a powerful automation and continuous integration/continuous delivery (CI/CD) tool built directly into GitHub. It allows developers and MLOps engineers to define workflows that run automatically in response to repository events, such as pushes, pull requests, or releases.

## 📝 Simple Example Flow

1. Event: push to main

2. Triggers Workflow: ci.yml

3. Workflow has Job: build

4. Job runs on: ubuntu-latest runner

5. Steps:

- Checkout code

- Install dependencies

- Run tests


## Basic Core Concepts

###1. Workflow

A workflow is an automated process defined in a YAML file inside `.github/workflows/.`

**Example:** `ci.yml` for testing, deploy.yml for deployment.

We can have multiple workflows in the same repo.

###2. Event

An event is what triggers a workflow.

**Examples:**

`push` → when code is pushed to a branch

`pull_request` → when a PR is opened/updated

`schedule` → run at specific times (cron jobs)

`workflow_dispatch` → manually triggered

###3. Job

A job is a set of steps executed in a runner (virtual machine).

Workflows can have one or multiple jobs (e.g., build, test, deploy).

Jobs can run in parallel or sequentially.

###4. Step

Each job is made up of steps.

A step can run a script (run:) or use an action (uses:).

**Example:**

Steps:
  - run: `echo "Hello World"`
  - uses: `actions/checkout@v3`


5. Action

An action is a reusable unit of code that performs a task.

**Example:**

`actions/checkout@v3` → checks out repo code

`actions/setup-python@v4` → sets up Python

We can also write our own custom actions.

###6. Runner

- A runner is the machine where jobs are executed.

- GitHub provides free hosted runners (Linux, Windows, macOS).

- We can also use self-hosted runners for custom environments.

# Amazon S3 (Storage)
Amazon S3 is a highly scalable, durable, and secure object storage service provided by AWS. Instead of organizing files in a traditional filesystem, S3 stores data as objects inside buckets. Each object consists of the data itself, metadata, and a unique key (like a filename).

**S3** provides the **storage backbone for an MLOps workflow**, making datasets, models, and logs easily accessible, versioned, and shareable across the pipeline.

**S3 Design [3]**

Buckets and objects can be managed via the console, SDKs, or REST API, with features like versioning, access control lists, static web hosting, authenticated URLs, BitTorrent support, and integration with file-system-like tools, though not fully POSIX-compliant.

**S3 Broad scope and competing [3]**
The Amazon S3 API defines standard rules for interacting with storage (uploading, downloading, managing files), and its wide adoption has made it a common standard in the industry. As a result, competitors like Cloudian, Backblaze B2, and Wasabi offer S3-compatible storage, allowing the same API commands to work across different providers.

**Amazon S3 Storage Classes:[3]**
S3 offers nine storage classes with varying durability, availability, and performance. Standard is for general-purpose frequent access, Express One Zone provides low-latency single-zone storage, and Glacier classes are for archival storage, separate from Amazon Glacier.

**File Size Limits:**
S3 objects can range from 0 bytes to 5 TB. Files larger than 5 GB must use multipart uploads, and objects above 5 TB must be split into chunks before uploading.

**POSIX (Portable Operating System Interface) [3]** is a family of standards specified by the IEEE for maintaining compatibility between operating systems. In essence, it's a rulebook that defines how a "proper" Unix-like operating system should behave, especially how it manages files, processes, and command-line utilities.
