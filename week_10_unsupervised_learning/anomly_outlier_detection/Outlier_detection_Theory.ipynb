# Outlier detection theory
When we fit models to training data, our hope is that the model will capture the real essence of the data while ignoring the noise. Noise is almost inevitable in real world datasets. It might arise from predictable phenomena like random fluctuations in sensors, or unimaginable events that are beyond our control.

If left unchecked, the model might interpret this noise as real data, and change itself accordingly to accomodate it. We do not want this. By definition, noise is random fluctuation in the data. So we do not want our model to do worse on actual data to perform better on outliers.

Some models are especially sensitive to outliers.

![model](https://i.postimg.cc/13ntDxWC/image.png)

Suppose you have data that is represented by the grey ellipse above. A regression model is fitting it well for now.

![model](https://i.postimg.cc/m2FfwbsH/image.png)

When it learns on the same dataset, but with one outlier (shown in blue), the outlier might "pull" the model towards it, thus making the model fit a little worse. Of course, this outlier might have been produced accidentally, and will not occur again. But, since it has changed our model, we will not be able to predict on new data as well as we would have done, if the outlier had not influenced our model.




## Robust covariance matrix

This is one of the ways we can deal with outliers. We assume that the data is coming from a gaussian distribution, and from this assumption, we try to define the "shape" of data. During this process, we build a robust covariance matrix by ignoring points that are outside the central body of the data.

*Note:* If the data is not gaussian distributed, this algorithm will not work.

![Robust covariance matrix](https://i.postimg.cc/pTWbP83N/image.png)

Consider the distribution shown above that has a mean shown by the yellow circle. There are two outliers, one red and the other blue. The red is very far away from the mean, so it's distance in the euclidean sense will be greater than the blue  dot.

Euclidean distance = $\sqrt{(x-\mu)^T(x-\mu)}$

Assume that this is a gaussian distribution. If we draw the covariance of this data, it will have the shape of the grey ellipse.

![Robust covariance matrix2](https://i.postimg.cc/s26dD1zv/image.png)

Now that we have taken the covariance of the data into account, we can see that the blue outlier is actually further than the red outlier. This type of distance where we also take into account the covariance of the data is called the Mahalanobis distance.

Mahalanobis distance = $\sqrt{(x-\mu)^T\Sigma^{-1}(x-\mu)}$

In other words, the blue dot is further away in Mahalanobis distance but closer in Euclidean distance.



## Novelty detection - One class SVM

This algorithm does not do as well with outlier detection, but works best with novelty detection. Novelty detection is when we assume that the training data does not have outliers, and we are more interested in new data that comes to us. We are more interested in knowing if this new datapoint is an outlier or not.

![Novelty detection](https://i.postimg.cc/xCzxDd1r/image.png)

In the figure above, the SVM has made a decision boundary shown in red, which is very close to the training observations shown in white. As new data comes in, it will classify them as regular observations (purple), or abnormal observation (yellow) with respect to how far these new instances lie from the decision boundary.



## Isolation forest
![Isolation forest](https://i.postimg.cc/2jdG4Mqx/image.png)

As the name suggests, our goal is to "isolate" outliers. This algorithm is very similar to the Extra-trees algorithm. It selects a feature at random and then randomly splits at a point between its minimum and maximum value. The maximum number of splits required to isolate a given instance is a measure of how normal this value is.

Similar to the random forest algorithm, we also average this out over many different trees, hence giving a better result over a dataset.

## Local outlier factor
![Local outlier factor](https://i.postimg.cc/1zxV7R2V/image.png)

This algorithm draws from the Nearest Neighbors algorithm. It computes a score based on the deviation of a given instance with respect to its neighbors. Outliers will have the lowest density of neighbors around them, so they can be identified.

The score of an observation is equal to the ratio of the average local density of his k-nearest neighbors, and its own local density: a normal instance is expected to have a local density similar to that of its neighbors, while abnormal data are expected to have much smaller local density.

In the figure, you can see that the outliers have small scores, represented by the large red circles, while the inliers have a large score.

*Note:* To use this algorithm on unseen data, we have to set the parameter `novelty` to `True`
