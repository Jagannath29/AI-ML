## Association Rule Mining (ARM)

Association Rule Mining is a data mining technique used to **find hidden patterns and relationships between items in large datasets**, typically in the form of "if-then" rules (e.g., if item A is bought, item B is likely to be bought).


Case Study:-
A retail chain analyzed its point-of-sale transaction data using association rule mining. Surprisingly, they discovered a frequent association between the purchase of beer and diapers, especially during evening hours.

Key Findings:

* **Rule discovered:** If a customer buys diapers, they are also likely to buy beer.
* **Customer profile:** Typically, young fathers buying diapers after work hours also picked up beer for themselves.
* **Time pattern:** This correlation was strongest on weekday evenings.

They placed beer near diapers in the store, which led to increased sales of both items.




## Fundamentals of Association Rule Mining

The process involves three core metrics:

- **Support** measures how frequently an itemset appears in the dataset.
- **Confidence** indicates the probability of an item appearing given another item has already appeared.
- **Lift** evaluates the strength of a rule by comparing it against random co-occurrence.

ARM Workflow

Find frequent itemsets (using minimum support) → Generate rules (based on confidence and lift) → Get useful insights (like customer behavior, cross-selling, and better decisions in retail, healthcare, web mining).

**Algorithms** such as **Apriori, FP-Growth, Partition, and Top-k Association Rules** are widely used for Association Rule Mining (ARM) tasks. In the following sections, we will provide a brief overview of each algorithm and explain how they are utilized in practice.


## Building a Association Rule Mining

```python
import requests  # For sending HTTP requests
from mlxtend.preprocessing import TransactionEncoder  # Encodes transactions to binary matrix
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules  # For frequent itemsets and association rules
import pandas as pd
```

```python
# Getting the dataset through internet
url = 'https://drive.google.com/file/d/1R7UqfKzufvXFFE3Aw6Rj5T3fEbyNUP_w/view?usp=sharing'
response = requests.get(url)
data = response.text.strip().split('\n')

# Convert each transaction line into a list of items
transactions = [line.split(',') for line in data]
```

One-hot encoding converts transactions into a binary format where each row represents a transaction and each column an item, marked True if present. This helps the Apriori algorithm easily identify frequent item combinations and patterns.

```python
#  Transform transactions into one-hot encoded DataFrame
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df = pd.DataFrame(te_ary, columns=te.columns_)

df.head() # Let's look at how our data is
```

### Apriori Algorithm

**Apriori algorithm** works by finding frequent itemsets in a dataset by iteratively exploring item combinations. It starts with single items, counts their occurrences, and eliminates those below a minimum support threshold. Then it generates larger itemsets by joining frequent smaller ones, pruning any that contain infrequent subsets. This process repeats until no more frequent itemsets are found, helping to discover association rules in transactional data.


```python
# Find frequent itemsets with minimum support of 1%
# i.e. 1% minimum support means the itemset appears in at least 1% of all transactions, filtering out rare patterns.

frequent_itemsets = apriori(df, min_support=0.01, use_colnames=True)
```

We generate association rules from frequent itemsets with at least 30% confidence, meaning the chance of buying the second item after the first is 30% or more. This filters for strong, reliable item relationships.

```python
# Generate association rules with minimum confidence of 30%
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.3)

# Show the top 5 rules
rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head()
```

Lets look at the row 0 and see what this rule is indicating. The rule says:
If a customer buys beef, they often buy other vegetables too.

* **Support (0.019725 or 1.97%)** means that about 2 out of every 100 shopping baskets have both beef and other vegetables together.

* **Confidence (37.6%)** means that when someone buys beef, there’s a 38% chance they also buy other vegetables.

* **Lift (1.94)** means buying beef makes it almost 2 times more likely to buy other vegetables compared to just buying vegetables randomly.

So, beef and other vegetables are frequently bought together, more than you would expect by chance.

#### Limitations of Apriori

1. Generates many candidates → high computation time.
2. Not suitable for very large datasets.

### FP-Growth Algorithm

**FP-Growth** finds frequent itemsets by building a compact data structure called an FP-tree that stores transactions without generating candidate itemsets explicitly. It recursively mines the tree to extract frequent patterns, making it faster and more scalable than Apriori, especially for large datasets.

```python
# Find frequent itemsets with minimum support of 1%
frequent_itemsets = fpgrowth(df, min_support=0.01, use_colnames=True)

# Generate association rules with minimum confidence of 30%
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.3)

# Show the top 5 rules
rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head()
```

Let's explain the most relevant rule — row 2. Rule: If a customer buys whole milk and citrus fruit, they often buy yogurt too.

1. **Support (0.010269 or 1.03%):** About 1 out of every 100 transactions include all three items together: whole milk, citrus fruit, and yogurt.

2. **Confidence (33.67%):** When a customer buys both whole milk and citrus fruit, there’s a 33.7% chance they also buy yogurt.

3. **Lift (2.41):** Buying whole milk and citrus fruit makes it 2.4 times more likely to buy yogurt compared to random chance.

Customers who buy whole milk and citrus fruit are about 2.4 times more likely to also buy yogurt, with a 34% chance of this happening. This shows these items are often bought together.

#### Limitaitons of FP-Growth

1. FP-tree construction is computationally heavy.
2. Inefficient for sparse transactions.

For better efficiency, we can use modern techniques like **Eclat** or **deep learning-based models** because they are faster, handle large/sparse data better, and support advanced pattern discovery.
