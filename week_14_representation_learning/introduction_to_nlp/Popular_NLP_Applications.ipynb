### Spam Classification

Figure 1: Spam classification example

 Spam classification falls under the text/content classification domain. Content classification refers to the task of classifying a large amount of textual data in which spam classification is an application engine that classifies either the content of an email, or any document is spam or not.

The most popular example of spam classification is Gmail using its spam classification engine to filter spams from emails and making our life secured. You might have noticed that in Gmail, we have a spam folder where all the irrelevant emails like offers, invitations, or email contents flagged as spam by many other users are filtered.

As you can see in the example image above, we have a spam email on the right side. It was flagged as spam because of various reasons. There was an unusual subject in the email, it was sent from a weird address, and the amount was mentioned in the email. Most of the spams are enforced to manipulate our minds by showing huge amounts or lottery winning information. Spam detection engines filter all those contents. It also looks at the database for past data if similar emails were reported as spam.

Social media analysis is the most popular application of NLP in industries. There's a wide variety of NLP applications that use data from social platforms. Some of them are:
- Sentiment Analysis
- Opinion Mining
- Trending topic detection
- Fake news detection
- Content filtering

In this section, we will discuss a few of them in brief.

**Sentiment Analysis and Opinion Mining**


All companies advertise their product and services on social media. Promoting the product is not sufficient. They have to know whether their advertising made some impact or not. To see the report, they have to analyze each advertisement and look for feedbacks, reactions of the customers/users. This analysis is the phase where NLP based applications come to rescue.

We can use some sentiment analysis or emotion recognition of the user in the particular products from their feedback, comments, or tweets to know how they feel about the product.

Sentiment Analysis is the process of analyzing the user's sentiment being positive, negative, or neutral based on the text. It has mostly helped e-commerce examine the customers' sentiment in popular e-commerce based companies like Amazon, eBay, etc. There are millions of customers who leave reviews for a range of products. It is nearly impossible to read each feedback and analyze how the product is doing in the market. So, in such scenarios, applications like sentiment analysis/emotion recognition come in rescue.


Similarly, opinion mining is a technique to get the sentiment/emotion from the opinioned data. Here, opinion data can be some opinion expressed tweets or comments of a user. These opinions expressed data can be extracted using text mining techniques to filter words that express opinions. It can also be applied to analyze all those feedback from users and prepare some reports of customers' opinions on a particular product. This analysis helps the company in marketing and setting future targets.

**Trending topic detection**

Social media is where every person from different interests gather together and discuss certain topics of their interests. If we analyze those discussion data correctly, we can find some hidden groups and detect some most trending topics.

Trending topics can be seasonal or recent from which companies can take benefit and make Ad campaigns or build strategy focusing on those topics to get more customers.

Google trends are the best live example of trending topic detection. It is not a social media-related but trending search detection engine that helps us explore what most people are searching in google search engines with various filter options.

 **Fake news detection**

 Social media is filled with all kinds of peoples. In the past few years, there have been several misusing social media cases to spread fake news. Social media are used by peoples to sway the opinion of the masses using fake and misleading rumors. There is a lot of work going on to detect and filter such fake news.

 **Content filtering**

 Similar to fake news, social media is filled with irrelevant content. People are using social media to spread inappropriate content. NLP is used extensively to identify and filter out those contents to help companies make efficient strategies.

### Retail Catalog Extraction

Large e-commerce enterprise needs an easy-to-access products catalog. A product catalog is a database of the products that the enterprise deals with or a user can purchase. This database contains product meta descriptions like descriptions, attributes, images, etc. Better product descriptions help the customer choose the right product from the catalog. Such information also helps in better product categorization and helps a recommendation, search engines. User-specific attribute extraction from the product helps to get users to get engaged more.

 For example, you have been planning to get on a diet, and for the past few weeks, you have been shopping for diet foods slowly. What if the particular store's recommendation engine senses your interest and stops showing junk foods, snacks, and instead recommending you with all the best diet foods. People are more tend to buy what they see in front of them instead of always searching. It is because sometimes even we don't know what we want. This kind of recommendation will be possible due to proper catalog extraction.

 Some of the applications in retail catalog extraction are:

 - **Review analysis:** Using NLP to analyze the user-specific reviews and know the customer better.

 - **Product search:** Using NLP to help and guide customers for effective and efficient product search.

 - **Product recommendation:** Using NLP to better recommend users with the product which they will be most interested.

### Health records analysis

Health records data are mostly collected and stored somewhere in unstructured text formats. These records include medical notes, prescriptions, audio transcripts, audio and text data log of experiments, etc. This unstructured storing of data makes it hard to organize, search, and categorize. NLP can help health peoples to explore and analyze this data better and even automate some workflows like:
health-related question answering, relevant patent information extraction, and storing on some structured data storage.

Some of the application in health record analysis are:

- **Patient prioritization and billing:** Prioritize patients according to his health conditions.

- **Clinical information extraction:** Automated the information extraction from patients' unstructured health records.


- **Health assistants:** Assistants for less critical tasks like FAQ bot, Question answering, scheduling, and other assistance.
