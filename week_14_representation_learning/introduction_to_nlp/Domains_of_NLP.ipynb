## Text Representation and Classification






Text representation and classification is one of the most common domains of NLP. Most applications are based in this domain, and it is also involved in the important step of the NLP pipeline.

In the text representation and classification, we deal with full-textual data, which can be anything from a tweet to a huge book corpus, and develop some model to classify them into one or more categories for a given set of possible categories.

Text representation and text classification are two different methods in an NLP pipeline.

Feature engineering is one of the most important steps in Machine learning. No matter how well your model is, if you have bad features, then the model will not perform well. Like feature engineering in ML, text representation is an essential step in NLP, which involves transforming text data into a numeric form that our model can understand and process.

In the example `figure 1` above, we prepared a vocabulary for all unique words in the whole dataset, and for each sample, we are putting one if the word exists in that sample else 0. This kind of representation referred to as the simple bag of words model.

There are various other text representation methods like Tf-Idf vectorization, ASCII representation, etc. to name a few. This example is just a simple one-hot vectors. You will learn more while implementing the NLP pipeline.

## Conversational Agent

Figure 4: Conversational AI workflow example

Since language is all about conversation, The conversational domain is the vast and most common NLP domain. All the other domains can be softly related inside this domain. A conversational agent is a dialogue system that conducts natural language processing and generates automatic responses in human language. These agents represent the practical implementation of language models, usually employed as chatbots. The inputs or response does not have to be conducted with just text. It can be a speech, as well.

If we have some speech-based model, we can transform that speech instead of text using some transformation, and the rest pipeline will be the same. Once we have a speech recognition model, then we can go from text to speech or speech to text according to our preference.

In recent years, conversational agents(chatbots in specific) have become more feasible and useful due to ML and DL's advancement. Chatbots can be used for many different tasks in many other industries, from retail to news and even in the medical field.
