## Introduction

NLTK, an acronym for Natural Language ToolKit, is a free, open-source platform for building python programming language based programs to work with human language data. It is cross-platform, which means that a program can be executed in Windows, Linux, or Mac OS X.

NLTK provides an interface to a large number of a corpus such as Gutenberg,  WordNet, etc. Corpus is a large collection of text. We will talk about corpus later. NLTK also provides many modules/functions for different tasks like classification, tokenization, and parsing, stemming, etc. NLTK is a fantastic tool to work with computational linguistics and play with the human (natural) language. Moreover, NLTK supports the largest number of languages.

### NLTK modules

NLTK is popular for simplicity, consistency, extensibility, and modularity. NLTK provides different modules to deal with different language processing tasks. Few of modules and their functionalities are listed below<sub>[[src]](http://www.nltk.org/book/ch00.html)</sub>:


|Modules| Tasks|
|--------:|:------|
|corpus|accessing different corpus and lexicons|
|tokenize, stem | string processing like tokenizing and stemming|
|tag | part of speech tagging  |
| classify, cluster| different ML algorithms like decision tree, naive Bayes, k-means, etc.  |
|probability | for frequency and probability distribution  |
|metrics| for different evaluation metrics like precision, recall etc. |


We can find the list of all available modules [here](https://www.nltk.org/py-modindex.html).

### Reading a corpus
Corpus is defined as a large collection of text documents, such as books, news articles, etc.

For instance, we can load corpus of Gutenberg corpus as:

`nltk.download('gutenberg')`

Gutenberg corpus is a large collection of free books. [This](https://www.gutenberg.org/) is the link for the site Project Gutenberg. Let's explore this corpus.

__Note :__ A large number of corpora (the plural of corpus) are already available in NLTK. To download all the corpus, you can use `nltk.download()`.



```
import nltk
from nltk.corpus import gutenberg  # corpus reader



# download gutenberg corpus
nltk.download('gutenberg')
```

The gutenberg corpus is now downloaded. Let's explore this corpus.

```
# list of all the methods and attributes
dir(gutenberg)
```

Few common methods in corpus reader are:

- `fileids()` lists the name of the text documents, i.e, each book in gutenberg.

-  `readme()` methods prints the readme file of the corpus.

- `encoding()` gives the encoding type of a text file in the corpus.

```
gutenberg.fileids()
```

Let's read the book `shakespeare-caesar.txt`.



```
nltk.download('punkt') # Corpus of punctuation required for tokenization


# read as raw text
print("*********************")
raw = gutenberg.raw("shakespeare-caesar.txt")
print("Raw texts:\n")
print(raw[: 40])
print("\n")
print("*********************")

# read as word tokens
words = gutenberg.words("shakespeare-caesar.txt")
print("Token of words:\n")
print(words)
print("\n")
print("*********************")

# read as sentence tokens
sents = gutenberg.sents("shakespeare-caesar.txt")
print("Token of sentence:\n")
print(sents[:2])
print("\n")
```

### Reading Lexicon

A lexicon is a collection of words. It is like a dictionary (list of vocabulary). Each word may have additional information like part of speech (noun, verb, etc.) and the meaning/definition.


Let's explore a few lexical resources in NLTK.

__stopwords__ is a list of high-frequency words like I, the, are, etc. These words are called so because they don't carry much semantic information and are filtered from the documents in some applications.

```
from nltk.corpus import stopwords


# Download stopwords corpus
nltk.download('stopwords')

#### Note: stopwords is a corpus reader and 'stopwords' is a corpus. ###

# stopwords from english language
stop_words = stopwords.words('english')

# Print stopwords
print(stop_words)
```

### Tokenization

Tokenization is the process of breaking the text into smaller units called tokens. You will learn tokenization in detail in upcoming chapters. In NLTK, we can tokenize the texts as:

```
from nltk import word_tokenize

text = "Tomorrow Tom and Elizabeth are getting married, I need to buy a gift for them."

# Tokenize
tokens = word_tokenize(text)
print(tokens)
```

Apart from the word, we can also perform sentence tokenization using the sent_tokenize function. This returns the list of individual sentences.

```
from nltk import sent_tokenize

text = "On my God! I almost fall down."
tokens = sent_tokenize(text)
print(tokens)
```

We are already familiar with the basics of a regular expression. We can also perform tokenization using a regular expression as:

```
from nltk.tokenize import RegexpTokenizer

text = "I bought a pair of watch for Tom and Elizabeth which costs $50 each."
tokenizer = RegexpTokenizer(r'[A-Za-z0-9]+')
tokens = tokenizer.tokenize(text)
print(tokens)
```

You can define complex regular expressions to filter out unnecessary patterns while performing tokenization.

### Frequency distribution

We can explore the frequency distribution of words using `nltk.FreqDist()`.
Let's see which is the most frequent words in Shakespeare-caesar



```
words = gutenberg.words('shakespeare-caesar.txt')
freq_dist = nltk.FreqDist(word.lower() for word in words if word.isalpha())
freq_dist.most_common(10)
```

### Encoding

Text is represented in different format in file and in memory. Encoding is used while writing text in the file. To read the text back into the memory (i.e, to decode), we must know how it was encoded (i.e, how it was written). Few popular encodings are UTF-8, ASCII, Latin-2. We will discuss about encoding in detail in the upcoming chapters.



```
# Unicode
unicode = '\u0915'
print("Unicode:", unicode)
encoded = unicode.encode('utf8')
print("After encoding:",encoded)
print("After decoding:",encoded.decode('utf8'))
```
