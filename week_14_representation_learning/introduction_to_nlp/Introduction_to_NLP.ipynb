### Human Language

Before anything of Natural Language Processing, let's understand what language is and how it is related to the fantastic field of Natural Language Processing.


Humans can communicate in different ways, like speaking, listening, making gestures, using signs, written text, etc. Language is a communication tool -- in verbal or written form -- and is used to share our thoughts and beliefs. It is a human language (which is highly sophisticated and advanced) that distinguishes humans from animals. Human language allows the speaker to express thought using sentences comprising subjects, verbs, and objects. For instance, the sentence, "__I love cat__" allows us to express love for cats.  Human language is unique from the language of other animals in its compositionality. This compositionality allows the endless capability for generating a new sentence. For instance, with just 25 different words for each role -- the role of subject, verb, and object -- it is possible to generate over 15,000 distinct sentences. <sub>[[1]](https://www.ncbi.nlm.nih.gov/pmc/articles/PMC5525259/)</sub>


While talking about the animals, they make a few different kinds of sounds/symbols to communicate. For instance, cats can make a few different sounds in different situations, like when encountered with the enemy, asking for food, being with their kids, etc. You can refer to this [video](https://www.youtube.com/watch?v=pjRs_WT8VzM) to get the idea of different sounds made by a cat in different situations.

## Natural Language Processing

We are now familiar with the human language, but what we are interested in is the communication between humans and machines using human language. This phenomenon comes under the umbrella of Natural Language Processing.


Natural Language Processing, NLP, is concerned with the interaction between humans and computers using natural (human) language. To be more specific, NLP is concerned with extracting meaningful information from human language input and producing output in a human-understandable form, such as in human language. NLP itself is a broad field and is closely related to different fields like linguistic, computer science, artificial intelligence, etc.

In a broad sense, NLP deals with understanding and the generation of a human language.

### History of NLP

The origin of NLP is rooted in the 1950s. A few of the impactful events in the history of NLP are listed below:

- __1950__: Alan Turing published an article "Computing Machinery and Intelligence," now called the Turing test -- a test of a machine's ability to evaluate if the device exhibits an intelligence indistinguishable from that of a human.

- __1954__: The period around this time is also called the age of machine translation. Machine translation refers to the task of conversion from one language to another. Attempts to automatic translation between Russian and English were made. The expectation was high, but the growth was slower, so the funding for Machine Translation was reduced.

- __1960__: Chomsky and others made a significant contribution to the field of linguistics.

- __Late 1980s__: Until the 1980s, most of the NLP system was based on complex hand-written rules. But in the late 1980s, NLP was revolutionized by probabilistic and data-driven machine learning models.

- __After 2000__: But after 2000, a humongous amount of textual data became available. And now, deep learning approaches are dominating the field of NLP.



## Why NLP?

As humans, we speak and write in human languages like English, Chinese, Spanish, Hindi, etc. But the computer has it's own language --machine language-- which is enigmatic for most people (except programmers). This is the major hurdle for communication between humans and computers.

Bridging this gap between humans and computers is the main motivative behind the study of natural language processing. This requires us to teach computers to understand human language.


Apart from this, a humongous amount of textual data is generated and is available over the internet. NLP enables machines to understand, interpret them, and hence generates some valuable information out of them. There are a bunch of applications of NLP, which we will discuss in the later chapters.

## Why is NLP difficult?

As a human, we can express ourself in infinite ways in written and verbal form. Human language is full of ambiguity and has no certain standardization, making it difficult even for a human.

Let's see with an example the difficulties with human language that make NLP systems even more challenging to understand.

$$\textbf{A man like him is hard to find.}$$

How do you interpret the above sentence? Think for a moment. How many interpretations does it have? The sentence has two interpretations. While the first interpretation is - the man is invaluable and is hard to find, the second one is - the man is absent/unavailable most of the time and is hard to find.

From the above example, we can see the sentence can have more than meaning/interpretation. Humans use multiple sentences, each with multiple meanings in daily life, and correctly identify the correct interpretation.


Let's see another example,

$$\textbf{When a hammer hits the glass table, it shatters.}$$

We can easily interpret this sentence and extract the correct sense. Here, the term "it" refers to the glass which shatters and not the hammer. But how do we know the glass, but not the hammer, is the shattered one?

Because we know the context and the background (or knowledge) of the world such as the hammer is usually made from metal (which is very hard) and is unlikely to shatter. On the other hand, we know that the glass is very delicate.

It's easy to teach the meaning/sense of these sentences to the human. However, it is challenging to teach computers what the word "it" actually refers to. This is because the computer lacks the background knowledge and the knowledge of the world to grab actual senses from these ambiguous sentences. We will study the different sources of ambiguity in the later part.


Apart from ambiguous sentences, there are other aspects that make NLP even harder. Ever since the evolution of humankind, the language is evolving and/or changing. That's why people call __language is alive__, and it will continue to do so. Apart from coining new words (for example, the word "selfie" is recently recognized as a dictionary word), the choice of words to express his idea or beliefs is different based on demographics like race, age, and socioeconomic class, etc., and changes over time.



Moreover, many words -- for example, in English -- have separate additional meaning apart from its literal meaning. For example, the word "donkey" refers to an animal in its literal meaning. However, it is also used in the sense of stupid.  Similarly, the use of sarcasm and metaphor makes the language more ambiguous.

If we are building an NLP system, we need to consider the text genre to train an algorithm. For example, our algorithm's text might come from Wikipedia, news articles, fiction or non-fiction books, telephone conversations, etc. Moreover, demographics characteristics of the writer (or speaker): their age, gender, race, socioeconomics class can all influence the linguistic properties of the text we are processing.<sub>[[2]](#Book)</sub>.


## Ambiguity

Human language is highly ambiguous instead of computer language (programming language), which is not vague by design.
 Let's discuss different types of ambiguity that occur in language.

__1. Phonological ambiguity:__


Phonological ambiguity can be described as words that sound similar (same) but are different in meaning. Such ambiguity arises in a dialogue-based system.

For example,

  - __I__ vs. __Eye__
  - __No__ vs. __Know__
  - __Peace__ vs. __Piece__

__2. Lexical ambiguity:__

Lexeme is the smallest or basic unit of the lexicon in language that bears some meaning. A lexicon is a list of all words (vocabulary) used in a particular language or dictionary. Lexical ambiguity can be described as words that have identical form but have unrelated or distinct meanings.

For example,

   - __bank__ - commercial bank vs. river bank
   - __mean__ - mathematical average vs. not nice


__3. Syntactic  ambiguity:__

Syntax refers to the structural relationship between words.
In syntactic ambiguity, ambiguity arises not due to the presence of lexeme with several meanings, but because the sentence's syntactic structure causes multiple interpretations. It is also called structural ambiguity.

For example,

$$\text{The tourist saw the child with telescope}$$

There is two possible interpretation of the sentence:
   1. The tourist has a telescope, and he saw a child when watching through it.
   2. A child has a telescope, and the tourist saw him.

__4. Semantic ambiguity:__

Semantics refers to the meaning of the words that make up a phrase or a sentence. Even after the syntax and the meaning of the individual words have been resolved, there is still ambiguity in the sentence. This is demonstrated with the following example.

$$\text{Mark and Elizabeth got married last week.}$$

The ambiguity here is that if Mark and Elizabeth got married to each other or a different person.

__5. Pragmatic ambiguity:__

Pragmatics refers to studying the relationship between the meaning of the words or phrases or sentences concerning the speaker's intention. Sometimes, the whole context is needed to understand the meaning of individual sentences. For example,

$$\text{I love chicken.}$$

The above sentence can be interpreted in two ways based on two different contexts. First, if the person talks about the food they like, then chicken refers to the meat. Second, if people are animals they want, then chicken refers to the chicken as an animal.

### How to avoid these ambiguities?

To avoid these ambiguities, we need an understanding of the world or context in which different words or sentences as a whole are used. For example, the word "bank" appears in two different contexts, and if we are familiar with the context, this ambiguity is not a problem. Similarly, in the sentence, "When a hammer hits the glass table, it shattered". If we have the common sense that that hammer is made from metal and is unlikely to shatter, an ambiguous referral of the word "it" is no longer a problem.


Apart from the understanding of the world and context, we will need to perform different levels of analysis like morphological analysis, syntax analysis, semantic analysis, pragmatic analysis, etc. to avoid ambiguities. We will discuss these in upcoming chapters.

NLP, however, is closely related to multiple fields apart from just computer science. This involvement of a multidisciplinary field is one of the reasons why NLP is hard. <sub>[[3]](#Slide)</sub>.

- __Linguistics:__

 Linguistics refers to the study of language. Under this framework, we study how words, phrases, and sentences are formed.

- __Psycholinguistics:__

 Humans use language to express their idea, beliefs, etc. Psycholinguist is a field that studies how people understand and communicate using human language.

- __Cognitive Modeling:__

 In cognitive modeling, we deal with different models and computational aspects of natural language. By models and computational aspects, we mean some algorithms.


- __Philosophy:__

 Philosophy refers to the study of the semantics of the language. We require the knowledge of the world to understand the semantic meaning.


- __Computer Science:__

 NLP deals with the use of computers to understand and generate human language. NLP is closely related to computer science, where different algorithms are implemented to process the natural language.


- __Artificial Intelligence:__

 In artificial intelligence, we deal with the aspect like knowledge representation and reasoning. In AI, we represent knowledge in some form and use it to perform reasoning and inference.


- __Statistics:__
 Many NLP problems are modeled using probabilistic models. For example, we use probabilistic models like N-gram language models to predict the next upcoming word.

- __Machine Learning:__

 Many NLP problems are modeled using machine learning models. In Machine Learning, based on the text's lexical, syntactic, and semantic features, we automatically learn different rules. For example, the use of decision tree to find if the punctuation "." is the end of a sentence or not.
