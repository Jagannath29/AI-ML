# Encoder-Decoder Architecture

Encoder-Decoder is a model with two parts, the encoder part, and the decoder part. This architecture is the most famous sequence to sequence architecture, which is widely used in industries. Most of the NLP tasks like chatbots, machine translation, speech recognition, question answering, etc. work best with the encoder-decoder models.

Encoder-decoder falls under the Many-To-Many RNN architecture where we have a fixed-length input and the fixed-length output but the input and output length can be different. We can see the high-level illustration of this architecture below:

## The Encoder

Encoder model is a stack of several RNN units(Can use vanilla RNN or LSTM/GRU units for better performance) where each unit accepts a single input and propagates. Let's take an example of a question-answer system, where the input sequence is the collection of all words from a particular question.

As you can see in the image above, similar to the earlier module's RNN examples, we feed each embedded word to an RNN unit. All these words are propagated forward, and in the end, we get a hidden state which encapsulates all the context of a word sequence as a context vector.


Let's understand how the encoder model works(Here, we will see an example with the LSTM unit for convenience).



 <!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1TaGzCOiXyZApumGBvnIlKEoJxZ3cwxz4"  > -->
 <img src="https://i.postimg.cc/SRmpf7Vb/image.png"  >
   
<figcaption>Figure 3: LSTM Unit</figcaption>


As we know that one LSTM unit takes only one element at a time, so $x_t$ is input at time step $t$. $h_t$ and $c_t$ are the internal states at timestep $t$ of the LSTM. Their initial values are set at random, and later those values are learned and optimized by the model.

Let's consider an example of question answering system as shown in the `figure 2` above. Here,
$$
\begin{align}
x_{t1} &= \mbox{how} \\
x_{t2} &= \mbox{are} \\
x_{t3} &= \mbox{you} \\
x_{t4} &= \mbox{?} \\
\end{align}
$$ We feed this input in our LSTM model and the hidden state and cell state from our last unit gives the summarization of whole sequence.

**Note**: Here, each $x_t$ is represented by a vector using the word embeddings.

```
class Encoder:
    def __init__(self, input_size, hidden_size):
        self.hidden_size = hidden_size
        self.embedding = Embedding(input_size, hidden_size)
        self.lstm = LSTM(hidden_size, hidden_size)

    def forward(self, input, hidden):
        embedded = self.embedding(input)
        output = embedded
        output, hidden = self.lstm(output, hidden)
        return output, hidden
```

### Decoder Model

Like the encoder model, decoder is also composed of many RNN units where each unit predicts an output at a timestep. The working of the decoder model is different from the encoder.

For the decoder to recognize the start and end of the sequence, we have to add some tokens like the `<START>` and the `<END>` token at the start and end of the sentece which you can see in the illutration below as well.

The decoder model's initial hidden state and cell state comes from the though vector, i.e., the last state of the encoder model. The decoder will be trained to generate the output based on the information gathered from the encoder.

During decoder training, we use the **teacher forcing** technique where the input at each time step is actual output and not the predicted output from the previous time step. The loss is calculated on the predicted outputs from each timestep and the errors are backpropagated through time to update the parameters of the model.

During the testing phase, the output from one timestep is passed as an input for another timestep in the decoder model.



You can see the summary of the encoder-decoder model from the illustration and a small code snippet below as well.

```
class Decoder:

    def __init__(self, hidden_size, output_size):
    
        self.hidden_size = hidden_size
        self.embedding = Embedding(output_size, hidden_size)
        self.gru = LSTM(hidden_size, hidden_size)
        self.out = FullyConnected(hidden_size, output_size)
        self.softmax = Softmax()

    def forward(self, input, hidden):
        output = self.embedding(input)
        output = ReluActivation(output)
        output, hidden = self.lstm(output, initial_state=[hidden])
        output = self.softmax(self.out(output[0]))
        return output, hidden


```

# Cons of Encoder-Decoder model

Let's look at some of the cons of the encoder-decoder model and possible solutions to tackle those problems.

**Note:** Some of the solutions can be beyond our course scope. Here we are just getting the idea of a solution.

- **Limited Memory:** As we know that RNN units have a small memory capacity, to overcome this issue, we use gated units like LSTM, but when it comes to longer sequences, LSTMs also not handle them. They also have limited memory. The final unit where we try to incorporate and compress all the sequences may not handle all the data, resulting in vanishing gradients and low compression.
    - Solution: We can use the attention mechanism and transformers to tackle memory issues. The attention mechanism stores the output from the previous RNN, and at each timestep, it ranks all outputs by relevancy. The word with the highest rank will be focused on.

- **Deep networks are harder to train:** When we stack many RNN layers to build a deeper network, it might work up to a specific limit. Still, after going beyond, the network's efficiency is decreased, which results in slow training time and poor performance.
    - Solution: We can add some residual connections to solve the issue in deep networks. The input of the current layer will be element-wise added with the output of the current layer.  The added output will be given as the input to the next layer. These residual connections will make the memory states more efficient.

- **Can not address the complexity of grammar:** As we have seen in the RNN module, normal RNNs are unidirectional. They cannot incorporate the information of later words in a sentence to predict the next word.
    - Solution: We can use the bi-directional RNNs to help the network incorporate the later words of a sentence to influence the next word.

# Key Takeaway

 - Input and the output length of the encoder-decoder model can be different.

 - Encoder model is a stack of several RNN units where each unit accepts a single input.

 - Decoder is also composed of many RNN units where each unit predicts an output at a timestep

 - The initial hidden state and cell state of the encoder model is set to random, and for the decoder model, they come from the though vector, i.e., the last state of the encoder model

 - We use the teacher forcing technique to train the decoder model.

 - During the testing phase, the output from one timestep is passed as an input for another timestep in the decoder model.
