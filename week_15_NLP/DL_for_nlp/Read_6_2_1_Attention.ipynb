## Drawbacks of Vanilla Encoder-Decoder Architecture


In an encoder-decoder network, it is assumed that the encoder's final hidden state ($h_{T_x}^e$) embodies all the context of the input sequence. That is, the context vector  is given as:
$$c = h_{T_x}^e$$
$c$ is the context of the input sequence, and $T_x$ is the length of the input sequence.

__Note :__ In $h_{T_x}^e$, $h$ represents the hidden vector, subscript $T_x$ represents final time step which is equal to the length of input sequence, superscript $e$ represents encoder. For decoder, the superscript is $d$.

This context is passed as an initial hidden state to the decoder.
$$h_0^d = c$$

And the successive hidden state at the decoder is
$$h_i^d = g(\hat{y}_{i-1}, h_{i-1}^d) \tag{1}$$

where, $g$ computes the current hidden state at decoder given the previous hidden state and previous output; and can be either a vanilla RNN, a GRU, or an LSTM unit.

The above equation $(1)$ can be represented in term of inputs and output as:

Despite encoder-decoder architecture improving the neural machine translation tasks, it has __some limitations__, which are discussed below.


- The context vector is supposed to represent the context of the whole input sequence but this __context vector alone can't represent all the information present in the input sequence__(especially when the input sequence is long). What actually happens in vanilla encoder-decoder network is that the context vector focus mostly at the latter part of the input sequence. The amount of information of invidual words in a sequence at the context vector(the hidden vector at timestep 5, here, is a context vector) is shown below as:

- Since the context is passed only at the decoder's initial hidden state, __the influence of the context vector fades out as the output sequence is generated__.

The solution to the first problem might be to use bi-directional RNNs(By RNNs, we mean either RNN, LSTM, or GRU). Despite using bi-directional RNNs, it is still difficult to represent the long input sequence by these two context vectors [ $\overrightarrow{h_{T_x}^e}, \overleftarrow{h_1^e}$ ]. Another solution is to use all the encoder hidden states for decoding. But the problem is that with variable input sequence length, we obtain a varying number of encoder hidden states. And we don't know how to incorporate these variable-length context vector for decoding.

The solution to the second problem might be to provide the context vector to every state in the decoder instead of just the first state.


These solutions are proposed just based on the understanding the problem with a context vector in encoder-decoder network and have many shortcomings/issues. The attention mechanism provides a much better and unified solution which somehow relates with our proposed solutions. The attention mechanims has proven to be most effective in dealing with the long term contextual dependency in the sequence. Let's discuss attention mechanism in the next section.


## Attention Mechanism

To overcome the issues in vanilla encoder-decoder architecture, Bahdanau et al. introduced a __joint learning and alignment concept__, commonly referred to as attention. The intuition behind this concept is to __allow the network to focus or pay attention to the most relevant part of input while decoding__.


In the attention mechanism, the decoder has access to the most relevant part of the input sequence while decoding. Each decoding state has access to the previous hidden state($h_{i-1}^d$), previous output($\hat{y}_{i-1}$), and the relevant contextual information from the encoder($c_i$).

$$h_i^d = g(\hat{y}_{i-1}, h_{i-1}^d, c_i) \tag{2}$$

The equation $(2)$ can be represented as:

Here, the context vector, $c_i$, is generated every time during each decoding step $i$ and is computed by considering all the encoder's hidden state. This ensures the $i^{ th}$ decoding step to get appropriate context information from the encoder side.


The contextual information in the attention mechanism is very robust compared to that with the typical encoder-decoder architecture. How? Think yourself. We have already discussed it.

Notice how the context vector $c_{i}$ is computed in attention mechanism. Instead of converting the whole input sequence into a single context vector, a new fresh context is computed for each target word directly from the encoder's hidden states.

Similarly, for the next target word prediction, the alignment weights $\alpha_{2 j}$ are recalculated, which tell where most of the attention be given.

### Mathematical Formulation of attention mechanism

Until now, we understood the working of an attention mechanism. Now, let us focus on the mathematical part involved in attention mechanism. As discussed, the context vector $c_i$ can be expressed as the sum of encoder's hidden state weighted by corresponding alignment weight $\alpha_{ij}$.

As seen in figure 5, the context vector $c$ can be expressed in term of encoder's hidden state as:

$$c_i = \sum_{j=1}^{T_x}\alpha_{ij}h_j^e$$

where $\alpha_{ij}$ is an alignment weight which determines how much weight or focus or importance or __attention__ be given to each encoder's hidden state during the $i^{th}$ decoding state.

The weight $\alpha_{ij}$ of each encoder hidden state is computed as:
$$\alpha_{ij} = \frac{\text{exp}(e_{ij})}{\sum_{k = 1}^{T_x}\text{exp}(e_{ik})}$$

where, $$e_{ij}= \text{score}(h_{i-1}^d, h_j^e)$$ is an alignment model. This alignment model gives the score(which can be thought of as similarity) how well the input around the position $j$ of encoder matches the output at position $i-1$ of a decoder.



One simple way to see how well these vector matches is to compute the similarity using dot product. i.e.,
$$\text{score}(h_{i-1}^d, h_j^e)= h_{i-1}^d.h_j^e \quad \text{ for all j= {1 to $T_x$} at $i^{th}$ decoding iteration }$$


Using a dot product as a similarity measure, the figure below shows the computation flow required to compute the context vector.

But rather than just examining how well these vectors(hidden state of encoder and decoder) match each other, we also want to learn how to align the decoder state with the most appropriate state of encoder during training. It means we want the score to be learnable/trainable, which learns which aspect of the encoder needs to give more importance or attention at the current decoding step.

__Note :__ Can you guess why the term $e_{ij}$ is called the alignment model? As the name suggests, $e_{ij}$ is a trainable model (neural network) that learns how to align the decoder state with the suitable encoder states. So it is called the alignment model.


The two popular alignment models are Luong's multiplicative style and Bahdanau's additive style. These models are mathematically represented as:


\begin{equation}
  \text{score}({h_{i-1}^d}, h_j^e) =
    \begin{cases}
    (h_{i-1}^d)^{\top}. h_j^e & \text{[dot product]}\\
      (h_{i-1}^d)^{\top}Wh_j^e & \text{[Luong's multipicative style]}\\
      v_{a}^\top \text{tanh}(W_1h_{i-1}^d + W_2h_j^e) & \text{[Bahdanau's additive style]}
    \end{cases}       
\end{equation}

Here, ${\top}$ represents the transpose operation.

In Luong's multiplicative style, the alignment model is just a dense layer with a parameter $W$. Similarly, in Bahdanau's additive style, the alignment model is a dense layer with learnable parameters $W_1$, $W_2$, and $v_a$.

 The figure shows the encoder at the bottom left and decoder at the top right side. It illustrates how to compute the context vector $c_i$ at $i^{th}$ decoding state.

As shown in the above figure, the score is calculated by taking the similarity between the previous hidden state of decoder $h_{i-1}^d$ and all the hidden state of encoder $(h_j^e)_{j=1}^{T_x}$. The score calculation is indicated by the dotted circle in the above figure. Note that the score function is a feed forward neural network that gives the score (similarity) and has learnable parameters.  The parameters help to learn how better to align the decoder state with the appropriate encoder states.

Once we compute the score, we then convert it into probability. For this, we use the softmax function, as shown in the above figure. Finally, we add all the encoder hidden states weighted by their corresponding weights to compute the context vector $c_i$. This context vector is used for the current decoding state, as shown in the above figure.

## Key Takeaways

- The context vector in the encoder-decoder network couldn't represent the context of a long sentence.
- The context vector is available only for the first hidden state of the decoder in encoder-decoder architecture.
- Attention mechanism allows which part of the encoder to focus while decoding.
- In the attention mechanism, a new fresh context vector is available to all the decoder's states.
