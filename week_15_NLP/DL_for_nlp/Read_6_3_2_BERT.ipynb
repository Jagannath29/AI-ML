Transfer learning is considerably prevalent in computer vision. We're accustomed to finetuning convolutional neural networks, pre-trained on large-scale classification datasets like Imagenet, on our small datasets. Also, the pre-trained CNN models can be finetuned for entirely different tasks like object localization, detection, segmentation, and so on. However, NLP didn't experience such incredible feats until the emergence of OpenAI GPT and BERT. Pre-training embeddings using CBOW, skip-gram models, Glove have increased the performance of many NLP tasks. The embeddings obtained from these methods are powerful but not context-dependent. Some architectures like ELMo, ULMFiT, OpenAI GPT, BERT have been trained on unlabeled text data in a task-agnostic, unsupervised fashion to process a sequence of input words into a robust context-dependent embedding. We call them pre-trained language models. The pre-training is unsupervised in NLP in contrast to vision applications, where pre-training is supervised. These pre-trained language models can be finetuned quickly to transfer the learned general linguistic knowledge to the downstream tasks without many architectural modifications. These escalated the performance to a new horizon, and more research leveraging the pre-trained language models has been ensued to improve NLU systems.

We'll focus mainly on BERT out of the four architectures we mentioned previously.

## Introduction to BERT

Language model pre-training has been effective for improving several natural language processing tasks. The standard language models are unidirectional. Several tasks like sequence tagging, question answering require context from both directions. Also, the unidirectional operations are sub-optimal for sentence-level tasks. Some prior works like ELMo use a combined left-to-right and right-to-left context to encode the sentences using bidirectional LSTMs. BERT alleviates the unidirectionality constraint and instead uses a bidirectional representation using a masked language model pre-training objective. Also, it exploits the powerful self-attention architecture.

**BERT** stands for **B**idirectional **E**ncoder **R**epresentations from **T**ransformers. The BERT framework consists of two steps: unsupervised pre-training and supervised finetuning. BERT model is pre-trained on a huge unlabeled corpus of text using the next sentence prediction, in addition to the masked language model. The pre-trained BERT model can be then finetuned easily on downstream NLP tasks with just one additional output layer.

BERT has successfully modeled both sentence-level such as paraphrasing and natural language inference and token-level tasks such as named entity recognition and question answering. BERT's arrival created a stir in the Deep Learning community with its state-of-the-art results on eleven NLP tasks. As given in the paper, these results include:
- GLUE score: 80.5% (7.7% improvement)
- MultiNLI accuracy: 86.7% (4.6% improvement)
- SQuAD v1.1 question answering Test F1: 93.2 (1.5 improvement)
- SQuAD v2.0 Test F1: 83.1 (5.1 improvement)

The BERT has been successfully embodied in Google search to both ranking and featured snippets. BERT's ability to better understand the search queries' nuances has assisted search in obtaining more relevant search results. In Google Search On event in 2020, the company announced that almost every English query is processed by the BERT model and expanded to work in dozens of other languages.


## Working of BERT

BERT's model architecture is the encoder from the Transformer architecture. It consists of bidirectional multiple layers with self-attention and fully connected sublayers. You may like to see **Attention is all you need** paper to go into this architecture's details. Figure 2 shows a representative stacked architecture of BERT.

A "sentence" can be an arbitrary span of contiguous text in the BERT framework rather than an actual linguistic sentence. Similarly, a "sequence" refers to the input token sequence to BERT, which may be a single sentence or two sentences packed together. Every sequence's first token is always a special classification token (**[CLS]**). Sentence pairs (say A and B) are packed together into a single sequence with a special token (**[SEP]**) in between. The tokens in the input sequence are shown with magenta-colored boxes at the bottom. These tokens are passed into the stacked encoder layers, and finally, we obtain the last hidden representations corresponding to each input sequence. They encode both left and right contexts. Thus unlike in embeddings from Word2Vec or GloVe, a single word can have different word embeddings based on the context. Hence, we call them **contextual embeddings**.

### Pretraining

Unsupervised pretraining is the first step in the BERT framework. The target is to create a high capacity deep linguistic representation from the large corpora of unlabeled text. The BERT uses two unsupervised learning objectives: the masked language model (MLM) and the next sentence prediction (NSP). Let's have a brief introduction to these topics.

#### Masked Language Model (MLM)

Before delving into the masked language model, what would you answer to fill in the blank for the next sentence?

<center> $\textbf{ We went to the library to ____ books.}$</center>

You may answer *read* or *study* or *fetch* or similar words but not *play*. Such a test is called cloze test (or occlusion test). Identifying the right replacement requires the ability to understand context and vocabulary. The masked language model is based on this test.

For the masked language model, some percentage of the input sequence is masked randomly, and the BERT model is pre-trained to predict these masked words. **[MASK]** token denotes the masking in the input sequence and is placed in the masked token place, as shown in figure 3. The BERT pretraining involves predicting these masked tokens only rather than reconstructing the entire input sequence. The hidden vectors corresponding to the masked tokens are passed to softmax over vocabulary to predict the masked words.

#### Next Sentence Prediction (NSP)

Language modeling doesn't capture the holistic relationship between sentences directly. To add this ability, the BERT model is pre-trained on the next sentence prediction task, which is formulated as a binary classification problem. Specifically, for each pretraining example, when choosing the sentences A and B   
- 50% of the time, B is the actual next sentence that follows A - labeled **IsNext**, and   
- 50% of the time, it is a random sentence from the corpus - labeled **NotNext**.

For this pretraining, the final hidden vector corresponding to the  **[CLS]** token is fed into a linear layer and softmax to predict the associated label. Figure 4(a) shows the training for **IsNext** case whereas 4(b) shows the case for **NotNext** case.

### Finetuning

BERT is famous for its ability to transfer its linguistic representation learned from pre-training to downstream NLP tasks quickly. We can take a pre-trained BERT model and finetune it on a variety of downstream NLP tasks. We may need to add some additional layers to the final hidden token representations depending on the problem. Figure 5 shows the BERT framework with pre-training on MLM and NSP tasks and finetuning the pre-trained model on different NLP tasks (namely MNLI - for natural language inference, NER, SQuAD - for question answering). And, then we finetune all the layers end-to-end for the downstream task with task-specific inputs and outputs. We can obtain noble models adept at the specific task with few epochs of finetuning only.

Let's see how BERT can be finetuned for other NLP tasks in general. We stated earlier that the pre-trained BERT model is finetuned end-to-end with task-specific inputs and outputs. Let's look into how we can format different task-specific inputs and outputs. At the input, sentence A and sentence B from pre-training are analogous to:
- sentence pairs in paraphrasing,
- hypothesis-premise pairs in entailment,
- question-passage pairs in question answering,
- a degenerate text-$\phi$ pair in text classification or sequence tagging.

Similarly, at the output:
- For token level tasks (like POS tagging, QA),  the final hidden token representations are fed into an output layer.
- For classification tasks (like entailment, sentiment analysis), the final hidden vector corresponding to the **[CLS]** representation is fed into an output layer.



## Influence of BERT

BERT is undoubtedly an influential work in the field of NLP. There has been much research that leverages the power of BERT on specific domains following its release. SCIBERT is one of them. The SCIBERT is pre-trained on a large multi-domain corpus of scientific publications. This architecture demonstrated significant improvements over BERT and achieved new state-of-the-art results on several tasks from scientific domains. Another on the line is BioBERT, with almost the same architecture as BERT. BioBERT is pre-trained on large-scale biomedical corpora to mitigate domain shift from common NLP tasks to the biomedical domain. BioBERT significantly outperforms the prior state-of-the-art models on the three representative biomedical text mining tasks: biomedical named entity recognition, biomedical relation extraction, and biomedical question answering.

Similarly, ViLBERT is a step further to multi-modal architecture. It attempts to combine the visual representation of an image with its text description in two separate streams. Thus, the architecture combines two data modals: image and text. Similarly, many other models have been developed, pushing the boundary of NLP further.



## Key Takeaways

- The BERT model takes in the input sequence and outputs the contextual embeddings corresponding to each input token.

- The BERT model architecture is the encoder of the transformers model.

- The BERT framework consists of two phases: pre-training in a large corpus of unlabeled text and finetuning for specific downstream NLP tasks.

- Pre-training is done with two tasks: Masked Language Model (MLM) and Next Sentence Prediction (NSP).

- The pre-trained BERT model can be finetuned with task-specific inputs and outputs by adding some additional layers on top quickly.
