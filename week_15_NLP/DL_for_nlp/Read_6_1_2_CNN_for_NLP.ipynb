## Convolution Filter

The main idea behind using CNN for NLP is to make use of its ability to extract features. In computer vision tasks, we apply CNN to extract important features from the images. Similarly, in NLP, we can use convolution filters on the **embedding vector** of a given sentence to extract useful features such as phrases and relationships between the sequence of words in a sentence. Embedding vectors helps to convert the text into spatial structure like images make it work with CNNs.

Unlike the traditional CNN used for images, **in NLP, CNN is usually made up of 1D convolutional** followed by a pooling layer. We will learn about the 1D convolution later but first, let's look at convolution filters and operations.

In the animation above, you can see how we are running a convolution filter from top to bottom where we have arranged the words of a sentence from top to bottom in a sequence, and each row is a vector of a word.


Here, we have used some padding to keep the output feature vector size similar to the input. Let's look at the simple formula below to see what will hapen to our output size if we do not use any padding and what can be done to keep the same padding.


$$
\mbox{output_size} = \frac{\mbox{input_size} - \mbox{filter_size} + 2(\mbox{padding}) + 1}{\mbox{stride}}
$$


$$
\begin{align}
\mbox{output_size}&= \frac{6 - 3 + 2*0 + 1}{1}\\
&= 4
\end{align}
$$

As you can see above, if we used no padding then our output size will be `4`. Now, to determine what padding we should set, we can use the following formula:

$$
\begin{align}
\mbox{padding_amount} &= \frac{\mbox{filter_size} - 1}{2}\\
&= \frac{3 - 1}{2}\\
&= 1.
\end{align}
$$

You can see that, to keep the output size same, we have to use `1` padding which can be seen in the animation above.



## Max pooling Layers

In CNN for NLP, we use a convolution layer followed by a max-pooling layer to get the representation's abstract form. This procedure is similar to CNN for computer vision.

In the max-pooling layer, we create a pool of some height, and from the pool, we pick the max value.


 <!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1aqorTyiqgtb0UHCjdZcd6j29mERLUMBB" > -->
 <img src="https://i.postimg.cc/hj7sjkq2/image.png" >

<figcaption>Figure 5: Convolution followed by max pooling</figcaption>



In the image above, we use the same size pool as our feature vector and select a max value from there. This process is also known as max-over-time pooling.

The max-pooling results in CNN for images are downsampled or pooled feature maps that highlight the most present feature in the patch. In the max-over-time pooling, our input is a feature vector computed over a 1-dimensional vector of words, and output is the most important feature, one with the highest value for each feature vector.

## Variants of CNN for NLP

Using convolution filters to find the pattern in text sequence is a general idea. As we saw in the embedding chapters, we can use sentence embedding, word embedding, and character embedding for the NLP task. Examples that we saw above was for the word-level ConvNet, where we applied CNN on word-vectors.

Depending on the embedding method we use, we can further expand our CNN and use it on those embeddings.

### Character-level ConvNet

In the character level ConvNet, the model accepts a sequence of encoded characters as an input. The encoding is done by prescribing an alphabet of size $m$ for the input language, and then quantizing each character using 1-of-$m$ encoding or one-hot encoding. The characters' sequence is then transformed to a sequence of such $m$ sized vectors with a fixed length. Any character exceeding length is ignored, and any characters
that are not in the alphabet, including blank characters, are quantized as all-zero vectors. The character quantization order is backward. The latest reading on characters is always placed near the beginning of the output, making it easy for fully connected layers to associate weights with the latest reading.

In the paper *Comparative Study of CNN and RNN for Natural Language Processing* from (Wenpeng Yin, 2017) and his colleagues, where they experimented RNN and CNN on a wide range of representative NLP tasks, concluded that RNNs perform well and robust in a broad range of tasks except when the task is essentially a keyphrase recognition task as in some sentiment detection and question-answer matching settings. In addition, CNNs are faster than RNNs for several tasks and the hyperparameters like hidden size and batch size can make DNN performance vary dramatically. This experiment suggested that the optimization of these two parameters is crucial to the good performance of both CNNs and RNNs.

**Note:** *You can learn more about the experiment and results from the paper from additional materials.*

To summarize the comparision of CNNs and RNNs:

- CNNs are better at capturing semantic of sentence and classification.

- Comparatively CNNs are faster than RNNs in tasks like recognition and classification.

- RNNs performs better where the text's length is important, like machine translation.

- CNNs and RNNs both can be used on a wide range of representative NLP tasks.
