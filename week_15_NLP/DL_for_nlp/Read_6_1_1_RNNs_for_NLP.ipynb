## Recurrent Neural Network(RNN)

A recurrent neural network is different from our usual neural network. In a recurrent neural network, we have a loop through which we use multiple copies of the same old network which shares the parameters.

In our earlier chapters, we saw that we could use Machine Learning based techniques for NLP. Those approaches included using bag-of-words, tf-idf models with/out n_grams language model to process text, and later use ML algorithms like Naive Bayes, SVM, etc. according to the use-case.

The ML-based approach is effective, but we have long sequences and huge data; it is not efficient in some cases. In earlier methods, to remember the text sequence, we used the $n$-grams model, where we used them to compute the probability of a sequence of words. They worked fine for short sequences, but as the length of sequence increases, we have to increase the $n$-grams to keep track of those sequences, which exponentially increases the feature dimensions, so these approaches are not quite efficient.

To recall the n_grams model and its limitations, let's look at the small example below, and later we will see how RNN based approaches are efficient.



### N-grams language model and its limitation

As we know in the $n$-grams model, we have to get the probability of a sequence of words.


$$
\begin{align}
p(w_2 | w_1) &= \frac{\mbox{count}(w_1, w_2)}{\mbox{count}(w_1)} \tag{bi-grams} \\
\end{align}
$$

As you can see in the equation above, to get the bi-grams(2-words sequence), we have to calculate the probability of some word $w_2$ given previous word $w_1$.

$$
\begin{align}
p(w_3 | w_1, w_2) &= \frac{\mbox{count}(w_1, w_2, w_3)}{\mbox{count}(w_1, w_2)} \tag{tri-grams}\\
&\dots
\end{align}
$$

Similarly to get the tri-grams(3-words sequence) we have to calculate the probability of some word $w_3$ given previous word $w_1$ and $w_2$. In this way the chain goes upto $n - 1$ words for $n$-grams.
$$
p(w_n|w_1 \dots w_{n-1}) $=\frac{\mbox{count}(w_1, w_2, w_3, \dots, w_n)}{\mbox{count}(w_1, w_2 \dots, w_{n-1})} \tag{n-grams}\\
$$

After that, to get the probability of a sentence, we have to multiply all those probabilities for $n$ words from its previous $n-1$ words.

$$
p(w_1, w_2, w_3, \dots, w_n) = p(w_1) \times p(w_2|w_1) \times p(w_3 | w1, w2) \dots \times p(w_n|w_1 \dots w_{n-1})
$$

Let's say we have a small corpus of general convertations. Now, given a new starter words:
```
where are we ______
```
we have to find the next word after `we`. If we used a bi-gram model then we have to calculate the probability of all words in a vocabulary given the word `we`. But the only word `we` will not be able to give the context of `where` so we have to take all other words in consideration. In this case we have to use $n$-grams and calculate the probability of next word given whole previous words.

$$
p(\mbox{sentence}) = p(\mbox{where}) \times p(\mbox{where | are}) \times p(\mbox{we | where are}) \times p(\mbox{going | where are we}) \dots
$$

You can see how quickly the $n$-gram approach becomes inefficient. If we consider bi-grams only, then it will also be ineffective in most cases because not every time the next word in a sentence depends on previous words; sometimes, we have to consider very distant words.


### Advantages of RNN
To capture those long word sequences, we use RNN models. You might already be familiar with the building blocks of RNN. Let's see how RNN memorizes the sequence.

Since all those cells are just a copy of the same network, RNN shares the same weights parameter. Below you can see a simple building block of vanilla RNN.

 Above, we have a forward propagation of many-to-many RNN. In each timestep of total timestep $\mathbf T$, it takes $\mathbf{x}$ and $\mathbf{h}$  as an input and outputs $\mathbf{\hat y}$. Let's look at the single-cell in a timestep below.

Here we have an RNN cell, and as you can see, we take an input $\mathbf{x}$ from a current timestep $\mathbf t_1$ and hidden state $\mathbf{h}$ of the previous timestep $\mathbf t_0$. Our inputs are passed in the cell multiplied with some weights, and we sum those vectors element-wise(indicated by the circle with a plus symbol). After getting a sum, we pass it through a hyperbolic tanh activation function. Now from the resulting value, we can calculate a $\mathbf{\hat y}$ on the same timestep $\mathbf t_1$ by multiplying it with some weights and passing it through some activation function of our choice(here we are using a softmax, but if the problem statement is of binary classification we can use sigmoid as well).

Let's look at the small code snippet of the RNN cell.

```
class RNN:
    def __init__(self, prev_h):
        self.h = prev_h

    def step(self, x):
        # update the hidden state
        self.h = np.tanh(np.dot(self.W_hh, self.h) + np.dot(self.W_hx, x))
        # compute the output vector
        y = softmax(np.dot(self.W_yh, self.h))
        return y, self.h

```



The similar process is repeated in other cells as well and we propagate forward in time. After the forward propagation, we have to calculate a loss value using some loss function $\mathbf{L}$ and some timestep $\mathbf{t}$. After getting the loss value we can now propagate backward in time to update the weights w.r.t the loss.

## Limitations of vanilla RNN

Vanilla RNNs are not powerful enough to memorize very long sequences. They suffer from the vanishing gradients problem, i.e., while learning, as the word sequence increases, it starts forgetting about the previous words. As figure 2 above illustrates, more weights are given to the latest words in the last layer, and the first words are given fewer weights. They are about to vanish.


 <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1LA02XcTK5OF4Q9G_nOsNrnxb7M6FSMq2" >
<figcaption>Figure 10: RNNs giving less weight to the first word in a sequence </figcaption>


There are other types of RNNs which are designed to solve the vanishing gradients problem. We will learn more about them and see how they preserve the memory later.

Another limitation of vanilla RNN is that it is unidirectional i.e., they propagate information in one direction only. It works fine for most of the tasks for not all tasks. For example, we have a fill in the problem of the blank where we have to predict a word in a sequence by considering the later words as shown in the example below:

```
We will be eating a _______ today. Our sister has gone shopping to buy all the required ingredients for pizza.
```

As humans, we can easily read and fill `pizza` as a missing word in the sentence. But in a unidirectional RNN, information flows in one direction only, so there is no context of pizza initially. The idea of filling the word pizza in the blank comes from the later sentence where the `ingredients for pizza` is referenced. To handle such problems, we need bi-directional RNNs.

# Long Short Term Memory(LSTM)

To address the limitations of vanilla RNN, the gated variants like LSTMs can be used. LSTM is the most popular and widely used variant of RNN and GRU is another variant of LSTM.


As the name suggests, LSTMs are the type of RNN with a little longer short term memory. It is similar to the network, which we saw earlier in RNN, but instead of those RNN cells, it has gated LSTM cells. LSTMs can process longer sequences, which are possible due to that gated architecture. LSTMs have different gates that learn to keep the relevant information and forget what is not necessary.

Let's look at a small dummy example for some intuition on what exactly is meant by the "relevant information."

Say we have a sentence:
- `" Yesterday I bought this book which was highly rated by other users. I tried that book, and it turned out to be awesome."`

Now our task is to predict the sentiment. Our LSTM model will learn to keep the relevant information like:

- `book highly rated by other users`
- `really awesome.`

And drop other non-important information.



LSTMs are very popular in industries for several use-cases. Mostly in NLP, LSTMs are used for almost all kinds of NLP tasks. LSTMs have been quite matured in this field and have been used in various industrial projects as well, and most of the state of the art (SOTA) results in NLP are achieved using LSTMs.


 <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=19KuOLMYt28LH6BDyvwj1wAXI6YJjn_3_"  >
  
<figcaption>Figure 12: LSTM Cell</figcaption>


In the image above, you can see a single cell of an LSTM in a certain time step. In the LSTM, we take inputs $\mathbf x$ of the current time step and the hidden state $\mathbf h$ and a cell state $\mathbf C$ from the previous timestep. In LSTM we have a forget gate(denoted by $\mathbf f$ at current timestep $\mathbf t$), an input gate(denoted by $\mathbf i$ at current timestep $\mathbf t$), a cell state(denoted by $\mathbf C$ at current timestep $\mathbf t$) and an output gate(denoted by $\mathbf O$ at current timestep $\mathbf t$). All the gates in RNNS have their purpose through which they keep relevant information only in their memory.

Let's look at the code snippet for LSTM cell and we will see all the gates separately below:

```
class LSTM:
    def __init__(self, prev_h, prev_c):
            self.h = prev_h
            self.c = prev_c
    
    def step(self, x):
        combine = np.dot(self.W_hh, self.h) + np.dot(self.W_hx, x)
        ft = forget_gate(combine)
        candidate = candidate_gate(combine)
        it = input_gate(combine)
        Ct = self.c * ft + candiate * it
        ot = output_gate(combine)
        ht = ot * tanh(Ct)
        yt = softmax(np.dot(W_yh, ht))
        return Ct, ht, yt


```

### Forget Gate

First of all, there is a forget gate which decides which information should be kept and dropped.



 <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1aMNvsqhvtZA0FsNRclbGno4xnBnYteZV" >
  
<figcaption>Figure 13: LSTM forget gate</figcaption>


 As you can see in the image above, forget gate takes **combined** input from previous hidden state and current input. Those inputs are passed through the sigmoid function($\sigma$), which gives outputs between 0and 1. A value near to 0 means to drop, and a value near to 1 means to keep.

### Input gate

Another gate that we have is the input gate.


 <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1NzZKGGIFQPvsBjUpK6aWK-3cRRdMyN9Q"  >

<figcaption>Figure 14: LSTM Cell</figcaption>


In the image above, similar to the forget gate, we pass our previous hidden state and current input in a sigmoid function that decides if the inputs are important or not, and the same inputs are passed to the tanh function. Tanh function outputs the values between -1 to 1, which helps in regulating the network. After that, we multiply the outputs element-wise. Here, multiplying the output from the sigmoid function decides which value to keep and drop.


### Cell state

Now we have enough information to update the cell state.


 <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1BqieyDfXjL2KrfKiN2cONjJRJg7xHmFC"  >

<figcaption>Figure 15: LSTM Cell</figcaption>



You can see in the image above; first, we get the element-wise multiplication of the previous cell state and our output from the forget gate. This gate decides which information is important, and after that, we add(element-wise) those outputs to the output from our input gate, which updates the cell state.



 <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1mtdNYqxtVG9LlkhetDoX5iiNcDlWqlc5"  >

<figcaption>Figure 16: LSTM Cell</figcaption>


 After the addition, we get our new cell state.

 ### Output gate

 Now, in the end, we have an output gate.


 <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1EVbRrGbImF8SHaDIFqCSgxS48I-a6mNG" >
<figcaption>Figure 17: LSTM Cell</figcaption>


 The output gate decides what our next hidden state will be. Our hidden state holds the information of previous inputs, and this is also used for the next input.  We directly pass our previous hidden state and the current input into a sigmoid function in the image above. After that, we pass our new updated cell state to the tanh function and multiply both outputs. This multiplication decides what information the hidden state should carry. The output is our new hidden state.


 In the end, we get the updated cell state and the new hidden state as final output. We can even use the new hidden state for prediction. We can pass it through any activation function(softmax in the image above) of our choice(according to the use-case) and get the prediction output.


## Gated Recurrent Unit(GRU)

Gated Recurrent Unit is pretty similar to LSTM. In GRUs, we get rid of the cell state and use the hidden state instead. It also has two gates, reset and update gate. They have less tensor operation and are faster than LSTMs. There are no specific best. They both can be tried and use whichever gives the best performance.


 <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1jKhp5LQLSA-limdUGPVbumiZG5AyGko6" >

<figcaption>Figure 18: GRU Cell</figcaption>

 In the image above, you can see a single GRU cell at a certain time step. GRU also takes input and a hidden state like vanilla RNN. Unlike LSTM, in GRU, we have a minimal number of gates, which means less tensor operation and fast computation. Here we have a reset gate(denoted by $\mathbf r$ at certain timestep $\mathbf t$) and an update gate(denoted by $\mathbf Z$ at certain timestep $\mathbf t$). The reset gate and update gate in GRU do similar task as that of LSTM, i.e., they keep only the relevant information of the current time step.

# Key takeaway

- RNNs works well with modeling a relationship among distant sequence words.

- In RNN, we propagate forward and backward in time.

- Each cell in RNN is a copy, so they share parameters.

- Vanilla RNN cannot capture long sequences, so other variants like LSTM, GRU is used.

- Bi-directional RNN captures the information flow both way, starts to end, and ends to start independently.
