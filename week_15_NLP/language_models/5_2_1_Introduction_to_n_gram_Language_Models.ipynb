## Intuition and Motivation

Language Models are the probabilities models which assign probabilities to sequences of words. In this chapter we introduce the simplest language model namely **n-gram model**. In the later part of this unit we will discuss **Neural Language Models** and in the later part of the course we will discuss more complex language models such as **RNNs**.


**Why n-gram is relevant to NLP?**

In NLP it is extremely important to capture the context of the words in the sentence. By knowing the context of the words the machine learns to get a better understanding of the meaning of the words in the sentences.

## n-gram

An n-gram is a sequence of n-words which are adjacent to each other in a given sample of text. Dedending on the number of consecutive words there are different variants of n-grams such as unigram, bigram, trigram etc.

Let's understand this with an example of a sample text.

**```I am eating food peacefully```**

* Unigram

An unigram is a sequence of just a single word.

**```{i}, {am}, {eating}, {food}, {peacefully}```**

* Bigram

A bigram is a sequence of two words.

**```{i am}, {am eating}, {eating food}, {food peacefully}```**

* Trigram

A trigram is a sequence of three words.

**```{i am eating}, {am eating rice}, {eating food peacefully}```**

## n-gram Model
An n-gram model is a type of probabilistic language model for predicting the next word in a sequence from (n-1) order Markov model.

Let's understand the defination with the example in the next section.


### Problem Formulation

Let's take an example of a sequence of words **```the waves were crashing on the```**. We want to find the probability that the next word after these sequence of words is **`shore`**.

### How do we calculate this?

This above probability that the word **``shore``** to come after the occurrence of the sequence of words **``the waves were crashing on the``**,  can be computed as follows. We count the number of occurance of the words **```the waves were crashing on the shore```**
and we divide it by the number of occurance of the sequence of words
**```the waves were crashing on the```**.

Therefore,

$$
P(shore | the\; waves\; were\; crashing\; on\; the) =
\frac {count(the \; waves \; were \; crashing \; on \; the \; shore)}{count(the \; waves \;were \; crashing \; on \; the)} \tag{i}
$$
<br>

Note this method for probability calculation is also known as Maximum Likelihood Estimate(MLE) which we will discuss in the next section.

<br>

**This is rather a simple way of estimating probabilities directly from counts of the occurance of the sequences of words. But the problem is that the language is artistic in nature and new sentences are being generated over the time, we can not always be able to count the entire sequence of words or sentences. And even the web is not big enough to give the good estimates and sometimes it's possible that the count of the entire sequence of words or sentences might be zero.**

<br>

Also, if we are interested in knowing the joint probability of occurance of the entire sequence of words **```the waves were crashing on the shore```**, we could do it as follows.

<br>

**Recall Joint Probability from previous chapter**:
$$
P(X_1, X_2, ...., X_n) = \prod \limits_{i=1}^{n} P(X_i| X_1,... , X_{i-1})
$$

Therefore,
<br><br>
$$
P(the\; waves\; were\; crashing\; on\; the \; shore)
$$
$$=$$
$$
P(the) * P(waves | the) \;* \;P(were | the\; waves)\; * \;P(crashing | the\; waves\; were)\; * \;P(on | the\; waves\; were\; crashing ) \;*\; P(the | the\; waves \;were \; crashing\; on) \;*\; P(shore| the \;  waves \;  were\; crashing\; on\; the)   \;\;\;\;\;   \;\;\;\;\; ..... \;\;\;\;\; \;\;\;\;\; (ii)
$$
<br><br>
As you can see, it involves a lot of intermediate steps and it is rather a lot to estimate and it needs to calculate the probability for unigram upto the probability for 7-grams. It is impractical to estimate the probability of larger sequences with this approach. It causes the intractibility issue and we need to collect more statistics as we discussed in the earlier chapter. We need to simplify the problem to a simpler one.

### Intuition of n-gram models

The intuition of n-gram model is that to compute the probability of a word given previous sequences of words we simply approximate the previous sequence of words with recent few words.

For example to estimate the probability of **shore** given  **the waves were crashing on the**, we could approximate this by calculating the probability of **shore** given only the last three words **crashing on the**.
<br><br>
$$
P (shore | the \;waves\; were\; crashing\; on\; the) \approx P(shore | crashing\; on\; the) \tag{iii}
$$

<br><br>
Let's take one more example. To estimate the probability of **shore** given  **the waves were crashing on the**, we could approximate this by calculating the probability of **shore** given only the last two words **on the** as follows:
<br><br>

$$
P (shore | the \;waves\; were\; crashing\; on\; the) \approx P(shore |  on\; the) \tag{iv}
$$

<br><br>
The formal defination of an n-gram model is:

**An n-gram model is a type of probabilistic language model for predicting the next word in a sequence from (n-1) order Markov model.**


By this defination, equation $(iii)$ is a 4-gram model and equation $(iv)$ is a trigram model.

### Bigram Model

A bigram model considers only the last word in the sequence to estimate the probability for the next word. This model is a first order Markov model.

Let's take an above example and approximate the probability by bigram model:

<br><br>
$$
P (shore | the \;waves\; were\; crashing\; on\; the) \approx P(shore | the) \approx \frac{count(the \;shore)}{count(the)} \tag{v}
$$
<br><br>

**Compare equation $(v)$ with equation $(i)$**


As you can see, we have simplified the problem to much simpler one. In this case we only need the counts of **```the shore```** and **``the``**, rather than the whole sequence of words.

Now, also let's calculate the probability of the entire sequence of words using bigram.

**Recall derivation from Markov Assumption for Joint Probability**
<br><br>
$$
P(X_1, X_2, ...., X_n) \approx \prod \limits_{i=1}^{n} P(X_i | X_{i-1})
$$
<br><br>

$$
P(X_1, X_2, X_3, ..., X_n) \approx P(X_1 | X_0) * P(X_2 | X_1) * P(X_3 | X_2) *...* P(X_n | X_{n-1})
$$
<br><br>

Let's apply this to calculate the probability of the complete sequence **```the waves were crashing on the shore```**.
<br><br>
$$
P(the\; waves\; were\; crashing\; on\; the \;shore)
$$
$$
\approx
$$
$$
P(the | < s >) * P(waves | the) * P(were | waves) * P(crashing | were) * P(on | crashing) * P(the | on) * P(shore | the) \tag{vi}
$$
<br><br>

**Compare equation $(vi)$ with equation $(ii)$**

Notice here $X_0 = <s>\;$  which means that this is the beginning of the sentence, we can assume that there is a start $< s >$ token at the beginning of each sentences. We will discuss importance of these tokens in the later sections. For now you can think that there is a pseudo word $<s>$ at the beginning of the sentence as follows also notice there is psuedo word $</s>$ to represent the end of the sentence as below.

**```<s> the waves were crashing on the shore </s>```**

<br>

In this way, we have sucessfully simplified the model and we now only need to estimate the bigram probabilities even for the estimation of a larger sequences of words or sentences.

In the next section we will discuss how to estimate these bigram probabilities using Maximum Likelihood Estimate.

## Maximum Likelihood Estimate

**Maximum likelihood estimation (MLE) is a method of estimation of probability by maximizing a likelihood function, so that under the assumed statistical model the observed data is most probable.**

We get the MLE estimate for the parameters of an n-gram model by taking the ratio of the count of the sequence of words to the count of its preceding words from the corpus so that they lie between 0 and 1.

Let's understand MLE with an example:
Suppose we have the following mini-corpus of three words. Each sentence starts with token $< s >$ meaning the start of the sentence to give the context to the first word and $</s>$ token to represent that it is the end of the sentence.

```
<s> he is good </s>
<s> he is nice </s>
<s> good is the new bad </s>
<s> is he a bad person </s>
```

For Bigram, we could compute MLE as count of present word and word to predict divided by count of all bigrams that starts with the present word.

In mathematical representation,
<br><br>
$$
P(w_n | w_{n-1}) = \frac{count(w_{n-1}{w_n})}{\sum \limits_w count(w_{n-1}w)}
$$
<br><br>

Let's apply this formula to calculate the estimate of the word **is** given **he**. MLE estimate is count of simultaneous occurance of words **he is** divided by the count of all the bigrams that start with **he**.

<br><br>
$$
P(is | he) = \frac{count(he \; is)}{\sum \limits_{w} count(he \; w)}
$$
<br><br>

Here, $w$ refers to all the words that follows the word $he$.

Therefore,
<br><br>
$$
P(is | he) = \frac{2}{count(he \;is) + count(he \; is) + count (he \; a)}
$$

<br><br>
$$
P(is | he) = \frac{2}{3}
$$
<br><br>

We can show that this estimate is similar if we divide count of occurance of the words $w_{n-1} w_n$ by the occurance of word $w_{n-1}$.

<br><br>
$$
P(w_n | w_{n-1}) = \frac{count(w_{n-1}{w_n})}{count(w_{n-1})}
$$
<br><br>

Let's calculate the estimate of **is** given **he**, which is simply the division of the count of occuarnce of **he is** divided by the count of occurance of the word **he**.

<br><br>
$$
P(is | he) = \frac{count(he \; is)}{count(he)}
$$

<br><br>
$$
P(is | he) = \frac{2}{3}
$$
<br><br>

### Log Probabilities

We always represent our model probabilities as log probabilities. The reason behind this is that probabilities are by defination less than or equal to 1 and if we multiply more probabilites together we would end up with numeric underflow. Instead of raw probabilities, if we represent it as log probabilities we get the numbers that are not as small. Also adding in log space is similar to multiplying in linear space. And addition is also computationally efficient than multipication.

We can always get the raw probabilites back taking the exponent of the log probabilities:

$$p1 × p2 × p3 × p4 = exp(log p1 +log p2 +log p3 +log p4)$$

## Key Takeaways
* An n-gram is a sequence of n-words which are adjacent to each other in a given sample of text.
* An n-gram model is a type of probabilistic language model for predicting the next word in a sequence from (n-1) order Markov model.
* MLE estimate/relative frequency is the of n-gram probability estimated by dividing the count of a sequence by the count of its preceding words.
* We generally use trigrams instead of bigrams in practical applications.
* We represent model probabilities as log probabilities to avoid numeric underflow.
