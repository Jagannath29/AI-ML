## Co-Occurrences or Context Windows
In the count-based models of representation of words by embeddings, we used words as contexts in a given size of the context window.    
_For example:_ If the context window is of size $\pm2$, then we will be looking into two words left and right of the current word to calculate the embeddings.

The selection of such context window size depends on the objective of representation. If we take a shorter context window, then the embedding representations are more semantically similar to the same speech parts. The information stored in the embeddings comes from immediate neighbors.

The embeddings from a longer context window for a word, `w` show an only topical association but not similarity with the other words.

_For an example:<sub>[[1.1]](#Book)</sub>_   
**With a window of $\pm2$, the most similar words to the word _Hogwarts (a school from Harry Potter series)_ were the names of other fictional schools: _Sunnydale (from Buffy the Vampire Slayer)_ or _Evernight (from a Vampire Series)_**.

**With a context window of $\pm5$, the most similar words to _Hogwarts_ were other topically related words like _Dumbledore (Professor at Hogwarts), Malfoy (A Protagonist Character), and half-blood.(One of the episodes of the series Harry Potter)_**

In this example, in the first context, the embeddings seem to have captured the semantic that _Hogwarts_ is a fictional school, and words similar to it will be other fictional schools. Still, in the latter context, the embeddings just captured words that are most frequently occurring around the word _Hogwarts_, which also can be necessary for some scenarios.



**We can define the following two semantic association or similarity or co-occurrences between the words:**

#### **1. First-order (1$^o$) co-occurrence**
This type of association is also known as a syntagmatic association and occurs when the words are in the same sentence or are typically near. Such relations are more concerned with the positioning of the words and are focused on capturing the intuition whether the two words are used in the same context or not.

_For example:_ The word `computer` is a first-order associate of `digital or machine` as the words `digital and machine` occur in a similar context to the word `computer.`

#### **2. Second-order (2$^o$) co-occurrence**
This type of association is also known as the paradigmatic association or paradigmatic parallels and occurs when the words have similar neighbors. Such relations are more concerned with substitution and focus on capturing the intuition of whether one word can be interchanged for the another. Such associations are widely used in thesaurus and such systems.

_For example:_ The word `computer` is a second-order associate of `laptop` represents the idea of whether the word computer can be interchanged with `laptop` or not.

>Note: _It is not mandatory that two words must demonstrate only one form of co-occurrences. Even paradigmatic relations can be syntagmatic._

## Analogy Reasoning

Similarly, another one of the most intriguing and crucial semantic properties of word embeddings is their ability to help in analogy reasoning.

`A man is to woman as a king is to what?`   
Many of us will quickly say that: `A man is to woman as a king is to a "queen."` but can an algorithm understand this?

Using word embeddings, we can capture this intuition. It is observed that **"The relative distance or offsets between vector embeddings can capture some analogical relationship between the words."** <sub>[[3.1][3.2.]](#Paper)</sub>

Mathematically,

$$\vec{king}-\vec{man}+ \vec{woman} \approx \vec{queen}\tag{1}$$

or

$$\vec{king}-\vec{queen } \approx \vec{ man}-\vec{woman}\tag{2}$$

Equation 2 implies that **if we semantically remove a queen's properties from a king, it will be analogous to removing a woman's properties from a man**, which intriguingly represents a difference in gender in two different contexts.

Similarly, we have some more examples of such:

`Paris is to France as Rome is to Italy`

`Uncle is to an aunt as nephew is to niece.`

`Bark is to Dog as Meow is to Cat`

`Doctor is to hospital as Teacher is to school.`

`Grass is to green as Sky is to blue.`

The word `Man` is located in the $5391^{th}$ position in the vocabulary, $V$ and is represented by:

`Man (5391)`=$E_{5391}=[-1, 0.01, 0.03, 0.09]$   

Similarly,

`Woman (9853)`= $E_{9853}$ = $[1, 0.02, 0.02, 0.01]$

`King(4914)`=$E_{4914}=[-0.95, 0.93, 0.70, 0.02]$

`Queen(7175)`=$E_{7175}=[0.97, 0.95, 0.69, 0.01]$

Now,

The distance between the vector embeddings of the words `Man` and `Woman` is given by:


$$E_{5391} - E_{9853}=[-1, 0.01, 0.03, 0.09]-[1, 0.02, 0.02, 0.01] \approx [ -2, 0, 0, 0]$$

<br>

Similarly, if we approximate the separation between the words `King` and `Queen,` we get:

$$E_{4914} - E_{7175}=[-0.95, 0.93, 0.7, 0.02]-[0.97, 0.95, 0.69, 0.01] \approx [ -2, 0, 0, 0]$$

Hence,

$$E_{Man}-E_{Woman} \approx E_{King} - E_{Queen}$$

Both the pairs seem to have an equal separation with its pair, giving a sense of similarity in semantic meaning.   

\

Therefore, **Word Embeddings and its resulting vector space give directions of semantic and syntactic meaning interpreted using simple vector operations.**


## Key Takeaways
- Shorter context window results in learning more semantic features, whereas more extended contexts learn only the topical association
- Two types of co-occurrences: First Order and Second Order
- First-order occurs when words are typically near to each other
- Second-order occurs when words have similar neighboring words
- Word Embeddings have the intriguing property of analogical reasoning, suggesting that offsets between word embeddings can capture the syntactic and semantic relationship between words.
