## Intuition and motivation

As we discussed in earlier chapters, we train n-gram model in the training set and we evaluate it's performance in the test set. We want our model to learn in training data and capture it's distribution well and adapt to new and unseen data drawn from the same distribution.

**Generalization is the model's ability to adapt to new and unseen data which are drawn from the same distribution as the training data.**

In this chapter, we will discuss generalization and issues of generalization in ngram models.

## Generalization in n-grams
There are following generalization issues in n-gram models:
* Training on one genre of corpus and testing on another genre
* Effect of value of N in n-grams
* Zeros as generalization types

Let's look at these issues one by one.

### Effect of different genres in training and test set

Suppose, our training corpus is a sport related corups and it contains sentences related to sports such as:
```
Messi scores a goal. Ronaldo took an hat-trick. Dhoni hits a Six.
```

The probability for sports related pharses/words in this senerio will be high. However, after training the n-gram model on the sport corpus, if we evaluate it in a politics corpus, then it assigns a very low probability to the politics related words or pharses such as:
```
GPD is increased by 10 percent.
```

n-gram model is thus dependent on training corpus and probabilities often encode specific facts about the training corpus. Infact, the semantics and grammer in the n-gram depends on the training set. n-grams only work well for word prediction if the test corpus looks like the training corpus. So, we should always be careful to match the genre of training and test set for the better generalization of n-gram models.

### Effect of N in n-gram

As N in n-gram increases, it does a better job of generalization as it is capable of learning the context of the words. The lower value of N in gram captures less context and higher value of N captures more context of the word.

Let's look at an example:
```
                 men
             for men
        step for men
  small step for men
A small step for men
```

Suppose, we have a prediction task for occurance of word **men**. For this task there is no unigram context, the word **for** is the bigram context, sequence of words **step for** is the trigram context is , sequence of words **small step for** is the 4-gram context and sequence of words **A small step for** is the 5-gram context. Clearly, in terms of context 5-gram is the winner as it captures all the information of the word **men**.

However, if we capture all the context in the training set, it does a better generalization in the training corpus but it can not adapt in the new unseen scenarios of word predictions and it will thus lead to overfitting.

```
...
with Amy
with Country
with dog
with dreams
...
``` -->

**Sparsity is the condition of not having enough of something.**

Sparsity in n-grams happen because of limited corpus. As we discussed earlier, language is artistic in nature and new data are being generated over time, we can never be assured of enoguh data. And also, even if we have a huge corpus some perfectly acceptable words will be missing from it.

Let's look at real case senerio of **Shakespear corpus**:

Total Number of words appeared in the corpus are, $N = 787,434$

The unique vocab size is, $V = 26,431$

The possible bigram from the vocabulary will be of size:
2-gram, $V^2= 698,597,761$

<!-- 4-gram, $V^4 = 4.8803* 10^{17}$ -->

Let's look at a particular senerio of bigrams taken from the **Shakespear corpus**. While creating bigrams all texts were converted in lowercase and puntuation marks were treated as words. The table shows the bigram count for a particular sentence **```my mistress with a monster is in love```**.


<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1T3vl2oyh221QQbQ8Q2ZdmpKkuHyF9Are" >  -->
<img src="https://i.postimg.cc/bwgWQmTH/image.png" >
<Figcaption>Fig 1: Bigram for a test sentence from Shakespear corpus and valid bigrams with zero counts</Figcaption>




You can see from the table that, even in this large corpus, bigrams such as **is monster**, **in monster** etc have zero counts. It is because not all the bigram sequence from the vocabulary appear in the training set. The individual words are in the vocabulary, but the related bigram sequence are missing in the training corpus. In the figure you can see some perfectly acceptable bigrams which have zero counts in the training corpus.

<!-- ```
is monster 0
in monster 0
love monster 0
with monster 0
``` -->

Zeros happen because of the missing of n-grams words in the training data.

## Zeros as a generalization type

In the above example, bigrams such as **{is monster}**, **{in monster}** etc have never occured in training data. But they are perfectly acceptable sequence of words, and they can appear in the test set.

However if we calculate their probability, we get 0 as their count in the training set.

i.e
<br>
$$
P(monster | is) = 0
$$

$$
P(monster | in) = 0
$$


Let's see at another example. Suppose our training set contains 4-gram sequences like in figure below:


<!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1WJogeVtUMebqIrjltejraWkqieRBYYDM" >  -->
<img src="https://i.postimg.cc/GpcrRBMz/image.png" >
<Figcaption> Fig 2: Zeros as a generalization type example</Figcaption>



We can correctly estimate the joint probability as well as conditional probability such as,


$$
P(bus | looked \;at\; a)
$$

$$
P(looked \;at\; a\; bus)
$$


but if our test set contains sequence of 4-gram sequence which are missing in the training set such as **...looked at a tiger...**:

The joint probability of the sequence becomes zero.
$$
P(tiger | looked \;at\; a) = 0  
$$

$$
P(looked \;at\; a\; tiger) = 0  
$$

As their probability is zero, the perplexity of the test set containing these n-grams becomes undefined.


This is an generalization issue called zeros which we will address in our next chapter called **Smoothing and Backoff**.

## Unknown words
In the previous section, we saw the issue zero probability for bigrams. These words were in the train set. But there are situations where we have the words which we have never seen before. These words are called unknown words. Let's understand the intuition of unknown words with the concept of closed and open vocabulary.

### Closed Vocabulary

In a closed vocabulary system, the test set contains words only from the known vocabulary so there will be no unknown words. for eg. machine translation uses a look up tables for the translation of the words and these lookup table have a fixed set of vocabulary.

### Open Vocabulary
In most of the cases, we have to deal with words which we have not seen before. These words are called unknown words or **Out of Vocabulary(OOV)** words.

In **Open vocabulary** system, we model these unknown words or OOV words appearing in the test set by adding a $<unk>$ token.

There are two ways of estimating the probabilities of OOV words.

#### First Method
1. Choose a vocabulary (word list fixed in advance)
2. In the text normalization step, assign any words which is not in the the fixed vocabulary we created in the step 1 by $<unk >$ token.
3. Estimate probabilities for $<unk>$ from its count like any other word in the training set.


#### Second Method
If we do not have prior fixed vocabulary, we could implicitly create such vocabulary from the training set based on the frequency.


For example, we could replace by $<unk>$ token if the word count occurs lesser than n times in the training set. Or we could fix a vocabulary size in advance, let's say $V = 50,000$ unique words and we could select top $V$ words by frequency and replace the other by $<unk>$.

<!-- **A language model can achieve low perplexity by choosing a small vocabulary and
assigning the unknown word a high probability. For this reason, perplexities should only be compared across language models with the same vocabularies **(Buck et al., 2014).** -->

## Key Takeways

* Generalization is the model's ability to adapt to new unseen data

* n-gram has mainly three types of generalization issues, namely training and test set of different genres, effect of N in n-grams and zeros as generalization types.

* n-gram is dependent on training corpus and probabilities often encode specific facts about the training corpus

* n-grams only work well for word prediction if the test corpus looks like the training corpus

* High value of N tends to generalize the training data well but it will lead to overfitting.

* Sparsity happens because of limited corpus.

* Unknown words are the words which are not in the training set.

# References

<a name="Book"></a>
1. Book
   * [Dan Jurafsky, James H. Martin, Speech and Language Processing, (2019) - Chapter 3 ](https://web.stanford.edu/~jurafsky/slp3/3.pdf)

<a name="Youtube videos"></a>
2. Youtube Videos
  * Yeung Hau. Sep 24, 2019. Natural Language Processing by Michael Collins, Columbia University P9 01. Line retrieved from https://www.youtube.com/watch?v=uqsJ3ZS1J6I
