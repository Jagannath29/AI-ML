## Intuition and Motivation

Markov models are the probabilistic models used to model randomly changing systems. Markov models assume that future states depend only on the present state and not on the previously occured events. This assumption is also known as Markov assumption. This assumption is the base for reasoning and computation of the model that would otherwise be intractable. That is why Markov models can be applied in the field of predictive modeling. Markov Models are hugely used in the fields like NLP, Time Series Prediction and Reinforcement Learning etc.

In the field of NLP,  n-grams are one of the significantly used models for prediction tasks such as word prediction, spelling correction and speech recognization etc. N-gram models are based on the the Markov assumptions. In this chapter, we will introduce Markov assumptions and in our next chapters, we will use the concepts of Markov assumptions to understand the workings of n-gram models.

## Markov Models and Markov Assumption

Markov assumption states that the probability of future given past can be approximated as the probability of future given some recent past events or the present.

Intuitively,
<br>
Markov Approximation of future based on present can be expressed as:
<br><br>
 $$
 P(Future | Past, Present) \approx P(Future | Present)
 $$
<br><br>

Markov Approximation of future based on present and some recent past events can be expressed as:
<br><br>
 $$
 P(Future | Past, Present) \approx P(Future | Recent\; Past, Present)
 $$
<br><br>



 Markov Models are the probabilistic models which assumes that future can be predicted without looking too far into the past.

## Why Markov Models?

Markov models solve the problem of intractability and still keeps the reasoning and computation in predictive modeling systems. Let's understand this with an example.

Suppose that we have three types of weather sunny, rainy and foggy and suppose that one kind of weather persists for the whole day.

Our aim is trying to predict what the weather will be the next day based on a history of observations of weather.

Let's say we collected a sample from the observations we have taken. We observed in that sample for Day 1 the weather was sunny, for Day 2 the weather was cloudy, for Day 3 it was foggy, for Day 4 it was sunny and for Day 5, it was foggy.

We want to find the probability for the weather to be sunny at Day 6, given the past weather observations.
<br><br>
$$
P(W_6=sunny | W_1=sunny,W_2=rainy,W_3=foggy,W_4=sunny,W_5=foggy)
$$
<br><br>
Here's the problem in finding this probability:
As we take account for more past events, we have to collect more statistics.
For instance, to calculate the probability of weather for the $6^{th}$ day, the statistics we must collect is $3^6 = 729$ past histories.

This particular example might not be a problem for a modern computer to solve but imagine a situation where you are trying to predict the next word in a NLP task, where your vocab size is 10,000 and the sentence length is 10. In this case you need to collect the statistics for $10,000^{10}$ past histories, which is a very very large number.

Let's address this issue by Markov Assumption. Instead of looking at the whole past weather events, we will now be looking only at the weather at Day 5 to predict the weather of Day 6.

The simplified assumption thus is:
<br><br>
$$ P(W_6=sunny | W_1=sunny,W_2=rainy,W_3=foggy,W_4=sunny,W_5=foggy) \approx P(W_6=sunny | W_5=foggy) $$

<br><br>
In this case statistics we need to collect statistics of $3^2=9$ past histories which is a very significant drop in figures. The probabilities below are the dummy observed probability for weather at day 6 given day 5.

### Conclusion
So we can conclude that there is a trade off between computation and statistics we need to collect versus simplicity. Markov models greatly reduce statistics and computation, simplifies the model and are highly suitable for tasks which are otherwise very difficult to solve. Markov models can be applied to systems where present or recent past is sufficient representation of the whole past.

## Mathematical Formulation of Joint Probability using Markov Assumptions

### Joint Probability

The conditional probability of B given A is given as:
<br><br>

$$
P(B|A) = \frac{P(A, B)} {P(A)}
$$

<br>

Where $P(A, B)$ is the joint probability of $A$ and $B$ i.e probability that event $A$ and event $B$ both occurs simultaneously.

and $P(A)$ is the probability of occurance of event $A$.

<br>

Thus, the joint probability of simultaneous occurance of event $A$ and event $B$ is:

<br>

$$
P(A, B) = P(A) * P(B|A)
$$
<br><br>
To calculate the joint probability of $(A, B)$ we need to know the probability of event $A$ and probability of $B$ given $A$
<br><br>
Similarly we can formulate the joint probability for events $A, B, C$

<br><br>
$$
P(C | A,B) = \frac {P(A, B, C)} {P(A, B)}
$$
<br><br>
i.e
$$
P(A, B, C) = P(A, B) * P(C | A,B)
$$

### Chain Rule of Probability



For a random variable $X_1, X_2$, the joint probability of  $X_1, X_2$ is calculated as the product of probability of event $X_1$ and conditional probability of $X_2$ given $X_1$.
<br><br>

$$
P(X_1, X_2) = P(X_1) * P(X_2 | X_1) \tag{i}
$$
<br><br>

For a random variable $X_1, X_2, X_3 $, the joint probability is calculated as :
<br><br>

$$
P(X_1, X_2, X_3) = P(X_1, X_2) * P(X_3 | X_1, X_2)  \tag{ii}
$$
<br><br>

You can see that the joint probability of $X_1, X_2, X_3$  depends on the joint probability of $ X_1, X_2 $ and the conditional probability of $X_3$ given events $X_1, X_2$ have already occured.
<br><br>


Substuting (i) to (ii)
<br><br>

$$
P(X_1, X_2, X_3) = P(X_1) * P(X_2 | X_1) * P(X_3 | X_1, X_2)  \tag{iii}
$$
<br><br>

### First Order Markov Approximation

The first order markov approximation states that the probability of an observation at timestep $i$ depends only on the observation at timestep $i-1$
<br><br>

$$
P(X_i| X_1,... , X_{i-1}) \approx P(X_i | X_{i-1}) \tag{v}
$$
<br><br>

For joint probability, substuting (vii) into (vi), we get
<br><br>

$$
P(X_1, X_2, ...., X_n) \approx \prod \limits_{i=1}^{n} P(X_i | X_{i-1}) \tag{vi}
$$

If we expand the Pi notation we get,
<br><br>

$$
P(X_1, X_2, ...., X_n) \approx P(X_1) * P(X_2|X_1) * P(X_3|X_2)*...*P(X_n|X_{n-1}) \tag{vii}
$$



### Second Order Markov Approximation (Optional)

The second order markov approximation states that the probability of an observation at timestep $i$ depends on the observation at timestep $i-1$ and timestep $i-2$.
<br><br>


$$
P(X_i| X_1,... , X_{i-1}) \approx P(X_i | X_{i-1}, X_{i-2}) \tag{viii}
$$
<br><br>

For joint probability, substituting (ix) into (vi), we get
<br><br>

$$
P(X_1, X_2, ...., X_n) \approx \prod \limits_{i=1}^{n} P(X_i | X_{i-1}, X_{i-2}) \tag{ix}
$$

## Key Takeaways
* Markov models assume that future is independent of past given present
* Markov Assumption significantly reduces statistics and solves the problem of intractability in predictive modeling systems
* Markov Models can be applied to processes when present or recent past is sufficient representation of the whole past.
