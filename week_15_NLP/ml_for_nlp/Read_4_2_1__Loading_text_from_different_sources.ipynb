## Loading Data from Disk

If you glance at different folders/locations in your computer, you will come across different files like pdfs, word documents, spreadsheets, etc. that have textual data in them. We can use python to read data from these files. We can also write or append data to these files, but those are not our concerns for now.

Reading a file in python involves three steps:
1. Opening a file
2. Reading the content of the file
3. Closing the file

**1. Opening a file**
To open a file in python, we use the `open(file_name, mode)` method, which takes in two arguments: `file_name` and `mode`.

 * `file_name`: It is the name/location of the file.
 * `mode`: It specifies the task that we need to perform. To read a file, we use `r` as the `mode`, whereas to write to a file, we use `w` as the `mode`.

**2. Reading the content of a file**
We can read the content of a file in two ways: each line and entire file at a time. To read each line, we use the `readline()` method, and to read the entire file at a time, we use the `read()` method.

**3. Closing the file**
Finally, after we are done with the file, we need to close it using the `close()` method.

The following example opens a file named `demo.txt` and reads the file's content using both `readline()` and `read()` method.



````
# Reading each line of a file at a time

print("#"*10+ " Reading each line at a time " +"#"*10)
f = open('demo.txt', 'r') # open demo.txt in read mode
print(f.readline())  # reads the first line
print(f.readline()) # reads the second line
print(f.readline()) # reads the third line
f.close() # closes demo.txt file

print("#"*10+ " Reading entire file at once " +"#"*10)

# Reading entire file at a time
f = open('demo.txt', 'r') # open demo.txt in read mode
print(f.read())  # reads the entire file
f.close() # closes demo.txt file
````

However, it is better to use the `with` keyword when dealing with file objects because it automatically takes care of closing the file. Using the `with` statement, the above code blocks can be written as:

````
print("#"*10+ " Reading each line at a time " +"#"*10)

# Reading each line of a file at a time
with open('demo.txt', 'r') as f:
  print(f.readline())
  print(f.readline())
  print(f.readline())

print("#"*10+ " Reading entire file at once " +"#"*10)

# Reading entire file at a time
with open('demo.txt', 'r') as f:
  print(f.read())
````

## Loading Data from Web

The web is a gigantic ocean having all sorts of data like text, images, videos, etc. The presence of humongous data makes the web a perfect place to look for textual data. Two popular methods for gathering data from the web are:
* Using API   
 Many organizations and entities like Twitter, Google, Facebook, etc., provide public access to their data through their private APIs. The access may be free or paid.

* Web Scraping   
 Web scraping is a technique to extract data from the web automatically. It is used to extract any data like images, text, videos, etc. One important thing to consider while extracting text data from the web is that the text's formatting is not preserved during the scraping process. All texts have the same style/formatting after scraping.

In this notebook, we will focus on web scraping using python, requests, and Beautiful Soup. `requests` is a python library used to make HTML requests, and `Beautiful Soup` is a library used to extract data from HTML and XML documents. We won't go into the details on how to use each of these libraries, but you may refer to their documentation ([`requests`](https://requests.readthedocs.io/en/master/) and [`Beautiful Soup`](https://www.crummy.com/software/BeautifulSoup/bs4/doc/)) if you like to know more. We will use these libraries to extract text from the **About** page of <https://fusemachines.com>.

````
# Get the about page
import requests


URL = 'https://fusemachines.com/about'
response = requests.get(URL)
print(response.status_code)
````

A status code of 200 means that we successfully got the response back. The entire HTML page is now stored in the `response` variable.

Now, we will use `Beautiful Soup` to extract the content from the page. For this, we will create an object of the `BeautifulSoup()` class and initialize it with two parameters, the content of our response and the HTML parser.



````
from bs4 import BeautifulSoup


soup = BeautifulSoup(response.content, 'html.parser')

# Get title of the page along with the <title>...</title> tags
print("\nTitle with tags")
print(soup.title)


# Get title of the page along without the <title>...</title> tags
print("\nTitle without tags")
print(soup.title.string)

# Get first 10 links from the document
print("\nFirst 10 Links")
for link in soup.find_all('a')[:10]:
    print(link.get('href'))
````

````
for script in soup(["script", "style"]):
    script.extract()    # removes all the javascript and css code from the html document

text = soup.getText() # gets all text from the html document
text = text.replace("\n\n", "") # Optional step, replaces two consecutive newline character with empty string
print(text)
````
