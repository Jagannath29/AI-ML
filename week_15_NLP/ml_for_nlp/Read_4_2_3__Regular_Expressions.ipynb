### Regular Expressions

**Regular expressions** or simply regex is a combination of characters and symbols that represent a pattern. They are widely used for search and replace operations, information extraction, and validation.

In regex, characters have special meanings. Let's start with the most basic ones.

Apart from these basic patterns, there are some common operators that capture repeated occurences. They are presented below.

#### Regex pattern that capture repetitions

* **kleene star(\*)**    
 This is used to capture 0 or more occurrences of the character/group it follows.    
 [abc]* - captures 0 or more occurrences of any of the characters a, b, and c.   

* **kleene plus(+)**   
 This is used to capture 1 or more occurrences of the character/group it follows.    
 [abc]+ - captures 1 or more occurrences of any of the characters a, b, and c.

* **optional(?)**   
 It matches zero or one occurrence of the character/group it follows.    
 [abc]? - captures zero or one occurrence of any of the characters a, b, and c.

  
    
In regex pattern matching, **the parenthesis-() and square brackets-[ ]** have a special role. **The square brackets are used to indicate a set or a range of characters.** Like mentioned in the table above, [abc] matches a set of characters 'a', 'b', and 'c'. [a-z] all characters from a to z. Other patterns like \w, \s, etc. work inside square brackets too, but the period (.) loses its meaning and just means a literal period.

**Parenthesis-() is used for extracting parts of the matched text**. For example, ([a-z]+)(\d+) matches a text in which digits follow alphabetical characters. The parenthesis then groups all alphabetical characters in one group and all numbers in a separate group.    

Now, let's discuss some common use of regular expressions with an example.



### Regex in Action

Now let's use our knowledge of regex to find patterns in the given text:    

> Whatsapp was bought by facebook.com and is now very popular throughout the world. Contact number for facebook's office is +1-934-343-4356.   

* **How can we find the first word of each sentence?**   
 Note that the first word of each sentence has a pattern. It starts with a capital letter and then is followed by one or more small letters. The equivalent regex is:   
 * [A-Z][a-z]+  

* **How can we extract all URLs from the given text?**   
 Note that the url in the text has a word followed by a period and a word. The equivalent regex is:   
 * [a-z]+\\.[a-z]+  
 * \w+\\.\w+   

* **How can we extract phone number from the given text?**    
 Note that the phone number in the text starts with a + sign followed by a number and then has a - sign followed by three or four numbers, and this pattern is repeated thrice. The equivalent regex is:    

 * \\+[0-9]+-[0-9]+-[0-9]+-[0-9]+    

**Note**: If a character represents a regex operator, it should be escaped with a '\\'. In the last two patterns, note \\ symbol before the +.    


Apart from pattern matching and information extraction, regex is also used for validation. For example, if you have an online form where the user submits information like first name, last name, email, etc. you want to make sure that names don't contain numbers; email is in the correct format(user@domain.com), etc. This can be done with respective regular expressions.

For the first name and last name validation, we can use the pattern: [a-z]+

For email validation we can use the pattern: \w+@\w+\\.\w+ .

## Regex in Python

To perform pattern matching using python, python provides a module called **re**. This module provides us various methods to perform pattern matching. Some of the most commonly used methods are:
* search()
* findall()
* match()
* split()
* sub()

Let's discuss each of them briefly with an example. If you need a detailed description, you can go through [Regular Expression HOWTO](https://docs.python.org/3/howto/regex.html) in the official python documentation.

We will perform pattern matching on the given text:

> Whatsapp was bought by facebook.com and is now very popular throughout the world. Contact number for facebook's office is +1-934-343-4356.




````python
text = "Whatsapp was bought by facebook.com, and is now very popular throught \
the world. Contact number for facebook's office is +1-934-343-4356."
````

* **`match()`**    
 `match()` is used to detect if the regular expression is matched at the beginning of the string or not.

Syntax:
```
pattern.match(string, start_pos, end_pos)
Parameters:
 pattern : The regex pattern to match.
 string : The piece of text to look for in order to find the occurrence of the given pattern.
 start_pos : An index in the string where the search is to start. The default value is 0.
 end_pos : An index in the string where the search is to stop.
```

````python
import re
# IMPORTANT: compile regex before using it with any method
regex = re.compile('[A-Z][a-z]+') # matches word starting with a capital letter
m = regex.match(text)
print(m)
````

Since our text starts with a word starting with a capital lettered alphabet, we find a match, i.e., `Whatsapp` spanning from index 0 to 8.

However, if our regex represented a number, we would not get any match because our text doesn't start with a number.

````python
m = regex.match(text)
print(m)
````

* **`search()`**    
 `search()` is used to find if there is a match for the given text or not.   

Syntax:   
```
pattern.search(string, start_pos, end_pos)
Parameters:
 pattern : The regex pattern to search for.
 string : The piece of text to look for in order to find the occurrence of the given pattern.
 start_pos : An index in the string where the search is to start. The default value is 0.
 end_pos : An index in the string where the search is to stop.
```

````python
m = regex.search(text)
print(m)
````

If you observe the input text, you will find two matches for the given pattern: "Whatsapp" and "Contact". The `search()` method only returns the first string that matches the provided pattern. If you want to find all possible matches for a given pattern, you need to use the `findall()` method.

* **`findall()`**    
 `findall()` returns a list containing all strings that match the provided regex.    

Syntax:    
```
pattern.findall(string, start_pos, end_pos)
Parameters:
 pattern : The regex pattern to find.
 string : The piece of text to look for to find the occurrence of the given pattern.
 start_pos : An index in the string where the search is to start. The default value is 0.
 end_pos : An index in the string where the search is to stop.
```

````python
m = regex.findall(text)
print(m)
````

We see that `findall()` correctly returns all the strings that match the given pattern.

* **`sub()`**   
 The `sub()` method substitutes the matched string with any string of our choice.

Syntax:
```
pattern.sub(repl, string, count)
Parameters:
 pattern : The regex pattern to substitute.
 repl : The new string/text used to substitute any string that matches pattern.
 string : The piece of text to look for in order to find the occurrence of the given pattern.
 count : The number of occurrences to substitute. By default, replaces all texts matching the pattern with repl.
```

````python
# Replaces only the first string that matches the regex
substitute_only_1 = regex.sub("Hello", text, 1)
print("Text with the first word starting with a capital letter replaced by Hello:\n\n{}"
      .format(substitute_only_1))

print("\n\n")

# Replaces all strings that match the regex
substitute_all = regex.sub("Hello", text)
print("Text with all words starting with a capital letter replaced by Hello:\n\n{}"
      .format(substitute_all))
````

* **`split()`**   
 `split()` method splits the text whenever it finds a match for the given regex and returns a list of string.


Syntax:
```
pattern.sub(string, maxsplit)
Parameters:
 pattern : The regex pattern used for splitting.
 string : The text to look for pattern and split.
 maxsplit : The maximum number of splits to create. By default, it creates all possible splits using the pattern.
```



````python
splitted_text = regex.split(text)
print(splitted_text)
````
