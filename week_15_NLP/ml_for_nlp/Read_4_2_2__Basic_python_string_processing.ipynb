## Search and Replace

### Search

The search operation is used to find the occurrence of a particular string/substring in a text. Searching may be helpful in scenarios when we want to detect if a certain string is present in a text.

To perform a search operation, python provides `find()` method. It returns the first position/index at which the specified substring occurs.


 <!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1TS27vD8H0ydtYKVqgWu3cqZ0_-GGO8xp" > -->
 <img src="https://i.postimg.cc/g2X4sMQ0/image.png" >
 Figure 1: Use of python find() method</figcaption>





The syntax for `find()` method is as follows:



```
string.find(substring, start, end)
Parameters:
 substring -> substring to search for inside `string`.
 start -> Starting index of the search. 0 is default.
 end -> ending index for the search. Default is the length of entire `string`.
Returns:
 The lowest index at which the substring starts in the `string`.

Note: start and end parameters are optional.
```



````python
complete_string = "fusemachines is a great place to work. At fusemachines we work on ML and AI"
print(complete_string.find('fusemachines'))

# searches for 'fusemachines' starting at index 30
complete_string = "fusemachines is a great place to work. At fusemachines we work on ML and AI"
print(complete_string.find('fusemachines', 30))
````

### Replace

Replace is used to search for a particular substring inside the given string and replace it with a new value.

For this, python provides a `replace()` method.

 <!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1Y_7uBMQOwp-NQr9HXUorXTetVdVpPij8"> -->
 <img src="https://i.postimg.cc/hjPdyvv0/image.png">
 Figure 2: Use of python replace() method</


Syntax:


```
string.replace(old_substring, new_substring, count)
```
The above syntax is self-explanatory. By default, all occurrences of `old_substring` are replaced by `new_substring` inside the string. If `count` is specified, only the first `count` occurrences are replaced.



````python
# replace all occurences of the word "fusemachines" by FM
complete_string = "fusemachines is a great place to work. At fusemachines we work on ML and AI"
print(complete_string.replace('fusemachines', 'FM'))

# replace the first occurences of the word "fusemachines" by FM
complete_string = "fusemachines is a great place to work. At fusemachines we work on ML and AI"
print(complete_string.replace('fusemachines', 'FM', 1))
````

## Split, Join, and Strip

### Split

Split operation is used to convert a text/string into a sequence of sub-strings.


 <!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1VAj58YbEhf4uFbEvUa7WubaA-Ghzc0gv"> -->
  <img src="https://i.postimg.cc/15cNQDFK/image.png">
 Figure 3: Use of python split() method
Syntax for the `split()` method:


```
string.split(sep, maxsplit)
Parameters:
 sep: Separates substrings from one another. The default value is whitespace.
 maxsplit: Maximum number of splits to create.
```



````python
complete_string = "fusemachines is a great place to work. At fusemachines we work on ML and AI"
# Each word in the text forms an item of the list
print(complete_string.split())


# Creates 5 splits and the list will contain 5+1 = 6 elements
print(complete_string.split(maxsplit=5))

# Use a period(.) to split
print(complete_string.split(sep="."))

# Use character 'a' to split
print(complete_string.split(sep="a"))
````

````python
p = "fusemachines is a great place to work at."
p.split(sep="a")
````

### Join

Join operation is used to combine a sequence of substrings into a single string/text.


 <!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1yjIgcnbe0S-e24TOO0KjBaLnvgpAwFkN"> -->
 <img src="https://i.postimg.cc/7Ysw06rk/image.png">
 Figure 4: Use of python join() method


Syntax:
```
sep.join(sequence)
 Parameters:
 sep: Separator to be placed in between two substrings while joining them.
 sequence: Sequence of substrings to combine into a single string.
```




````python
sequence_of_string = ['fusemachines', 'is', 'a', 'great', 'place', 'to', 'work.',
                      'At', 'fusemachines', 'we', 'work', 'on', 'ML', 'and', 'AI']

# combines above sequence using whitespace as a separator
print(" ".join(sequence_of_string))

# combines above sequence using hyphen as a separator
print("-".join(sequence_of_string))

# combines above sequence using newline character
print("\n".join(sequence_of_string))
````

### Strip

Strip is used to remove desired characters from the left and right of the string.

 <!-- <img src="https://doc.google.com/a/fusemachines.com/uc?export=download&id=1lsmhIdjoqeaYxPtR_q5ZrHstoMUf_dI9"> -->
 <img src="https://i.postimg.cc/85JtKZj4/image.png">
Figure 5: Use of python strip() method


Syntax:
```
string.strip(chars)
Parameter:
 chars : string specifying a set of characters to be removed.
 All combinations of characters in 'chars' parameter are removed from right and left.  
 By default, it removes the whitespace.
```



````python
text = "   Fusemachines    "
#removes whitespace, the default character
print(text.strip())

text = "Hi, this is fusemachines. How are you doing?"

#removes 'Hi,' and 'doing?'
print(text.strip('Hi,?doing'))
````

The `strip()` method eliminates a combination of given characters from the left and right of the original string. If you want to remove some set of characters only from the left or the right end, you need to use the 'lstrip()' and 'rstrip()' method, respectively.

````python
text = "             Fusemachines          "
#removes whitespace from the right using rstrip()
print(text.rstrip())
#removes whitespace from the left using lstrip()
print(text.lstrip())

text = "Hi, this is fusemachines. How are you doing?"

#removes 'Hi,' and 'doing?' from right usign rstrip()
print(text.rstrip('Hi,?doing'))

#removes 'Hi,' and 'doing?' from left usign lstrip()
print(text.lstrip('Hi,?doing'))
````

**Note:**   
All the methods specified above return a copy of the original string. They don't make any changes to the original string.

## Unicode

A numerical value represents all characters(English text, french texts, emojis, etc.) you see on digital mediums. For example: 'a' is represented by 0061, '😀' is represented by 1F600, and so on. Numerical equivalents for these characters are called **code points**. Unicode specifies code points for all characters. **Unicode** is a standard for the consistent encoding, representation, and handling of text.

In natural language processing, we need to work with different languages and characters. We need Unicode representation to represent characters that are not present in normal QWERTY keyboards. For the demo, we will show you how you can represent the Devanagari script using Unicode. [This document](http://unicode.org/charts/PDF/U0900.pdf) contains Unicode code points for different Devanagari characters.

Let us construct some text in Devanagari using Unicode representation and python. In python, we need to add a prefix '\u' to code points to instruct python that the following characters represent a Unicode character.



````python
text = "\u0928\u0947\u092a\u093e\u0932"
print(text)
````

Apart from Unicode, there are other encoding techniques like utf-8 and ascii. Text in python 2 is stored in `str`, which supports ascii data. The main disadvantage of ascii representation is that it only supports 128 unique characters. When working with languages like french, german, devanagari, etc. it is not possible to use characters like é, ल, ñ, etc. as they are not supported by ascii.    
Python 3 elevates this issue by using unicode as the default encoding technique. This makes it possible to use characters like é, ल, ñ, etc. in python code and text. However, you can still use the `encode()` and `decode()` methods to convert text to and from representation like ascii and utf-8.

````python
# Convert text to utf-8
utf_8_text = text.encode('utf-8')
print("UTF-8 representation = {}".format(utf_8_text))

# Convert text from utf-8 back to unicode
text = utf_8_text.decode('utf-8')
print("".format(text))
````

Python operations discussed above(search, replace, split, etc.) can be used with unicode characters as well. Let us use the `replace()` method to remove emoticons 😀 and 😅from the dummy tweet below.
> Yoshua is very funny, 😀. He often uses emojis in his tweet. 😅😅

````python
# writing the above tweet in python
text = "Yoshua is very funny, \U0001F600. He often uses emojis in his tweet. \
\U0001F605 \U0001F605"
print(text)
````

Let us remove all emojis from this text.

````python
# replaces all occurences of "\U0001f600" and \U0001F923" with an empty string
text.replace("\U0001f600", "").replace("\U0001F605", "")
````

All unicodes from \U0001F600 to \U0001F64F represent emoticons. This makes it easier to remove all emoticons at once using a regular expression. You will be learning about regular expressions in detail in the next chapter but for now, know that we can efficiently remove emoticons from the text using regular expressions. The function below removes all emoticons from the input text without needing to use a replace method for each unique emoticon like the above method.

````python
import re

def remove_emoticons(text):
  RE_EMOJI = re.compile('[\U0001f600-\U0001f64f]', flags=re.UNICODE)
  text = RE_EMOJI.sub(r'', text) # Replace emoticon with an empty string
  return text

text = "Yoshua is very funny, \U0001F600. He often uses emojis in his tweet. \
\U0001F605 \U0001F605"
print(remove_emoticons(text))
````
