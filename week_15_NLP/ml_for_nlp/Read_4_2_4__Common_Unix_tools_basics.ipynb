## GREP



GREP stands for global regular expression print. It is a common command-line tool found on unix and linux systems. It is used to search file for a given pattern. If a match is found, it prints the respective line in the standard output(default).    

Syntax for using the grep tool:    
`grep [option...] [patterns] [file...]`   

Example:    
`grep fusemachines about_us.txt`    
The above example searches for the word 'fusemachines' in the file 'about_us.txt'. If a match is found, it prints the respective line to the standard output.    

By default, the search process is case sensitive. You can pass the option `-i` to perform a case insensitive search.    
`grep -i fusemachines about_us.txt`    
The above command searches for all occurrences of the word 'fusemachines' in the file 'about_us.txt' without worrying about the case. It will return a positive match for 'fusemachines', 'FUSEMACHINES', 'FuSeMachiNES', or any other combination of cases.    

Instead of displaying the result to the standard output, we can also write the result to a file using the `>` or `>>` operators. `>` overwrites the existing content(if any) of the file, whereas `>>` appends the result to the existing file.       

`grep -i fusemachines about_us.txt >> match_found.txt`       

Grep supports regular expressions. We can specify a regular expression pattern, and grep will return all the lines that match the given pattern.    
  
**Note**   
By default, in linux, grep uses the POSIX compatible regular expression. But regular expressions we studied earlier are Perl compatible regular expressions. So we need to provide -P option while performing a regular expression based search if we want to work with perl compatible regular expressions.
On a mac, things are different. By default, mac has a BSD version of grep, which doesn't support perl compatible regular expressions. You will need to install gnu grep if you want to work with perl compatible regular expressions. Then you can use -P option to evaluate perl compatible regular expressions.

Let us consider a file `regex_test.txt` contains the following text in three separate lines:  

> Whatsapp was bought by facebook.com.  
It is now very popular throught the world.  
Contact number for facebook's office is +1-934-343-4356.





* How do you extract  phone numbers from the file `regex_test.txt`?
  * `grep -P '\+[0-9]-[0-9]+-[0-9]+-[0-9]+' regex_test.txt`    

* How do you extract urls having format `domain.extension` from the file `regex_test.txt`?
  * `grep -P '\w+\.\w+' regex_test.txt`   

* How do you extract all capital lettered words from the file `regex_test.txt`?
  * `grep -P '[A-Z]\w+' regex_test.txt`    

## TR

TR stands for translate characters. It is a command-line utility that copies standard input to standard output with substitution or deletion of selected characters. If you want to specify input from a file instead of the standard input, you can do so using the `<` operator.    

Syntax:   
`tr [option] [string1] [string2]`   

Example:   
`echo hello | tr "[:lower:]" "[:upper:]"`    
 The above example converts the string 'hello' to 'HELLO' and displays on the standard output.
 Here, you can see the use of a pipe(|) operator. The command following the pipe operator takes the output of the command before the pipe operator as the input.   

`echo "Hello jack" | tr -d "jl"`    
 The option "-d" is used to delete characters. The given command deletes all occurrences of "j" and "l" from "Hello jack".    

`echo "Hello jack" | tr -c -d "jl"`    
 The option "-c" is used to complement characters in the string. The given command deletes all characters except "j" and "l" from "Hello jack".    

Now, let us use tr to do some basic operations on the content of the file `regex_test.txt` specified in the earlier section.    

* How do you compress the file `regex_test.txt` by replacing multiple occurrences of characters 'o' and 'p' with a single character?  
 * `tr -s "op" < regex_test.txt`    
 The "-s" option squeezes multiple simultaneous occurence of characters 'o' and 'p' into 1. 'facebook' becomes 'facebok' and 'Whatsapp' becomes 'Whatsap'.    
 Here, `<` is a redirection operator which is used to read the input from a file.    

* How do you remove all occurrences of characters 'o' and 'p' from the file `regex_test.txt`?    

 * `tr -d "op" <regex_test.txt`     
 Deletes characters 'o' and 'p' from the given input text. 'facebook' becomes 'facebk'.    

* How do you remove all numbers from the file `regex_test.txt`?     
 * `tr -d "[:digit:]" < regex_test.txt OR tr -d "[0-9]+" <regex_test.txt`     
 Here, `[:digit:]` represents numerical characters.    

* How do you convert all whitespace character from the file `regex_test.txt` to a newline character?     
 * `tr "[:space:]" "\n" < regex_test.txt`     
 Here `[:space:]` represents whitespace characters. The above command results in each word printed in a new line, mimicking word tokenization.     


* How do you remove everything except digits or numbers from the file `regex_test.txt` ?    
 * `tr -cd "[:digit:]" < regex_test.txt`    



## SED

SED stands for Stream Editor. It is a command-line utility that reads the specified file or the standard input, modifies the input as specified by the set of commands, and writes the output to standard output by default.   

Syntax:   
`sed [options] command [file]`    

You may specify a single command(default), multiple commands by using the -e parameter or a file containing commands using -f parameter.    

You can also use the pipe operator to provide input instead of using a file.   

Example:    

`echo "I love linux" | sed 's/linux/unix/'`    
 Finds the word "linux" and replaces its first occurrence with "unix".    

`echo "I love linux" | sed 's/linux/unix/g'`    
 Finds the word "linux" and replaces its all occurrences with "unix".    



Let's use sed to do some basic text manipulation on the file `regex_test.txt`.    

* How do you print the content of the file `regex_test.txt` to the standard output(monitor)?   
 * `sed '' regex_test.txt`    

* How do you print all content except the first line of the file `regex_test.txt` to the standard output?   
 * `sed '1d' regex_test.txt`   

* How do you print all content except the first two lines of the file `regex_test.txt` to the standard output?    
 * `sed -e '1d' -e '2d' regex_test.txt`    


* How do you take all commands stored in the file `commands.txt`, execute them on the file `regex_test.txt` and display the result in the standard output?   
 * `sed -f commands.txt regex_test.txt`     

* How do you display all lines in the file `regex_test.txt` containing the word 'Contact'?   
 * `sed -n '/Contact/ p' regex_test.txt`     
 The parameter `-n` is used so that the content of the file `regex_test.txt` is not printed before printing the output. If this parameter is not used, the entire content of the file is displayed first, and finally, the line matching the pattern is displayed.     


* How do you display all lines in the file `regex_test.txt` containing phone number?    
 * `sed -r -n '/\+[0-9]-[0-9]+-[0-9]+-[0-9]+/ p' regex_test.txt`     
 Parameter `-r` is used to specify perl compatible regex. If you are on a mac use `-E` instead of `-r`.     

* How do you display all lines in the file `regex_test.txt` that starts with a capital letter?    
 * `sed -r -n '/[A-Z][a-z]+/ p' regex_test.txt`
 The given command prints lines containing words that start with a capital letter.     

* How do you display all lines in the file `regex_test.txt` containing urls in the form [domain].[extension]?    
 * `sed -r -n '/[a-z]+\.[a-z]+/ p' regex_test.txt`     

## AWK

AWK is a pattern scanning and processing language. It uses patterns heavily. Common awk workflow involves searching specified pattern within the specified input file and then performing the specified actions in the words that match the pattern.  

AWK is a programming language and has all the features like conditional, loops, functions, etc. However, we will only cover its basic functionalities.


Syntax:  
`awk 'command' filename.extension`  

AWK works with fields. The field \$0 represents the entire file, \$1 represents the first field text, \$2 represents the second field text, and so on.  

Example:  
`awk 'length($0) > 10' file_name.txt`  
  Prints all lines which have more than 10 characters from the file `file_name.txt`.  

`awk '{ if($3 == "B6") print $0;}' file_name.txt`   
  This command shows the use of 'if-statement' in awk. The command prints the line if the third word in the line is 'B6'.  

Now, let us use awk to do some text manipulations on the file `regex_test.txt`.

* How do you print all the contents of the file `regex_test.txt`
  * `awk {'print $0'} regex_test.txt`

* How do you print first word(field) from each line of the file `regex_test.txt`
  * `awk {'print $1'} regex_test.txt`    

* How do you print first two words(fields) from each line of the file `regex_test.txt`
  * `awk {'print $1, $2'} regex_test.txt`  
  Note that fields should be separated with a comma.  

* From the file `regex_test.txt`, how do you print the second word of the line in which the first word is 'Whatsapp'?  
  * `awk '$1=="Whatsapp"{print $2}' regex_test.txt`  
  *  ```
  awk '{
         if($1 == "Whatsapp"){
            print $2
         }
       }' regex_test.txt
```

  The second example demonstrates the use of if-statement in awk.
   
* From the file `regex_test.txt`, how do you print the second word of the line in which the first word starts with a capital letter?
  * `awk '$1~/[A-Z][a-z]+/ {print $2}' regex_test.txt`  
  Here, the pattern `[A-Z][a-z]+` represents a word starting with a capital letter.    

* From the file `regex_test.txt`, how do you print all lines containing phone number?  
  * `awk '/\+[0-9]-[0-9]+-[0-9]+-[0-9]+/' regex_test.txt`   
  The given regex represents a phone number of the form +X-XXX-XXX-XXXX. Each 'X' represents a digit.      

  
* How do you print all contents of the file `regex_test.txt` with capital-lettered words replaced by the word "fusemachines"?
  * `awk '{sub(/[A-Z][a-z]+/,"fusemachines")}1' regex_test.txt`    





This notebook serves as a guide to help you get started with the above command-line tools. The notebook is not a complete manual for these tools and, in no way, presents the complete potential of these tools. If you want to know more about these tools, you can refer to their manual page using the command `man tool_name` in your shell. For your ease, some additional resources are linked at the end of this notebook.

## Key Takeaways
* Command-line provides powerful tools that help you manipulate text right from the terminal/shell.
* GREP, TR, SED, and AWK are some powerful command-line tools.
* We can use regex to perform a complex operation on text using command-line tools.
* AWK is a powerful programming language that you can use to manipulate text using command-line.
* Command-line provides a manual for common tools, and the manual can be accessed using the command `man tool_name`.



## Additional Resources

### Tutorials and Manuals
* [Tutorial on grep, awk, sed any many more](https://learnbyexample.gitbooks.io/command-line-text-processing/content/gnu_sed.html)   
  Refer to this site to learn about AWK, SED, and GREP.

* [GREP Manual by GNU](https://www.gnu.org/software/grep/manual/grep.html)  

* [SED Manual by GNU](https://www.gnu.org/software/sed/manual/sed.html)  
* [AWK Manual by GNU](https://www.gnu.org/software/gawk/manual/gawk.html)
