## Raw Text Processing

Before going into details of this stage, it is noteworthy that **depending upon the data, whether the data is noisy or not, this step may be entirely optional** and operations in this step sometimes overlap with text preprocessing. In this step, we work with the text directly obtained from the source(by scraping from the web or the local storage). Such text may contain unnecessary entities like links, image bytes, unwanted dates, etc. This step's primary goal is to get familiar with the text, learn to manipulate and clean the text, and learn essential tools like regular expressions and command-line utilities (like awk, sed, tr, and grep).

We describe this step as optional but have dedicated an entire unit for raw text processing. What might be the reason? The reason is that tools and techniques that you will learn about in raw text processing can help you gather information like phone numbers, contact addresses, urls, etc. from the text. Furthermore, you could use these tools and techniques to augment text cleaning. You will learn more about raw text processing in chapters that follow but for now, let's move on to text preprocessing.

## Text Preprocessing

The input to this stage is a cleaned file obtained after raw text processing. However, this cleaned file may still not be suited to feed a machine learning model. Depending upon the model we are trying to create, the text might still need some processing. **Text preprocessing is an application of a predefined set of operations performed on a standard dataset(clean text)**. Some common operations performed during text preprocessing are:

* Text cleaning   
 Text cleaning refers to eliminating unnecessary words/symbols from the text. Example: Removing any links, image bytes, etc.

* Text tokenization (word and sentence tokenization)   
 Tokenization refers to breaking a large chunk of text into smaller ones. Example: Breaking a paragraph into sentences, breaking sentences into words, etc.


* Text normalization   
 Text normalization is the process of converting all the text to a single canonical form, for example, converting entire text to lowercase, converting all words to their base form (changing 'being' and 'been' to 'be').

* Stopwords removal   
 Stopwords removal is the process of removing the most common words from the text. For an English language text, words like 'and', 'the', 'a', etc. are the stopwords. We can find them in almost all documents.

* Text vectorization   
 Text vectorization refers to the conversion of text into numerical features. Some standard text vectorization algorithms are Bag of Words, Term frequency-inverse document frequency, etc.



Depending upon the model we are trying to create, we may need to perform all or only some of the above-mentioned operations.

Now, let us discuss some standard tools that can help us perform the operations mentioned above. Python is the most common language used for text preprocessing because of its advanced libraries that help accelerate the process. Some of the most common libraries used for text processing in python are:

* NLTK   
 NLTK provides packages like [`tokenize`](https://www.nltk.org/api/nltk.tokenize.html?highlight=word%20tokenize) for tokenization, [`stem`](https://www.nltk.org/api/nltk.stem.html?highlight=stem#module-nltk.stem) for normalization, and many more.
* SpaCy   
 SpaCy provides classes like [`Tokenizer`](https://spacy.io/api/tokenizer) for tokenization, [`Lemmatizer`](https://spacy.io/api/lemmatizer) for normalization, and many more.
* scikit-learn   
 scikit-learn provides [`feature_extraction`](https://scikit-learn.org/stable/modules/classes.html#module-sklearn.feature_extraction.text) module to extract numerical features from text.

## Machine Learning

Machine Learning is an application of artificial intelligence and uses statistical methods to extract patterns from data. By now, you should be familiar with machine learning algorithms like decision trees, support vector machines, random forests, etc. Machine learning is not to be confused with deep learning, which is a subfield of machine learning and uses deep neural networks to extract information/patterns from data.

Machine Learning is the final step of a natural language processing project, which involves building an actual model. All the steps discussed earlier build a foundation on which a machine learning model is built. This step uses the output of the text preprocessing stage as the input. Depending on the model we are trying to build, we may manipulate features: add new features or remove unwanted features. This manipulation of features is called **feature engineering**, and it requires domain knowledge. Finally, we use the dataset to train the model and fine-tune it to map the input to the desired output. The model is then ready to make inferences on unseen data.


By now, you should have trained some models in the **Machine Learning** module. In this module, you will create models that operate on text data. An example of such models could be **music genre classifier**. A music genre classifier takes the lyrics of a song as input, generates features from the input text, and then uses machine learning models like SVM, decision tree, naive bayes, etc to classify the song into one of the predefined genres like rock, pop, etc.

Some commonly used libraries/frameworks for building machine learning models are:
* scikit-learn   
 It is a machine learning library which provides a ready to use implementation of algorithms like support vector machines, decision trees, random forest, linear regression, etc. It provides good documentation and tutorials to help beginners get started with using the library. A major drawback of scikit-learn is that it doesn't support execution on GPU, which causes the model building process to slow down.
* PyTorch   
 It is a machine learning library built by Facebook and is commonly used for deep learning applications. It is built using python with a focus on rapid experimentation and research.
* TensorFlow and Keras   
 TensorFlow is an end to end machine learning platform to build and deploy machine learning models. It was built by Google with a focus to bridge the gap between development and deployment. Keras is another beginner-friendly machine learning library built on top of TensorFlow 2.0 and takes complete advantage of deployment capabilities of the TensorFlow platform.

## Key Takeaways
* Raw Text Processing, Text Preprocessing, and Machine Learning are three major stages of an NLP project.
* Raw text preprocessing cleans text files by removing unnecessary image bytes, URLs, etc.
* Text preprocessing operations depend on the machine learning model of choice.
* Normalization, Augmentation, Tokenization, and Vectorization are common operations performed during text preprocessing.
* Machine learning involves building and fine-tuning the model to do the relevant task.
