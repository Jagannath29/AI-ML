## Normalization


### Introduction

The word _normalization_ generally refers to bringing something to its standard or common form. In the context of texts, normalization is the process of **transforming multiple forms of the same word to a common standard form**.
A text might contain the same word in different forms, such as 'Like', 'like', 'liking', 'liked', etc. If these words are not transformed into a standard form, then the machine learning model treats them differently. This will adversely affect the performance of the model. Eventually the words present in the documents of a dataset become the features of the dataset. If multiple forms of the same word exist in a dataset then the number of unnecessary features will increase as well. This makes the model inefficient and slow. So you need to **normalize** the text document. In this chapter, you will learn about the three main steps involved in normalization.




Since normalization is generally done after tokenization, you will have a list of tokens instead of a single document. Let's take the following tokens as a sample that each step of text normalization will work on.

['Formulated', 'and', 'formulating', 'are', 'the', 'two', 'tenses', 'of', 'Formulate'].

This list of token is corresponding to the sentence *Formulated and Formulating are the two tenses of Formulate*.


`````
tokens = ['Formulated', 'and', 'formulating', 'are', 'the', 'two', 'tenses', 'of', 'Formulate']
`````

### Case Folding

Consider two words: _Table_ and _table_. These two are exactly the same word having different cases of _t_. A human brain can easily realize this, but a machine learning model cannot. A machine learning model will interpret these words as two different words. **Case folding simply refers to transforming all the letters in a text document into a single case.** It can either be the upper case or the lowercase as long as the whole document is in the same case. A common strategy for case folding is to convert all the letters to lowercase.

You can use the [`string.lower`](https://docs.python.org/2/library/stdtypes.html#str.lower) method to convert a particular string to lowercase. Let's illustrate this on the sample tokens.

`````
tokens_lower = []
for token in tokens:
  token = token.lower()
  tokens_lower.append(token)

print(tokens_lower)
`````

`````
tokens = tokens_lower
`````

### Stemming

Consider two words: played and playing. Both of these words are different variants derived from the same root word *play*. **Stemming simply chops off the ends of the different variants of a word to reduce them to their common root form.** This common root form, called **stem**, may not even have an actual meaning. For example, the words *formulation* and *formulated* will be reduced to *formul*. In this case, the suffixes: *ating* and *ated* are chopped off from the words, respectively. The stem *formul* doesn't have any actual meaning but now both the words: *formulation* and *formulated* will be represented using this stem.

You can use the `PorterStemmer` from [`nltk.stem`](https://www.nltk.org/api/nltk.stem.html) package to stem a given string as:

````
stemmer = PorterStemmer()
stemmer.stem(string)
````

Let's illustrate this on the sample tokens.


`````
# Importing PorterStemmer
from nltk.stem import PorterStemmer

# Instantiating PorterStemmer
stemmer = PorterStemmer()

# Stemming each token in tokens
tokens_stem = [stemmer.stem(token) for token in tokens]

print(tokens_stem)
`````

Some other popular types of stemmers available in [nltk.stem package](https://www.nltk.org/api/nltk.stem.html) are:

* Lancaster Stemmer (`nltk.stem.LancasterStemmer`)
* Snowball Stemmer (`nltk.stem.SnowballStemmer`)
* RegExp Stemmer (`nltk.stem.RegexpStemmer`)

### Lemmatization

Lemmatization also aims at reducing a word to its root form but in a different way. **Lemmatization is the process of reducing a word to its lemma.** A **lemma** is the base form of a word using which it is recorded in the dictionary. As such, a lemma always has a meaning. For example, the words *formulation* and *formulated* will now be reduced to *formulate*. The lemma *formulate* does have a meaning.

You can use the `WordNetLemmatizer` from `nltk.stem` to lemmatize a given string as:

````
lemmatizer = WordNetLemmatizer()
lemmatizer.lemmatize(string, pos)
````
The argument `pos` takes in the value corresponding to which part of speech the string belongs to.


The `WordNetLemmatizer` uses the [WordNet](https://wordnet.princeton.edu/) database to find the lemmas of a word. It is a large and publicly available lexical database of English. You will first need to download it to use the `WordNetLemmatizer`.




`````
import nltk
nltk.download('wordnet')
`````

Now that you have imported the wordnet database, let's lemmatize the words *formuating* and *formulated* using `WordNetLemmatizer`.

`````
# Importing Wordnet Lemmatizer
from nltk.stem import WordNetLemmatizer

# Instantiating the WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

# Lemmatizing each token in tokens
tokens_lem = [lemmatizer.lemmatize(token, pos='v') for token in tokens]

print(tokens_lem)
`````

### Putting it all together

Now that you know the necessary tasks generally used to normalize the data, let's create a function `normalize_text` that combines all of them. For the sake of simplicity and computational efficiency, you will be using stemming instead of lemmatization. But note that lemmatization often has its own advantages.

`````
def normalize_text(tokens):

    from nltk.stem import PorterStemmer

    # lowercasing all the tokens
    tokens = [token.lower() for token in tokens]

    # stemming all the tokens
    stemmer = PorterStemmer()
    tokens_stem = [stemmer.stem(token) for token in tokens]

    return tokens

tokens = normalize_text(['Formulated', 'and', 'formulating', 'are', 'the', 'two', 'tenses', 'of', 'Formulate'])
print(tokens)
`````

## Stopwords Removal



Stop words are a set of commonly used words in any language. For example: *a, an, the, they, we, is, are, were, not* and *so on* are common stop words in English langauge. We remove stop words from our corpus because they do not provide any important information. Removing stop words allows us to look at only meaningful data or information.

`nltk` has a set of stop words for every language. You can get the list of stopwords for the English language from `nltk` as:

```
stopwords.words('english')
```

But before using the stopwords, you first need to download it as follows:

`````
#download `stopwords` for nltk
nltk.download('stopwords')
`````

Let’s store the stopwords as a list in variable `eng_stopwords` and print the first 15 english stopwords.

`````
from nltk.corpus import stopwords
eng_stopwords = stopwords.words('english')

#print first 15 stopwords from the list
print(eng_stopwords[:15])
`````

Now that you have the list of english stopwords, you can simply check for each token and remove it if it is a stopword. Let's illustrate this by removing the stopwords from the lemmatized tokens `tokens_lem`.

`````
print("Before removing stopwords:\n", tokens_lem)

tokens = []
for token in tokens_lem:
  if token not in eng_stopwords:
    tokens.append(token)

print("After removing stopwords:\n", tokens)
`````

## Getting back the original text

Tokenization has served its purpose after allowing us to normalize and remove stopwords from the text. Now you will need to get back the original text by joining the tokens. You can do this by using the `join` method of string.

````
string = " ".join(tokens)
````
This creates a string by concatenating the tokens with space.

Let's get the original text back from our tokens.

`````
string = " ".join(tokens)
print(string)
`````

## Normalization and Stopwords Removal on a Real World Dataset

Now that you know what Normalization and Stopwords Removal are, let's implement them on a real world dataset. You will use the same [SMS Spam Collection](https://archive.ics.uci.edu/ml/datasets/SMS+Spam+Collection) dataset that you used earlier for text cleaning and tokenization.

`````
import pandas as pd
dataset = pd.read_csv('https://storage.googleapis.com/codehub-data/1-lv1-7-1-spam.tsv', sep='\t', names=['spam', 'sms'])
dataset.head()
`````

Let's first quickly define the `clean_text` function from the previous chapter as you will require it here as well before applying normalization and removing stopwords.


`````
def clean_text(string):

    import re
    from bs4 import BeautifulSoup

    # Removing URLs
    string = re.sub("http\S+", " ", string)

    # Removing html tags
    string = BeautifulSoup(string, 'lxml')
    string = string.get_text()

    # Decontraction
    string = re.sub(r"n\'t", " not", string)
    string = re.sub(r"\'re", " are", string)
    string = re.sub(r"\'s", " is", string)
    string = re.sub(r"\'d", " would", string)
    string = re.sub(r"\'ll", " will", string)
    string = re.sub(r"\'t", " not", string)
    string = re.sub(r"\'ve", " have", string)
    string = re.sub(r"\'m", " am", string)

    # Removing Punctuations and numbers
    string = re.sub('[^A-Za-z]+', ' ', string)

    return string
`````

You'll also need the `word_tokenize` function from `nltk`to tokenize the messages before normalizing and removing the stopwords.

`````
import nltk
nltk.download('punkt')
from nltk.tokenize import word_tokenize
`````

In the previous chapter, you cleaned and tokenized the messages. Here, in addition to the cleaning and tokenization, you will be normalizing and removing stopwords from the messages.

`````
corpus = [] # list containing all the preprocessed messages in dataset

# for each message in the document.
for message in dataset['sms']:

  # cleaning the message
  message = clean_text(message)

  # tokenizing the message
  tokens = word_tokenize(message)

  # text normalization
  tokens = normalize_text(tokens)

  # stopwords removal
  tokens = [token for token in tokens if token not in eng_stopwords]

  # getting back the original text
  string = " ".join(tokens)

  corpus.append(string)

# printing top 10 documents
for document in corpus[:10]:
  print(document)
`````

As you can see, the list `corpus` contains all the messages in the dataset in their cleaned and normalized form. You can also see that none of the messages contain the stopwords.

## Key Takeaways

* Normalization is the process of transforming multiple forms of the same word to a common standard form.

* Case folding simply refers to transforming all the letters in a text document into a single case.


* Stemming reduces a word to its root form by simply chopping off its end.

* Lemmatization reduces a word to its lemma.

* Stopwords removal is the process of removing commonly used words in any language.
