## Text Cleaning

### Introduction

In normal machine learning problems, you used datasets with numerical values as features. If categorical values were present, they were also encoded to numerical values. However, while dealing with text data, you don't have any features. All you have is a document in text format. You will need to convert this document into numerical features that can be fed into a machine learning model. All this starts with cleaning the available text document. The document might contain anything from unnecessary symbols and numbers to URLs and HTML tags. These unnecessary strings do not provide any information useful for machine learning models, and so you will learn to remove them in this section. You will be using regular expressions extensively for this chapter. So let's quickly import the ['re`](https://docs.python.org/3/library/re.html) module.

`````python
import re
`````

Let's take the following text as a sample text. We will apply the steps involved in text cleaning to this sample text.

*I love Facebook.https://www.facebook.com I wanted an induction cooker and facebook just showed me one at just $<bold> Rs. 1000 </bold>$. They cannot be reading my private messages, can they? Wouldn't it violate my right to privacy?*

`````python
string = """
I love Facebook.https://www.facebook.com I wanted an induction cooker and facebook
just showed me one at just <bold> Rs. 1000 </bold>. They cannot be reading
my private messages, can they? Wouldn't it violate my right to privacy?"""
`````

### Removing the  URLs from the document

Sometimes, the data might contain traces of the website's address from where the data was extracted from. These unnecessary, broken URLs might not contribute any relevant information required by a machine learning model. So it is a good practice to remove these URLs from the texts.   

You can use the `sub()` method from [`re`](https://docs.python.org/3/library/re.html) module to substitute the URLs with a space as:

````
re.sub("http\S+", " ", document)
````

Let's illustrate this in the sample text.

`````python
string = re.sub("http\S+", " ", string)
print(string)
`````

As you can see, the URL part of the string has been removed. More information [Removing URLs from string](https://stackoverflow.com/questions/11331982/how-to-remove-any-url-within-a-string-in-python).
   

### Removing HTML tags from the text

The text document might be extracted from an HTML document. So it may contain HTML tags. HTML tags are just redundant texts and do not provide any necessary information to be used by a machine learning model.

You can instantiate an object of BeautifulSoup class with `lxml` parser from [`Beautiful Soup`](https://www.crummy.com/software/BeautifulSoup/bs4/doc/) library to remove html tags. Let's illustrate this on the sample text.

`````python
from bs4 import BeautifulSoup

string = BeautifulSoup(string, 'lxml')
string = string.get_text()
print(string)
`````

As you can see, the html tags present in the string have been removed. More information [Removing tags from Strings (parsing)](https://stackoverflow.com/questions/16206380/python-beautifulsoup-how-to-remove-all-tags-from-an-element
)

### De-contraction of the words

Contraction refers to the shortening of words by removing one or more letters indicated by an apostrophe ('). For example *cannot to can't, you are to you're, are not to aren't* and so on. You can check out different contractions used in the english language [here](https://en.wikipedia.org/wiki/Wikipedia%3aList_of_English_contractions). It is a good practice to decontract these types of word to their original form before removing the punctuations. For example, consider the two sentences:

1. I'm having a good day.

2. I am having a good day.

If the first sentence is not decontracted, a machine learning model will treat these two sentences differently, even if they mean the same thing.

You can use the same sub method from ['re`](https://docs.python.org/3/library/re.html) module to decontract the words. For example, you can decontract *aren't to are not* as:

````
re.sub("n\'t", " not", document)
````

Let's illustrate this in the sample text.

`````python
string = re.sub("n\'t", " not", string)
string = re.sub("\'re", " are", string)
string = re.sub("\'s", " is", string)
string = re.sub("\'ll", " will", string)
string = re.sub("\'ve", " have", string)
string = re.sub("\'m", " am", string)

print(string)
`````

As you can see, the contracted word *Wouldn't* has been decontracted to *Would not*.

### Removing punctuations and numbers

The text document might contain different types of punctuation marks. These marks do not provide any important information about the text that a machine learning model can use. So you need to remove these punctuation marks. Similar is the case with numbers. There might be some instances where numbers are important for a machine learning model, but generally, we remove them.

You can use the same sub method from [`re`](https://docs.python.org/3/library/re.html) module to replace all the punctuation marks, symbols and numbers with space in a text document as:

````
re.sub('[^A-Za-z]+', ' ', document)
````

This will replace everything other than characters A-Z and a-z with space in the document. Let's illustrate this on our sample text.

`````python
string = re.sub('[^A-Za-z]+', ' ', string)
print(string)
`````

### Putting it all together

Now that you know the necessary tasks generally used to clean the data, let's create a function `clean_text` that combines all of them together.

`````python
def clean_text(string):

    import re
    from bs4 import BeautifulSoup

    # Removing URLs
    string = re.sub("http\S+", " ", string)

    # Removing html tags
    string = BeautifulSoup(string, 'lxml')
    string = string.get_text()

    # Decontraction
    string = re.sub(r"n\'t", " not", string)
    string = re.sub(r"\'re", " are", string)
    string = re.sub(r"\'s", " is", string)
    string = re.sub(r"\'d", " would", string)
    string = re.sub(r"\'ll", " will", string)
    string = re.sub(r"\'t", " not", string)
    string = re.sub(r"\'ve", " have", string)
    string = re.sub(r"\'m", " am", string)

    # Removing Punctuations and numbers
    string = re.sub('[^A-Za-z]+', ' ', string)

    return string

string = clean_text(string)
print(string)
`````

## Tokenization

### Introduction

In the previous section, you studied text cleaning to remove unnecessary strings and symbols from the text document. In this section, you will learn about the next step in text preprocessing called **tokenization**. Let's start with an analogy of how human brains deal with a text document. Consider the following sentence: " _I go to the college everyday._" To understand the meaning of this sentence, you first identify all the constituent words in that sentence. In this case, the constituent words are *I, go, to, the, college, everyday*. Then you analyze the meaning of each of those words and finally derive the meaning of the whole sentence. As you might have noticed, identifying the constituent elements of any text document is a fundamental step in understanding a Natural Language. This step is known as **tokenization**. Now let's move into a more formal definition of tokenization.

Tokenization is the process of splitting a larger text document into several smaller units called **tokens**. These tokens can be anything from single characters to words, subwords, and sentences. This allows you to analyze each token individually, which is required by the following steps of text preprocessing.

### Types of Tokenization

There are different types of tokenizations available in [nltk.tokenize](https://www.nltk.org/api/nltk.tokenize.html) package. In this section, you will be learning about two of the most popular types of tokenization:

* Word Tokenization

* Sentence Tokenization

To use both of these tokenizers from nltk, you'll first need to download `punkt` from `nltk`.


`````python
import nltk
nltk.download('punkt')
`````

#### Word Tokenization:
Word tokenization splits a text document into each individual words. You can use the `word_tokenize` method from `nltk.tokenize` package to perform word tokenization. Let's tokenize the cleaned sample text stored in the variable `string`.

`````python
from nltk.tokenize import word_tokenize

word_tokens = word_tokenize(string)
print(word_tokens)
`````

#### Sentence Tokenization

Sentence tokenization splits a text document into each individual sentences. It identifies the constituent sentences of a text using the presence of various punctuation marks such as fullstops $(.)$, exclamation marks $(!)$, and question marks $(?)$.  You can use the `sent_tokenize` method from `nltk` to perform sentence tokenization.

**Note:** You cannot use the same cleaned sample text as the cleaning process has removed all the punctuation marks from it. So let's illustrate it on a small new piece of text.

`````python
from nltk.tokenize import sent_tokenize

text = "Is it raining? Do not forget to close the window if it is raining. I hate mopping the floor."

sentence_tokens = sent_tokenize(text)
print(sentence_tokens)
`````

## Text Cleaning and Tokenization on a Real World Dataset

Now that you know what text cleaning and tokenization are, let's implement them on a real world dataset.

### Dataset Description

You will be cleaning and tokenizing the messages present in the [SMS Spam Collection](https://archive.ics.uci.edu/ml/datasets/SMS+Spam+Collection) dataset. It is a public set of SMS labeled messages that have been collected for mobile phone spam research. It contains 5,574 SMS labeled as 'spam' or 'ham'.

**Author:** Tiago A. Almeida and José María Gómez Hidalgo <br>
**Reference:** http://www.dt.fee.unicamp.br/~tiago/smsspamcollection/

### Attribute Information:

The collection of the SMSs is contained in a single file, where each line has the correct class followed by the raw message. For example:

* ham:    What you doing?how are you?

* ham:    Cos i was out shopping wif darren jus now n i called him 2 ask wat present he wan lor. Then he started guessing who i was wif n he finally guessed darren lor.

* spam:    FreeMsg: Txt: CALL to No: 86888 & claim your reward of 3 hours talk time to use from your phone now! ubscribe6GBP/ mnth inc 3hrs 16 stop?txtStop
* spam:    Sunshine Quiz! Win a super Sony DVD recorder if you canname the capital of Australia? Text MQUIZ to 82277. B
* spam:    URGENT! Your Mobile No 07808726822 was awarded a L2,000 Bonus Caller Prize on 02/09/03! This is our 2nd attempt to contact YOU! Call 0871-872-9758 BOX95QU

**Note:** For the current purpose you are concerned only with the raw message and not their label. You will use the labels in later units where you will apply a machine learning model to this data.







`````python
import pandas as pd
dataset = pd.read_csv('https://storage.googleapis.com/codehub-data/1-lv1-7-1-spam.tsv', sep='\t', names=['spam or ham', 'sms'])
dataset.head()
`````

Now let's clean the messages in the dataset and tokenize them.

`````python
document_tokens = [] # list containing the tokens of each of the messages in dataset

# for each message in the document.
for message in dataset['sms']:

  # cleaning the message
  message = clean_text(message)

  # tokenizing the message
  document_tokens.append(word_tokenize(message))

# printing the tokens of top 10 documents
for tokens in document_tokens[:10]:
  print(tokens)
`````

As you can see, the list `tokens` contains all the tokens for all the messages in the dataset. The token's are cleaned and does not contain any unnecessary symbols and strings.
