## 1. spaCy

Like NLTK, spaCy is a free, open-source library for Natural Language Processing in python. Unlike NLTK -- designed for research and teaching -- spaCy is used in production and getting things done. spaCy is one of the fastest NLP tools. Since spaCy is more object-oriented, it is easy to learn. Unlike NLTK, it provides built-in word vectors. However, it doesn't support as many languages as NLTK does.

spaCy is widely used to pre-process the large volume of text in deep learning. It comes with a bunch of features like part of tokenization, lemmatization, part-of-speech tagging, sentence boundary detection, named entity recognition, text classification, etc. We will get familiar with these terms in the later chapters.

### Installation

We can install spaCy as:

`pip install -U spacy` or `pip3 install -U spacy`

You can refer [here](https://spacy.io/usage) on a full installation guide.

### Models

spaCy has around 46 statistical models in over 15 languages, which can be used for different NLP tasks like part-of-speech tagging, dependency parsing, named entity recognition, etc. There are two models: core models and starter models. Core models are general-purpose models that can be used out of box for the different applications. Starter models are one whose weights can used to initialize our models.

We can load model `en_core_web_sm` as:

`python -m spacy download en_core_web_sm`.

All models follow the naming convention i.e, [lang_[name]]; the name itself has three components:

- Type: Type defines the model type, which in term defines the model capability. It can be __core__( is a general-purpose model with vocab, syntax, entities, and word vectors) or __depent__(for vocab, syntax, and entities).
- Genre: Genre defines the type of text on which the model is trained. For example, __web__ or __news__.
- Size: Size defines the size of the model, which can be __sm__, __md__, or __lg__ for small, medium, and large, respectively.

```
# Loading models
!python -m spacy download en_core_web_sm
```

We can now load the downloaded model as `spacy.load()`

```
import spacy

model = spacy.load("en_core_web_sm")
```

Now which ever text we need to process, we can pass it into the model.

```
text = "I bought a pair of watch for Tom and Elizabeth which costs $50 each."
processed = model(text)


print("text -- POS\n ---------")
for token in processed:
    print(f"{token.text} -- {token.pos_}")
```

Above, we printed out the tokens and part of speech they correspond to.

__Note :__ We used _ at the end of `pos` because spaCy encodes all string into hashes. This reduces the memory requirement. To get a readable string representation of an attribute, we use _ at the end of its name.

In the above example, the variable `processed` holds not just the individual words and their annotation. It holds all the text information, such as white space, which makes it possible to reconstruct the original text. That means spaCy performs non-destructive tokenization.

```
print("text -- POS -- POS hash\n ---------")
for token in processed:
    print(f"{token.text} -- {token.pos_} -- {token.pos}")
```

## 2. Gensim

Like spaCy, Gensim is a free, open-source python library used to process text, work with word vector models, and build unsupervised topic modeling. Topic modeling is a technique to extract desired topics from a large volume of text. Gensim is generally used when we need to work with large datasets and support deep learning. However, Gensim doesn't have enough modules/packages. So we need to use it with other tools like spaCy, NLTK, etc.

### Usage

Gensim is used for various tasks such as building word vector, topic identification/modeling, document comparison, and semantic analysis. Gensim provides the implementation of the following:

- fastText
- word2vec
- Latent Semantic analysis
- tf-idf

And many more.



### Installation

We can install Gensim as:

`pip install -U gensim` or `pip3 install -U gensim`

You can refer [here](https://radimrehurek.com/gensim/index.html) on full installation guide.

### Terminologies
Following are the frequent terms that we encounter with when using Gensim

- __Document:__ Document refers to some texts, which may be a list of sentences or paragraphs. For example, book, article, etc.
- __Corpus:__ We are already familiar with the corpus. Corpus is simply the collection of documents. For example, collection of books.
- __Dictionary:__ While working with texts, Gensim converts words to the unique ids. For this, we need to create a dictionary that maps from word to a unique id. Gensim provides `corpora.Dictionary()` object for this task.

- __Vector:__ Vector is simply a vector representation of Document.

- __Models:__ Once documents are vectorized, we can perform different mathematical transformations using models. For instance, we can compute tf-idf of the Document using tf-idf model in Gensim.

### Dictionary

```
import gensim
from gensim.corpora import Dictionary

document = ["Tomorrow Tom and Elizabeth are getting married, I need to buy a gift for them.",
            "I bought a pair of watch for Tom and Elizabeth which costs $50 each."]


# Tokenization
tokens = [[token for token in docs.split()] for docs in document]

# Create dictionary
dictionary = Dictionary(tokens)
print(dictionary)
```

The above dictionary shows that we have 23 unique tokens in the document. To get the unique id out of these tokens we can do:

```
dictionary.token2id
```

### Vector
Now we will convert a document into vector using bag of word.

```
new_document = "I hope they like my gift"
vector = dictionary.doc2bow(new_document.lower().split())
print(vector)
```

The method `doc2bow()` returns the list of tuples. The first item (number) in the tuple represents the id of the word in the dictionary, and the second item (number) is the number of occurrences of a given word in the text. The words which don't appear in the dictionary are discarded. As seen in the previous cell, the vector representation of text `I hope they like my gift` using bag-of-words is $[(10,1)]$. Here, the number 10 in the tuple represents the id of the word `gift` as seen in the previous cell. Similarly, the number 1 in the tuple indicates that the word `gift` has occurred once in the text.


### Model

Let's compute the tf-idf of a document and compare it with the its vector representation.

```
from gensim import models
import numpy


documents = ['This is first line',
            'These are second lines',
            'This is third line']

# Token
tokens = [[token for token in docs.split()]for docs in documents]

# Dictionary
dictionary = Dictionary(tokens)
print("Dictionary : \n ", dictionary.token2id)
print("\n")


# Vector
print('Vector of each document: ')
vector = [dictionary.doc2bow(token) for token in tokens]
for vect in vector:
    print(vect)
print("\n")


# BOW of each documents
print("Vector of each document in term of token:")
for document in vector:
    print([[dictionary[id], freq] for id, freq in document])



# tfidf model
tfidf = models.TfidfModel(vector)


# Output of tfidf model
print("\n")
print("tf-idf assigned to each token:")
for document in tfidf[vector]:
    print([[dictionary[id], numpy.around(freq, decimals=3)] for id, freq in document])
```
