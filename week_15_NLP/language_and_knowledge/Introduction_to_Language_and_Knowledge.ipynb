# Introduction

**Language and Knowledge**

Language is a **medium** by which we communicate with others, prove our intellect, and express our emotions. A form of **spoken or written language** abides with a particular set of grammatical rules or structure of words which might differ from one place to another.   

A language is the **communication of human experiences and ideas** that others with a similar linguistic background can understand. The form of a language is not restricted as it can be some hand gestures, a sequence of words to sets of symbols that eventually interpreted an idea. This interpretation of an idea is called knowledge.

### 1. Phonetics and Phonology

Phonetics and Phonology is the study of **knowledge on linguistic sounds.**

More specifically, Phonetics deals with the study of **generation of verbal sounds** corresponding to the word without any forehand knowledge about the language being spoken. We pronounce words as a sequence of symbols representing phones or segments or phonetic symbols.

### 2. Morphology

Morphology is the **study of structure and formation of words** from smaller units called morphemes.

More specifically, morphemes can be divided into two classes: **stems and affixes.**

More specifically, it is responsible for defining the arrangement of words in a sentence acceptable in a language (or arrangement of words that makes sense.)

### 4. Semantics
Semantics deals with **knowing the meaning of words** that make up a phrase or a sentence.

More specifically, semantic analysis is responsible for creating a representation of the meaning of individual words and words in combination to make a phrase or a sentence.

_We will be talking about lexemes, word relations, wordnets, and word similarities in the upcoming chapters_

The following shows the sample of meaning representation for the sentence "He likes Ali" using First-order Predicate Logic (FOPL), which is a form of semantics analysis.


$$∃ \space x \space \space likes(x, Ali) $$   _It reads : For some x, there is a relationship "likes" with Ali._

>_Note: FOPL is out of scope for L1 in NLP, but this example is included to clarify the concept of semantic analysis._




### 5. Pragmatics
Pragmatics deals with the study of **relationship between the meaning of the words or phrases or sentences concerning the speaker's intention.**

More specifically, Pragmatics studies the context in which the sentence is spoken and infers the meaning based on that particular conversational exchange. Without the context or shared knowledge about the exchange situation, the sentence's meaning can become unclear.

_For example:_    
In a conversation between James and Will,

`James: Will, have you seen Avery?`   
`Will: The red car is over there.`

**What does Will mean by pointing the red car over there?**

This particular example would be apparent if we knew that Avery owns a red car else will seem like two incoherent sentences.

_Further more in pragmatics, we will be talking about linguistic and physical contexts with examples in the upcoming chapters._

### 6. Discourse
Discourse is the **knowledge of linguistic units larger than a single utterance.**

More specifically, discourse deals with units that are **larger than a sentence with the concepts of text, textuality, and genre** to understand better and examine spoken or written language in actual communication. Analysis of a group of **sentences for their coherent and conversational relationship** in paragraphs, conversations, or interviews is the primary objective of discourse analysis.

**Pragmatics Vs. Discourse:** Pragmatics deals with a sentence's context and determines its meaning, referencing that particular context where discourse refers to understanding the language in actual communication with larger units like paragraphs.

**Basically, Discourse is analyzed using Pragmatics.**


_Further more in the discourse, we will be talking about the need for discourse analysis and the methods to perform such analysis in the upcoming chapters._

## Basic Process of NLU
NLU stands for Natural language Understanding and is defined as the processing model's capability to understand the sentence/language it is processing.

More specifically, NLU structures an **unstructured input text to the format that makes it comprehensible to the algorithms** and can analyze it. A machine can process textual information only after understanding what the information means and contains. Some of the common examples of NLU are Machine Translations, Automated Reasoning, Question Answering, and so on.

## Key Takeaways
* Phonetics and Phonology deals with knowledge of linguistic sounds.
* Morphology is the study of the formation and structure of words.
* Syntax studies the relationship between those words.
* Semantics studies the meaning of individual words and a collective phrase or sentence.
* Pragmatics deals with the meaning of phrases or sentences in the context of conversational exchange.
* Discourse analyses the coherent and conversational relationship in larger units like paragraphs, interviews, and so on.
* NLU is the process to comprehend the meaning of the input text before processing it.
