## Introduction
Before moving forward with the chapter in hand, let us recap what we have completed so far.

In this previous chapter, we discussed the linguistic terminologies in language, its relationship with knowledge, morphology, and formation of words, and different classes/part-of-speech in which the words can be classified.

Since the words are now formed, we will be discussing the structural arrangement of words, also known as **Syntax**. A combination of words to capture the essence of an idea being expressed using a language must follow a set of grammatical rules. **These grammatical rules justify the combination of words belonging to the language**.

For example, _Subject + Verb + Object_ is the basic structure of a sentence in English grammar that accepts sentences like _He is studying machine learning. The man bought the flowers._

The principle idea behind Syntax is that **words or phrases are combined to form constituents that act as a single unit**. These constituents eventually combine to form sentences. Apart from the constituency, Syntax also includes **grammatical relations, subcategorizations, and dependencies**.   
One of the most common methods to model constituents is Context-Free Grammar (CFG).   
_We will be talking all about context-free-grammar it in this reading material._

## Syntax and Syntactic Analysis
Syntax focuses on combining words or phrases to form constituents, whereas syntactic analysis checks whether the constituents/sentence agrees with the language's grammar. Syntactic analysis is also known as "Parsing" and can be defined as the process of filtering constituents to match with the grammar of the language. Not all the available constituents will be accepted in the grammar and it is upto parsing to determine which constituents are accepted and which are not.

_For Example:_ A grammar defined for English Language will reject words written in any other language.

The fundamental notion of combining words/phrases to model a single abstract unit or constituency acts as a backbone for modeling ***Context-Free Grammars(CFGs)***.

## What is Context-Free Grammars (CFG) or Phrase Structure Grammar (PSG)?
CFG, in layman's terms, is the modeling of constituents to decide if it is incorporated in language or not. **Context-Free** because such grammar doesn't need any context to apply production rules. It doesn't depend on any symbols that may be or may not be around it to apply the production rule. The single non-terminal on the left handside of the production rule can always generate the symbols on the right.

Formally, **A context-free grammar is a recursive set of production rules used to generate patterns of strings.**   
It consists of the following components or "4-tuples", also represented in <sub>[[Figure 1]](#figure1)</sub>

* a set of terminal symbols, $T$
* a set of non-terminal symbols (variables), $V$
* a set of production rules, $P$
* a start symbol, $S$

A language is defined using the **phenomena of derivation**. In our case, derivation of the specific pattern of terminal symbols. Each generation includes the following steps.
1. Each iteration starts at the start symbol, $S$
2. Apply one of the production rules having $S$ on the left-hand side and corresponding terminal or non-terminals on the right
3. Iterate this process of selecting non-terminal strings based on particular production until all the non-terminal symbols have been replaced with terminal symbols.

In the above flowchart, The production rule `Non Terminal => Terminal` states that one terminal symbol can be generated using a non terminal symbol. Similarly, `Non-Terminal => Terminal | Non Terminal` states that a non terminal symbol can generate either a terminal symbol or another non terminal symbol.


In this way, a CFG defies a formal language. Generally, it can be thought of as **a device for generating strings or assigning structure to a sentence**.      
Example of CFG: _contains three sets of production rules with $NP$ as start symbol and determinants (a, an, the) as terminal symbols._

$$\text{NP} \rightarrow \text{Det Nominal} $$
$$\text{NP} \rightarrow \text{ProperNoun}$$
$$\text{Nominal} \rightarrow \text{Noun } | \text{ NominalNoun}$$
$$\text{Det} \rightarrow \text{a |an |the}$$

In this example, the term NP stands for Noun Phrase, representing a phrase with a noun; a phrase can act as a subject or object in a sentence. Similarly, Nominal groups together Nouns or adjectives based on their shared properties. **The intuition of NP and Nomial comes from the English Language**. Still, in respect to CFG, they can be considered only as a symbol of a non-terminal, which derives either some more non-terminals or ends in terminals and avoid diving deeper into the domain of English Language.

Also, the space in between $\text{Det} \space \space \text{Nominal}$ simple represents two separate units or two different branches of a tree as shown in the <sub>[[Figure 3]](#figure3)</sub>. The OR symbol `|` represents two or more symbols that can be derived.

## CFG for English: Examples
Let us take an example of CFG for the English language and generate a sentence using the listed production rules:

1. Noun $\rightarrow$ flight | breeze | trip | morning | ...    
2. Verb $\rightarrow$ is | prefer | like | need | want | fly |...   
3. Ajective $\rightarrow$ cheapest | non-stop | first | latest | other | direct | ..   
4. Pronoun $\rightarrow$ me | I | you | it | ...      
5. ProperNoun $\rightarrow$ Alaska | Baltimore | Los Angeles | Chicago | United | ...   
6. Det $\rightarrow$ the | a | an | this | these | that | ...   
7. Prepostion $\rightarrow$ from | to | on | near | ...   
8. Conjunction $\rightarrow$ and | or | but | ...   

and,

9. S $\rightarrow$ NP $\space \space$ VP      
10. NP $\rightarrow$ Pronoun | ProperNoun | Det $\space$ Nominal      
11. Nominal $\rightarrow$ Noun $\space \space$ Nominal | Noun      
12. VP $\rightarrow$ Verb | Verb $\space \space$ NP | Verb $\space$ NP $\space$ PP      
13. PP $\rightarrow$ Preposition $\space \space$ NP   

Let's generate the sentence "**I prefer a morning flight**" using parse tree for more exact intuition. We will be looking into a parse tree in the upcoming chapter.

**The numbering in the parse tree represents the productioon rule applied to generate the corresponsing symbols.**

## Key Takeaways
- The principle idea behind syntax is that words or phrases are combined to form constituents that act as a single unit.
- Syntax analysis is the process of filtering constituents to match with the grammar of the language.
- A context-free grammar is a recursive set of rules of productions used to generate patterns of strings.
- CFGs can be thought of as a device for generating strings or a device to assign structure to a sentence.
