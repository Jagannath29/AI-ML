# Introduction


**Syntactic parsing** is the task of recognizing a sentence and assigning a syntactic structure (parse tree) to it. This chapter will discuss the structures assigned by a context-free grammar discussed in the previous chapter. However, context-free grammars do not specify how the parse tree for a given sentence would be computed. Therefore, we need to specify algorithms that employ these grammars to efficiently produce correct trees <sub>[[1]](#Book)</sub>.



# Applications of Parse Trees

Parse trees are useful in applications such as:

- **Grammar checking:** In word-processing systems: a sentence that cannot be parsed may have grammatical errors (or at least be hard to read).

- **Semantic analysis:** Parse trees serve as an essential intermediate stage of representation for semantic analysis.

- **Machine translation:** Parse trees can translate a sentence in the source language to a sentence in the desired language.

- **Question answering and Information Extraction:** Parse trees are used to answer questions like "What movies were directed by Cristopher Nolan in the period 2000-2010?" or extract specific information from a collection of information.


# Parsing as Search

Parsing is viewed as searching through the space of all possible parse trees to find the correct parse tree for the sentence. The search space of possible parse trees is defined by a grammar (CFG).

For parsing an input sentence, we need a grammar (CFG). Consider the sentence "Book that flight." We will try to construct the correct parse tree for it using the grammar and lexicons defined in Figure 1.

|S $\rightarrow$ VP| Verb $\rightarrow$ book \| include \| prefer

|Nominal $\rightarrow$ Noun|
|Nominal $\rightarrow$ Nominal Noun|Proper-Noun $\rightarrow$ Houston \| TWA
|NP $\rightarrow$ Proper-Noun|   
|VP $\rightarrow$ Verb||
|VP $\rightarrow$ Verb NP||

A parsing search's primary goal is to find all the trees whose root is the start symbol S and cover precisely the input sentence's words. There are two kinds of constraints that help to guide the search <sub>[[1]](#Book)</sub>.

1. **The input sentence:**

    For example, the parse tree for the input sentence "Book that flight" must have three leaves corresponding to the words "book," "that," and "flight."

2. **The grammar:**

    The final parse tree must have one root, which must be the start symbol S.

**Parsing Strategies**

There are two search strategies based on the two constraints discussed above.

1. Top-down or goal-directed search
2. Bottom-up or data-directed search






## 1. Top-Down Parsing

In top-down parsing, we begin with the start symbol S and work down the parse tree down to the leaves using the grammatical rules defined in CFG. We continue this process of successive derivation of the symbols to generate all possible parse trees until the given sentence can be successfully parsed by any generated parse tree.

The algorithm begins from the start symbol S. Next; we look for all the
grammar rules with S on the left-hand side in the CFG rules defined in Figure 1. There are three such rules, producing three partial trees on the second level of the search space, as shown in Figure 3 below.

Figure 3: Top-down parsing example

The next step is to expand the symbols in the three new trees, similar to expanding of S. At each level of the search space; we use the right-hand sides of the rules to derive new sets of expectations for the parser, which are then used to generate the rest of the trees recursively. We keep expanding the trees downward until they reach the part-of-speech categories at the bottom of the tree. At this stage, we reject the trees whose leaves fail to match all the words in the input. Hence, we are left with the successful parse trees for our input sentence.

In Figure 3, only the fifth parse tree in the third level (the one which has expanded the rule `VP → Verb NP`) will eventually match the input sentence "Book that flight." The algorithm continues in the same way until the tree leaves match the input sentence exactly. Note that Figure 1 does not show the complete algorithm to produce the final thriving parse tree <sub>[[1]](#Book)</sub>.

## Bottom-Up Parsing

In bottom-up parsing, we begin with the input sentence's words and try to build trees from the words up to the start symbol S by applying the grammatical rules defined in the CFG. With the start symbol S as root, the resulting parse tree is considered the thriving parse tree for the input sentence. Figure 4 shows the bottom-up search space for the sentence "Book that flight."

The algorithm begins by searching for each word of the input sentence in the lexicon defined in Figure 1 and producing three partial trees with the part-of-speech for each word. Note that the word "book" is ambiguous since it can be a verb or a noun. Thereupon, we must consider two possible sets of trees. We can see this initial division of the search space in the first two levels of Figure 4.

At each level, all the trees are expanded by looking for the right-hand side of the rules matching the symbols and replacing them with the left-hand side, just opposite to what we did in the top-down parser.

In the fifth level, we remove the tree, interpreting the book as a noun, from the search space as there is no rule in the grammar with the right-hand side `Nominal NP`. Continuing in this way, in the final level, we will be able to find a correct parse tree in the search space <sub>[[1]](#Book)</sub>.

## Comparisons between Top-Down and Bottom-Up Parsing

|Top-Down Parsing|Bottom-Up Parsing|
|--|--|
|It does not waste time to explore trees that cannot result in S|It may try to generate trees that have no hope of leading to S|
|It spends considerable effort on S trees that are not consistent with the input|It does not suggest trees that are not at least locally grounded in the input.|

For example, the left branch of Figure 4 in the bottom-up parsing example is a wasted effort. Similarly, in the top-down parsing example in Figure 3, the first four of the six trees in the third level have left branches that cannot match the word book. None of these trees can be used in parsing the sentence correctly.

# Ambiguity in Parsing

It is one of the most challenging problems in parsing algorithms. The ambiguity arising in parsing from many commonly used rules in CFG is called structural ambiguity. It occurs when the parser assigns more than one thriving parse tree to a sentence. A well-known example of structural ambiguity is a line by Groucho Marx as Captain Spaulding in Animal Crackers, which is "I shot an elephant in my pajamas."

The sentence is ambiguous because the phrase "in my pajamas" can be part of the noun phrase `NP` headed by elephant or a part of the verb phrase `VP` headed by shot. Figure 5 illustrates these two analyses of Marx’s line <sub>[[1]](#Book)</sub>.

Two common kinds of ambiguity are:

1. Attachment ambiguity
2. Coordination ambiguity

## 1. Attachment ambiguity

A sentence has an attachment ambiguity if a particular constituent can be attached to the parse tree at more than one place. The previous example of Marx's line is an example of PP-attachment ambiguity. It is also common in various kinds of adverbial phrases. Another similar example is the sentence, "We saw the Eiffel Tower flying to Paris." This sentence is anexample of the attachment ambiguity because the phrase `flying to Paris` can have `the Eiffel Tower` or `saw` as its subject.

## 2. Coordination ambiguity

Coordination ambiguity arises from different sets of phrases that can be conjoined by a conjunction like 'and.' For example, the phrase `old men and women` can be perceived as `[old [men and women]],` meaning old men and old women, or as `[old men] and [women],` meaning that old refers to the men alone.

# Parsing methods for removing ambiguity

Algorithms that effectively remove ambiguity, i.e., disambiguation algorithms, require statistical, semantic, and contextual information of the input.

Dynamic programming parsing methods and Statistical parsing methods can solve the problems associated with ambiguous grammars. We will only introduce these methods but will not go into the details as they are beyond this chapter's scope.


## Dynamic Parsing Methods

The dynamic programming approach has an advantage due to the context-free nature of the grammar rules, i.e., we can record the presence of a constituent discovered in a segment of the input and use it later in subsequent derivations
which require it. As a result, we can improve the parse's time and storage efficiencies by avoiding reanalysis of the parse tree.

Although the dynamic parsing methods represent the ambiguities efficiently, they cannot fully resolve them.


### 1. Cocke-Kasami-Younger (CKY) Algorithm

It is based on the bottom-up parsing method and requires first normalizing the grammar to Chomsky Normal Form (CNF). It uses a table of partial parses to parse ambiguous sentences efficiently.






### 2. Earley Algorithm

It is based on top-down parsing and does not require normalizing grammar. However, it is a complex algorithm in comparison to the CKY algorithm. It consists of three stages: Predictor, Scannor and Completor.

## Statistical Parsing Methods

In contrast to the dynamic parsing methods, a probabilistic parser solves the ambiguity problem completely. It computes the probability of each parse tree and chooses the most probable parse tree.

**Probabilistic context-free grammar (PCFG)** is the most commonly used grammar for the computation of the probabilities. PCFG is an extension of context-free grammars in which each rule is associated with a probability. Hence, using independent assumptions, each complete parse tree's probability can be computed and compared at the end.

# Key Takeaways

1. Parsing is an essential component of NLP, used in applications like grammar checking, semantic analysis, machine translation, question answering, information extraction, etc.

2. Parsing can be described as a search problem guided by a CFG.

3. Top-down parsing begins with the root S and moves down to the input words, whereas bottom-up parsing starts with the input words and moves up toward the root S.

4. A sentence is structurally ambiguous if the grammar assigns more than one thriving parse tree to it.

5. Dynamic parsing methods and statistical parsing methods are used for disambiguation.
