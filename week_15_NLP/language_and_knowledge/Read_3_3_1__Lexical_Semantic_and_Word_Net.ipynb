### 2. Lexical Semantics

Lexical Semantics is defined as the study of the systematic meaning of individual words along with their sub-words, affixes, stems, and the connections between each word.

A **lexeme** is defined as a pairing of a particular form (orthographic or phonological) with its meaning, whereas a **lexicon** is a list of lexemes. A lexeme can have multiple inflections to form a set of several words, also known as inflected variants of that lexeme.    
_For an example: eat $\rightarrow$ eating, eats, eaten_

Similarly, a **lemma** is simply a dictionary/thesauruses version of a lexeme. Like in a dictionary, we have a generic word and its inflected forms along with its corresponding examples, a lemma is like that generic word.

A lemma or citation form is the grammatical form of a lexeme. It can be understood as the base form of any word, e.g., the drive is the base form of driving. These forms of a base word are also known as wordforms, which gives us the definition of lemmatization. The process of deriving a lemma from its inflected word forms is known as **lemmatization**.

>_Students are encouraged to explore the subtle difference between lemmatization and stemming here_ <sub>[[3.1.]](#Tutorials)</sub>

There are several ways to represent meaning or semantics; in this case, lexical semantics using:

- Word Senses
- Word Relations
- Word Similarity

We will only be talking about the first two in this reading material. Word Similarity will be revisited in the upcoming chapter.

### 2.1. Word Senses:
A lemma can have different contextual meanings. Such meanings are known as senses or word senses. So, a **sense** is a discrete representation of aspects of a word's meaning. The different senses of the word might not share a mutual relationship and are used in different contexts.
_For example, the word **bank** represents two senses like_
- _**bank**: a financial instuition_
- _**bank**: side of a river_

_These two senses are said to be homonyms (word relation)._

###  2.2. Word Relations
Two or more words can share some form of relationship between them, like semantic, orthographic or phonological. With that let us discuss some of the most common word relations:
#### 2.2.1. Homonymy
Words with different meanings but are spelled or pronounced the same way, and their relationship is known as Homonumy.     
If the words are identical in spelling, they are called **homographs**, and if the words are pronounced the same, they are called **homophones**.

#### 2.2.2. Polysemy
Words that can be used to express different meanings based on the context are called Polysemous words.

_for an example:_
- Calculate the **solution** in your head
- Stir the **solution** slowly until solutes dissolve.   
Here the word **solution** expresses different meanings based on the context of use.

_Polysemous words have co-existence of different but intended meanings for a word unlike homonyms (homographs)_

>Students are encouraged to explore the difference between homographs and polysemy.

#### 2.2.3. Metonymy
Metonymy is defined as using one aspect of a concept to refer to another aspect of that concept or the concept itself. _For an Example:_ White house, show a metonymy relation and describes the administrative office inside the White house. This relation, metonymy, is a distinct subtype of polysemy. Similarly, in the quote **"The pen is mightier than the sword, "**, pen signifies a metonymy relation with written words, whereas the sword signifies war and aggression.

#### 2.2.4. Synonymy
When two senses of two different words(lemmas) have the same meanings, then the relation is Synonymy. Synonymous words can be substituted for each other in all situations with acceptability based on notions of politeness, slang, register, genre, etc.: **Notion of substitutability**

_Examples_
- Couch/Sofa
- Water / H$_2$O
- Vomit/ Throwup

Synonymy is more of a relation between senses than words, and there are very few examples of perfect Synonymy.

#### 2.2.5. Antonymy
Antonymy, in contrast to Synonymy, includes senses that are opposite in meaning to each other. These senses will represent the opposite end of some scale.

_Examples:_
- long/short
- wide/narrow
- up/down
- cold/hot

#### 2.2.6. Hyponymy
A sense is said to be hyponymy of another if the first sense is more specific, denoting it belongs to a sub-class of another.    
_For Example_ A dog is hyponymy of an animal. A car is hyponymy of a vehicle, and a banana is hyponymy of a fruit. It is also known as subordinate.  
**Hypo is Sub**


>Students are encouraged to explore the transitive, extensional, and entailment property of hyponymy.


#### 2.2.7. Hypernymy
In contrast to hyponymy, senses denoted by Hypernymy extensionally includes the senses belonging to hyponymy. It is also known as superordinate instead of Hypernymy.   
**Hyper is Super**

_For Example:_
An animal is a hypernymy of a dog. A vehicle is a hypernymy of a car and so on.

Hyponymy and Hypernymy also show **IS-A Hierarchy**. Hyponymy IS-A Hypernymy like A car IS-A vehicle.

#### 2.2.8. Meronymy
Also known as the part-whole relation, Meronymy relation is exhibited when one sense is a part of another. _For Example:_ A wheel is a meronym of a car, or a finger is a meronym of a hand. On the contrary, we also define A car as a **holonym** of the wheel.   
**Part of Relation**

### 3. WordNets
A WordNet is defined as a hierarchically organized lexical database containing a set of lemmas, where each one of them is referenced with a set of senses.   
A WordNet consists of three separate databases: nouns, verbs, and the third for adjectives and adverbs. A standard WordNet can be easily accessed using the internet and downloaded to a local machine.

#### 3.1. Synsets
The synonym set is the set of near-synonyms senses, concepts, or instances with a standard gloss (gloss is another term for definition).

_For example:_ Consider a synset :   
{chump, fool, gull, mark, patsy, fall guy, sucker, mug}   
_The gloss of each sense in this set is defined as a person who is easy to take advantage of._   
Each of the senses can be used to express this concept.

#### 3.3. Structure of WordNet
The primary relation among the words in a WordNet is the Synonymy, which is grouped into unordered synsets and corresponding gloss. These synsets are connected with a small number of "conceptual relations." There are about 117 000 synsets, each linked to synsets.

## Key Takeaways
* Study of the systematic meaning of individual words is called Lexical Semantics, and terminologies associated with it are lexemes, lexicon(collection of lexemes), a lemma (representation of lexemes).
* Semantics represented using senses, relations, and similarity
* Homonymy, Polysemy, Metonymy, Synonymy, Antonymy, Meronymy are some word relations we talked about
* WordNet is a hierarchically organized lexical database containing a set of lemmas referenced with senses.
* Collection of Synonyms is called synsets.
