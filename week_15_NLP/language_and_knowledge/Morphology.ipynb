# Introduction

Morphology is the study of the way words are built up from smaller meaning-bearing units called morphemes <sub>[[1]](#Book)</sub>.


# Morphemes

The minimal meaning bearing units in language are called morphemes.

## Classes of Morphemes

Morphemes can be divided into two classes : **stems and affixes.**

### Stems

The central morpheme of the word, supplying the main meaning are called stems.

### Affixes

Affixes add “additional” meanings of various kinds to the stems and modify their meanings and grammatical functions.

### Forms of affixes

There are four forms of affixes <sub>[[2]](#Slide)</sub>. They are:

- **Prefixes:** They precede the stem. Example: un-, dis-, im-, a-.

- **Suffixes:** They follow the stem. Example: -ly, -ous, -ing, -fy.

- **Infixes:** They are inserted inside the stem. They are not common in the English language—examples: -s- in cupsful, spoonsful, and passersby.

- **Circumfixes:** They both precede and follow the stem. They are not common in the English language. Example: en- -en in enlighten, em- -en in embolden.

**Hence, on the basis of the use of affixes to develop a word we have the following types of morphology:**

**Concatenative morphology**

The usage of prefixes and suffixes concatenated to the stem creates a concatenative morphology. It is common in the English language.

**Non-Concatenative morphology**

When morphemes are combined in more complex ways, we call it a non-concatenative morphology. It is more common in other languages like Arabic and Hebrew.

# Types of Morphology

Based on the class of the word generated, generally there are two types of morphology:

## 1. Inflectional Morphology

The combination of a word stem with a grammatical morpheme, usually resulting in a word of the same class as the original stem, and usually filling some syntactic function like agreement.

## 2. Derivational Morphology

The combination of a word stem with a grammatical morpheme, usually resulting in a word of a different class, often with a meaning hard to predict exactly.

## 1. Inflectional Morphology

In English, only nouns, verbs, and sometimes adjectives can be inflected.

Let's see examples of inflections based on these word classes.

### i. Nominal inflection (Nouns)

This is a common type of inflection in English language which involves nouns.

### ii. Verbal Inflection

Comparatively, verbal inflection is more complicated since there are three classes of verbs in English which are <sub>[[2]](#Slide)</sub> :

- Main verbs: drink, stand, run
- Modal verbs: should, shall, will, can
- Primary verbs: have, be, do

## 2. Derivational Morphology

Let's see some examples of derivational morphology.



### i. Nominalization

The formation of new nouns from verbs or adjectives is called nominalization.

|Suffix|Base Verb / Adjective|Derived Noun|
|--|--|--|
|-ation|computerize (V)|computerization|
|-ee|appoint (V)|appointee|
|-er|kill (V)|killer|
|-ness|fuzzy (A)|fuzziness|



### ii. Formation of adjectives from Nouns / Verbs

|Suffix|Base Noun / Verb|Derived Adjective|
|--|--|--|
|-al|computation (N)|computational|
|-able|embrace (V)|embraceable|
|-less|clue (V)|clueless|

Derivation in English is more complex than inflection because

- Generally less productive
    - A nominalizing affix like –ation can not be added to every verb. Example: eat + ation = eatation (**X**)

- There are subtle and complex meaning differences among nominalizing suffixes. For example, sincerity has a subtle difference in meaning from sincereness.


# Morphological parsing

In general, parsing means taking an input and producing a structure for it. Morphological parsing takes words as input and produces a structure that reveals its morphological features <sub>[[2]](#Slide)</sub>.

Finite State Automata (FSA) is used for morphological parsing, but it is beyond the scope of this course. So, we will skip this topic.


# Key takeaways

1. Morphology is the study of the way words are built up from smaller meaning-bearing units called morphemes

2. The minimal meaning bearing units in language are called morphemes.

3. English mainly uses **prefixes** and **suffixes** to express inflectional and derivational morphology.

4. Inflectional morphology is simple and does not involve a word-class change, whereas derivational morphology is complex and involves a word-class change.

5. Morphological parsing is the process of producing the constituent morphemes in an input word.
