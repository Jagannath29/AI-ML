## Pragmatics

Pragmatics is the study of the meaning of words and sentences based on the context of conversational exchange.
**Context, Intentions, and Shared Knowledge** define Pragmatics.   
In the absence of the context of the conversation, the actual meaning might be "invisible."

In comparison to semantics, a **pragmatics attempts to relate the meaning as a fundamental notion of senses-relations and similarity** within the language whereas pragmatics relates the **meaning to the context of utterance.**

**"Pragmatics is the study of language that focuses on the users' attention and the context of language use rather than on reference, truth or grammar."**  <sub>[[2.2.]](#University)</sub>

_For and Example: (In a conversation)_   
A: I am married   
B: I guess that's fine   
A: I also have a cat   
B: Oh, I am so sorry   

What do you understand from this conversation?

It would be challenging to infer one meaning to the conversation unless you knew that A is trying to rent a house from B. This contextual information binds the exchange with perfect meaning.

Hence, Pragmatics deals with studying meanings of words and sentences in the context they are used.

### Context
We can differentiate the context of utterances into four categories, namely:    **Physical Context, Epistemic Context, Linguistic Context, Social Context.**

_We will only be looking into Linguistic and Physical context as the remaining two are out of scope for this reading material._

#### 1. Physical Context and Examples
Physical context is defined by the location of the given the word, **objects surrounding the conversational exchange**, time and place, and even the situations taking place around the conversation.

_For an Example:_    

SLOW DOWN. PUPILS CROSSING THE STREET   
These indicate vehicles to slow down as students cross the street not to slow down because you might drive over eye pupil.
_ I want that toy. _Accompanied by POINTING_

\

#### 2. Linguistic Context (Co-Text)
The linguistic context (co-text) is the **set of other words used in the same sentence**. More specifically, it references other words or senses said before in the conversation.

_For an Example:_    
Linda stole the glass vase yesterday. She thought no one is going to notice.   
To understand the reference for "she though no one will notice," we need the previous sentence's cotext.

## Discourse
Any language consists of a structured, collocated, and coherent group of sentences that contribute to a general meaning/understanding. This is where discourse comes into play.

Discourse is studying units of languages larger than phrases and utterances and contributes to a general understanding/meaning.   
It includes speech, writings, texts, genre, and the concepts of pragmatics also.    
**Study of Conversation**

The types of discourse are Monologues and Dialogue. Dialogue can be further classified into Human-Human interaction and Human-Computer interaction. An example of human-computer interaction is brilliantly portrayed in the movie "Her."   
Spoken discourse includes conversational exchange, speech, monologues, dialogues.

Discourse Analysis is:
- the study of how sentences in spoken and written language form larger meaningful units such as paragraphs, conversations,
interviews, etc.
- how the choices of articles, pronouns, and tenses affect the structures of the discourse

To understand discourse, one must also understand two critical phenomena of discourse:
- **Cohesion**:    
It refers to how the components of the surface text (words we see) are connected within a sequence of sentences.
- **Coherence**:   
It refers to the deeper structures (not surface sentential structures) in texts involving semantic and pragmatic analysis.

### Why do we need discourse analysis?

Discourse analysis includes examining how a language is communicated and how meaning is interpreted from different social contexts to a common understanding.

We conduct discourse analysis mainly to:
- Understand the purpose and effects of different types of linguistic structures of language to provide a foundation for several NLP applications
- Extract pieces of information in the form of reference resolution, summarization, and textual coherence.   
_ Example  of reference resolution:_   
   **Mahabir Pun** is a Nepalese teacher, social entrepreneur, and activist. **He** is known for his extensive work in applying wireless technologies to remote areas of the Himalayas._

_He is referenced to Mahabir Pun in the above ExampleExample._

- To address cultural rules, idioms, and conventions in any language
- Include the significance of how values, beliefs, and understanding are communicated
- Relate language based on its political, historical, and socio-cultural aspects

###  Performing Discourse Analysis
#### 1. Scope
The scope of discourse analysis fundamentally focuses on understanding the coherent and cohesive structures of the words and the contextual information. Apart from that, discourse focuses on semantics & pragmatics (meaning), text linguistics (interpretations), genre analysis (communication), and stylistics.

#### 2. Influences
Influences of Discourse goes a long way. It has always been a useful and reliable tool to critique several disciplines, including:
- Phonology and Morphology
- Politics
- Literature and Journalism
- Business, Economics, and Finance
- Nursing and Medicine
- Marketing and Advertisement
- Psychology
- Religion

Discourse analysis influences the in-depth evaluation of socio-linguistic factors to critique language and its understanding.


#### 3. Approaches
The approaches to discourse analysis can be summarised as follows:

- The Speech Act Approach
- Interactional Sociolinguistics
- The Ethnography of communication
- Pragmatics
- Conversational Analysis
- Variational Approach

## Pragmatics Vs. Discourse

Pragmatics focuses on studying language from the point of view of users, especially of their choices, obstructions they encounter while using language as a form of social interaction, and effects of the use of language in participants.

Whereas, Discourse fundamental focuses on the understanding and examination of spoken or written language in a conversational exchange. It includes a broader range of analysis like text, textuality, and genre.

**Pragmatics is a primary tool to analyze discourse.**

## Key Takeaways
- Pragmatics is the study of the meaning of words and sentences based on the context of conversational exchange
- Context can be divided into four categories: Physical Context, Epistemic Context, Linguistic Context, Social Context.
- Discourse is the study of units of languages larger than phrases and utterances and contributes to a general understanding/meaning.
- Performing discourse analysis includes: Scope, Approaches, Influences, and Devices
