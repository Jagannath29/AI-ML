## Types of Word Classes
Based on how frequently new words are added, we can divide the types of word classes into: Closed and Open.


### 1. Closed Class:   
These set of classes have **relatively fixed membership to words**, meaning there wont be much addition or deletion of words from these classes.

Closed class includes **Prepositions, Determiners, pronouns, Conjunctions, Auxillary Verbs, Particles and Numerals.** A new preposition or a new determiners are rarely added or the existing one are rarely removed. Closed class words are frequently used and used to structure in grammar.
Let's discuss each of the categories in brief:

* **Prepositions**: Prepositions occur before noun and is often used to express the relation (temporal or spatial) of noun with another words in the phrase.   
 _For example: The vessel is **on** the top of the table_ shows the positional relationship of the vessel with the table.   
Examples of prepositions are: on, under, with, near, by ,at, from ,to

* **Determiners**: Determiners occurs in the begining of a noun phrase (i.e. preceding the noun) which mainly focuses on illustrating whether it is something specific or something general.    
Articles (_a, an, the_) are a subtype of determiners. Determiners also include this, that, Many, My, ours, your, its.   
_For example: **The** cat ate the milk (specific)_   
_Many dogs were barking last night (general)_

>_Like articles, determiners are of 3 more types: demonstratives, possessives, and quantifiers. The study of these types are out of this scope but the students are encouraged to explore themselves here_ <sub>[[1.1]](#book)</sub>

* **Conjunctions**: This word class (conjoins) connects two phrases, clauses or sentences. Conjunctions are futher divided into coordinating conjunctions (like and, but, or, yet, for, nor) or Subordinating conjunctions (like after, although, as if, because, before) or Correlative conjunctions (like both...and, not only ... but, also).   
_For example: Will submitted his resume to the company **and** waited by the phone for the response._

* **Pronouns**: This class of words acts as a reference to some noun phrase or entity. Basically, it takes palce of the repeating noun in a subordnate sentence (followup sentence). Pronouns can be sub categorised into 7 caegories but we will only be talking about Personal pronouns (person or entities), Possessive pronouns (indicates actual possession) and Wh-Pronouns(what, who, whom, whoever).    
_For Example: John is studying computer science. **He** also works in a company._
_Other categories are out of scope for this reading material_

* **Auxillary Verbs**: This subclass of words marks the semantic feature of the main verb. Together with main verb it shows whether the action takes place in the present, past or the future (verb's tense), aspect of the main verb (if it is completed or not), mood (whether the action verb is necessary, suggested or desired) and finally polarity (whether it is neglected or not).       
_For example: The bank **was** robbed yesterday._





 ### 2. Open Class:   
Unlike closed class, open class categories have a large number of words and new ones are easily invented and added. Nouns, verbs, adjectives and adverbs are the examples of open class categories. More specifically open class nouns can be proper nouns and common nouns.

* **Nouns**: identifies persons or entities (Types: Common nouns and Proper Nouns)(eg: shoe, basketball, Paris, Book)

* **Verbs**: Refers to the actions or processes generally taken by nouns (Types: Action verb, Stative verb, Transitive Verb, modal Verb and so on) (eg: explain, describe, illustrate, understand, am, likes, eats )

* **Adjectives**: Refers to the properties or qualities of nouns, generally. they modify or describe other words (Types based on degree of comparison: Absolute, Comparative and Superlative) (eg: old, young, strong, cute)

* **Adverbs**: This class of words modify adjectives or verbs or another adverbs to provide more information about how it is done. An adverb generally ends up with -ly like carefully, slowly, terribly. (Types: Adverbs of manner, Adverbs of place, Adverbs of Frequency and so on) (eg: daily, monthly, normally, anywhere, badly, correctly)

### POS /Word Classes/Morphological Class/Lexical Tags

Like we discussed earlier, Part-of-speech tagging involves assigning specific tags/class to each words in the sentences. A POS tagging algorithm is fed with tokenized words in sequence and the ouputs the sequence of best tags corresponding the words. Along with the words tagging is also done on punctuation marks like comma and period.

The process of tagging is not trivial, meaning there is not a definite tag for anyword as the degree of ambuiguity for each word in English language is very high (40% based on brown corpus.)   
_For example_:   
* _It used to be different **back/RB** then._
* _Carol was determined to buy **back/RP** her shares._
* _Nice leather on the **back/JJ** seat._

Hence, The vital things here to understand is that words are ambigious in nature and the process of assigning tags to them is a disambiguation task.

The objective of POS tagging of sequence of words is to select the best tag among many. Now how do we fullfill this objective? How do we assign a particular tag from a  set of tags to a word?

The most trivial and baseline concept will be to assign the tag that is most frequent with that word in that corpus. This approach is also known as **Most Frequent Class Baseline**. But the obvious problem with this is that with difference in corpus the baseline tags also differs. Lets dive deeper down into the types of POS tagging algorithms available.





## Types of POS Taggers with examples
_Discussion on all the types of POS taggers is out of this scope yet we will be talking about the most general ones._

POS tagging algorithms can be divided into the following categories.



<!-- <img src="https://drive.google.com/uc?id=1gmsSky5iwLxeSO3M1Lq3AuV-4P_fMzXo" >  -->
<img src="https://i.postimg.cc/d1M4JKfz/image.png" > Figure 5: Approaches to POS Tagging

1. **Rule based POS Taggers**:   

As the name suggests rule based POS taggers removes the disamiguations of tags for a word based on a **large set of handwritten disambiguation rules** (can be some contextual rules or even regular expressions in the forms of finite state automata). These rules specify which tag to prioritize in which condition.

_For an example (one of the rule can be):      
 the tag for an ambigious word must be a  noun rather than a verb if it follows a determiner._      
This is actually the second stage in rule based POS tagging, initially the tagging uses the standard dictionay to assign each word with its respective part-of-speech.
Hence, The tagging is based on manually built rules with a limit of number of rules to be generally 1000.



2. **Stochastic POS Taggers**:   

Unlike rule based, stochastic POS taggers resolve ambuguities in tags for a word based upon the **probability of a given tag in the given context for that word**. This model incorporates frequency of the word with a particular tag and probability of given sequence of tags occuring.    
More specifically, in frequency of words approach, we calculate the **probability of the given word occurring with a tag**. We select the tag that occurs most frequently with that word in the given corpus of study. With the frequency of tags as the main parameter to evaluate, the **sequence of tags might not always be admissible.**   

Similarly, in tag sequence probability calculation, we calculate the probability of occurance of a sequence of tags trained on the given corpus. This approach is also known as **n-gram approach**. n determined the tags referenced to determine n+1 tag.    
Hidden Markov Models (HMM), Maximum Entrophy Markov Model (MEMM), Conditional Random Field (CRF) are the types of stochastic POS tagging.  

>_The study of these classifications are beyond the scope of this reading material but students are highly encouraged to explore themselves here_<sub>[[1.2]](#Book)</sub>



3. **Transformation based POS Taggers or Brill Tagging**:

This approach **combines both previously discussed methods to resolve tagging ambugiuty**. Like rule based tagging, this approach is also made up of rules that assigns tags to a given word but like stochastic tagging, these rules are not handwritten rather induced by machine learning from the given training corpus, hence the name transformation and learning.    
The learning is an **iterative process of selecting the most beneficial transformation to the word in the given training courpus** and is stopped when there is not change in the tag for that word. At this point of time the machine is believed to have learned the best transformation rule to assign a tag give a word. The learning of rules makes it very time consuming in larger corpora and also results are not probabilities yet just rules.

On the brighter side, the ouput set of rules are enough to assign the appropriate tags and this model is comparatively faster than HMM.

## Implementation of POS Tagging on a given sentence
We will be using nltk library to tokenize the words from sentences and assign POS tags to them

```python
# Importing NLTK
import nltk

nltk.download("punkt")
nltk.download("averaged_perceptron_tagger")
nltk.download("tagsets")
```

```python
given_text = "She sells seashells on the seashore."

# Tokenization
tokens = nltk.word_tokenize(given_text)
print(tokens)
```

```python
# Assigning tags to each tokenized words
tags = nltk.pos_tag(tokens)
print(tags)
```

```python
# Documentation for each tags can be queried in the following way
nltk.help.upenn_tagset("VBZ")
print("\n")
nltk.help.upenn_tagset(".")
```

**Note:** nltk.pos_tag is NLTK's currently recommended POS tagger which uses PerceptronTagger <sub>[[3.1.]](#Documentation)</sub>

## Key Takeaways
- Part-of-speech tagging (POS tagging) (also called POST) is the process of assigning a part of speech or word class to each word available in the corpus.
- Words can be divided into two classes: Closed and Open, based upon how frequently new words are added.
- Rule based taggers use handwritten rules to determine context and assign tags.
- Stochastic Taggers calculate the probability of a tag to assign to the given word based upon the training corpus.
- Transformation based learning makes the use of both rules based approach and stochastic approach as well. Learns transformation rules from the corpus rather than hand written rules.
