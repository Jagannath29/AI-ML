## Introduction
Convolution is the measure of the effect of one signal on the other signal. It is a mathematical operation between two signals in which one signal operates on another to produce the third signal as output. The output signal indicates how one is altered or modified by another.

Apart from image processing, the term convolution is also widely used in the signal, system, and signal processing. The signal passes through some system, and the resultant output is a response or output signal. Let’s understand the signal and system with the help of the analogy of communication.


Suppose you want to make a call with your friend over the telephone. When you make a call, your voice signal must pass through several operations like amplification, modulation, analog to digital conversion, digital to analog conversion, attenuation, etc., before it reaches your friend. In other words, the input voice signal has to pass through the different systems or devices like an amplifier, noise remover,  low pass filter, ADC, etc., performing different operations.  Each of these devices receives an input signal, performs some operation on it, and produces a response or output. How these devices act on the input signal is defined by their impulse response. The impulse response of a system is a reaction of a system when presented with a short impulse. The impulse response is expressed as a function of time.
For instance, the impulse response of some arbitrary device is shown in the figure. In computer vision, the impulse response is famous by the name filter or kernel and is generally represented by F or h.

When some arbitrary input signal, $I(x)$ is passed through a system with the given impulse response, the output/response of a system, $O(x)$ obtained, is shown in the above figure. This response, in other words, can be expressed as a convolution of given two signals.
The convolution operation is represented by $*$(star) symbol. The output signal $O(x)$ is given as a convolution between input signal $I(x)$ and filter $F(x)$ and can be expressed as:
$$O(x)=I(x)*F(x)$$

## Convolution
Until now, we have discussed convolution from the perspective of a signal and system and discussed a few of its applications. Now let's discuss convolution operation mathematically and see and how it is related to the correlation.

The convolution operation is very similar to that of correlation operation except that we flip either the input signal or a filter just before correlating. It means in convolution, we first flip the filter and then slide the flipped form of filter over the input signal, followed by multiplication and addition of the overlapped elements. The convolution operation between any two signal is defined as:

$$O(x)=F(x)*I(x)=\sum_{i=-N}^NF(i)I(x\color{red}{-}i)$$

Where, $I(x)$ is an input signal, $F(x)$ is a filter, $N$ is an integer that determines the size of a filter. Notice that the expression of convolution differs from correlation only in the "$-$" sign. The -ve sign here represents the flip operation.

The figure below shows how the the convolution operation is related with the correlation operation. The convolution between input signal $I(x)$ and filter $F(x)$ is given as correlation between $F(-x)$ and $I(x)$ or correlation between $F(x)$ and $I(-x)$ as shown in the figure below.

## 2D convolution

In 1D convolution, both filter and input signals are of one dimension. However, in 2D convolution, both filters and input are of two-dimension; other operations are the same as for 1D convolution. Image is a perfect example of a two-dimensional signal. The convolution in 2D is also very similar to the correlation, except that we flip either the filter or input signal both horizontally and vertically before correlating.

The convolution between 2D input signal $x(i,j)$ and filter $h(m,n)$ is mathematically given as:
$$y(i,j)= h(m,n)*x(i,j)=\sum_n\sum_mh(m,n)x(m\color{red}{-}i, n\color{red}{-}j)$$

Notice that the expression for 2D convolution is very similar to that of correlation. The only difference is in "$-$" sign indicating the flip of a filter.

The figure below shows how the convolution in 2D can be expressed in terms of correlation. In the left-hand side equation, the convolution is expressed as a correlation of flipped form of filter and input signal. While in the right-hand side equation, the convolution is expressed as a correlation of a flipped form of the input signal and the filter.

## Convolution vs Correlation

By this point, you might be already familiar with the major difference between convolution and correlation.

Convolution allows us to measure the effect of filter/kernel on the input signal. We can cause some desired effect on the input signal, for example, edge detection, image sharpening, etc., with the choice of an appropriate filter.

In the case of correlation, you are expecting similarity. Therefore, if you have the image with a specific pattern and want to check whether that pattern is present in this image or not, you will apply correlation. In doing so, you will know where your desired patterns or similar patterns are present.

Unlike correlation, the kernel/filter is reflected by 180 degrees(in other words, flipped) in convolution.

```
# Load the image
!curl "https://upload.wikimedia.org/wikipedia/commons/3/39/Lichtenstein_img_processing_test.png" -o "test.png"
```

```
import cv2
import matplotlib.pyplot as plt
import numpy as np
from google.colab.patches import cv2_imshow

# Read and show the image
img = cv2.imread("test.png")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(7, 7))
plt.xticks([]), plt.yticks([])
plt.imshow(img)
```

Can you tell how many edges are present in the above figure? Well, there are many many edges. To detect the edges in the image, we can convolve it with the special filter as shown.
\begin{align}
\text{filter} =
 \begin{bmatrix}
-1 & -1 & -1\\
-1 &  8 & -1\\
-1 & -1 & -1\\
\end{bmatrix}
\end{align}

Notice that the sum of all the elements in the filter is equal to zero $($i.e., $-1-1-1-1-1-1-1-1+8 =0)$ which is a necessary condition for the edge detection filter.

Different filters' choice results in other effects like image smoothing, image blurring, and image sharpening. We will discuss how to choose the different filters for different applications in the upcoming chapter. For now, let's take it for granted that the above filter detects the edges.

We want to convolve the image with the given kernel/filter.


```
# Kernel/filter for edge detection
kernel = np.array([[-1, -1, -1],
                   [-1, 8, -1],
                   [-1, -1, -1]])

# Perform convolution using filter2D method
dst = cv2.filter2D(img,-1, kernel)

# Plot the original image
fig = plt.figure(figsize=(15, 15))
plt.subplot(121)
plt.imshow(img)
plt.title('Original Image')
plt.xticks([])
plt.yticks([])

# Plot the detected edges.
plt.subplot(122)
plt.imshow(dst)
plt.title('Edge Detected')
plt.xticks([])
plt.yticks([])
plt.show()

```

## Key Takeaways
- Convolution mathematically defines how the input signal gets altered or modified when passed through a system. In other words, convolution measure the effect of one signal on another signal.

- Convolution is like correlation, except we flip either the filter or an input signal just before correlating.

- The convolution has a wide range of applications ranging from detecting the edges, removing noise from the image, sharpening the image, and so on.

- The choice of filter plays a very crucial role in determining the nature of the output signal.
