In contrast to the point operator, there is another type of operator called neighborhood operator in which the output value of a pixel depends on a collection of input pixel values. In this chapter, we will discuss about the neighborhood operator, its types, and applications.


## Neighbourhood operator
The neighborhood operator uses a collection of pixel values around the given pixel to determine its final output value.  The neighborhood operators can be used for a wide range of applications like image smoothing, blurring, sharpening, noise removable, etc. The neighborhood operator is slightly different from the point operator -- the output pixel value depends only on the corresponding input pixel.

The above figure shows the point operation and neighborhood operation and their corresponding mathematical expression for a two-dimensional data/signal. Both of these operations can be used for one-dimensional signals, e.g., audio. Notice that in point operation,  the output pixel value at location $(x, y)$ depends only on the input's corresponding pixel value.

In contrast, in neighbourhood operator, the output pixel value at location $(x, y)$ depends not only on the corresponding pixel vale of input but also on the neighbourhood pixel values of input. Here, the pixel value at location $(x, y) $of output are dependent on input pixel value at nine $(3X3)$ different locations $(x-1, y-1), (x, y-1), (x+1, y-1), (x-1, y), (x, y), (x+1, y), (x-1, y+1), (x, y+1)$, and $(x+1, y+1)$.

The input-output relationship in the neighborhood operator can be mathematically expressed as:
$$O(x, y) = \sum_{i=-N}^N\sum_{j=-N}^Na_{ij}I(x+i, y+i) + b_{ij}$$

Here, $I$ is an input signal, $a_{ij}$ is the weighting factor applied to each pixel. In computer vision, $a_{ij}$ is called filter and common notation is $F(i,j)$. Similarly, $b_{ij}$ is a bias applied to each pixel. $N$ determines the number of neighborhoods considered. In other words, $N$ determines the size of a filter. In the above example, the value of $N$ is one giving a filter of size $3\times3$.

There are two most common types of neighborhood operators. They are:
1. Correlation
2. Convolution


In this chapter, we will discuss correlation operation, and in the next chapter, we will discuss convolution.


## Correlation
Correlation is a measure of the similarity between two different signals. Such signals can be one-dimensional, e.g., audio, or two-dimensional, e.g., image.  In Computer Vision, the correlation is used to locate, or match the specific region or pattern in the image.  Such an application is commonly known as template matching. If we have an image and a mask(template), and want to know where the mask has the highest similarity, we can use correlation.

Correlation operation is a straightforward operation that can be summarized in three words: __Sliding, Multiplication, and Addition__.  In correlation, we have a filter $F(x)$ which we scan/slide through the input signal, $I(x)$. During sliding, the overlapped part of the filter and an image are multiplied, followed by addition.

### Mathematical expression of 1D correlation

Until now, we have discussed how the correlation works for a one-dimensional dataset with the help of an example. Now let's see how we can mathematically represent the one-dimensional correlation.


The correlation for a one-dimensional signal is given by the sum of the product of filter, $F(i)$ and the input signal, I(x), as:

$$O(x)= F\otimes I(x)=\sum_{i=-N}^NF(i) I(x+i)$$

Where, the symbol $\otimes$ represent the correlation operation. $N$ is an integer that determines the size of a filter. For the above example, the value of N is 1, and the size of filter is given as $2N+1$. For $N=1$, the size of filter is $2.1+1=3$. Similarly, for $2$, the size of the filter is 5. This shows that the size of the filter is always odd.

In the above expression,  the multiplication and addition operations are apparent. You might be wondering how the sliding operation is incorporated in the above equation. The answer is that the sliding operation is represented by $x$. We perform the above multiplication and addition for every value of x. So different value of $x$ is modeling the slide operation.

There are generally two types/variants of correlation: __auto-correlation__ and __cross-correlation__. Auto-correlation is the correlation of a signal with itself. It means both input signal and filter are the same. However, in cross-correlation, the input signal and filter are different signals. The example that we discussed above is a cross-correlation.

Radar is a detection system that uses radio waves to detect the distance, location, and velocity of objects like an airplane, ships, spacecraft, missiles, etc.  A radar system consists of a transmitter antenna that radiates a radio wave of some prespecified amplitude and frequency. Similarly, the radar system consists of a receiver antenna that receives the signal reflected from the object, as shown in the figure.

When a target object like an airplane is flying in the radar zone, the transmitter's signal will reflect back to the receiver. But the receiver may also receive the signal transmitted by other sources like BTS tower, cell towers, etc. So how to make sure that the received signal is the one actually transmitted by the transmitter? Well, to ensure that we compute the correlation between the transmitted and reflected signal. A high value of correlation indicates the presence of a target object. ## 2D Correlation

Until now, we discussed correlation in one-dimensional space with the help of an example. The concept of correlation can be extended to two-dimensional space, as seen in the pattern matching example. Let's discuss how the correlation operation works for two-dimensional space with the help of an example.

In 2D correlation, both input signal and filter are of two-dimension. Image is an example of two-dimensional signal. Suppose we have following $(3\times3)$ input signal and $(2\times2)$ filter (generally the filter has odd size, but for simplicity we choose a small filter of size $(2\times2)$) and we want to find the correlation between them.

Initially, we place the filter at the top-left position of an input signal and perform element-wise multiplication of the overlapped part followed by addition. In the above animation, we start with the empty table representing the output of the correlation operation. We fill the cells in the table step by step by placing the filter over the input signal. The overlapped elements are multiplied and then added. We repeat these steps until we reach the end of an input signal.


Notice that the shape of output signal is reduced while performing the correlation operation. The width ($W$) and height($H$) of the output signal is given by the following expression:
$$W_{\text{output}} = W_{\text{input}}-W_{\text{filter}}+1$$
$$H_{\text{output}} = H_{\text{input}}-H_{\text{filter}}+1$$

The above relation shows that the larger the filter, the smaller the size of the output signal. Sometimes we want to maintain the output signal size the same as that of an input signal. In such a case, we can pad the input signal by some appropriate value, say 0.


### Mathematical expression of 2D correlation

Until now, we have discussed how the correlation works for a two-dimensional signal with the help of an example. Let's now understand the mathematical expression of 2D correlation.

The mathematical expression of 2D correlation is very similar to that for one-dimension. The only difference is that both the input signal and filter are of two-dimension.


The correlation for two-dimensional signal is given by the sum of product of filter, $h(m,n)$ and the input signal, $x(i,j)$ as:

$$
    y(i, j) = h(m, n)\otimes x(i,j)= \sum_n \sum_m h(m,n) \cdot x(i+m, j+n)
$$

Where, the symbol $\otimes$ represent the correlation operation.

## Implementation

Until now, we discussed how the correlation works for both one-dimensional and two-dimensional signals with the help of an example. In this section, we will implement a correlation and use it to match a specific pattern in the given image.

```
# Load the image
!curl "https://upload.wikimedia.org/wikipedia/commons/3/39/Lichtenstein_img_processing_test.png" -o "test.png"
```

```
import cv2
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
from google.colab.patches import cv2_imshow

# Original image
img = cv2.imread("test.png")
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img = img - img.mean()

# Template pattern is extracted from the original image
pattern = img[80:180, 190:210]
pattern = pattern - pattern.mean()

# Height and width of a pattern
h, w = pattern.shape[0:2]

# Plot the template and original image
fig = plt.figure(figsize=(10, 5))
ax1 = fig.add_subplot(121)
ax1.imshow(img, cmap='gray',)
ax1.set_title("Original Image")
ax2 = fig.add_subplot(122)
ax2.imshow(pattern, cmap='gray')
ax2.set_title("Pattern")
fig.show()
```

```
from scipy import signal


# Compute 2D correlation between original image and template pattern.
corr = signal.correlate2d(img, pattern, boundary='symm', mode='valid')
```

```
# Location with highest correlation value
y, x = np.unravel_index(np.argmax(corr), corr.shape)
print(f'y, x = ({y}, {x})')

# Highlight the matched pattern
fig = plt.figure(figsize=(10, 5))
ax1 = fig.add_subplot(121)
ax1.imshow(img, cmap='gray',)
ax1.set_title("Pattern Matched")

# Add rectangular bounding box at the matched location with highest correlation value.
rect = patches.Rectangle((x,y), w, h, linewidth=2, edgecolor='r', facecolor='none')
ax1.add_patch(rect)

ax2 = fig.add_subplot(122)
ax2.imshow(pattern, cmap='gray')
ax2.set_title("Pattern")
fig.show()
```

The output of the above cell shows that the maximum correlation(match) between the original input image and the template pattern occurs at the location (80, 190). The above plot shows the rectangular bounding box obtained at the matched location.
