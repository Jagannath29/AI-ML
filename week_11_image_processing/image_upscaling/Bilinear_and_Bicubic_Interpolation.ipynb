#Bilinear Interpolation

Bilinear interpolation extends linear interpolation in 2D. In linear interpolation, we sampled the new point between any two closest points. In the case of bilinear interpolation, we estimate the missing value by approximating from four nearest points to the given point. When we do this, our interpolation becomes a little bit complex.

If $Q_{11}, Q_{12}, Q_{21},$ and $Q_{22}$ are four different pixels that are close to pixel at $I(x,y)$, we can interpolate $I(x,y)$ using these pixels. A general idea can be shown in the figure below:

Using this interpolated values, we again interpolate at $I$ and thus obtain bilinearly interpolated value. Here, "Bi" stands for 2 and what we just performed is a 2 times interpolation. Thus the name.

Consider previously the distance between the pixels was 1. Now, after expansion by 3 times, the distance between the pixels is 1/3. And, for updating the value of pixel P we will have to first interpolate to find the values of $X_1$ and $X_2$. And, on the basis of these values we interpolate again to find the value of P.

We can find the value of $X_1$ as:

$$X_1 = 10 * \frac{2}{3} + 20 * \frac{1}{3}$$
$$X_1 = \frac{40}{3}$$

Similarly, we obtain the value of $X_2$ as:
$$X_2 = 30 * \frac{2}{3} + 40 * \frac{1}{3}$$
$$X_2 = \frac{100}{3}$$


Now, using these values of $X_1$ and $X_2$, we update the value of P through linear interpolation as:

$$P = \frac{40}{3} * \frac{2}{3} + \frac{100}{3} * \frac{1}{3}$$
$$P = 20$$

#Bicubic Interpolation

Bicubic interpolation extends the idea of cubic interpolation from 1D to 2D. Unlike bilinear interpolation that samples from the closest 4 points, bicubic interpolation samples from the nearest 16 locations. In the case of our image, if we want to get the value of the pixel at $I(x,y)$ we will have to consider 16 other pixels surrounding it. The math is a bit complex and is out of the scope.

Also, a new curve is obtained from the points at W, X, Y and Z in a similar way the point P is interpolated.

So, this is how a bicubic interpolation occur.


#Implementation using OpenCV

```
!curl -o demo.jpg https://upload.wikimedia.org/wikipedia/commons/thumb/f/f0/ARSENE_WENGER.jpg/195px-ARSENE_WENGER.jpg
```

```
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```
img = cv2.imread("demo.jpg", 0)
fig, ax = plt.subplots(nrows=1, ncols=4, figsize=(15, 7), dpi=80, sharex=True, sharey=True)
ax[0].imshow(img, cmap="gray")
ax[0].set_title("Original")
height, width = img.shape
resized = cv2.resize(img,(2*width, 2*height), interpolation=cv2.INTER_NEAREST)
ax[1].imshow(resized, cmap="gray")
ax[1].set_title("Nearest neighbor")


resized2 = cv2.resize(img,(2*width, 2*height), interpolation=cv2.INTER_LINEAR)
ax[2].imshow(resized2, cmap="gray")
ax[2].set_title("Bilinear")
fig.show()

resized3 = cv2.resize(img,(2*width, 2*height), interpolation=cv2.INTER_CUBIC)
ax[3].imshow(resized3, cmap="gray")
ax[3].set_title("Bicubic")
fig.show()
```

```
fig, ax = plt.subplots(nrows=1, ncols=4, figsize=(15, 7), sharex=True, sharey=True)
ax[0].imshow(img[150:175, 100:125], cmap="gray")
ax[0].set_title("Original")
ax[1].imshow(resized[300:350, 200:250], cmap="gray")
ax[1].set_title("Nearest Neighbor")
ax[2].imshow(resized2[300:350, 200:250], cmap="gray")
ax[2].set_title("Bilinear")
ax[3].imshow(resized3[300:350, 200:250], cmap="gray")
ax[3].set_title("BiCubic")
fig.show()
```

In a 2D image, we only apply the nearest neighbour, bilinear or the bicubic implementation to interpolate on intermediate values.

Bicubic interpolation unlike the bilinear interpolation which samples from nearest 4 points, samples from 16 closest points. Because of this fact, the result image obtained after bicubic interpolation are smoother. We can see this in the above plot too.

The nearest neighbour pads the image with the same pixels so the image interpolated with these techniques consists large number of sharp edge pixels.

The images interpolated using the Linear interpolations are smoother but they also consist of small sharp edge pixels due to difference in pixel intensity during interpolation.

Moreover, the bicubic interpolation smoothens the curve and makes the pixel intensity smoother.



#Key Takeways

1. Bilinear interpolation extends the idea of linear interpolation to 2D signals, whereas bicubic extends the idea of cubic interpolation.

2. Bicubic smoothens the image but introduces haloing effect.

3. Bicubic is computationally expensive than bilinear interpolation.
